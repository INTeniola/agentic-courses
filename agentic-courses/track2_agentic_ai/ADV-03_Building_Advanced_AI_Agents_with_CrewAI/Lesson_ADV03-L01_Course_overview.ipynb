### Setting Up Your Environment

If you're using Google Colab, many dependencies are pre-installed, but we'll still run an `!pip install` command to ensure we have the latest versions and specific CrewAI components. For local development, create a virtual environment and install the packages there.

Let's get started by installing the necessary libraries and setting up our API keys.


```python
# Install necessary libraries
# We'll use langchain_openai as a common LLM provider for demonstration.
# Feel free to substitute with your preferred provider (e.g., langchain_anthropic, langchain_google_genai).
!pip install -q crewai crewai-tools langchain_openai

# Import os for environment variables
import os

# --- Set your API Key --- 
# Replace 'YOUR_OPENAI_API_KEY' with your actual OpenAI API key.
# If using another provider, adjust the environment variable accordingly.
# For example, for Anthropic: os.environ["ANTHROPIC_API_KEY"] = "sk-ant-..."
# For Google Gemini: os.environ["GOOGLE_API_KEY"] = "AIzaSy..."

os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY"

# Optional: Set your preferred LLM model. Default is 'gpt-4o' or 'gpt-4-turbo' for CrewAI.
# os.environ["OPENAI_MODEL_NAME"] = "gpt-4o"

# Verify installation and basic setup with a simple CrewAI component
from crewai import Agent, Task, Crew, Process
from langchain_openai import ChatOpenAI

# Initialize a dummy LLM to check if the API key is recognized
# This won't make an actual API call yet, but checks the setup.
try:
    llm = ChatOpenAI(model="gpt-4o", temperature=0.7)
    print("\n✅ LLM initialized successfully. Your environment is ready!")
    print(f"Using model: {llm.model_name}")
except Exception as e:
    print(f"\n❌ Error initializing LLM: {e}")
    print("Please ensure your API key is correctly set and has access to the specified model.")

# A very basic 'hello world' agent and task to confirm CrewAI imports
# This won't run a full crew, just demonstrates the objects can be instantiated.
try:
    dummy_agent = Agent(role='Test Agent', goal='Verify setup', backstory='A simple agent for testing.', llm=llm, verbose=False)
    dummy_task = Task(description='Perform a simple check.', agent=dummy_agent)
    print("✅ CrewAI Agent and Task objects instantiated successfully.")
except Exception as e:
    print(f"❌ Error instantiating CrewAI objects: {e}")
    print("This might indicate an issue with the CrewAI installation or a deeper dependency problem.")
```
