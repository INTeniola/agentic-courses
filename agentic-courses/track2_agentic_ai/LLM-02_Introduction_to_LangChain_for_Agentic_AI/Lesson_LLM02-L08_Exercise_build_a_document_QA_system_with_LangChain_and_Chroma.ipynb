## Exercise: Build a Document QA System with LangChain and Chroma

```python
# Standard library imports
import os
import uuid

# LangChain imports
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_community.llms import Ollama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser

# --- Configuration and Setup ---

# Set up Ollama base URL if it's not the default
# Ensure your Ollama server is running and you have models pulled (e.g., 'llama3', 'nomic-embed-text')
# Example: ollama run llama3, ollama run nomic-embed-text
os.environ["OLLAMA_BASE_URL"] = os.getenv("OLLAMA_BASE_URL", "http://localhost:11434")

# Define the directory for Chroma persistence
CHROMA_DB_DIR = "./chroma_db_llm02_l08"

# --- Mock Data ---
# These are the documents your QA system will be built upon.
raw_documents_text = [
    "Agentic AI systems are designed to autonomously perform complex tasks by breaking them down into smaller, manageable sub-tasks. They leverage large language models (LLMs) as their reasoning engine, often integrating with various tools to interact with the environment. This approach allows for more dynamic and adaptable solutions compared to traditional rule-based systems.",
    "LangChain is a powerful framework for developing applications powered by large language models. It simplifies the process of chaining together different components, such as LLMs, prompt templates, retrievers, and tools. LangChain 0.3 introduced significant improvements in modularity and performance, making it easier to build complex agentic workflows and retrieval-augmented generation (RAG) systems.",
    "Retrieval-Augmented Generation (RAG) is a technique that enhances the capabilities of LLMs by providing them with external, relevant information during the generation process. Instead of relying solely on the LLM's pre-trained knowledge, a RAG system first retrieves pertinent documents from a knowledge base (like a vector store) and then uses these documents as context to generate more accurate and up-to-date answers. This mitigates issues like hallucination and outdated information.",
    "Chroma is an open-source embedding database that makes it easy to build LLM applications. It allows you to store embeddings and their associated metadata, and then query them efficiently for similarity search. Chroma integrates seamlessly with LangChain, serving as a robust vector store for RAG applications. It supports both in-memory and persistent storage options, making it versatile for development and production environments.",
    "The core components of a LangChain RAG system typically include a Document Loader to ingest data, a Text Splitter to break documents into manageable chunks, an Embedding Model to convert text into numerical vectors, a Vector Store (like Chroma) to store and search embeddings, and an LLM to generate answers based on retrieved context. LCEL (LangChain Expression Language) is used to compose these components into a coherent and efficient pipeline.",
    "In 2026, agentic AI is becoming mainstream, with frameworks like LangChain providing the foundational tools for developers. The emphasis is on creating reliable, scalable, and observable AI applications. The integration of advanced retrieval techniques and sophisticated prompt engineering are key to achieving high-performance agentic systems."
]

# Convert raw text into LangChain Document objects
# In a real-world scenario, you'd use DocumentLoaders (e.g., WebBaseLoader, PyPDFLoader)
# Here, we simulate by creating Document objects directly.
documents = [Document(page_content=text) for text in raw_documents_text]

print(f"Loaded {len(documents)} raw documents.")

# --- Initialize Components (Partial) ---

# Initialize the embedding model
# Ensure 'nomic-embed-text' or a similar embedding model is pulled in Ollama
embeddings = OllamaEmbeddings(model="nomic-embed-text")

# Initialize the LLM for generation
# Ensure 'llama3' or another suitable LLM is pulled in Ollama
llm = Ollama(model="llama3", temperature=0.0)

print("OllamaEmbeddings and Ollama LLM initialized.")
print(f"Chroma DB will be stored in: {CHROMA_DB_DIR}")

# --- Helper function to clean up Chroma DB directory ---
def cleanup_chroma_db(db_path):
    if os.path.exists(db_path):
        import shutil
        shutil.rmtree(db_path)
        print(f"Cleaned up Chroma DB directory: {db_path}")

# Optional: Clean up previous run's Chroma DB for a fresh start
# cleanup_chroma_db(CHROMA_DB_DIR)
```

Now it's your turn to complete the exercise. Using the provided setup code and the requirements above, implement the full RAG pipeline. Your solution should:

1.  **Split the `documents`** into chunks.
2.  **Create a `Chroma` vector store** from these chunks and the `embeddings` model, ensuring persistence.
3.  **Define a retriever** from your vector store.

```python
# --- Reference Solution ---

# 1. Text Splitting
# Initialize the RecursiveCharacterTextSplitter
# We choose a chunk size and overlap to ensure context is maintained but chunks are not too large.
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, # Max characters per chunk
    chunk_overlap=200, # Overlap between chunks to maintain context
    add_start_index=True # Add metadata about the start index of the chunk
)

# Split the loaded documents into chunks
splitted_documents = text_splitter.split_documents(documents)
print(f"Split {len(documents)} documents into {len(splitted_documents)} chunks.")

# 2. Create and Populate Chroma Vector Store
# Check if the Chroma DB directory already exists to avoid re-embedding if not needed
if not os.path.exists(CHROMA_DB_DIR) or not Chroma(persist_directory=CHROMA_DB_DIR, embedding_function=embeddings)._collection.count():
    print(f"Creating new Chroma DB at {CHROMA_DB_DIR}...")
    # Create a new Chroma vector store from the split documents and embeddings
    # This will embed the documents and store them, persisting to disk.
    vectorstore = Chroma.from_documents(
        documents=splitted_documents,
        embedding=embeddings,
        persist_directory=CHROMA_DB_DIR
    )
    print("Chroma DB created and populated.")
else:
    print(f"Loading existing Chroma DB from {CHROMA_DB_DIR}...")
    # Load the existing Chroma vector store
    vectorstore = Chroma(persist_directory=CHROMA_DB_DIR, embedding_function=embeddings)
    print("Chroma DB loaded.")

# 3. Define a Retriever
# Convert the vector store into a retriever. This allows us to fetch relevant documents.
# We can specify the search type (e.g., 'similarity') and the number of results (k).
retriever = vectorstore.as_retriever(search_type="similarity", search_kwargs={"k": 3})
print(f"Retriever initialized to fetch top {retriever.search_kwargs['k']} similar documents.")

# 4. Construct a ChatPromptTemplate for RAG
# This prompt guides the LLM on how to use the retrieved context to answer the question.
# It's crucial for effective RAG.
rag_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are an AI assistant for Agentic AI and LangChain. Use the following context to answer the question. If you don't know the answer, just say that you don't know, don't try to make up an answer. Keep your answer concise and to the point.\n\nContext: {context}"),
    ("human", "{question}")
])
print("RAG ChatPromptTemplate created.")

# 5. Build the LCEL RAG Chain
# This is the core of our RAG system, composed using LangChain Expression Language (LCEL).
# The chain defines the flow: retrieve context -> format prompt -> generate answer.
rag_chain = (
    {"context": retriever, "question": RunnablePassthrough()} # Retrieve context based on the question
    | rag_prompt_template # Format the prompt with context and question
    | llm # Pass the formatted prompt to the LLM
    | StrOutputParser() # Parse the LLM's output into a string
)
print("LCEL RAG chain constructed.")

# 6. Test the Chain
print("\n--- Testing the QA System ---")

questions = [
    "What are agentic AI systems designed to do?",
    "What is LangChain and what improvements did version 0.3 bring?",
    "How does Retrieval-Augmented Generation (RAG) mitigate issues like hallucination?",
    "What are the core components of a LangChain RAG system?",
    "What is Chroma used for in LLM applications?"
]

for i, question in enumerate(questions):
    print(f"\nQuestion {i+1}: {question}")
    # Invoke the RAG chain with the question
    answer = rag_chain.invoke(question)
    print(f"Answer {i+1}: {answer}")

# --- Optional: Clean up Chroma DB directory after testing ---
# Uncomment the line below if you want to remove the persistent Chroma DB after running.
# cleanup_chroma_db(CHROMA_DB_DIR)
```
