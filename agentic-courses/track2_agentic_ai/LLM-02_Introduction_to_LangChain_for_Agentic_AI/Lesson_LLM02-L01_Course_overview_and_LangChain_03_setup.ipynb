### 3. Obtain and Configure API Keys

To interact with LLMs, you'll need an API key from a provider like OpenAI. If you don't have one, sign up at [platform.openai.com](https://platform.openai.com/) and generate a new secret key.

**Security Best Practice:** Never hardcode API keys directly into your code. We'll use `python-dotenv` to load them from an environment file.

#### For Local Development:

1.  Create a file named `.env` in the root directory of your project.
2.  Add your OpenAI API key to this file in the following format:
    ```
    OPENAI_API_KEY="your_openai_api_key_here"
    ```
    Replace `"your_openai_api_key_here"` with your actual key.

#### For Google Colab:

1.  Click on the 'Secrets' icon (key icon) in the left sidebar.
2.  Click '+ Add new secret'.
3.  For 'Name', enter `OPENAI_API_KEY`.
4.  For 'Value', paste your OpenAI API key.
5.  Ensure 'Notebook access' is toggled on.

Now, let's load these environment variables in our Python code.


````python
# Load environment variables from .env file (for local development)
from dotenv import load_dotenv
import os

load_dotenv()

# Verify that the API key is loaded
openai_api_key = os.getenv("OPENAI_API_KEY")

if openai_api_key:
    print("OpenAI API key loaded successfully!")
else:
    print("Error: OpenAI API key not found. Please check your .env file or Colab secrets.")
    print("Ensure you have set OPENAI_API_KEY.")

# Import necessary LangChain components
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# --- Hello, LangChain! --- 
# This simple example demonstrates a basic LLM call using LangChain 0.3

try:
    # 1. Initialize the Chat Model
    # We'll use gpt-4o for its advanced capabilities, but you can use gpt-3.5-turbo if preferred.
    llm = ChatOpenAI(model="gpt-4o", temperature=0.7, api_key=openai_api_key)

    # 2. Define a Prompt Template
    # This allows us to structure our input to the LLM.
    prompt = ChatPromptTemplate.from_messages([
        ("system", "You are a helpful AI assistant that introduces itself."),
        ("user", "Introduce yourself and mention LangChain 0.3.")
    ])

    # 3. Create a Simple Chain
    # LangChain Expression Language (LCEL) allows chaining components easily.
    chain = prompt | llm | StrOutputParser()

    # 4. Invoke the Chain
    # Send the prompt to the LLM and get the response.
    response = chain.invoke({})

    print("\n--- LangChain 0.3 Setup Verification ---")
    print("Response from LLM:")
    print(response)
    print("\nCongratulations! Your LangChain 0.3 environment is set up and working correctly.")

except Exception as e:
    print(f"\nAn error occurred during the LangChain setup verification: {e}")
    print("Please ensure your API key is correct and your internet connection is stable.")
    print("Also, double-check the installed library versions.")
````
