````python
# Capstone Project: Production-Ready RAG System - Detailed Solution

import os
import uuid
import logging
from typing import List, Dict, Any
from pydantic import BaseModel, Field

# --- Configuration --- 
# Set your API keys here or as environment variables
# Ensure you have OPENAI_API_KEY set in your environment or uncomment and set it here.
# os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY"

# Configure basic logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

# --- Mock Data (Replace with actual files in a real scenario) ---
document_texts = [
    "The AgenticSupport Pro plan includes unlimited chat support, 24/7 phone assistance, and a dedicated account manager. It costs $99/month.",
    "Our Basic plan offers chat support during business hours (9 AM - 5 PM EST) and email support. It costs $29/month.",
    "To reset your password, visit our login page and click 'Forgot Password'. Follow the instructions sent to your registered email address.",
    "AgenticSupport Solutions was founded in 2023 with the mission to revolutionize customer service through AI-powered tools.",
    "We support integrations with Salesforce, Zendesk, and HubSpot. Custom integrations are available for enterprise clients.",
    "Our Enterprise plan offers custom pricing, dedicated engineering support, and on-premise deployment options. Contact sales for a quote."
]

# --- Evaluation Dataset (Pydantic Model for clarity) ---
class EvaluationItem(BaseModel):
    query: str
    ground_truth: str
    contexts: List[str] = Field(default_factory=list) # Expected relevant contexts

# In a real scenario, this would be a carefully curated, versioned dataset.
eval_dataset_items = [
    EvaluationItem(query="What are the features of the Pro plan?", 
                   ground_truth="The Pro plan includes unlimited chat support, 24/7 phone assistance, and a dedicated account manager.", 
                   contexts=[document_texts[0]]),
    EvaluationItem(query="How much does the Basic plan cost?", 
                   ground_truth="The Basic plan costs $29/month and offers chat support during business hours and email support.", 
                   contexts=[document_texts[1]]),
    EvaluationItem(query="How do I reset my password?", 
                   ground_truth="You can reset your password by visiting the login page, clicking 'Forgot Password', and following the instructions sent to your registered email.", 
                   contexts=[document_texts[2]]),
    EvaluationItem(query="What integrations does AgenticSupport offer?", 
                   ground_truth="AgenticSupport integrates with Salesforce, Zendesk, and HubSpot. Custom integrations are available for enterprise clients.", 
                   contexts=[document_texts[4]]),
    EvaluationItem(query="When was AgenticSupport Solutions founded?", 
                   ground_truth="AgenticSupport Solutions was founded in 2023.", 
                   contexts=[document_texts[3]])
]

# --- 1. Data Ingestion & RAG Core ---
from langchain_community.document_loaders import TextLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate
from langchain.chains import create_retrieval_chain
from langchain.chains.combine_documents import create_stuff_documents_chain

def load_and_chunk_documents(texts: List[str]) -> List[Document]:
    """Loads and chunks documents for the RAG system."""
    logger.info("Loading and chunking documents...")
    # In a real scenario, you'd load from files, databases, etc.
    docs = [Document(page_content=text) for text in texts]
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
    chunked_docs = text_splitter.split_documents(docs)
    logger.info(f"Loaded {len(docs)} documents and split into {len(chunked_docs)} chunks.")
    return chunked_docs

def create_vector_store(documents: List[Document]) -> Chroma:
    """Creates and populates a vector store with document embeddings."""
    logger.info("Creating vector store...")
    # Using OpenAIEmbeddings and ChromaDB for simplicity and local persistence
    embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
    vectorstore = Chroma.from_documents(documents, embeddings, persist_directory="./chroma_db")
    vectorstore.persist()
    logger.info("Vector store created and persisted to ./chroma_db.")
    return vectorstore

def create_rag_chain(vectorstore: Chroma) -> Any:
    """Creates the RAG chain using LangChain."""
    logger.info("Creating RAG chain...")
    llm = ChatOpenAI(model="gpt-4o", temperature=0.1, streaming=False) # Use gpt-4o for better quality
    
    system_prompt = (
        "You are an AI customer support assistant for AgenticSupport Solutions. "
        "Use the following retrieved context to answer the user's question accurately and concisely. "
        "If the context does not contain enough information to answer the question, state that you don't have enough information. "
        "Do not make up answers.\n\n{context}"
    )
    prompt = ChatPromptTemplate.from_messages([
        ("system", system_prompt),
        ("human", "{input}"),
    ])

    question_answer_chain = create_stuff_documents_chain(llm, prompt)
    retriever = vectorstore.as_retriever(search_kwargs={"k": 3}) # Retrieve top 3 relevant documents
    rag_chain = create_retrieval_chain(retriever, question_answer_chain)
    logger.info("RAG chain created.")
    return rag_chain

# --- 2. Automated Evaluation ---
from ragas import evaluate
from ragas.metrics import faithfulness, answer_relevance, groundedness
from datasets import Dataset

def run_automated_evaluation(rag_chain: Any, dataset_items: List[EvaluationItem]) -> Dict[str, Any]:
    """Evaluates the RAG system using Ragas framework."""
    logger.info("Running automated evaluation with Ragas...")
    
    # Prepare Ragas dataset format
    ragas_data = {
        "question": [],
        "answer": [],
        "contexts": [],
        "ground_truth": []
    }

    for item in dataset_items:
        logger.debug(f"Evaluating query: {item.query}")
        response = rag_chain.invoke({"input": item.query})
        generated_answer = response["answer"]
        retrieved_contexts = [doc.page_content for doc in response.get("context", [])]
        
        ragas_data["question"].append(item.query)
        ragas_data["answer"].append(generated_answer)
        ragas_data["contexts"].append(retrieved_contexts)
        ragas_data["ground_truth"].append(item.ground_truth)

    ragas_dataset = Dataset.from_dict(ragas_data)
    
    # Evaluate with Ragas
    # Using a dedicated LLM for evaluation (can be different from the RAG LLM)
    eval_llm = ChatOpenAI(model="gpt-4o", temperature=0.0)
    eval_embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

    score_results = evaluate(
        ragas_dataset,
        metrics=[faithfulness, answer_relevance, groundedness],
        llm=eval_llm,
        embeddings=eval_embeddings
    )
    
    avg_faithfulness = score_results["faithfulness"]
    avg_relevance = score_results["answer_relevance"]
    avg_groundedness = score_results["groundedness"]

    logger.info(f"Ragas Evaluation Results:")
    logger.info(f"  Average Faithfulness: {avg_faithfulness:.4f}")
    logger.info(f"  Average Answer Relevance: {avg_relevance:.4f}")
    logger.info(f"  Average Groundedness: {avg_groundedness:.4f}")

    # Define a simple pass/fail threshold
    PASS_THRESHOLD = 0.75 # Average score across metrics
    overall_score = (avg_faithfulness + avg_relevance + avg_groundedness) / 3
    test_status = "PASS" if overall_score >= PASS_THRESHOLD else "FAIL"

    logger.info(f"Overall Evaluation Score: {overall_score:.4f} (Threshold: {PASS_THRESHOLD})")
    logger.info(f"Automated Evaluation Status: {test_status}")

    # In a real CI/CD, these results would be logged to MLflow, Weights & Biases, etc.
    # For simplicity, we'll just return them.
    return {
        "overall_status": test_status,
        "overall_score": overall_score,
        "metrics": {
            "faithfulness": avg_faithfulness,
            "answer_relevance": avg_relevance,
            "groundedness": avg_groundedness
        },
        "raw_results": score_results.to_pandas().to_dict(orient='records') # Convert to dict for JSON compatibility
    }

# --- 5. Observability (Mock Implementation) ---
def mock_send_to_observability_platform(data: Dict[str, Any], data_type: str):
    """Mocks sending data to a centralized observability platform."""
    logger.info(f"[OBSERVABILITY_MOCK] Sending {data_type} data: {data}")
    # In a real system, this would use client libraries for Datadog, Honeycomb, Grafana, etc.

def start_trace(operation_name: str) -> str:
    """Mocks starting an OpenTelemetry trace and returns a trace ID."""
    trace_id = str(uuid.uuid4())
    logger.info(f"[TRACE_MOCK] Starting trace '{operation_name}' with ID: {trace_id}")
    # In a real system, this would use opentelemetry.trace.get_tracer().start_as_current_span()
    return trace_id

def end_trace(trace_id: str, status: str = "success", attributes: Dict[str, Any] = None):
    """Mocks ending an OpenTelemetry trace."""
    logger.info(f"[TRACE_MOCK] Ending trace {trace_id} with status: {status}. Attributes: {attributes}")
    # In a real system, this would end the span and export it.

# --- Main Execution --- 
if __name__ == "__main__":
    logger.info("Starting Capstone Project Execution...")

    # 1. Data Ingestion & RAG Core
    documents = load_and_chunk_documents(document_texts)
    vectorstore = create_vector_store(documents)
    rag_chain = create_rag_chain(vectorstore)

    # --- Demonstrate a single RAG interaction with Observability ---
    test_query = "What are the benefits of the Enterprise plan?"
    trace_id = start_trace("user_query_processing")
    try:
        logger.info(f"Processing user query: '{test_query}'")
        rag_response = rag_chain.invoke({"input": test_query})
        generated_answer = rag_response["answer"]
        retrieved_contexts = [doc.page_content for doc in rag_response.get("context", [])]
        
        # Log the interaction
        log_data = {
            "query": test_query,
            "retrieved_docs": [doc[:100] + '...' for doc in retrieved_contexts],
            "response": generated_answer,
            "trace_id": trace_id,
            "llm_model": "gpt-4o",
            "timestamp": os.getenv("CURRENT_TIMESTAMP", "2026-01-01T12:00:00Z") # Mock timestamp
        }
        mock_send_to_observability_platform(log_data, "rag_interaction_log")

        print(f"\nUser Query: {test_query}")
        print(f"RAG Response: {generated_answer}")
        print(f"Retrieved Contexts (first 100 chars):\n  - " + "\n  - ".join([c[:100] + '...' for c in retrieved_contexts]))
        end_trace(trace_id, status="success", attributes={"query_length": len(test_query), "response_length": len(generated_answer)})
    except Exception as e:
        logger.error(f"Error during RAG invocation for query '{test_query}': {e}")
        end_trace(trace_id, status="error", attributes={"error_message": str(e)})

    print("\n" + "-"*50 + "\n")

    # --- 2. Automated Evaluation ---
    evaluation_results = run_automated_evaluation(rag_chain, eval_dataset_items)
    mock_send_to_observability_platform(evaluation_results, "evaluation_results")

    print("\nEvaluation Summary:")
    print(f"Overall Evaluation Status: {evaluation_results['overall_status']}")
    print(f"Overall Average Score: {evaluation_results['overall_score']:.4f}")
    print(f"Individual Metrics: {evaluation_results['metrics']}")

    # --- 3. CI/CD Integration (Conceptual Explanation) ---
    print("\n" + "-"*50 + "\n")
    print("### CI/CD Integration Strategy (Conceptual)")
    print("\n**Triggering the Pipeline:**")
    print("A robust LLMOps CI/CD pipeline for this RAG system would be triggered by:")
    print("1.  **Code Changes**: Any push to the RAG system's codebase (e.g., changes to `create_rag_chain`, `evaluate_rag_system` functions). Git hooks or webhook integrations with platforms like GitHub Actions, GitLab CI, or Jenkins would initiate the pipeline.")
    print("2.  **Data Updates**: Changes to the underlying product documentation (e.g., new markdown files, updates to existing ones) in a data version control system (e.g., DVC, LakeFS). This would trigger re-embedding and vector store updates.")
    print("3.  **Prompt Changes**: Modifications to the system prompt or retrieval instructions. Prompt versioning (e.g., storing prompts in a Git repository or a dedicated prompt management tool like PromptLayer, LangChain Hub) ensures traceability and allows for A/B testing.")
    print("4.  **Model Updates**: New versions of the base LLM or embedding model (e.g., `gpt-4o` to `gpt-4o-mini`). These are typically managed via model registries (e.g., MLflow Model Registry, Hugging Face Hub, custom registries).")

    print("\n**Pipeline Stages:**")
    print("1.  **Build & Test**: Linting, unit tests for code logic, dependency checks. Containerization (Docker) of the RAG application.")
    print("2.  **Data Preparation & Versioning**: If data changed, fetch new data, re-process documents, chunk, embed, and update the vector store. Data versioning (e.g., DVC, LakeFS) ensures reproducibility of the exact data used for a given model version.")
    print("3.  **Automated Evaluation**: Run the `run_automated_evaluation` function against a versioned evaluation dataset. If evaluation metrics (e.g., Ragas scores) fall below predefined thresholds, the pipeline fails, preventing deployment. Results are logged to an experiment tracking platform (e.g., MLflow, Weights & Biases).")
    print("4.  **Model/Prompt Versioning & Registration**: Tag successful RAG configurations (code commit, prompt version, data snapshot, LLM version, embedding model version) with a unique version ID. Register the RAG pipeline as a new 'model' in a model registry.")
    print("5.  **Deployment**: If all checks pass, deploy the new RAG system (or update the vector store) to a staging environment, then production. Strategies include: ")
    print("    *   **Blue/Green Deployment**: Deploy new version alongside old, switch traffic.")
    print("    *   **Canary Deployment**: Gradually shift a small percentage of traffic to the new version, monitor, then full rollout.")
    print("    *   **A/B Testing**: Route different user segments to different RAG versions to compare performance in production.")
    print("6.  **Post-Deployment Monitoring**: Continuously monitor performance, quality, and drift in production using the observability stack. Automated alerts trigger rollbacks if issues are detected.")

    # --- 4. Model Serving Considerations (Conceptual Explanation) ---
    print("\n" + "-"*50 + "\n")
    print("### Model Serving Strategy (Conceptual)")
    print("\n**Serving Framework Choice:**")
    print("For a RAG system, we have two main components to serve: the embedding model and the LLM, plus the orchestration logic. Given the scale and need for efficiency, a robust serving framework is essential:")
    print("*   **vLLM / TGI (Text Generation Inference)**: Excellent choices for serving the LLM itself, offering state-of-the-art optimizations like continuous batching, PagedAttention, and efficient KV cache management. These drastically improve GPU utilization, throughput, and reduce latency for LLM inference.")
    print("*   **Triton Inference Server / KServe**: Ideal for serving both the embedding model and potentially orchestrating the RAG pipeline as an ensemble. Triton supports dynamic batching, model ensembles, and various backends (TensorRT, ONNX Runtime). KServe provides a serverless experience on Kubernetes, abstracting away much of the infrastructure.")
    print("*   **Ray Serve**: A highly flexible and scalable option for building complex, distributed inference graphs. It can orchestrate the retriever, embedding model, and LLM as a single, fault-tolerant service, making it suitable for intricate RAG pipelines.")
    print("A common approach would be to use vLLM for the LLM and Triton for the embedding model and RAG orchestration, all deployed on Kubernetes with KServe.")

    print("\n**Optimization Techniques:**")
    print("1.  **Quantization**: Reduce model size and memory footprint (e.g., FP16, INT8, AWQ, GPTQ) for both embedding and LLM. This allows running more models per GPU, using smaller/cheaper GPUs, or achieving higher throughput due to faster memory access.")
    print("2.  **Continuous Batching / PagedAttention (LLM)**: vLLM's core innovation. Instead of processing requests one by one or in fixed-size batches, it dynamically batches requests as they arrive and manages KV cache efficiently, drastically improving GPU utilization and throughput, especially under variable load.")
    print("3.  **Semantic Caching**: Implement a semantic cache for frequently asked questions or common RAG outputs. Before hitting the LLM, check if a semantically similar query has been answered recently. This significantly reduces latency and LLM costs for repetitive queries.")
    print("4.  **Speculative Decoding**: Use a smaller, faster 'draft' model to predict a sequence of tokens, then verify these predictions with the larger, more accurate 'target' model in parallel. This can speed up generation by 2-3x without sacrificing quality.")
    print("5.  **Hardware Acceleration**: Utilize specialized hardware like NVIDIA H100s/L40s for LLM inference, which offer superior performance for transformer workloads due to their Tensor Cores and high memory bandwidth.")
    print("6.  **Distributed Inference**: For very large models or extreme loads, shard the model across multiple GPUs or machines (e.g., using DeepSpeed, Megatron-LM) to parallelize computation.")
    print("7.  **Asynchronous Processing**: Use asynchronous I/O (e.g., `asyncio` in Python) to handle multiple requests concurrently, especially when waiting for external services like vector databases or LLM APIs.")

    # --- 5. Observability (Conceptual Explanation) ---
    print("\n" + "-"*50 + "\n")
    print("### Observability Strategy (Conceptual)")
    print("\n**Key Metrics to Track:**")
    print("1.  **Performance**: Latency (p50, p90, p99 for retrieval, generation, end-to-end), throughput (queries/second), error rates, GPU/CPU utilization, memory usage.")
    print("2.  **Quality**: Automated evaluation scores (faithfulness, relevance, groundedness) in production, human feedback rates (e.g., thumbs up/down), user satisfaction scores (e.g., from surveys or implicit signals).")
    print("3.  **Cost**: Token usage (input/output), API costs (if using external LLMs), infrastructure costs (GPU hours, storage).")
    print("4.  **Safety & Ethics**: Guardrail trigger rates, hallucination rates (detected by LLM-as-a-judge or specific models), toxicity scores, PII detection rates, jailbreak attempts.")
    print("5.  **Data**: Vector store freshness (last update time), document retrieval success rates, embedding drift (monitoring changes in embedding space over time), query distribution drift.")

    print("\n**Tools and Implementation:**")
    print("1.  **Logging**: Use structured logging (e.g., Python's `logging` module with JSON formatters) to capture every RAG interaction: query, retrieved documents, generated response, LLM parameters, latency, user ID, session ID, trace ID. Ship logs to a centralized platform like **Grafana Loki**, **Datadog Logs**, **Honeycomb**, or **Elasticsearch/Kibana**.")
    print("2.  **Tracing**: Implement distributed tracing using **OpenTelemetry**. Instrument the RAG pipeline to create spans for individual steps: incoming request, embedding generation, vector store lookup, LLM API call, post-processing. This helps pinpoint latency bottlenecks and understand the flow of a request. Traces would be sent to **Grafana Tempo**, **Jaeger**, or **Datadog APM**.")
    print("3.  **Monitoring**: Collect metrics (e.g., using **Prometheus/Grafana**, **Datadog Metrics**, **New Relic**) on API calls, token usage, latency, error rates, and GPU utilization. Create comprehensive dashboards to visualize these metrics in real-time.")
    print("4.  **Alerting**: Set up automated alerts on critical metrics. Examples: P99 latency exceeding 5 seconds, error rate > 1%, automated evaluation score dropping below threshold, sudden spike in guardrail triggers, or significant model/data drift detected. Alerts can be sent via PagerDuty, Slack, email.")
    print("5.  **Feedback Loops**: Integrate user feedback mechanisms (e.g., thumbs up/down buttons, free-text feedback) directly into the application UI. This human feedback is crucial for continuous improvement and identifying issues that automated evaluations might miss. This data should be collected and analyzed as part of the observability stack.")
    print("6.  **LLM-Specific Monitoring**: Utilize specialized LLM observability platforms like **Arize AI**, **Galileo**, or **Langfuse** to get deeper insights into LLM-specific issues like hallucination, bias, and prompt effectiveness.")

    print("\nCapstone Project Execution Complete.")
````
