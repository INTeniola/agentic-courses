````python
# Capstone Project: Detailed Solution

import os
import requests
from bs4 import BeautifulSoup

# --- Mock LLM API Client (for demonstration purposes) ---
# In a real scenario, you would use an actual SDK like openai.OpenAI() or anthropic.Anthropic()
class MockLLMClient:
    def __init__(self, api_key):
        self.api_key = api_key
        self.chat = self.Chat()

    class Chat:
        def __init__(self):
            self.completions = self.Completions()

        class Completions:
            def create(self, model, messages, temperature=0.7, max_tokens=150):
                # print(f"\n--- Mock LLM Call to {model} ---")
                # print(f"Messages: {messages}")
                # print(f"Temperature: {temperature}, Max Tokens: {max_tokens}")
                
                # Simulate a response based on prompt content and simple logic
                last_user_message = messages[-1]['content']
                system_message = messages[0]['content'] if messages and messages[0]['role'] == 'system' else ''

                if "summarize" in system_message.lower() or "summarize" in last_user_message.lower():
                    return type('obj', (object,), {'choices': [{'message': {'content': 'This article discusses the future of AI agents in 2026, highlighting advancements in reasoning, multimodal capabilities, and ethical deployment. It emphasizes the need for robust frameworks to manage complex agentic behaviors and ensure responsible AI development.'}}]})()
                elif "answer the user's question strictly based on the provided context" in system_message.lower():
                    if "AI agents expected by 2026" in last_user_message:
                        return type('obj', (object,), {'choices': [{'message': {'content': 'According to the article, key advancements in AI agents expected by 2026 include enhanced reasoning, multimodal capabilities, and more sophisticated ethical deployment frameworks.'}}]})()
                    elif "capital of France" in last_user_message:
                        return type('obj', (object,), {'choices': [{'message': {'content': 'The provided article does not contain enough information to answer this question.'}}]})()
                elif "tool_call_llm_coin_price" in last_user_message.lower(): # This would typically be handled by an agent orchestrator, not the LLM directly
                    return type('obj', (object,), {'choices': [{'message': {'content': 'The current price of LLM-Coin is $123.45.'}}]})()
                
                return type('obj', (object,), {'choices': [{'message': {'content': 'Mock response: I processed your request based on the prompt.'}}]})()

# --- Configuration ---
# Set your actual API key as an environment variable or replace with your key for testing
LLM_API_KEY = os.getenv("LLM_API_KEY", "sk-mock-api-key-12345") # Using a mock key for this example
llm_client = MockLLMClient(LLM_API_KEY)

# Define the LLM model to use (hypothetical 2026 model)
LLM_MODEL = "AgenticAI/GPT-5-Turbo-2026-03-15" # A powerful, efficient model for general tasks

# --- Functions Implementation ---

def fetch_article_content(url: str) -> str:
    """
    Fetches the main textual content from a given URL.
    Uses BeautifulSoup to parse HTML and extract readable paragraphs, headings, and list items.
    Includes basic error handling and content cleanup.
    """
    print(f"Attempting to fetch content from: {url}")
    try:
        headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/108.0.0.0 Safari/537.36'}
        response = requests.get(url, timeout=15, headers=headers)
        response.raise_for_status() # Raise an HTTPError for bad responses (4xx or 5xx)
        soup = BeautifulSoup(response.text, 'html.parser')

        # Prioritize common content containers. Adjust selectors for specific websites if needed.
        content_tags = soup.find_all(['p', 'h1', 'h2', 'h3', 'h4', 'li', 'blockquote'])
        article_text_parts = []
        for tag in content_tags:
            text = tag.get_text(separator=' ', strip=True)
            if text: # Only add non-empty text
                article_text_parts.append(text)

        article_text = "\n\n".join(article_text_parts)

        # Basic cleanup: remove multiple newlines and excessive whitespace
        article_text = os.linesep.join([s for s in article_text.splitlines() if s]) # Remove empty lines
        article_text = ' '.join(article_text.split()) # Normalize whitespace

        return article_text
    except requests.exceptions.Timeout:
        print(f"Error: Request to {url} timed out.")
        return ""
    except requests.exceptions.RequestException as e:
        print(f"Error fetching URL {url}: {e}")
        return ""
    except Exception as e:
        print(f"An unexpected error occurred during content fetching: {e}")
        return ""

def summarize_content(content: str, model: str = LLM_MODEL) -> str:
    """
    Generates a concise summary of the provided content using an LLM.
    Employs a system message for persona and clear user instructions.
    Handles potential content truncation for very long articles to stay within token limits.
    """
    if not content:
        return "No content to summarize."

    # Truncate content if it's excessively long to fit within typical LLM context windows.
    # A 2026 model might have a context window of 200k-1M tokens, but it's good practice to manage.
    MAX_CONTENT_LENGTH = 10000 # Characters, roughly corresponds to ~2500 tokens
    truncated_content = content[:MAX_CONTENT_LENGTH]
    if len(content) > MAX_CONTENT_LENGTH:
        truncated_content += "\n\n[Content truncated for brevity...]"

    messages = [
        {"role": "system", "content": "You are an expert technical summarizer. Your task is to condense complex technical articles into a concise, 3-5 sentence summary. The summary should highlight the main arguments, key findings, and overall significance for a technically proficient audience. Maintain an objective and informative tone."},
        {"role": "user", "content": f"Please provide a concise summary of the following article content:\n\nArticle Content:\n```\n{truncated_content}\n```\n\nSummary:"
        }
    ]

    try:
        response = llm_client.chat.completions.create(
            model=model,
            messages=messages,
            temperature=0.3, # Lower temperature for factual, concise summaries
            max_tokens=150   # Limit summary length
        )
        return response.choices[0].message.content.strip()
    except Exception as e:
        print(f"Error during summarization: {e}")
        return "Error generating summary."

def answer_question_with_context(question: str, context: str, model: str = LLM_MODEL) -> str:
    """
    Answers a question based *only* on the provided context using an LLM.
    If the answer is not in the context, states that explicitly.
    Uses a strict system prompt to enforce grounding.
    """
    if not context:
        return "No context provided to answer the question."

    # Truncate context if it's too long for the LLM's context window
    MAX_CONTEXT_LENGTH = 10000 # Adjust based on model capabilities
    truncated_context = context[:MAX_CONTEXT_LENGTH]
    if len(context) > MAX_CONTEXT_LENGTH:
        truncated_context += "\n\n[Context truncated for brevity...]"

    messages = [
        {"role": "system", "content": "You are a highly accurate question-answering assistant. Your sole purpose is to answer the user's question *strictly and exclusively* based on the provided CONTEXT. If the CONTEXT does not contain sufficient information to answer the question, you MUST respond with: 'The provided article does not contain enough information to answer this question.' Do not use any prior knowledge or make assumptions. Be concise and direct."},
        {"role": "user", "content": f"CONTEXT:\n```\n{truncated_context}\n```\n\nQUESTION: {question}\n\nANSWER:"
        }
    ]

    try:
        response = llm_client.chat.completions.create(
            model=model,
            messages=messages,
            temperature=0.1, # Very low temperature for factual, non-creative answers
            max_tokens=200   # Allow enough tokens for a detailed answer if available
        )
        return response.choices[0].message.content.strip()
    except Exception as e:
        print(f"Error during contextual question answering: {e}")
        return "Error answering question."

def mock_llm_coin_price_tool() -> str:
    """
    A mock tool function that simulates fetching the real-time price of LLM-Coin.
    In a real agentic system, an LLM might decide to call this function via function calling capabilities.
    """
    # Simulate a real-time API call or database lookup for external, dynamic data
    import time
    time.sleep(0.5) # Simulate network latency
    current_price = "$123.45" # This would come from an external API
    last_updated = "March 15, 2026, 10:30 AM UTC"
    return f"The current price of LLM-Coin is {current_price} (Last updated: {last_updated})."

# --- Main Execution Block (Demonstration) ---
if __name__ == "__main__":
    # Using a real, accessible URL for demonstration. Ensure it has substantial text.
    # Example: A blog post about AI trends or a technical documentation page.
    sample_url = "https://www.agenticlabs.ng/blog/the-future-of-ai-agents-2026-outlook" 
    # If the above URL is not accessible or suitable, use a placeholder or local file content
    # For robust testing, ensure the URL provides substantial text content.

    print("\n--- Step 1: Fetch Article Content ---")
    article_content = fetch_article_content(sample_url)
    if article_content:
        print(f"Successfully fetched {len(article_content)} characters of content.")
        # print(article_content[:1000] + "...\n") # Print first 1000 chars for preview
    else:
        print("Failed to fetch article content. Using placeholder content for demonstration.")
        # Fallback placeholder content if URL fetching fails
        article_content = (
            "This is placeholder content about the future of AI agents in 2026. "
            "It highlights several key advancements: enhanced reasoning capabilities, "
            "seamless multimodal interaction, and robust ethical frameworks for deployment. "
            "The article emphasizes that agents will move beyond simple task execution to "
            "proactive problem-solving and complex decision-making. Challenges include "
            "ensuring transparency, mitigating bias, and developing secure interaction protocols. "
            "LLM-Coin, a hypothetical cryptocurrency, is not discussed in this article." 
            "The capital of France is Paris, but this information is not relevant to AI agents." # Add irrelevant info to test grounding
        )

    print("\n--- Step 2: Summarize Content ---")
    summary = summarize_content(article_content)
    print(f"Summary:\n{summary}")

    print("\n--- Step 3: Answer Contextual Questions ---")
    question_1 = "What are the key advancements in AI agents expected by 2026 according to the article?"
    answer_1 = answer_question_with_context(question_1, article_content)
    print(f"\nQuestion: {question_1}\nAnswer: {answer_1}")

    question_2 = "What is the capital of France?" # Question not directly in context about AI agents
    answer_2 = answer_question_with_context(question_2, article_content)
    print(f"\nQuestion: {question_2}\nAnswer: {answer_2}")

    print("\n--- Step 4: Demonstrate Mock Tool Integration ---")
    # In a full agentic system, an orchestrator would determine if a tool call is needed.
    # Here, we simulate that decision based on the question content.
    question_3 = "What is the current price of LLM-Coin?"
    if "llm-coin" in question_3.lower() or "price" in question_3.lower():
        print(f"\nLLM system detected a need for an external tool to answer: '{question_3}'.")
        print(f"Calling `mock_llm_coin_price_tool()`...")
        tool_answer = mock_llm_coin_price_tool()
        print(f"Question: {question_3}\nAnswer: {tool_answer}")
    else:
        # If the question doesn't trigger a tool, fall back to contextual Q&A or a default response
        print(f"\nQuestion: {question_3}\nAnswer: The provided article does not contain enough information to answer this question, and no relevant tool was found.")

    print("\n--- End of Capstone Project Demonstration ---")
````
