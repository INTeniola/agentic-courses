## LLM01-L02: How LLMs work: tokens, context windows, next-token prediction

Welcome to the foundational lesson on how Large Language Models (LLMs) operate at their core. While LLMs appear to understand and generate human-like text, their underlying mechanism is surprisingly simple yet incredibly powerful: they are sophisticated next-token predictors. To truly build robust LLM applications, understanding this mechanism is paramount.

### 1. Tokens: The Building Blocks of Language for LLMs

Imagine you're building with LEGOs. You don't just have giant blocks; you have various shapes and sizes, from large bricks to tiny studs. Similarly, LLMs don't process raw characters or entire words directly. Instead, they break down text into smaller units called **tokens**.

*   **What are tokens?** Tokens are numerical representations of common sequences of characters. They can be whole words (e.g., "hello"), parts of words (e.g., "ing", "tion"), punctuation marks (e.g., ".", "!"), or even special characters. For example, the word "unbelievable" might be tokenized as "un", "believe", "able".
*   **Why tokens?**
    *   **Efficiency:** Processing individual characters is too granular, and processing entire words leads to an impossibly large vocabulary (think of all possible words, including misspellings and new coinages). Tokens strike a balance, allowing LLMs to handle a vast range of text efficiently.
    *   **Handling rare words:** By breaking down rare or complex words into common sub-word units, the model can still understand and generate them, even if it hasn't seen the full word during training.
    *   **Numerical Representation:** LLMs are mathematical models. Text must be converted into numbers (token IDs) for the model to process them. Each unique token has a unique ID.

### 2. Context Window: The LLM's Short-Term Memory

Think of an LLM's **context window** as its short-term memory or a stage where only a limited number of actors can perform at any given time. When you interact with an LLM, it doesn't remember everything you've ever said or everything it has ever generated. It only 'sees' and processes a fixed number of tokens from the current conversation or input.

*   **What is it?** The context window is the maximum number of tokens (input + output) an LLM can consider when generating its next token. This limit is defined by the model's architecture and training.
*   **Implications:**
    *   **Coherence:** A larger context window generally allows for more coherent and contextually relevant responses over longer interactions.
    *   **Limitations:** If a conversation exceeds the context window, the LLM will effectively 'forget' the earliest parts of the conversation, leading to a loss of coherence or an inability to reference past information.
    *   **Cost & Performance:** Processing more tokens within the context window requires more computational resources, leading to higher inference costs and potentially slower response times.

In 2026, context windows have grown significantly, with many state-of-the-art models offering windows of 128K, 256K, or even larger, enabling much longer and more complex interactions than ever before. However, the fundamental limitation remains.

### 3. Next-Token Prediction: The Core Mechanism

At its heart, an LLM is a highly sophisticated autocomplete engine. Its primary task is **next-token prediction**.

*   **How it works:** Given a sequence of input tokens (the prompt and previous conversation history within the context window), the LLM calculates the probability distribution over its entire vocabulary for what the *next most likely token* should be. It then samples a token from this distribution and appends it to the sequence. This process repeats, token by token, until a stop condition is met (e.g., a special `[EOS]` token, reaching a maximum length, or generating a natural end to a sentence).
*   **Analogy:** Imagine you're typing a sentence, and your phone's keyboard suggests the next word. An LLM does this on a massive scale, considering billions of parameters and patterns learned from its vast training data to predict the most probable next token.
*   **Creativity vs. Determinism:** While it's predicting the 'most likely' token, LLMs often introduce a degree of randomness (controlled by parameters like `temperature` or `top_p`) to make the output less deterministic and more creative or varied. A `temperature` of 0 makes the model highly deterministic (always picking the most probable token), while a higher `temperature` allows for more diverse and surprising outputs.

These three concepts—tokens, context windows, and next-token prediction—are the bedrock upon which all LLM applications are built. Understanding them is crucial for effective prompting, managing costs, and designing robust LLM-powered systems.


```python
# To run this code, you'll need to install the transformers library:
# pip install transformers sentencepiece

from transformers import AutoTokenizer

# --- 1. Tokenization Demonstration ---

# In 2026, many LLMs use advanced tokenizers like SentencePiece or BPE variants.
# We'll use a tokenizer from a modern model (e.g., Google's Gemma) to demonstrate.
# Note: The actual model weights are not loaded, only the tokenizer.

model_name = "google/gemma-2b"
tokenizer = AutoTokenizer.from_pretrained(model_name)

print(f"--- Tokenizer for {model_name} ---")

text_example_1 = "Hello, AgenticLabs! How are you doing today?"
print(f"Original Text 1: '{text_example_1}'")

# Encode the text to get token IDs
token_ids_1 = tokenizer.encode(text_example_1)
print(f"Token IDs 1: {token_ids_1}")

# Decode the token IDs back to see the tokens (often with special prefixes like ' ' for spaces)
tokens_1 = tokenizer.convert_ids_to_tokens(token_ids_1)
print(f"Tokens 1: {tokens_1}")
print("\n")

text_example_2 = "Large Language Models are revolutionizing AI."
print(f"Original Text 2: '{text_example_2}'")
token_ids_2 = tokenizer.encode(text_example_2)
print(f"Token IDs 2: {token_ids_2}")
tokens_2 = tokenizer.convert_ids_to_tokens(token_ids_2)
print(f"Tokens 2: {tokens_2}")
print("\n")

text_example_3 = "Unbelievable! This is a very long word."
print(f"Original Text 3: '{text_example_3}'")
token_ids_3 = tokenizer.encode(text_example_3)
print(f"Token IDs 3: {token_ids_3}")
tokens_3 = tokenizer.convert_ids_to_tokens(token_ids_3)
print(f"Tokens 3: {tokens_3}")
print("\n")

# --- 2. Context Window Concept (Conceptual Demonstration) ---

# We can't directly 'set' a context window on a tokenizer, but we can simulate its effect
# by truncating input to a hypothetical maximum length.

hypothetical_context_window_size = 10 # A very small window for demonstration
long_text = "This is a very long sentence that will definitely exceed our small hypothetical context window size for demonstration purposes. LLMs have much larger windows, but the principle is the same."

print(f"--- Context Window Simulation (Hypothetical size: {hypothetical_context_window_size} tokens) ---")
print(f"Original Long Text: '{long_text}'")

# Tokenize the long text
long_text_token_ids = tokenizer.encode(long_text)
print(f"Total tokens in long text: {len(long_text_token_ids)}")

if len(long_text_token_ids) > hypothetical_context_window_size:
    # Simulate truncation: only the last 'hypothetical_context_window_size' tokens are kept
    # In reality, LLMs often truncate from the beginning or use more sophisticated methods.
    truncated_token_ids = long_text_token_ids[-hypothetical_context_window_size:]
    truncated_text = tokenizer.decode(truncated_token_ids)
    print(f"Truncated tokens (last {hypothetical_context_window_size}): {truncated_token_ids}")
    print(f"Text visible to LLM (after truncation): '{truncated_text}'")
    print("Note: The LLM would 'forget' the beginning of the original text.")
else:
    print("Text fits within the hypothetical context window.")
print("\n")

# --- 3. Next-Token Prediction (Conceptual Explanation) ---

# We cannot run a full LLM for next-token prediction in this simple script,
# but we can illustrate the input it would receive.

input_for_prediction = "The quick brown fox jumps over the"
input_token_ids = tokenizer.encode(input_for_prediction)

print(f"--- Next-Token Prediction Input ---")
print(f"Input text for prediction: '{input_for_prediction}'")
print(f"Input token IDs: {input_token_ids}")
print(f"Input tokens: {tokenizer.convert_ids_to_tokens(input_token_ids)}")

# Conceptually, an LLM would take `input_token_ids`,
# process them through its neural network, and output a probability distribution
# over its entire vocabulary for the *next* token.
# For example, it might predict:
#   'the': 0.85 probability
#   'a': 0.10 probability
#   'its': 0.03 probability
#   ... (for all other tokens in its vocabulary)

# Then, based on sampling parameters (like temperature), it would select one token.
# If temperature is low, it might pick 'the'. If high, it might pick 'a' or 'its'.
# Let's assume it picks 'the'.
predicted_next_token_id = tokenizer.encode(" the")[-1] # Get the ID for ' the'
predicted_next_token = tokenizer.decode([predicted_next_token_id])

print(f"\nConceptual next token prediction (example): '{predicted_next_token}' (ID: {predicted_next_token_id})")
print("The LLM would then append this token and repeat the process.")
```

### Interpreting the Code Output and Practical Implications

#### Tokenization Output

From the code output, you'll observe:

*   **Subword Tokenization:** Words like "AgenticLabs" or "revolutionizing" are often broken down into smaller, more common subword units (e.g., `[' Agent', 'ic', 'Labs']` or `[' revolution', 'izing']`). This is a key feature of modern tokenizers like SentencePiece (used by Gemma) or BPE, allowing them to handle a vast vocabulary efficiently and generalize to unseen words.
*   **Special Characters:** Punctuation and spaces are often treated as distinct tokens or integrated into the preceding word's token (e.g., `[' Hello', ',', ' Agent', 'ic', 'Labs', '!']`). Notice the leading space ` ` in many tokens; this helps the model distinguish between words that start with a space and those that don't (e.g., `[' word']` vs. `['word']`).
*   **Token IDs:** Each token corresponds to a unique numerical ID. These IDs are the actual input the LLM processes. The size of the vocabulary (the number of unique tokens) can range from tens of thousands to hundreds of thousands.

**Performance Trade-offs & Use Cases:**

*   **Cost:** LLM API calls are almost universally priced per token. Understanding how your input text translates into tokens is crucial for managing costs. A seemingly short prompt might expand into many tokens due to subword tokenization, especially for complex or non-English text.
*   **Efficiency:** Tokenization is a pre-processing step. Efficient tokenizers are vital for fast inference. Different models might use different tokenizers, leading to varying token counts for the same text.
*   **Multilingual Support:** Subword tokenization is particularly effective for multilingual models, as it can break down words from various languages into shared subword units.

#### Context Window Simulation

Our conceptual demonstration highlights the critical limitation of the context window:

*   **Truncation:** When input text (or conversation history) exceeds the context window, the LLM must truncate it. Our example showed truncating from the beginning, meaning the LLM would 'forget' the initial part of the `long_text`. This is a common strategy, though some models might employ more advanced methods like summarization or attention mechanisms that prioritize recent tokens.

**Performance Trade-offs & Use Cases:**

*   **Coherence vs. Cost:** Larger context windows allow for more coherent and context-aware responses over longer interactions (e.g., chatbots, document analysis). However, they come at a higher computational cost (more memory, slower inference, higher API charges).
*   **Information Loss:** Developers must be acutely aware of context window limits to prevent critical information from being 'forgotten' by the LLM. Strategies like **Retrieval Augmented Generation (RAG)** (which you'll explore in later lessons) are designed to overcome this by dynamically injecting relevant information into the context window.
*   **Prompt Engineering:** Effective prompt engineering involves crafting prompts that fit within the context window while providing all necessary information for the LLM to perform its task.

#### Next-Token Prediction Concept

The conceptual explanation of next-token prediction underscores the probabilistic nature of LLM generation:

*   **Probabilistic Output:** LLMs don't 'know' the next word; they predict the most statistically probable next token based on their training data. This is why they can sometimes generate factually incorrect information (hallucinations) or exhibit biases present in their training data.
*   **Controlling Creativity:** Parameters like `temperature` and `top_p` are crucial for controlling the randomness and creativity of the LLM's output. For tasks requiring factual accuracy (e.g., data extraction), a low `temperature` is preferred. For creative writing or brainstorming, a higher `temperature` can yield more diverse results.
*   **Iterative Generation:** The LLM generates text token by token, making each new token prediction based on all preceding tokens in the context window. This iterative process is what allows for long, coherent generations.

In 2026, the sophistication of next-token prediction has evolved to include multi-modal tokens (for models that process text, images, audio, etc.) and more nuanced control over generation parameters, but the fundamental mechanism remains the same.
