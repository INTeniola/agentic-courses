### The Scenario: Shelf Organizer Bot

Imagine you are building a simple **Shelf Organizer Bot** for a small, simulated warehouse. Its primary task is to find a specific `target_item` among various items placed on different shelves. The bot operates in a loop:

1.  **Perceive:** It looks at the items currently on a shelf.
2.  **Reason:** It decides if the `target_item` is present on the current shelf.
3.  **Act:** If the item is found, it "picks it up" and its task is complete. If not, it "moves to the next shelf" to continue its search.

### Task Requirements:

Your goal is to implement the `perceive`, `reason`, and `act` functions, and then integrate them into a main `agent_loop` function. You must adhere to the following:

1.  **`perceive(environment_state)` function:**
    *   Takes `environment_state` (a list of strings representing items on a shelf) as input.
    *   Returns the perceived data (which, for this simple case, will be the `environment_state` itself).

2.  **`reason(perceived_data, target_item)` function:**
    *   Takes `perceived_data` (the output from `perceive`) and the `target_item` (a string) as input.
    *   Determines if the `target_item` is present in the `perceived_data`.
    *   Returns a string indicating the decision, e.g., `'found'` or `'not_found'`.

3.  **`act(decision, target_item, current_shelf_items)` function:**
    *   Takes the `decision` (from `reason`), the `target_item`, and the `current_shelf_items` as input.
    *   If `decision` is `'found'`:
        *   Prints a message indicating the item was found and picked up.
        *   Returns `(True, None)` to signal success and no further environment state needed.
    *   If `decision` is `'not_found'`:
        *   Prints a message indicating the item was not found on the current shelf and the bot is moving to the next.
        *   Returns `(False, simulate_environment(target_item))` to signal continuation and provide the next shelf's items.

4.  **`agent_loop(target_item, max_attempts)` function:**
    *   Orchestrates the entire PRA cycle.
    *   Initializes the `current_shelf_items` by calling `simulate_environment(target_item)` once before the loop starts.
    *   Runs for a maximum of `max_attempts` iterations.
    *   Inside the loop, it calls `perceive`, `reason`, and `act` in sequence.
    *   It should terminate the loop early if the `act` function signals that the `target_item` has been found.
    *   Prints a final status message (e.g., "Task completed!" or "Task failed after X attempts.").

### Evaluation Criteria:

*   **Correctness:** The agent successfully finds the `target_item` when it's available within `max_attempts`.
*   **Modularity:** Clear separation of concerns into `perceive`, `reason`, and `act` functions.
*   **Readability:** Well-structured code with meaningful variable names and comments.
*   **Adherence to PRA:** The `agent_loop` clearly demonstrates the Perception-Reasoning-Action cycle.

```python
import random

# --- Environment Setup (Provided - Do Not Modify) ---

def simulate_environment(target_item: str, current_attempt: int) -> list[str]:
    """
    Simulates a shelf in a warehouse environment.
    Randomly places items, sometimes including the target item.
    """
    all_possible_items = [
        "wrench", "screwdriver", "hammer", "tape_measure", "level",
        "pliers", "saw", "drill", "safety_glasses", "gloves",
        "nuts", "bolts", "washers", "nails", "screws",
        "paint_can", "brush", "roller", "ladder", "toolbox"
    ]

    # Ensure the target item appears more frequently in later attempts
    # to simulate the agent eventually finding it, but not always immediately.
    if current_attempt > 3 and random.random() < 0.7: # 70% chance after 3 attempts
        shelf_items = random.sample(all_possible_items, k=random.randint(3, 7))
        if target_item not in shelf_items:
            shelf_items[random.randint(0, len(shelf_items) - 1)] = target_item
    else:
        shelf_items = random.sample(all_possible_items, k=random.randint(3, 7))
        # Small chance to include target even in early attempts
        if random.random() < 0.2 and target_item not in shelf_items:
             shelf_items[random.randint(0, len(shelf_items) - 1)] = target_item

    print(f"[Environment] Shelf {current_attempt}: {shelf_items}")
    return shelf_items

# --- Agent Parameters (Provided - Do Not Modify) ---
target_item_to_find = "drill"
max_search_attempts = 7

print(f"Agent initialized. Target item: '{target_item_to_find}'. Max attempts: {max_search_attempts}\n")

# --- Reference Solution (Your Implementation Goes Here) ---

def perceive(environment_state: list[str]) -> list[str]:
    """
    Perception phase: The agent observes the current state of the environment.
    In this simple case, it directly receives the list of items on the shelf.
    """
    print(f"[Perception] Observing items: {environment_state}")
    return environment_state

def reason(perceived_data: list[str], target_item: str) -> str:
    """
    Reasoning phase: The agent processes perceived data to make a decision.
    It checks if the target item is present in the perceived list.
    """
    if target_item in perceived_data:
        print(f"[Reasoning] Decision: '{target_item}' found!")
        return "found"
    else:
        print(f"[Reasoning] Decision: '{target_item}' not found.")
        return "not_found"

def act(decision: str, target_item: str, current_attempt: int) -> tuple[bool, list[str] | None]:
    """
    Action phase: The agent performs an action based on its decision.
    Returns (is_task_complete, next_environment_state).
    """
    if decision == "found":
        print(f"[Action] Picking up '{target_item}'. Task complete!")
        return (True, None) # Task complete, no next environment state needed
    else:
        print(f"[Action] '{target_item}' not here. Moving to the next shelf...")
        # Simulate moving to the next shelf by requesting a new environment state
        next_shelf_items = simulate_environment(target_item, current_attempt + 1)
        return (False, next_shelf_items) # Task not complete, provide next state

def agent_loop(target_item: str, max_attempts: int):
    """
    The main agent loop orchestrating Perception, Reasoning, and Action.
    """
    print("--- Starting Agent Loop ---")
    current_shelf_items = simulate_environment(target_item, 1) # Initial environment state
    task_completed = False

    for attempt in range(1, max_attempts + 1):
        print(f"\n--- Attempt {attempt}/{max_attempts} ---")

        # 1. Perception
        perceived_data = perceive(current_shelf_items)

        # 2. Reasoning
        decision = reason(perceived_data, target_item)

        # 3. Action
        task_completed, next_env_state = act(decision, target_item, attempt)

        if task_completed:
            print("--- Agent Loop Terminated: Task successfully completed! ---")
            break
        else:
            current_shelf_items = next_env_state # Update environment for next iteration

    if not task_completed:
        print(f"\n--- Agent Loop Terminated: Task failed after {max_attempts} attempts. '{target_item}' not found. ---")

# --- Run the Agent --- 
agent_loop(target_item_to_find, max_search_attempts)
```
