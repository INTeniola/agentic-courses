```python
import json
from typing import List, Dict, Any, Optional

# --- Mock LLM and Tools (DO NOT MODIFY) ---
class MockLLM:
    """A mock Large Language Model for simulating agent reasoning."""
    def __init__(self, model_name: str = "mock-gpt-4o-2026"):
        self.model_name = model_name

    def chat_completion(self, messages: List[Dict[str, str]], temperature: float = 0.7) -> str:
        """Simulates an LLM chat completion based on simple keyword logic."""
        prompt = messages[-1]['content'].lower()
        
        # Logic for generating search terms
        if "what are the most relevant search terms to find information" in prompt:
            if "latest ai agent frameworks in 2026" in prompt: return "latest AI agent frameworks 2026"
            if "quantum computing breakthroughs" in prompt: return "quantum computing breakthroughs 2026"
            if "future of work" in prompt: return "future of work AI impact"
            if "their focus on self-improvement" in prompt and "latest ai agent frameworks" in prompt: return "AI agent frameworks self-improvement features"
            if "capital of france" in prompt: return "NO_SEARCH" # Example of a query that doesn't need search
            if "fault-tolerant systems" in prompt and "quantum computing" in prompt: return "quantum computing fault-tolerant systems breakthroughs"
            return "NO_SEARCH" # If no specific search terms are identified, assume no search needed
        
        # Logic for summarizing/answering
        if "synthesize the following search results to answer the query" in prompt or "answer the question" in prompt:
            if "no relevant results found" in prompt: return "I couldn't find relevant information for your query. Please try rephrasing."
            
            # Specific answers based on search results and query
            if "latest ai agent frameworks 2026" in prompt and "langchain" in prompt and "llamaindex" in prompt:
                return "The latest AI agent frameworks in 2026, such as advanced versions of LangChain and LlamaIndex, are focusing on multi-modal reasoning, robust tool integration, and especially **self-improving loops**. These frameworks emphasize modularity and allow agents to learn and adapt over time. [Simulated Source: Web Search]"
            
            if "quantum computing breakthroughs 2026" in prompt and "superconducting qubits" in prompt:
                return "Recent quantum computing breakthroughs in 2026 highlight advancements in superconducting qubits achieving higher fidelity, new error correction techniques, and the development of more stable quantum processors, moving closer to fault-tolerant quantum computers. [Simulated Source: Web Search]"
            
            if "future of work ai impact" in prompt and "automation" in prompt:
                return "AI's impact on the future of work in 2026 is characterized by increased automation of routine tasks, the rise of AI-augmented human roles, and a greater demand for skills in AI development, ethics, and human-AI collaboration. [Simulated Source: Web Search]"
            
            if "their focus on self-improvement" in prompt and "ai agent frameworks" in prompt:
                return "Building on previous discussions, the focus on self-improvement in 2026 AI agent frameworks involves agents learning from their past actions, refining their internal models, and adapting their strategies to achieve better outcomes over time. This includes mechanisms for reflection and meta-learning. [Simulated Source: Memory & Web Search]"
            
            if "capital of france" in prompt: return "The capital of France is Paris. [Simulated Source: General Knowledge]"

            if "quantum computing fault-tolerant systems breakthroughs" in prompt and "error correction" in prompt:
                return "Yes, the breakthroughs in quantum computing, particularly in error correction protocols and increased qubit stability, are indeed crucial steps towards achieving fault-tolerant quantum systems. While fully fault-tolerant systems are still a research goal, 2026 advancements are bringing us significantly closer. [Simulated Source: Web Search & Memory]"
            
            # Generic fallback if specific keywords not matched
            return "Based on the information provided and my current context, a concise answer is: [Simulated Answer based on LLM's understanding]. [Simulated Source: LLM Synthesis]"
            
        return "I'm not sure how to respond to that specific prompt yet."

def web_search(query: str) -> List[str]:
    """A mock web search tool that returns predefined results."""
    print(f"[TOOL CALL] Performing web search for: '{query}'")
    query_lower = query.lower()
    if "latest ai agent frameworks 2026" in query_lower or "ai agent frameworks self-improvement features" in query_lower:
        return [
            "Result 1: LangChain 0.2.x: Advanced Agent Orchestration with self-correction modules",
            "Result 2: LlamaIndex 0.10.x: Enhanced RAG, Memory, and adaptive planning",
            "Result 3: Custom Agentic Frameworks: Focus on Self-Improvement through reflection and meta-learning"
        ]
    elif "quantum computing breakthroughs 2026" in query_lower or "quantum computing fault-tolerant systems breakthroughs" in query_lower:
        return [
            "Result A: Superconducting Qubits achieve 99.99% fidelity",
            "Result B: New error correction protocols for quantum processors",
            "Result C: IBM and Google unveil 1000+ qubit systems, focusing on error mitigation"
        ]
    elif "future of work ai impact" in query_lower:
        return [
            "Result X: Automation of routine tasks accelerates",
            "Result Y: New job roles emerge in AI supervision and ethics",
            "Result Z: Human-AI collaboration becomes standard practice"
        ]
    else:
        return ["No relevant results found for your specific query."]

# --- Agent Solution ---

class SmartResearchAssistant:
    def __init__(self, llm: MockLLM):
        self.llm = llm
        self.memory: List[Dict[str, str]] = [] # Simple memory to store past interactions
        print("SmartResearchAssistant initialized.")

    def _perceive(self, query: str) -> Dict[str, str]:
        """Perceives the user's query and any relevant memory.
        Combines the current query with a summary of past interactions to provide context for reasoning.
        """
        # Create a concise string from memory for the LLM to process
        context_from_memory = ""
        if self.memory:
            context_from_memory = "Previous interactions:\n" + "\n".join(
                [f"Q: {item['query']}\nA: {item['answer']}" for item in self.memory]
            )
        
        return {"current_query": query, "context_from_memory": context_from_memory}

    def _reason(self, perceived_data: Dict[str, str]) -> Dict[str, Any]:
        """Reasons about the perceived data to decide on actions and generate search terms/summaries.
        Uses the LLM to determine if a search is needed and what terms to use, or if it can answer directly.
        """
        query = perceived_data["current_query"]
        context = perceived_data["context_from_memory"]
        
        # Prompt the LLM to decide on search terms or if no search is needed
        # The LLM's response will be parsed to determine the next action
        reasoning_prompt = f"""You are a research assistant. Given the user query: '{query}' and the following past context:
{context}

Decide if a web search is necessary to answer this query. If yes, provide the most relevant search terms (e.g., 'keyword1 keyword2'). If no search is needed (e.g., if the answer is in the context or general knowledge), respond with 'NO_SEARCH'.

Example Search Response: 'latest AI agent frameworks 2026'
Example No Search Response: 'NO_SEARCH'
"""
        
        llm_decision = self.llm.chat_completion(messages=[{"role": "user", "content": reasoning_prompt}])
        
        if llm_decision == "NO_SEARCH":
            # If no search, the agent needs to formulate an answer directly from its internal knowledge/memory
            return {"action": "answer", "final_answer_prompt": f"Answer the question: '{query}' based on the following context: {context}. If the context is insufficient, state that you cannot provide a detailed answer without more information."}
        else:
            # If search is needed, provide the search query
            return {"action": "search", "search_query": llm_decision}

    def _act(self, reasoning_output: Dict[str, Any]) -> Dict[str, str]:
        """Executes actions based on reasoning, e.g., calling tools or formulating a final answer.
        This method handles calling the web_search tool or directly asking the LLM for a final answer.
        """
        action_type = reasoning_output["action"]
        
        if action_type == "search":
            search_query = reasoning_output["search_query"]
            # Call the external web_search tool
            search_results = web_search(search_query)
            # Return results as a JSON string to easily pass around
            return {"action_output_type": "search_results", "results": json.dumps(search_results)}
        elif action_type == "answer":
            # This path is for when no search is needed, and the LLM should directly answer
            final_answer_prompt = reasoning_output["final_answer_prompt"]
            final_answer = self.llm.chat_completion(messages=[{"role": "user", "content": final_answer_prompt}])
            return {"action_output_type": "final_answer", "answer": final_answer}
        else:
            return {"action_output_type": "error", "message": "Unknown action type"}

    def _update_memory(self, query: str, answer: str):
        """Updates the agent's memory with the latest interaction.
        Stores the user's query and the agent's final answer.
        Implements a simple FIFO (First-In, First-Out) strategy to keep memory size manageable.
        """
        self.memory.append({"query": query, "answer": answer})
        # Keep only the last 3 interactions to prevent memory bloat for this simple agent
        if len(self.memory) > 3:
            self.memory.pop(0)
        print(f"[AGENT] Memory updated. Current memory size: {len(self.memory)}")

    def run(self, user_query: str) -> str:
        """Orchestrates the agent's perception, reasoning, action, and memory update loop.
        This is the main entry point for interacting with the agent.
        """
        print(f"\n--- AGENT START: Processing Query ---")
        print(f"[AGENT] Received query: '{user_query}'")
        
        # 1. Perception: Gather current query and relevant context from memory
        perceived_data = self._perceive(user_query)
        print(f"[AGENT] Perceived current query and context from memory.")
        
        # 2. Reasoning (Initial decision: search or direct answer?)
        reasoning_output = self._reason(perceived_data)
        print(f"[AGENT] Reasoning led to action: {reasoning_output['action']}")
        
        final_response = ""
        if reasoning_output["action"] == "search":
            # 3. Action: Perform the web search using the determined query
            action_output = self._act(reasoning_output)
            search_results = json.loads(action_output["results"])
            print(f"[AGENT] Web search completed. Results: {search_results}")
            
            # 2. Reasoning (Synthesize results): Use LLM to summarize search results and answer the query
            synthesis_prompt = f"""Synthesize the following search results to answer the user's original query: '{user_query}'.
Search Results: {search_results}

Also consider the following past context for a more comprehensive answer:
{perceived_data['context_from_memory']}

Provide a concise and informative answer, citing the simulated web search as the source.
"""
            final_response = self.llm.chat_completion(messages=[{"role": "user", "content": synthesis_prompt}])
            
        elif reasoning_output["action"] == "answer":
            # 3. Action: Directly answer based on internal knowledge/memory (no search needed)
            action_output = self._act(reasoning_output)
            final_response = action_output["answer"]
            
        else:
            final_response = "An internal error occurred during agent processing. Please try again."
            
        # 4. Memory Update: Store the current interaction
        self._update_memory(user_query, final_response)
        
        print(f"[AGENT] Final Answer: {final_response}")
        print(f"--- AGENT END: Query Processed ---\n")
        return final_response

# --- Example Usage: Run the SmartResearchAssistant ---
llm_instance = MockLLM()
agent = SmartResearchAssistant(llm_instance)

print("\n--- First Query ---")
agent.run("What are the latest AI agent frameworks in 2026?")

print("\n--- Second Query (using memory for follow-up) ---")
agent.run("Tell me more about their focus on self-improvement.")

print("\n--- Third Query (new topic) ---")
agent.run("What are the recent breakthroughs in quantum computing?")

print("\n--- Fourth Query (no search needed, general knowledge/memory) ---")
agent.run("What is the capital of France?")

print("\n--- Fifth Query (another follow-up on quantum computing) ---")
agent.run("Are these breakthroughs leading to fault-tolerant systems?")
```
