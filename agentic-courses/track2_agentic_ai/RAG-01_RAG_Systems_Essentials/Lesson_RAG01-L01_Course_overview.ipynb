### Why RAG? The Foundation of Grounded AI

In the rapidly evolving landscape of AI, Large Language Models (LLMs) have demonstrated incredible capabilities in understanding and generating human-like text. However, they come with inherent limitations:

1.  **Hallucination:** LLMs can generate plausible-sounding but factually incorrect information.
2.  **Outdated Knowledge:** Their knowledge is limited to their training data cutoff, making them unaware of recent events or developments.
3.  **Lack of Domain-Specific Expertise:** Without specific fine-tuning, LLMs often lack deep knowledge in specialized domains (e.g., internal company policies, specific research papers).

This is where **Retrieval Augmented Generation (RAG)** steps in. Imagine giving an LLM an 


open-book exam.


RAG empowers LLMs by providing them with access to external, up-to-date, and domain-specific information *at the time of inference*. This dramatically reduces hallucinations, ensures factual accuracy, and allows LLMs to answer questions based on proprietary or real-time data.

## Initializing the Environment and a Basic LLM Interaction

Now that our environment is set up, let's perform a very basic LLM interaction. This will serve as our baseline, highlighting the kind of generic or potentially limited responses a standalone LLM might provide, thereby setting the stage for why RAG is so crucial.

**Important:** Replace `YOUR_OPENAI_API_KEY` with your actual OpenAI API key. For security, it's best practice to load this from environment variables, especially in production. For this notebook, we'll set it directly for simplicity.


```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# --- Configuration --- #
# Set your OpenAI API key here. In a real application, use environment variables.
# Example: os.environ["OPENAI_API_KEY"] = "sk-..."
os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY" # <<< REPLACE WITH YOUR ACTUAL KEY

# Initialize the LLM
# We'll use a recent, capable model for demonstration.
# As of 2026, 'gpt-4o-mini' or 'gpt-4o' are excellent choices for general tasks.
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)

# --- Basic LLM Interaction --- #
print("\n--- Performing a basic LLM query ---")

# Let's ask a question that requires very recent or specific, non-public knowledge.
# A base LLM might struggle with this, providing a generic answer or stating it doesn't know.
question = "Who won the 2026 AI Innovation Award for Agentic Systems, and what was their breakthrough?"

# Invoke the LLM with our question
response = llm.invoke([HumanMessage(content=question)])

print(f"\nQuestion: {question}")
print(f"LLM's Response: {response.content}")

print("\nNotice how the LLM might give a generic answer, state it doesn't know, or even 'hallucinate' a plausible but fictional response for such a specific, future-dated question. This is precisely the gap that RAG aims to fill by providing real-time, accurate, and grounded information!")
```
