---

### Question 1

**Correct Answer: C)** To find relevant documents or passages from a knowledge base that are related to the user's query.

**Explanation**: The retrieval component's core function is to search a vast corpus of information (like a vector database) and pull out pieces of data that are most pertinent to the user's input. This retrieved information then serves as context for the LLM.

---

### Question 2

**Correct Answer: C)** The retrieved documents are passed to the LLM as additional context alongside the user's query.

**Explanation**: After retrieval, the key step is to augment the LLM's prompt. The user's original query is combined with the relevant retrieved documents, forming an enriched prompt that guides the LLM to generate a more informed and grounded response.

---

### Question 3

**Correct Answer: D)** Over-reliance on LLM Pre-training: The LLM ignoring retrieved context in favor of its internal knowledge.

**Explanation**: While LLMs can sometimes exhibit a bias towards their pre-trained knowledge, the term "over-reliance on LLM pre-training" isn't typically listed as a *challenge* of RAG itself in the same vein as the others. RAG is specifically designed to *mitigate* the LLM's reliance on internal knowledge by providing external, up-to-date context. The other options (Hallucination, Context Window Limitations, Retrieval Relevance) are all well-known and active areas of research and development in RAG systems.

---

### Question 4

**Explanation**: **Chunking** is the process of breaking down large documents into smaller, manageable segments or "chunks." Each chunk is then independently embedded and stored in the vector database. It is important for several reasons:

1.  **Context Window Management**: LLMs have a limited context window. Chunking ensures that the retrieved information fits within this window, preventing truncation of vital context.
2.  **Improved Relevance**: Smaller chunks often represent more focused topics. This allows the retrieval system to pinpoint highly specific and relevant information more accurately, rather than retrieving an entire, potentially less relevant, large document.
3.  **Efficiency**: Processing and embedding smaller chunks is generally more efficient than handling entire large documents.

---

### Question 5

**Correct Answer: B)** The Embedder (Embedding Model)

**Explanation**: The embedder (or embedding model) is responsible for converting both the user's query and the chunks of the knowledge base into numerical vector representations (embeddings). These embeddings capture the semantic meaning of the text. The vector database then uses the query's embedding to find the most semantically similar document embeddings, facilitating the retrieval process.
