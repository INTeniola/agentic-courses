```python
# Capstone Problem: RAG System for AgenticPedia

# --- Setup and Imports ---
import os
# Ensure you have the necessary libraries installed:
# pip install langchain langchain-community chromadb sentence-transformers tiktoken

from langchain_community.document_loaders import TextLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain.chains import RetrievalQA
from langchain_openai import ChatOpenAI # Or use another LLM like Ollama, Anthropic, etc.

# --- Configuration ---
# Create a dummy knowledge base directory and files if they don't exist
KB_DIR = "knowledge_base"
if not os.path.exists(KB_DIR):
    os.makedirs(KB_DIR)
    with open(os.path.join(KB_DIR, "agentic_ai.txt"), "w") as f:
        f.write("Agentic AI refers to AI systems capable of autonomous decision-making, planning, and execution to achieve complex goals. They often involve reasoning, memory, and tool use. Examples include Auto-GPT and BabyAGI. They are designed to operate with minimal human intervention, often leveraging tools to interact with the environment.")
    with open(os.path.join(KB_DIR, "rag_overview.txt"), "w") as f:
        f.write("Retrieval Augmented Generation (RAG) combines information retrieval with large language models. It addresses LLM limitations by fetching external, up-to-date information to ground responses, reducing hallucinations and improving factual accuracy. RAG systems typically involve document loading, chunking, embedding, vector storage, retrieval, and generation.")
    with open(os.path.join(KB_DIR, "automation_tools.txt"), "w") as f:
        f.write("Automation tools streamline repetitive tasks. In the context of AI, these can be anything from robotic process automation (RPA) to sophisticated AI agents that manage workflows and interact with APIs. Modern automation often integrates AI for intelligent decision-making and adaptation.")
    with open(os.path.join(KB_DIR, "llm_limitations.txt"), "w") as f:
        f.write("Large Language Models (LLMs) can suffer from hallucinations, where they generate factually incorrect or nonsensical information. They also have a knowledge cutoff, meaning their training data doesn't include recent events. RAG systems are designed to overcome these limitations by providing external, up-to-date context.")

# Set your OpenAI API key here, or configure another LLM
# os.environ["OPENAI_API_KEY"] = "YOUR_API_KEY"

# --- Step 1: Load Documents ---
def load_documents(directory_path: str):
    print(f"Loading documents from {directory_path}...")
    documents = []
    for filename in os.listdir(directory_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(directory_path, filename)
            loader = TextLoader(file_path)
            documents.extend(loader.load())
    print(f"Loaded {len(documents)} documents.")
    return documents

# --- Step 2: Chunk Documents ---
def chunk_documents(documents):
    print("Chunking documents...")
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=500, # Optimal chunk size can vary, 500 is a good starting point
        chunk_overlap=50, # Overlap helps maintain context between chunks
        length_function=len,
        add_start_index=True,
    )
    chunks = text_splitter.split_documents(documents)
    print(f"Created {len(chunks)} chunks.")
    return chunks

# --- Step 3: Embed and Store ---
def create_vector_store(chunks, persist_directory="./chroma_db"):
    print("Creating embeddings and storing in vector database...")
    # Using a robust open-source embedding model suitable for 2026
    # 'all-MiniLM-L6-v2' is a good balance of performance and speed for many tasks.
    # For higher performance, consider larger models or specialized ones.
    embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

    # Initialize ChromaDB as the vector store
    # It's a good choice for local development and smaller-scale applications.
    vector_store = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        persist_directory=persist_directory
    )
    print(f"Vector store created and persisted to {persist_directory}.")
    return vector_store

# --- Step 4 & 5: Retrieve and Generate (RAG Chain) ---
def setup_rag_chain(vector_store):
    print("Setting up RAG chain...")
    # Initialize the LLM. Using ChatOpenAI as a common example.
    # For local LLMs, you might use `Ollama(model="llama2")` or similar.
    # Ensure OPENAI_API_KEY is set in your environment or replace with a local LLM setup.
    llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0.1) # Lower temperature for more factual answers

    # Create a RetrievalQA chain. This orchestrates retrieval and generation.
    # 'stuff' chain type puts all retrieved documents into the LLM's context.
    qa_chain = RetrievalQA.from_chain_type(
        llm=llm,
        chain_type="stuff",
        retriever=vector_store.as_retriever(search_kwargs={"k": 3}), # Retrieve top 3 relevant chunks
        return_source_documents=True # Useful for debugging and showing sources
    )
    print("RAG chain ready.")
    return qa_chain

# --- Main Execution ---
if __name__ == "__main__":
    # 1. Load documents
    documents = load_documents(KB_DIR)

    # 2. Chunk documents
    chunks = chunk_documents(documents)

    # 3. Embed and store
    vector_store = create_vector_store(chunks)

    # 4. & 5. Setup RAG chain
    rag_chain = setup_rag_chain(vector_store)

    print("\n--- AgenticPedia RAG System Ready ---")
    print("Enter your queries. Type 'exit' to quit.")

    while True:
        query = input("\nYour Query: ")
        if query.lower() == 'exit':
            break

        if not os.getenv("OPENAI_API_KEY") and isinstance(rag_chain.llm, ChatOpenAI):
            print("Warning: OPENAI_API_KEY not set. Cannot run LLM. Please set the environment variable or configure a local LLM.")
            continue

        try:
            result = rag_chain.invoke({"query": query})
            print("\nAnswer:")
            print(result["result"])
            print("\nSource Documents:")
            for i, doc in enumerate(result["source_documents"]):
                print(f"  {i+1}. Content: {doc.page_content[:150]}... (Source: {doc.metadata.get('source', 'N/A')}, Start Index: {doc.metadata.get('start_index', 'N/A')})")
        except Exception as e:
            print(f"An error occurred: {e}")
            print("Please ensure your LLM API key is correctly set or your local LLM is running.")

    print("\nThank you for using AgenticPedia RAG System!")

    # Optional: Clean up the ChromaDB persistence directory
    # import shutil
    # if os.path.exists("./chroma_db"):
    #     shutil.rmtree("./chroma_db")
    #     print("Cleaned up chroma_db directory.")
```

```python
# Detailed Solution: Capstone RAG System

# --- Setup and Imports ---
import os
import shutil # For cleanup

# Install necessary libraries if you haven't already:
# pip install langchain langchain-community chromadb sentence-transformers tiktoken openai

from langchain_community.document_loaders import TextLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain.chains import RetrievalQA
from langchain_openai import ChatOpenAI # Using OpenAI for demonstration, replace with local LLM if preferred
from langchain.prompts import ChatPromptTemplate

# --- Configuration ---
KB_DIR = "knowledge_base_solution"
PERSIST_DIR = "./chroma_db_solution"

# Create a dummy knowledge base directory and files for the solution
if os.path.exists(KB_DIR): # Clean up previous run if exists
    shutil.rmtree(KB_DIR)
os.makedirs(KB_DIR)

# Populate with more detailed content for better RAG performance
with open(os.path.join(KB_DIR, "agentic_ai_deep.txt"), "w") as f:
    f.write("Agentic AI systems are advanced forms of artificial intelligence designed to operate autonomously, making decisions and executing actions to achieve complex, high-level goals. They typically incorporate components like planning modules, memory (short-term and long-term), tool-use capabilities (interacting with APIs, databases, web), and self-reflection mechanisms. Prominent examples include frameworks like Auto-GPT, BabyAGI, and more recent developments in multi-agent orchestration platforms. These agents excel in dynamic environments where tasks require adaptive strategies and interaction with external systems. Their core strength lies in breaking down complex problems into manageable sub-tasks and iteratively solving them, often learning from past experiences.")
with open(os.path.join(KB_DIR, "rag_evolution.txt"), "w") as f:
    f.write("Retrieval Augmented Generation (RAG) has evolved significantly since its inception. Initially, RAG focused on simple keyword-based or vector similarity retrieval. Modern RAG systems (as of 2026) often incorporate advanced techniques such as re-ranking retrieved documents using cross-encoders, multi-hop retrieval for complex queries requiring information from multiple sources, and query expansion to improve initial retrieval recall. Hybrid search, combining semantic and keyword search, is also common. The goal is to provide the LLM with the most precise and comprehensive context possible, minimizing the risk of hallucination and maximizing factual accuracy. Evaluation of RAG systems now includes metrics like context relevance, faithfulness, and answer correctness, often automated using smaller, specialized LLMs or human feedback loops.")
with open(os.path.join(KB_DIR, "automation_future.txt"), "w") as f:
    f.write("The future of automation is deeply intertwined with AI, moving beyond Robotic Process Automation (RPA) to intelligent process automation (IPA) and hyperautomation. AI-powered automation tools can now understand unstructured data, make predictive decisions, and adapt to changing conditions. This includes AI agents that can manage entire business processes, from customer service to supply chain optimization, by integrating with various software systems and learning from operational data. The emphasis is on creating 'lights-out' operations where human intervention is minimal, allowing employees to focus on higher-value, creative tasks. Ethical considerations and robust governance frameworks are becoming paramount as these systems gain more autonomy.")
with open(os.path.join(KB_DIR, "llm_challenges_rag_solutions.txt"), "w") as f:
    f.write("While powerful, Large Language Models (LLMs) face challenges such as factual inaccuracies (hallucinations), outdated knowledge (knowledge cutoff), and limited context windows. RAG directly addresses the first two by grounding LLM responses in external, up-to-date information. To manage context window limitations, RAG systems employ strategies like intelligent chunking (e.g., using sentence transformers to identify semantic boundaries), summarization of retrieved documents before feeding to the LLM, and multi-stage retrieval where initial retrieval identifies broad topics, and subsequent stages fetch specific details. Advanced prompt engineering also plays a role in guiding the LLM to effectively use the provided context and avoid generating information outside of it.")

# Set your OpenAI API key here. For a production system, use environment variables.
# os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY"

# --- Step 1: Load Documents ---
def load_documents(directory_path: str):
    """Loads all .txt documents from the specified directory."""
    print(f"Loading documents from {directory_path}...")
    documents = []
    for filename in os.listdir(directory_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(directory_path, filename)
            loader = TextLoader(file_path, encoding="utf-8") # Specify encoding for robustness
            documents.extend(loader.load())
    print(f"Loaded {len(documents)} documents.")
    return documents

# --- Step 2: Chunk Documents ---
def chunk_documents(documents):
    """Splits documents into smaller, overlapping chunks."""
    print("Chunking documents...")
    # RecursiveCharacterTextSplitter is highly effective for maintaining semantic coherence.
    # The chunk_size and chunk_overlap are crucial hyperparameters.
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=700, # Slightly larger chunks for more context, adjust based on content
        chunk_overlap=100, # Increased overlap to ensure context flow
        length_function=len,
        add_start_index=True, # Helps in tracing back to original document location
    )
    chunks = text_splitter.split_documents(documents)
    print(f"Created {len(chunks)} chunks.")
    return chunks

# --- Step 3: Embed and Store ---
def create_vector_store(chunks, persist_directory: str):
    """Generates embeddings for chunks and stores them in a ChromaDB vector store."""
    print("Creating embeddings and storing in vector database...")
    # HuggingFaceEmbeddings provides access to many open-source models.
    # 'BAAI/bge-small-en-v1.5' is a strong performer for general-purpose embeddings as of 2026.
    # For even better performance, consider 'BAAI/bge-large-en-v1.5' or specialized models.
    embeddings = HuggingFaceEmbeddings(model_name="BAAI/bge-small-en-v1.5")

    # Initialize ChromaDB. It's file-based and excellent for local development and testing.
    # It automatically handles persistence if `persist_directory` is provided.
    vector_store = Chroma.from_documents(
        documents=chunks,
        embedding=embeddings,
        persist_directory=persist_directory
    )
    print(f"Vector store created and persisted to {persist_directory}.")
    return vector_store

# --- Step 4 & 5: Retrieve and Generate (RAG Chain) ---
def setup_rag_chain(vector_store):
    """Sets up the Retrieval Augmented Generation chain using an LLM and retriever."""
    print("Setting up RAG chain...")
    
    # Define a custom prompt for better control over LLM behavior.
    # This prompt guides the LLM to use the provided context and be concise.
    template = """You are an AI assistant for AgenticPedia, a knowledge base about Agentic AI and automation. 
    Use the following pieces of retrieved context to answer the question. 
    If you don't know the answer, just say that you don't know, don't try to make up an answer.
    Keep the answer concise and factual.

    Context: {context}
    Question: {question}
    Answer:"""
    prompt = ChatPromptTemplate.from_template(template)

    # Initialize the LLM. Using a modern OpenAI model (e.g., gpt-4o for 2026).
    # For local LLMs, you could use `Ollama(model="llama3")` or `LlamaCpp`.
    # Ensure your API key is set or the local LLM server is running.
    llm = ChatOpenAI(model_name="gpt-4o", temperature=0.1) # Low temperature for factual consistency

    # Configure the retriever. 'search_kwargs' allows tuning retrieval parameters.
    # 'k=4' retrieves the top 4 most relevant chunks.
    retriever = vector_store.as_retriever(search_kwargs={"k": 4})

    # Create a RetrievalQA chain with a custom prompt.
    # 'stuff' chain type is simple: it stuffs all retrieved documents into the LLM's context.
    # For more complex scenarios, consider 'map_reduce', 'refine', or 'map_rerank'.
    qa_chain = RetrievalQA.from_chain_type(
        llm=llm,
        chain_type="stuff",
        retriever=retriever,
        return_source_documents=True, # Essential for verifying sources and debugging
        chain_type_kwargs={"prompt": prompt} # Pass the custom prompt to the chain
    )
    print("RAG chain ready with custom prompt.")
    return qa_chain

# --- Main Execution ---
if __name__ == "__main__":
    # Clean up previous vector store if it exists to ensure a fresh start
    if os.path.exists(PERSIST_DIR):
        shutil.rmtree(PERSIST_DIR)
        print(f"Cleaned up previous vector store at {PERSIST_DIR}.")

    # 1. Load documents from the knowledge base directory
    documents = load_documents(KB_DIR)

    # 2. Chunk the loaded documents
    chunks = chunk_documents(documents)

    # 3. Create embeddings and store them in a local ChromaDB vector store
    vector_store = create_vector_store(chunks, PERSIST_DIR)

    # 4. & 5. Set up the RAG chain for retrieval and generation
    rag_chain = setup_rag_chain(vector_store)

    print("\n--- AgenticPedia RAG System Ready for Queries ---")
    print("Type your questions about Agentic AI and automation. Type 'exit' to quit.")

    # Example Queries for demonstration
    queries = [
        "What are the key components of Agentic AI systems?",
        "How has RAG evolved, and what advanced techniques are used today?",
        "What challenges do LLMs face, and how does RAG address them?",
        "Describe the future of automation and its connection to AI.",
        "What is the capital of France?" # Out-of-context query to test robustness
    ]

    # Process example queries
    for i, query in enumerate(queries):
        print(f"\n--- Query {i+1}: {query} ---")
        
        # Check if API key is set for OpenAI, otherwise skip LLM call
        if not os.getenv("OPENAI_API_KEY") and isinstance(rag_chain.llm, ChatOpenAI):
            print("Warning: OPENAI_API_KEY not set. Cannot run LLM for this query. Please set the environment variable or configure a local LLM.")
            continue

        try:
            result = rag_chain.invoke({"query": query})
            print("\nAnswer:")
            print(result["result"])
            print("\nSource Documents (first 100 chars):")
            if result["source_documents"]:
                for j, doc in enumerate(result["source_documents"]):
                    print(f"  {j+1}. Content: {doc.page_content[:100]}... (Source: {doc.metadata.get('source', 'N/A').split('/')[-1]}, Start Index: {doc.metadata.get('start_index', 'N/A')})")
            else:
                print("  No relevant source documents found.")
        except Exception as e:
            print(f"An error occurred during query processing: {e}")
            print("Ensure your LLM API key is correctly set or your local LLM is running and accessible.")

    # Interactive loop for additional queries
    print("\n--- Interactive Query Session ---")
    while True:
        user_query = input("\nYour Query (or 'exit' to quit): ")
        if user_query.lower() == 'exit':
            break

        if not os.getenv("OPENAI_API_KEY") and isinstance(rag_chain.llm, ChatOpenAI):
            print("Warning: OPENAI_API_KEY not set. Cannot run LLM. Please set the environment variable or configure a local LLM.")
            continue

        try:
            result = rag_chain.invoke({"query": user_query})
            print("\nAnswer:")
            print(result["result"])
            print("\nSource Documents (first 100 chars):")
            if result["source_documents"]:
                for j, doc in enumerate(result["source_documents"]):
                    print(f"  {j+1}. Content: {doc.page_content[:100]}... (Source: {doc.metadata.get('source', 'N/A').split('/')[-1]}, Start Index: {doc.metadata.get('start_index', 'N/A')})")
            else:
                print("  No relevant source documents found.")
        except Exception as e:
            print(f"An error occurred during query processing: {e}")
            print("Ensure your LLM API key is correctly set or your local LLM is running and accessible.")

    print("\nThank you for using AgenticPedia RAG System!")

    # Final cleanup of the knowledge base and vector store directories
    if os.path.exists(KB_DIR):
        shutil.rmtree(KB_DIR)
        print(f"Cleaned up knowledge base directory: {KB_DIR}.")
    if os.path.exists(PERSIST_DIR):
        shutil.rmtree(PERSIST_DIR)
        print(f"Cleaned up vector store directory: {PERSIST_DIR}.")
```
