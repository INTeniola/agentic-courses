## AG02-L03: Workflow Patterns in Agentic AI Architectures

Welcome to AG02-L03, where we delve into the essential workflow patterns that dictate how intelligent agents collaborate and execute complex tasks. As software architects and AI system designers, understanding these patterns is crucial for building robust, scalable, and efficient agentic systems. Just as a well-designed city has efficient traffic flow, a well-architected agent system has clear and effective communication and task delegation mechanisms.

Agentic AI systems are rarely monolithic. Instead, they often comprise multiple specialized agents, tools, and large language models (LLMs) that need to interact seamlessly. Workflow patterns provide the blueprints for these interactions, ensuring that tasks are processed logically, efficiently, and intelligently.

We will explore four foundational patterns:

1.  **Chaining**: Sequential execution, where the output of one agent becomes the input for the next.
2.  **Routing**: Dynamic selection of the next agent or workflow based on current context or input.
3.  **Parallelisation**: Concurrent execution of multiple agents or tasks to enhance speed or explore diverse options.
4.  **Orchestrator-Subagent (Hierarchical)**: A high-level agent delegates tasks to specialized sub-agents, manages their progress, and synthesizes results.

Let's dive into each pattern with intuitive analogies and practical examples.

### 1. Chaining: The Assembly Line

**Concept**: Chaining is the simplest workflow pattern, where agents or steps are executed in a predefined sequence. The output of one step serves as the input for the subsequent step. This creates a linear flow of information and processing.

**Analogy**: Imagine an **automobile assembly line**. A bare chassis moves to Station 1 (engine installation), then to Station 2 (body panel attachment), then to Station 3 (painting), and so on. Each station performs a specific task and passes the partially completed car to the next. The order is fixed, and each step builds upon the previous one.

**Real-world Agentic Example**: A document processing pipeline. An `ExtractorAgent` extracts key information from a document, then a `SummarizerAgent` summarizes the extracted data, and finally, a `CategorizerAgent` classifies the summary. Each agent specializes in one part of the overall task.
