### API Key Management:

Many agentic applications require API access to LLMs (e.g., OpenAI, Anthropic, Google Gemini). **Never hardcode your API keys directly into your code.** We will use `python-dotenv` to manage these securely.

1.  Create a file named `.env` in the root directory of your project.
2.  Add your API keys to this file, for example:
    ```
    OPENAI_API_KEY='sk-your_openai_api_key_here'
    ANTHROPIC_API_KEY='sk-ant-api03-your_anthropic_api_key_here'
    ```
3.  Ensure your `.env` file is added to your `.gitignore` to prevent accidental commits.

Let's verify our environment setup with a simple 


````python
# Import necessary libraries
import os
from dotenv import load_dotenv

# Load environment variables from .env file
# This makes your API keys available as environment variables
load_dotenv()

# Verify that an API key is loaded (e.g., OpenAI API key)
# This check assumes you have OPENAI_API_KEY set in your .env file
openai_api_key = os.getenv("OPENAI_API_KEY")

if openai_api_key:
    print("OPENAI_API_KEY successfully loaded from .env file. Environment is ready!")
    
    # A very basic LangChain setup to confirm LLM connectivity
    try:
        from langchain_openai import ChatOpenAI
        
        # Initialize a basic ChatOpenAI model
        # We'll use a simple model for this test, e.g., gpt-3.5-turbo-0125
        llm = ChatOpenAI(model="gpt-3.5-turbo-0125", temperature=0)
        
        # Invoke the LLM with a simple prompt
        response = llm.invoke("Hello, agent world! Confirm your presence.")
        
        print("\nLLM Test Response:")
        print(response.content)
        print("\nBasic LLM connectivity confirmed. You're all set to begin!")
        
    except ImportError:
        print("LangChain or OpenAI library not found. Please ensure 'pip install langchain-openai' was run.")
    except Exception as e:
        print(f"An error occurred during LLM test: {e}")
        print("Please check your OPENAI_API_KEY and network connection.")

else:
    print("Warning: OPENAI_API_KEY not found in environment variables. Please ensure your .env file is correctly set up.")
    print("You may need to restart your kernel or notebook environment after creating/modifying .env.")

print("\n--- End of Environment Setup Check ---")
````
