## LangGraph 'Hello World' - Confirming Setup

Let's run a minimal LangGraph example to ensure everything is installed correctly and you can define and execute a simple graph. This will introduce you to the core concepts of `StateGraph`, `Node`, `Edge`, and `Compilation`.


````python
from langgraph.graph import StateGraph, END
from typing import TypedDict, Annotated
import operator

print(f"LangGraph version: {langgraph.__version__}")
print(f"LangChain version: {langchain.__version__}")

# 1. Define the Agent State
# This is how information flows between nodes in our graph.
# 'messages' will accumulate a list of strings.
class AgentState(TypedDict):
    messages: Annotated[list[str], operator.add]

# 2. Define a Node Function
# A node is a step in our graph. It takes the current state and returns updates to the state.
def greet_node(state: AgentState) -> AgentState:
    current_messages = state.get("messages", [])
    print(f"\n--- Inside 'greet_node' ---")
    print(f"Received messages: {current_messages}")
    new_message = "Graph says: Hello from LangGraph!"
    print(f"Adding message: '{new_message}'")
    return {"messages": [new_message]}

# 3. Build the Graph
# Initialize a StateGraph with our defined state.
workflow = StateGraph(AgentState)

# Add our 'greet_node' to the graph, giving it a name.
workflow.add_node("greeter", greet_node)

# Set the entry point: where the graph execution begins.
workflow.set_entry_point("greeter")

# Add an edge from 'greeter' to 'END', meaning after 'greeter', the graph finishes.
workflow.add_edge("greeter", END)

# 4. Compile the Graph
# This finalizes the graph structure and makes it ready for invocation.
app = workflow.compile()

# 5. Invoke the Graph
# Provide an initial state to start the execution.
print("\n--- Invoking the LangGraph 'Hello World' ---")
initial_state = {"messages": ["User says: Start the greeting!"]}
final_state = app.invoke(initial_state)

print("\n--- Graph Execution Complete ---")
print(f"Final state of the graph: {final_state}")

assert "Graph says: Hello from LangGraph!" in final_state["messages"], "Hello World failed!"
print("\nLangGraph 'Hello World' executed successfully! Your environment is ready.")
````
