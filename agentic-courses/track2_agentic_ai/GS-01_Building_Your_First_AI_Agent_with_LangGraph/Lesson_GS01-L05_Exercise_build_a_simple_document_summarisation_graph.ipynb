```python
from typing import TypedDict, Annotated
from langchain_core.runnables import RunnableLambda
from langgraph.graph import Graph

# --- Setup Code (repeated for self-contained solution) ---

def mock_llm_summarizer(text: str) -> str:
    """
    A mock LLM function that simulates summarization.
    """
    print(f"\n--- Mock LLM Summarizer Called ---")
    print(f"Input text length: {len(text)} characters")
    return f"This is a concise summary of the provided document, highlighting key points such as its length ({len(text)} chars) and the general topic. The original document discusses various aspects of modern AI and automation tools, emphasizing their impact on industry and daily life. This summary serves as a brief overview of the main content."

sample_document = """
In the rapidly evolving landscape of 2026, Agentic AI and automation tools are no longer futuristic concepts but integral components of enterprise infrastructure. Organizations are increasingly leveraging sophisticated AI agents to automate complex workflows, from customer service and data analysis to supply chain optimization and personalized content generation. The shift towards agentic architectures, where AI systems can autonomously plan, execute, and adapt to achieve goals, marks a significant leap from traditional reactive AI models.

LangGraph, a powerful library built on LangChain, stands at the forefront of this revolution. It provides developers with the tools to build robust, stateful, and highly controllable AI agents. By enabling the definition of computational graphs with nodes representing specific actions or LLM calls, and edges dictating the flow of control and data, LangGraph allows for the creation of intricate decision-making processes. This includes implementing advanced patterns like ReAct (Reasoning and Acting), conditional routing based on LLM outputs, and managing persistent state across multiple turns of interaction. The ability to inspect and debug these complex workflows is crucial for developing reliable and trustworthy AI systems.

As AI developers, mastering frameworks like LangGraph is essential for building the next generation of intelligent automation solutions. These tools empower us to move beyond simple prompt engineering to architect truly autonomous and adaptive AI agents that can tackle real-world challenges with unprecedented efficiency and intelligence.
"""

# --- Reference Solution ---

# 1. Define the Graph State
# We use TypedDict for simplicity, but Pydantic BaseModel is also a great choice.
class GraphState(TypedDict):
    """
    Represents the state of our graph.
    - document: The original input document string.
    - summary: The generated summary string.
    """
    document: str
    summary: Annotated[str, "The generated summary of the document"] # Using Annotated for clarity

# 2. Define the Nodes

def prepare_document_node(state: GraphState) -> GraphState:
    """
    Node to prepare the document. For this exercise, it simply passes
    the document through, but in a real scenario, it might clean, chunk,
    or reformat the text.
    """
    print("\n--- Executing prepare_document_node ---")
    document = state["document"]
    # In a more complex scenario, you might add a prompt prefix here
    # For now, we just ensure the document is in the state for the next step.
    print(f"Document prepared. Length: {len(document)} characters.")
    return {"document": document, "summary": ""} # Ensure summary is initialized or reset

def summarize_document_node(state: GraphState) -> GraphState:
    """
    Node to call the mock LLM for summarization.
    It takes the document from the state and updates the state with the summary.
    """
    print("\n--- Executing summarize_document_node ---")
    document = state["document"]
    print("Calling mock LLM summarizer...")
    generated_summary = mock_llm_summarizer(document)
    print("Mock LLM summarizer returned.")
    return {"summary": generated_summary}

# 3. Build the Graph

# Initialize the Graph builder
workflow = Graph()

# Add nodes to the workflow
# Each node is given a unique name and a function to execute.
workflow.add_node("prepare_document", prepare_document_node)
workflow.add_node("summarize_document", summarize_document_node)

# Set the entry point of the graph
# This is where the graph execution begins.
workflow.set_entry_point("prepare_document")

# Add edges to define the flow between nodes
# An edge connects the output of one node to the input of another.
workflow.add_edge("prepare_document", "summarize_document")

# Set the finish point of the graph
# This indicates where the graph execution should stop and return the final state.
workflow.set_finish_point("summarize_document")

# Compile the graph into a runnable LangChain object
# This makes the graph ready for invocation.
app = workflow.compile()

# 4. Run the Graph

print("\n--- Invoking the Document Summarization Graph ---")

# Invoke the compiled graph with the initial state.
# The initial state must contain the 'document' field as expected by 'prepare_document_node'.
final_state = app.invoke({"document": sample_document, "summary": ""})

print("\n--- Graph Execution Complete ---")
print("\nFinal State:")
print(f"Original Document (first 100 chars): {final_state['document'][:100]}...")
print(f"Generated Summary:\n{final_state['summary']}")

# You can also inspect the graph's structure (requires graphviz and pydot)
# from IPython.display import Image, display
# try:
#     display(Image(app.get_graph().draw_mermaid_png()))
# except Exception as e:
#     print(f"Could not draw graph: {e}")
#     print("Ensure you have graphviz and pydot installed: pip install 'langgraph[mermaid]' graphviz pydot")
```
