```python
import autogen
import json

# --- Configuration --- #
# Ensure you have your OAI_CONFIG_LIST file set up correctly
# Example OAI_CONFIG_LIST.json:
# [
#   {
#     "model": "gpt-4-turbo-2024-04-09",
#     "api_key": "YOUR_OPENAI_API_KEY"
#   }
# ]
config_list = autogen.config_list_from_json(
    "OAI_CONFIG_LIST",
    filter_dict={
        "model": ["gpt-4-turbo-2024-04-09", "gpt-4-0125-preview", "gpt-4-1106-preview", "gpt-4", "gpt-3.5-turbo-0125", "gpt-3.5-turbo-1106", "gpt-3.5-turbo"]
    }
)

llm_config = {
    "timeout": 600,
    "cache_seed": 42,
    "config_list": config_list,
    "temperature": 0,
    "functions": [
        {
            "name": "get_city_metric",
            "description": "Simulates fetching a specific city metric for a given area.",
            "parameters": {
                "type": "object",
                "properties": {
                    "metric_name": {
                        "type": "string",
                        "description": "The name of the metric to fetch (e.g., 'population_density', 'traffic_impact', 'green_space_percentage')."
                    },
                    "area": {
                        "type": "string",
                        "description": "The specific area or district within the city (e.g., 'Downtown', 'Northside', 'Industrial Zone')."
                    }
                },
                "required": ["metric_name", "area"]
            }
        }
    ]
}

# --- Custom Tool Definition --- #
def get_city_metric(metric_name: str, area: str) -> dict:
    """
    Simulates fetching a specific city metric for a given area.
    Args:
        metric_name (str): The name of the metric to fetch (e.g., 'population_density', 'traffic_impact', 'green_space_percentage').
        area (str): The specific area or district within the city (e.g., 'Downtown', 'Northside', 'Industrial Zone').
    Returns:
        dict: A dictionary containing the metric value and units, or an error message.
    """
    print(f"\n[TOOL CALL] DataAnalystAgent is fetching '{metric_name}' for '{area}'...")
    # Simulate different data based on area and metric
    data = {
        "Downtown": {
            "population_density": {"value": 8500, "unit": "people/sq_km"},
            "traffic_impact": {"value": "high", "unit": "level"},
            "green_space_percentage": {"value": 10, "unit": "%"}
        },
        "Northside": {
            "population_density": {"value": 3200, "unit": "people/sq_km"},
            "traffic_impact": {"value": "moderate", "unit": "level"},
            "green_space_percentage": {"value": 30, "unit": "%"}
        },
        "Industrial Zone": {
            "population_density": {"value": 500, "unit": "people/sq_km"},
            "traffic_impact": {"value": "low", "unit": "level"},
            "green_space_percentage": {"value": 5, "unit": "%"}
        }
    }

    if area in data and metric_name in data[area]:
        return {"metric": metric_name, "area": area, **data[area][metric_name]}
    elif metric_name not in ["population_density", "traffic_impact", "green_space_percentage"]:
        return {"error": f"Metric '{metric_name}' not supported. Supported metrics are: population_density, traffic_impact, green_space_percentage."}
    else:
        return {"error": f"Data for area '{area}' not available for '{metric_name}'."}

# --- Agent Definitions --- #
urban_planner_agent = autogen.UserProxyAgent(
    name="UrbanPlannerAgent",
    human_input_mode="ALWAYS", # Allows for continuous human feedback
    max_consecutive_auto_reply=0, # Ensures human input is always sought after an agent reply
    is_termination_msg=lambda x: x.get("content", "").rstrip().endswith("TERMINATE"),
    code_execution_config=False, # The planner doesn't execute code
    system_message="You are the human urban planner. Provide development proposals and give feedback on the recommendations. Type 'TERMINATE' to end the conversation."
)

data_analyst_agent = autogen.AssistantAgent(
    name="DataAnalystAgent",
    llm_config=llm_config,
    system_message="You are a data analyst specializing in urban metrics. Your role is to fetch and interpret relevant city data using the 'get_city_metric' tool to support development proposal evaluations. Provide factual data points."
)

policy_advisor_agent = autogen.AssistantAgent(
    name="PolicyAdvisorAgent",
    llm_config=llm_config,
    system_message="You are an expert in urban policy, zoning laws, and sustainability goals. Evaluate development proposals against current regulations and provide policy-related insights and concerns."
)

infrastructure_engineer_agent = autogen.AssistantAgent(
    name="InfrastructureEngineerAgent",
    llm_config=llm_config,
    system_message="You are an infrastructure engineer. Assess the technical feasibility, resource requirements, and potential strain on existing utilities, transportation, and public services for development proposals. Highlight any infrastructure challenges."
)

reviewer_agent = autogen.AssistantAgent(
    name="ReviewerAgent",
    llm_config=llm_config,
    system_message="You are a senior urban planning reviewer. Your task is to synthesize the findings from the Data Analyst, Policy Advisor, and Infrastructure Engineer. Provide a clear, structured recommendation (Approve/Reject/Revise) for the Urban Planner, including a summary, pros, and cons. Ensure the recommendation is well-justified by the specialist reports."
)

# Register the custom tool with the DataAnalystAgent
urban_planner_agent.register_function(function_map={
    "get_city_metric": get_city_metric
})
data_analyst_agent.register_function(function_map={
    "get_city_metric": get_city_metric
})

# --- Group Chat Setup --- #
# The specialized agents will collaborate first, then the reviewer takes over.
# The selector_speaker_selection_method allows the LLM to decide who speaks next based on context.

group_chat = autogen.GroupChat(
    agents=[data_analyst_agent, policy_advisor_agent, infrastructure_engineer_agent, reviewer_agent],
    messages=[],
    max_round=15, # Increased rounds for complex discussions
    speaker_selection_method="selector", # Dynamic speaker selection based on context
    allow_repeat_speaker=False # Prevent agents from speaking consecutively without new input
)

group_chat_manager = autogen.GroupChatManager(
    groupchat=group_chat,
    llm_config=llm_config,
    system_message="Manage the collaboration between Data Analyst, Policy Advisor, Infrastructure Engineer, and Reviewer to evaluate urban development proposals. Ensure all relevant aspects are covered before the Reviewer provides a final recommendation."
)

# --- Initiate Conversation --- #
# Example proposal from the Urban Planner
proposal = "I propose a new mixed-use development in the Downtown area. It will include 600 residential units, 15000 sq meters of commercial space, and a small public park. I need an assessment of its impact on population density, traffic, green space, and compliance with current zoning laws. Also, assess the strain on existing infrastructure."

print("\n--- Initiating Smart City Planning Assistant Conversation ---\n")

urban_planner_agent.initiate_chat(
    group_chat_manager,
    message=proposal
)

print("\n--- Conversation Ended ---\n")

# Example of a follow-up or refinement if human_input_mode="ALWAYS" was used:
# After the initial chat, if the UrbanPlannerAgent provides feedback, the conversation can continue.
# For instance, if the planner asks for revisions:
# urban_planner_agent.send(
#     recipient=group_chat_manager,
#     message="The recommendation is good, but can you explore options to mitigate the high traffic impact?"
# )
```
