### API Key Setup

To interact with LLMs, you'll need an API key. We recommend using OpenAI's API for this course. Follow these steps to set up your API key securely:

1.  **Obtain an API Key:** If you don't have one, sign up at [platform.openai.com](https://platform.openai.com/) and generate a new secret key.
2.  **Create a `.env` file:** In the root directory of your project (or wherever you run your Jupyter Notebooks), create a file named `.env`.
3.  **Add your API Key:** Open the `.env` file and add the following line, replacing `YOUR_OPENAI_API_KEY` with your actual key:
    ```
    OPENAI_API_KEY=YOUR_OPENAI_API_KEY
    ```
    **Important:** Never commit your `.env` file to version control (e.g., Git). Add `.env` to your `.gitignore` file.

This method keeps your sensitive keys out of your code and allows `python-dotenv` to load them as environment variables.


## Environment Check: Your First LLM Interaction

Let's perform a quick check to ensure everything is set up correctly. We'll load your API key and make a very simple call to an LLM. This "hello world" interaction will confirm your environment is ready for the exciting lessons ahead.


````python
import os
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables from .env file
load_dotenv()

# Retrieve the API key
openai_api_key = os.getenv("OPENAI_API_KEY")

# Check if the API key is loaded
if not openai_api_key:
    raise ValueError("OPENAI_API_KEY not found. Please set it in your .env file.")

# Initialize the OpenAI client
# In 2026, client initialization is typically straightforward.
client = OpenAI(api_key=openai_api_key)

print("OpenAI client initialized successfully!")

# --- Your First LLM Interaction ---
# Let's ask the LLM a simple question to confirm connectivity.

try:
    # Using a common chat model available in 2026, e.g., 'gpt-4o' or similar.
    # Adjust the model name if a newer, more capable model is preferred.
    response = client.chat.completions.create(
        model="gpt-4o", # Or 'gpt-3.5-turbo', 'claude-3-opus-20240229', 'gemini-1.5-flash'
        messages=[
            {"role": "system", "content": "You are a helpful assistant."}, # System message for context
            {"role": "user", "content": "Explain prompt engineering in one concise sentence."} # User's prompt
        ],
        max_tokens=50 # Limit response length for conciseness
    )

    # Extract and print the LLM's response
    llm_response = response.choices[0].message.content
    print("\n--- LLM Response ---")
    print(llm_response)
    print("\nEnvironment check complete! You're ready to start your prompt engineering journey.")

except Exception as e:
    print(f"\nAn error occurred during the LLM interaction: {e}")
    print("Please ensure your API key is correct and you have an active internet connection.")
    print("You might also check the OpenAI status page for any service outages.")
````
