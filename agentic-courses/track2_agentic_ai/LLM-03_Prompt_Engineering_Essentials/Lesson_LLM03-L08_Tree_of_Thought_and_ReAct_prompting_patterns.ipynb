## Tree of Thought (ToT) and ReAct Prompting Patterns: Unleashing Advanced LLM Reasoning

As developers and product builders in 2026, we're moving beyond simple one-shot prompts. The frontier of LLM interaction lies in enabling models to perform complex, multi-step reasoning and interact dynamically with their environment. This lesson dives into two pivotal advanced prompting patterns: **Tree of Thought (ToT)** and **ReAct (Reasoning and Acting)**. These techniques empower LLMs to tackle problems that require planning, exploration, and external tool use, pushing them closer to truly autonomous agents.

### 1. Tree of Thought (ToT): Navigating Complex Decisions

Imagine you're solving a complex puzzle or planning a strategic move in a game. You don't just jump to a conclusion; you explore different possibilities, evaluate their potential outcomes, and backtrack if a path seems unpromising. This is the essence of Tree of Thought.

**Core Idea:** Instead of generating a single, linear sequence of tokens (like Chain-of-Thought), ToT encourages the LLM to explore multiple *intermediate thoughts* or *reasoning steps* in parallel, forming a tree-like structure. Each 'node' in this tree represents a partial solution or a distinct line of reasoning. The LLM then evaluates these thoughts, prunes less promising branches, and expands on the most viable ones.

**Analogy:** Think of it like a sophisticated brainstorming session combined with a decision tree. You generate several ideas (thoughts), assess their feasibility, discard the weak ones, and elaborate on the strong ones, potentially generating sub-ideas, until you reach a satisfactory solution.

**How it works (Simplified):**
1.  **Decomposition:** Break down the main problem into smaller, manageable sub-problems or decision points.
2.  **Thought Generation:** For each sub-problem, prompt the LLM to generate multiple diverse *thoughts* or *potential next steps*. These are not final answers but intermediate reasoning steps.
3.  **Evaluation:** Design a mechanism (either another LLM call, a heuristic, or a human feedback loop) to evaluate the quality or promise of each generated thought. This helps in scoring or ranking the branches.
4.  **Selection & Expansion:** Select the most promising thoughts/branches. Prompt the LLM to expand on these selected thoughts, generating further sub-thoughts, effectively traversing deeper into the 'tree'.
5.  **Backtracking:** If a path leads to a dead end or a low-quality solution, the system can backtrack to a previous decision point and explore alternative branches.
6.  **Aggregation:** Once a satisfactory depth or solution is found, aggregate the best path of thoughts to form the final answer.

**Why ToT?** It significantly improves performance on tasks requiring search, planning, and strategic reasoning, such as mathematical problem-solving, creative writing, and complex logical puzzles, by allowing the LLM to self-correct and explore a broader solution space.

### 2. ReAct (Reasoning and Acting): The Agent's Internal Monologue

ReAct stands for **Reasoning and Acting**. It's a powerful pattern that enables LLMs to perform dynamic tasks by interleaving *reasoning traces* (thoughts) with *task-specific actions*. This mimics how humans often solve problems: we think about what to do, then we do it, observe the result, and then think about the next step.

**Core Idea:** An LLM agent using ReAct doesn't just generate text; it generates a sequence of `Thought`, `Action`, and `Observation` steps. The `Thought` step is the LLM's internal monologue, where it plans, analyzes, and decides. The `Action` step is where the LLM invokes an external tool or API (e.g., a search engine, a calculator, a database query, a code interpreter). The `Observation` step is the result returned by that tool, which then feeds back into the LLM's next `Thought`.

**Analogy:** Imagine a detective solving a case. They `Thought`: "I need to find the suspect's last known location." Then they take an `Action`: "Check the police database." They get an `Observation`: "Database shows address X." Then they `Thought` again: "Now I need to go to address X and look for clues." This cycle continues until the case is solved.

**How it works (Simplified):**
1.  **Initial Prompt:** The LLM receives a query or goal.
2.  **Thought:** The LLM generates a `Thought` explaining its reasoning, what it needs to do, or what tool it plans to use.
3.  **Action:** Based on the `Thought`, the LLM generates an `Action` in a predefined format (e.g., `tool_name(arguments)`). This action is then executed by an external executor.
4.  **Observation:** The external executor runs the `Action` and returns an `Observation` (the result of the tool's execution) back to the LLM.
5.  **Loop:** The LLM receives the `Observation` and uses it to generate the next `Thought`, continuing the `Thought -> Action -> Observation` loop until the goal is achieved or a stopping condition is met.

**Why ReAct?** ReAct is fundamental for building robust, interactive, and tool-augmented LLM agents. It allows LLMs to overcome their inherent limitations (e.g., lack of up-to-date information, inability to perform calculations accurately) by delegating specific tasks to specialized tools. This makes agents more reliable, factual, and capable of dynamic interaction with the real world.

### Synergy and the Future (2026)

In 2026, ToT and ReAct are often combined within sophisticated agentic frameworks. A ToT approach might be used to plan a high-level strategy, exploring different sequences of actions, while ReAct handles the execution of individual steps within that plan, leveraging tools as needed. This creates powerful, multi-modal agents capable of tackling highly complex, open-ended problems by reasoning, planning, and interacting with their environment in a structured, iterative manner. Modern LLM APIs (like Google Gemini's function calling, OpenAI's Assistants API, or Anthropic's tool use capabilities) are designed to facilitate these patterns, making them easier to implement than ever before.


```python
import json
import random
import time

# --- Mock LLM Interface (Simulating a powerful LLM in 2026) ---
# In 2026, this would be an actual API call to Google Gemini, OpenAI GPT-4o, Anthropic Claude 3.5, etc.
# For this demonstration, we'll simulate its behavior.

def mock_llm_response(prompt, model_name="Gemini-1.5-Flash", temperature=0.7):
    """Simulates an LLM's response based on keywords in the prompt."""
    time.sleep(0.1) # Simulate API latency

    if "generate thoughts" in prompt.lower():
        # Simulate ToT thought generation
        problem = prompt.split("Problem:")[1].split("Current State:")[0].strip()
        current_state = prompt.split("Current State:")[1].split("Generate 3 diverse thoughts")[0].strip()
        
        thoughts = []
        if "solve a math problem" in problem.lower():
            if "2 + 3" in current_state:
                thoughts = [
                    "Thought 1: Directly calculate 2 + 3 = 5. This is the most straightforward approach.",
                    "Thought 2: Consider if there's a trick. Perhaps it's a string concatenation? No, it's a math problem.",
                    "Thought 3: Break it down: 2 + (1+1+1) = 2+1+1+1. This is overly complex but a valid thought process."
                ]
            elif "(5 * 2) - 7" in current_state:
                 thoughts = [
                    "Thought 1: First, calculate the multiplication: 5 * 2 = 10. Then subtract 7: 10 - 7 = 3.",
                    "Thought 2: Consider order of operations (PEMDAS/BODMAS). Multiplication before subtraction.",
                    "Thought 3: What if I subtract first? (2-7) is -5. Then 5 * -5 = -25. This is incorrect, so discard."
                ]
            else:
                thoughts = [
                    "Thought 1: Identify the operations and their order of precedence.",
                    "Thought 2: Break the problem into smaller, solvable sub-expressions.",
                    "Thought 3: Consider potential pitfalls or common errors in this type of problem."
                ]
        elif "plan a trip" in problem.lower():
            if "destination: Paris" in current_state:
                thoughts = [
                    "Thought 1: Research flights and accommodation options for Paris.",
                    "Thought 2: Look into local attractions and create a preliminary itinerary.",
                    "Thought 3: Check visa requirements and travel advisories for France."
                ]
            else:
                thoughts = [
                    "Thought 1: Define the destination and travel dates.",
                    "Thought 2: Estimate budget and preferred travel style.",
                    "Thought 3: Identify key activities or experiences desired."
                ]
        else:
             thoughts = [
                f"Thought 1: Consider a direct approach for {current_state}.",
                f"Thought 2: Explore alternative interpretations or strategies for {current_state}.",
                f"Thought 3: Identify potential sub-problems within {current_state}."
            ]
        return "\n".join(thoughts)

    elif "evaluate thought" in prompt.lower():
        # Simulate ToT thought evaluation
        thought = prompt.split("Thought:")[1].split("Score this thought")[0].strip()
        if "most straightforward" in thought.lower() or "correct" in thought.lower():
            return "Score: 9/10 - Highly promising, direct and correct."
        elif "overly complex" in thought.lower() or "incorrect" in thought.lower():
            return "Score: 3/10 - Less promising, potentially inefficient or wrong."
        else:
            return "Score: 7/10 - Moderately promising, worth exploring further."

    elif "your next thought and action" in prompt.lower():
        # Simulate ReAct thought/action generation
        history = prompt.split("Current conversation history:")[1].strip()
        if "current price of AAPL" in history.lower() and "get_stock_price" not in history.lower():
            return """Thought: The user is asking for the current stock price of AAPL. I need to use a tool to get this information.
Action: get_stock_price("AAPL")"""
        elif "current price of GOOG" in history.lower() and "get_stock_price" not in history.lower():
            return """Thought: The user is asking for the current stock price of GOOG. I need to use a tool to get this information.
Action: get_stock_price("GOOG")"""
        elif "what is 15 * 3" in history.lower() and "calculate" not in history.lower():
            return """Thought: The user is asking for a mathematical calculation. I should use the calculator tool.
Action: calculate("15 * 3")"""
        elif "Observation: 45" in history:
            return """Thought: I have calculated the result. Now I can provide the answer to the user.
Action: finish("The result of 15 * 3 is 45.")"""
        elif "Observation: 175.23" in history:
            return """Thought: I have retrieved the stock price. Now I can provide the answer to the user.
Action: finish("The current price of AAPL is $175.23.")"""
        elif "Observation: 1700.50" in history:
            return """Thought: I have retrieved the stock price. Now I can provide the answer to the user.
Action: finish("The current price of GOOG is $1700.50.")"""
        else:
            return """Thought: I am not sure how to proceed. I will ask for clarification.
Action: finish("I need more information to help you with that.")"""
    
    return """Thought: I am a general purpose LLM. I can answer your question directly.
Action: finish("I don't have a specific action for this, but I can respond directly.")"""

# --- Mock Tools for ReAct (Simulating external APIs/functions) ---

def get_stock_price(symbol):
    """Mocks fetching a stock price."""
    print(f"[TOOL CALL] Fetching stock price for {symbol}...")
    if symbol.upper() == "AAPL":
        return "175.23"
    elif symbol.upper() == "GOOG":
        return "1700.50"
    else:
        return "N/A"

def calculate(expression):
    """Mocks a calculator tool."""
    print(f"[TOOL CALL] Calculating {expression}...")
    try:
        return str(eval(expression)) # WARNING: eval() is dangerous in real apps, use a safe parser
    except Exception as e:
        return f"Error: {e}"

# Define available tools for the ReAct agent
AVAILABLE_TOOLS = {
    "get_stock_price": get_stock_price,
    "calculate": calculate
}

# --- Tree of Thought (ToT) Implementation Example ---

class ToTNode:
    def __init__(self, state, thought=None, score=None, parent=None):
        self.state = state
        self.thought = thought
        self.score = score
        self.parent = parent
        self.children = []

    def __repr__(self):
        return f"Node(State: '{self.state}', Thought: '{self.thought[:30]}...', Score: {self.score})"

def run_tree_of_thought(problem_description, initial_state, max_depth=2, num_thoughts_per_step=3):
    print(f"\n--- Running Tree of Thought for: {problem_description} ---")
    root = ToTNode(state=initial_state)
    current_level_nodes = [root]
    best_path = []
    best_score = -1

    for depth in range(max_depth):
        print(f"\n--- Depth {depth + 1} ---")
        next_level_nodes = []
        for node in current_level_nodes:
            print(f"Expanding from: {node.state}")
            
            # Step 1: Generate multiple thoughts
            prompt_generate_thoughts = f"""You are an expert problem solver. Given the problem: {problem_description}
            Current State: {node.state}
            Generate {num_thoughts_per_step} diverse thoughts or next steps to progress towards the solution. Each thought should be concise.
            """
            llm_thoughts_raw = mock_llm_response(prompt_generate_thoughts)
            generated_thoughts = [t.strip() for t in llm_thoughts_raw.split('\n') if t.strip()]

            for thought_text in generated_thoughts:
                # Step 2: Evaluate each thought
                prompt_evaluate_thought = f"""You are an expert evaluator. Given the problem: {problem_description}
                Thought: {thought_text}
                Score this thought on a scale of 1-10 for its promise and relevance to solving the problem. Provide only the score and a brief justification.
                """
                llm_evaluation_raw = mock_llm_response(prompt_evaluate_thought)
                score_match = [int(s) for s in llm_evaluation_raw.split() if s.isdigit()]
                score = score_match[0] if score_match else 5 # Default score if parsing fails
                
                print(f"  - Thought: '{thought_text[:50]}...' (Score: {score})")

                # Create a new node for this thought
                new_state = f"{node.state} -> {thought_text}" # Simplistic state update
                child_node = ToTNode(state=new_state, thought=thought_text, score=score, parent=node)
                node.children.append(child_node)
                next_level_nodes.append(child_node)

                # Keep track of the best path found so far
                if score > best_score:
                    best_score = score
                    current_node = child_node
                    path = []
                    while current_node:
                        path.insert(0, current_node.thought)
                        current_node = current_node.parent
                    best_path = path[1:] # Exclude the initial None thought

        # Prune less promising branches (simplified: only keep top N nodes for next level)
        next_level_nodes.sort(key=lambda x: x.score, reverse=True)
        current_level_nodes = next_level_nodes[:num_thoughts_per_step] # Keep top N for next iteration
        if not current_level_nodes: # If no promising nodes, stop
            break

    print("\n--- ToT Final Result ---")
    print(f"Best Path (highest score {best_score}):")
    for i, thought in enumerate(best_path):
        print(f"  Step {i+1}: {thought}")
    if not best_path:
        print("No clear best path found within the given depth.")

# --- ReAct Implementation Example ---

def run_react_agent(query, max_iterations=5):
    print(f"\n--- Running ReAct Agent for Query: '{query}' ---")
    conversation_history = [f"User: {query}"]
    final_answer = None

    for i in range(max_iterations):
        print(f"\n--- ReAct Iteration {i+1} ---")
        prompt = f"""You are an AI assistant that can reason and use tools to answer questions.
        You have access to the following tools:
        {json.dumps({name: func.__doc__ for name, func in AVAILABLE_TOOLS.items()}, indent=2)}

        Current conversation history:
        {'\n'.join(conversation_history)}

        Based on the conversation history, provide your next thought and action. 
        If you have the final answer, use the 'finish' action.
        Format your response as:
        Thought: [Your reasoning]
        Action: [tool_name(arguments) or finish("final answer")]
        """
        
        llm_response = mock_llm_response(prompt)
        print(f"LLM Raw Response:\n{llm_response}")
        conversation_history.append(f"LLM: {llm_response}")

        # Parse LLM's response for Thought and Action
        thought_match = [line.replace("Thought:", "").strip() for line in llm_response.split('\n') if line.startswith("Thought:")]
        action_match = [line.replace("Action:", "").strip() for line in llm_response.split('\n') if line.startswith("Action:")]

        thought = thought_match[0] if thought_match else "No thought provided."
        action_str = action_match[0] if action_match else "No action provided."
        
        print(f"Parsed Thought: {thought}")
        print(f"Parsed Action: {action_str}")

        if action_str.startswith("finish("):
            final_answer = action_str[len("finish("):-1]
            print(f"\n--- ReAct Final Answer: {final_answer} ---")
            break
        elif "(" in action_str and ")" in action_str:
            tool_name = action_str.split('(')[0]
            tool_args_str = action_str.split('(', 1)[1][:-1] # Get content inside parentheses
            
            # Simple parsing for arguments (can be more robust for complex args)
            try:
                # Attempt to parse as JSON if it looks like a dict/list
                if tool_args_str.startswith('{') and tool_args_str.endswith('}'):
                    tool_args = json.loads(tool_args_str)
                else:
                    # Assume single string argument for simplicity
                    tool_args = tool_args_str.strip('"\'') # Remove quotes if present
            except json.JSONDecodeError:
                tool_args = tool_args_str.strip('"\'')

            if tool_name in AVAILABLE_TOOLS:
                tool_func = AVAILABLE_TOOLS[tool_name]
                if isinstance(tool_args, dict):
                    observation = tool_func(**tool_args)
                else:
                    observation = tool_func(tool_args)
                print(f"Observation: {observation}")
                conversation_history.append(f"Observation: {observation}")
            else:
                observation = f"Error: Unknown tool '{tool_name}'"
                print(observation)
                conversation_history.append(f"Observation: {observation}")
        else:
            print("Error: Invalid action format.")
            conversation_history.append("Observation: Error: Invalid action format.")
            break
    
    if not final_answer:
        print("\n--- ReAct Agent did not reach a final answer within max iterations. ---")


# --- Run the Examples ---

# Example 1: Tree of Thought for a simple math problem
run_tree_of_thought(
    problem_description="Solve the mathematical expression (5 * 2) - 7.",
    initial_state="The problem is to calculate (5 * 2) - 7.",
    max_depth=3,
    num_thoughts_per_step=2
)

# Example 2: Tree of Thought for a planning task
run_tree_of_thought(
    problem_description="Plan a short weekend trip to Paris for a couple.",
    initial_state="Starting to plan a trip to Paris.",
    max_depth=2,
    num_thoughts_per_step=3
)

# Example 3: ReAct Agent using tools
run_react_agent("What is the current price of AAPL?")

# Example 4: ReAct Agent using a different tool
run_react_agent("What is 15 * 3?")

# Example 5: ReAct Agent with a slightly more complex query (will use the same tool)
run_react_agent("Can you tell me the stock price for GOOG?")
```

### Interpreting the Code Output and Practical Considerations

The code above provides simplified, runnable examples of both Tree of Thought and ReAct patterns. Since we're using a `mock_llm_response` function, the 'reasoning' and 'actions' are pre-programmed based on keywords. In a real 2026 application, these would be dynamic outputs from a powerful LLM like Google Gemini 1.5 Flash, GPT-4o, or Claude 3.5.

#### Tree of Thought Output Interpretation:

*   **`--- Running Tree of Thought ---`**: This section shows the iterative process of exploring different reasoning paths.
*   **`--- Depth X ---`**: Each depth represents a layer in the thought tree. The LLM (mocked here) generates new thoughts based on the `Current State` of the parent node.
*   **`Expanding from: ...`**: Shows the state from which new thoughts are being generated.
*   **`Thought: ... (Score: X)`**: For each generated thought, a score is assigned (simulated by our mock LLM). This score is crucial for ToT, as it guides the pruning and selection of promising branches.
*   **`Best Path (highest score X):`**: The final output traces the sequence of thoughts that led to the highest-scoring path found within the `max_depth` and `num_thoughts_per_step` constraints.

**Performance Trade-offs for ToT:**
*   **Pros:** Significantly improves reasoning quality, especially for complex, multi-step problems. Allows for self-correction and exploration of diverse solutions.
*   **Cons:** **High computational cost and latency.** Each thought generation and evaluation step requires an LLM call, leading to many API calls. This can be expensive and slow. The `max_depth` and `num_thoughts_per_step` parameters directly impact cost and latency.
*   **Typical Use Cases:** Strategic planning, complex mathematical or logical problem-solving, creative content generation (e.g., story outlines, game design), scientific hypothesis generation, code generation requiring multiple iterations.

#### ReAct Output Interpretation:

*   **`--- Running ReAct Agent ---`**: This section demonstrates the iterative `Thought -> Action -> Observation` loop.
*   **`LLM Raw Response:`**: Shows the full output from the LLM, which contains both the `Thought` and the `Action`.
*   **`Parsed Thought: ...`**: The LLM's internal reasoning for the current step.
*   **`Parsed Action: ...`**: The specific tool call identified from the LLM's response.
*   **`[TOOL CALL] ...`**: Indicates when an external tool (like `get_stock_price` or `calculate`) is invoked.
*   **`Observation: ...`**: The result returned by the tool, which is then fed back into the LLM's context for the next iteration.
*   **`--- ReAct Final Answer: ...`**: The agent's final response after successfully completing its task using tools.

**Performance Trade-offs for ReAct:**
*   **Pros:** Enables LLMs to interact with the real world, access up-to-date information, perform accurate calculations, and overcome their inherent limitations. Makes agents more factual, reliable, and dynamic.
*   **Cons:** **Increased latency** due to external tool calls. Requires robust tool definitions and careful prompt engineering to ensure the LLM correctly identifies and uses tools. Error handling for tool failures is critical. Each `Thought`/`Action` step is an LLM call.
*   **Typical Use Cases:** Autonomous agents, question answering with dynamic information retrieval (e.g., current events, stock prices), data analysis, code execution, interacting with APIs (e.g., booking systems, CRM), customer support bots requiring external data.

### Combining ToT and ReAct

While demonstrated separately, these patterns are often combined in advanced agent architectures. For instance, a ToT process might be used to *plan* a complex sequence of actions, where each 'thought' in the tree represents a high-level goal. Once a promising path is chosen, a ReAct agent could then be employed to *execute* each step of that plan, using tools to achieve the sub-goals and gather observations. This hierarchical approach allows for both strategic depth and practical execution, forming the backbone of many sophisticated AI agents in 2026.

### Key Takeaways for Developers:

*   **Prompt Engineering is Orchestration:** These patterns highlight that prompt engineering is less about a single perfect prompt and more about orchestrating a series of LLM interactions and external tool calls.
*   **Cost and Latency are Real:** Be mindful of the increased API calls. Optimize by carefully setting `max_depth`, `num_thoughts_per_step`, and `max_iterations`.
*   **Tool Design is Crucial:** For ReAct, well-defined, reliable, and robust tools are paramount. The LLM's ability to use them effectively depends on clear descriptions and predictable outputs.
*   **Error Handling:** Implement robust error handling for both LLM failures (e.g., malformed responses) and tool failures.
*   **Observability:** Logging the `Thought`, `Action`, and `Observation` steps is vital for debugging and understanding agent behavior.

By mastering ToT and ReAct, you equip yourself with the fundamental building blocks for creating highly capable and intelligent LLM-powered agents that can reason, plan, and interact with the world effectively.
