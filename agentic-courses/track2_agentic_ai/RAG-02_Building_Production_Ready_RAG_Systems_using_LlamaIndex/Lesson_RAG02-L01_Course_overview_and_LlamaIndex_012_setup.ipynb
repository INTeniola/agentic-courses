````python
# Initialize Environment and Run a Basic LlamaIndex 'Hello World'

import os
import nest_asyncio
from dotenv import load_dotenv

# Apply nest_asyncio for Colab/Jupyter compatibility with async operations
nest_asyncio.apply()

# Load environment variables from .env file
load_dotenv()

# Set your OpenAI API key from environment variables
# Ensure you have OPENAI_API_KEY='your_key_here' in your .env file
openai_api_key = os.getenv("OPENAI_API_KEY")

if not openai_api_key:
    raise ValueError("OPENAI_API_KEY not found. Please set it in your .env file or environment variables.")

# LlamaIndex 0.12 (or equivalent) setup for basic components
# We'll use a simple LLM and embedding model for this 'hello world'.
# In 2026, LlamaIndex's API is highly modular and configurable.

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.llms.openai import OpenAI
from llama_index.embeddings.openai import OpenAIEmbedding

# Configure the LLM and Embedding Model
# For 2026, we assume 'gpt-4o' is a standard, powerful model.
# 'text-embedding-3-large' is a robust embedding model.
llm = OpenAI(model="gpt-4o", api_key=openai_api_key)
embed_model = OpenAIEmbedding(model="text-embedding-3-large", api_key=openai_api_key)

# LlamaIndex's global settings for convenience
from llama_index.core import Settings
Settings.llm = llm
Settings.embed_model = embed_model

print("LlamaIndex environment initialized with OpenAI models.")

# --- Basic 'Hello World' RAG Example ---

# 1. Create a dummy document for demonstration
# In a real scenario, you'd load from files, databases, etc.
# For this 'hello world', we'll simulate a document.

# Create a temporary directory and file for SimpleDirectoryReader
if not os.path.exists("data"):
    os.makedirs("data")

with open("data/hello_rag.txt", "w") as f:
    f.write("The capital of France is Paris. Paris is known for its Eiffel Tower and delicious croissants.")
    f.write("\nLlamaIndex is a data framework for LLM applications, enabling RAG over private or domain-specific data.")

# 2. Load documents
documents = SimpleDirectoryReader("data").load_data()
print(f"Loaded {len(documents)} document(s).")

# 3. Create an index from the documents
# This step chunks the document, embeds the chunks, and stores them in a vector store.
index = VectorStoreIndex.from_documents(documents)
print("VectorStoreIndex created successfully.")

# 4. Query the index
query_engine = index.as_query_engine()
response = query_engine.query("What is LlamaIndex used for?")

print("\n--- LlamaIndex Hello World Query Result ---")
print(f"Query: What is LlamaIndex used for?")
print(f"Response: {response}")
print("-------------------------------------------")

# Clean up the dummy file and directory
os.remove("data/hello_rag.txt")
os.rmdir("data")

print("\nEnvironment setup and basic LlamaIndex functionality confirmed!")
````
