```python
# Part 2: Capstone Project - Detailed Solution

import os
from dotenv import load_dotenv

# --- Create dummy data files (run this cell once) ---
# Ensure directories exist
os.makedirs("eng_docs", exist_ok=True)
os.makedirs("support_kb", exist_ok=True)
os.makedirs("research_papers", exist_ok=True)

# Engineering Docs (Markdown)
with open("eng_docs/api_spec.md", "w") as f:
    f.write("# AgenticCorp API Specification v2.1\n\n## Endpoints\n\n- `/users`: Manage user accounts. Supports GET, POST, PUT, DELETE.\n- `/products`: Product catalog. Supports GET, POST.\n\n## Authentication\n\nAll API calls require an `Authorization` header with a Bearer token.\n\n## Error Codes\n\n- `400 Bad Request`: Invalid input.\n- `401 Unauthorized`: Missing or invalid token.\n- `404 Not Found`: Resource not found.\n")

with open("eng_docs/design_principles.md", "w") as f:
    f.write("# AgenticCorp Design Principles\n\n1.  **Modularity**: Components should be loosely coupled.\n2.  **Scalability**: Design for horizontal scaling.\n3.  **Security First**: Implement security at every layer.\n4.  **Observability**: Integrate logging, metrics, and tracing.\n")

# Customer Support Knowledge Base (Simulated CSV/JSON)
with open("support_kb/faqs.txt", "w") as f:
    f.write("Question,Answer\n")
    f.write("How do I reset my password?,Visit the 'Forgot Password' link on the login page and follow the instructions.\n")
    f.write("What are your support hours?,Our support team is available Monday to Friday, 9 AM to 5 PM EST.\n")
    f.write("How do I contact billing?,You can reach our billing department at billing@agenticcorp.com.\n")

# External Research Papers (Mock PDF content)
# For a real PDF, you'd use pypdf to create one. Here, we'll simulate content.
# LlamaIndex's SimpleDirectoryReader can handle .txt files as well, so this works for demonstration.
with open("research_papers/rag_advances.txt", "w") as f:
    f.write("Recent Advances in Retrieval-Augmented Generation\n\nThis paper explores novel techniques for improving RAG systems, including query reformulation, multi-hop reasoning, and fine-grained document chunking. We demonstrate significant improvements in factual consistency and answer relevance on complex question-answering benchmarks. The integration of large language models with sophisticated retrieval mechanisms is key to achieving human-level performance. Future work involves integrating knowledge graphs and dynamic context window management.\n")

print("Dummy data files created/ensured.")

# Load environment variables (e.g., OPENAI_API_KEY)
load_dotenv()

# Set up OpenAI API key (uncomment and replace if not using .env)
# os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY"

# --- LlamaIndex Imports ---
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
from llama_index.core.query_engine import RouterQueryEngine
from llama_index.core.selectors import LLMSingleSelector
from llama_index.core.tools import QueryEngineTool, ToolMetadata, FunctionTool
from llama_index.agent.openai import OpenAIAgent
from llama_index.core.evaluation import FaithfulnessEvaluator, RelevancyEvaluator
from llama_index.core.retrievers import QueryFusionRetriever
from llama_index.core.response_synthesizers import ResponseSynthesizer
from llama_index.llms.openai import OpenAI
from llama_index.embeddings.openai import OpenAIEmbedding

# Configure LLM and Embedding Model (for 2026 readiness, use modern models)
# Using gpt-4o-mini for cost-effectiveness and good performance in assessments.
Settings.llm = OpenAI(model="gpt-4o-mini")
Settings.embed_model = OpenAIEmbedding(model="text-embedding-3-small")

print(f"LlamaIndex Settings: LLM={Settings.llm.model}, Embed Model={Settings.embed_model.model}")

# --- 1. Data Ingestion and Indexing (Multi-Source) ---

# 1.1. Engineering Documentation Index
print("Indexing Engineering Documentation...")
eng_docs_reader = SimpleDirectoryReader(input_dir="./eng_docs")
eng_docs = eng_docs_reader.load_data()
eng_index = VectorStoreIndex.from_documents(eng_docs)

# Apply an advanced retriever to the engineering documentation query engine
# Using QueryFusionRetriever for better handling of complex queries.
eng_retriever = QueryFusionRetriever(
    nodes=eng_index.as_retriever().get_all_nodes(),
    similarity_top_k=4,
    num_queries=4, # Generate 4 diverse queries
    mode="reciprocal_rank_fusion",
    llm=Settings.llm
)
eng_query_engine = eng_index.as_query_engine(retriever=eng_retriever)

print("Engineering Documentation Index and QueryFusionRetriever set up.")

# 1.2. Customer Support Knowledge Base Index (or Tool)
# For structured, small data like FAQs, a custom tool can be more efficient than a full RAG index.
# This tool will directly parse the 'faqs.txt' file.
print("Setting up Customer Support Tool...")
def get_support_answer(question: str) -> str:
    """Provides answers to common customer support questions from the knowledge base. """
    faqs = {}
    try:
        with open("support_kb/faqs.txt", "r") as f:
            lines = f.readlines()
            for line in lines[1:]: # Skip header
                q, a = line.strip().split(',', 1)
                faqs[q.lower()] = a
    except FileNotFoundError:
        return "Customer support knowledge base not found."

    # Simple keyword matching for demonstration
    for q_key, answer in faqs.items():
        if question.lower() in q_key:
            return answer
    return "I couldn't find a direct answer in the FAQs. Please try rephrasing or contact billing@agenticcorp.com for billing inquiries."

support_tool = FunctionTool.from_defaults(fn=get_support_answer)
print("Customer Support Tool created.")

# 1.3. External Research Papers Index
print("Indexing External Research Papers...")
research_papers_reader = SimpleDirectoryReader(input_dir="./research_papers")
research_papers_docs = research_papers_reader.load_data()
research_papers_index = VectorStoreIndex.from_documents(research_papers_docs)
research_papers_query_engine = research_papers_index.as_query_engine()
print("External Research Papers Index set up.")

# --- 2. Intelligent Query Routing ---

# Define QueryEngineTools for each index
eng_tool = QueryEngineTool(
    query_engine=eng_query_engine,
    metadata=ToolMetadata(
        name="engineering_docs",
        description="Provides information about AgenticCorp's internal engineering documentation, API specifications, and design principles."
    )
)

research_tool = QueryEngineTool(
    query_engine=research_papers_query_engine,
    metadata=ToolMetadata(
        name="research_papers",
        description="Provides information from external AI/ML research papers, focusing on advanced RAG techniques and related topics."
    )
)

# Create a RouterQueryEngine using LLMSingleSelector
# The router will decide which query engine to use based on the query.
router_query_engine = RouterQueryEngine(
    selector=LLMSingleSelector.from_defaults(),
    query_engine_tools=[
        eng_tool,
        research_tool
    ],
    llm=Settings.llm # Ensure the router uses the configured LLM
)
print("RouterQueryEngine created with engineering and research tools.")

# --- 4. Agentic Capabilities ---

# Create an OpenAIAgent with access to the router query engine and the custom support tool.
# The agent can decide to use the router for general knowledge or the specific support tool.
agent = OpenAIAgent.from_tools(
    tools=[router_query_engine, support_tool],
    llm=Settings.llm,
    verbose=True # Set to True to see agent's thought process
)
print("OpenAIAgent created with access to router and customer support tool.")

# --- 5. Basic Evaluation ---

# Define evaluators
faithfulness_evaluator = FaithfulnessEvaluator(llm=Settings.llm)
relevancy_evaluator = RelevancyEvaluator(llm=Settings.llm)

print("Evaluators initialized.")

# --- Demonstrate the system ---
print("\n--- Demonstrating System Queries ---")

# Query 1: Engineering Docs (should use eng_tool via router)
query1 = "What are the authentication requirements for AgenticCorp's API?"
print(f"\nQuery 1: {query1}")
response1 = agent.query(query1)
print(f"Agent Response: {response1}\n")

# Evaluate response 1
print("Evaluating Query 1 response...")
eval_result_faithfulness_1 = faithfulness_evaluator.evaluate_response(response=response1)
eval_result_relevancy_1 = relevancy_evaluator.evaluate_response(query=query1, response=response1)
print(f"Faithfulness Score: {eval_result_faithfulness_1.score} (Reason: {eval_result_faithfulness_1.feedback})")
print(f"Relevancy Score: {eval_result_relevancy_1.score} (Reason: {eval_result_relevancy_1.feedback})")

# Query 2: Customer Support (should use support_tool directly)
query2 = "How do I reset my password?"
print(f"\nQuery 2: {query2}")
response2 = agent.query(query2)
print(f"Agent Response: {response2}\n")

# Evaluate response 2
print("Evaluating Query 2 response...")
eval_result_faithfulness_2 = faithfulness_evaluator.evaluate_response(response=response2)
eval_result_relevancy_2 = relevancy_evaluator.evaluate_response(query=query2, response=response2)
print(f"Faithfulness Score: {eval_result_faithfulness_2.score} (Reason: {eval_result_faithfulness_2.feedback})")
print(f"Relevancy Score: {eval_result_relevancy_2.score} (Reason: {eval_result_relevancy_2.feedback})")

# Query 3: Research Papers (should use research_tool via router)
query3 = "What are some recent advances in RAG systems according to research papers?"
print(f"\nQuery 3: {query3}")
response3 = agent.query(query3)
print(f"Agent Response: {response3}\n")

# Evaluate response 3
print("Evaluating Query 3 response...")
eval_result_faithfulness_3 = faithfulness_evaluator.evaluate_response(response=response3)
eval_result_relevancy_3 = relevancy_evaluator.evaluate_response(query=query3, response=response3)
print(f"Faithfulness Score: {eval_result_faithfulness_3.score} (Reason: {eval_result_faithfulness_3.feedback})")
print(f"Relevancy Score: {eval_result_relevancy_3.score} (Reason: {eval_result_relevancy_3.feedback})")

# Query 4: Complex/Multi-hop (agent might use multiple tools or router)
query4 = "Tell me about AgenticCorp's design principles and also what are the future directions for RAG systems mentioned in recent research?"
print(f"\nQuery 4: {query4}")
response4 = agent.query(query4)
print(f"Agent Response: {response4}\n")

# Evaluate response 4 (more complex, might require manual review for full accuracy)
print("Evaluating Query 4 response...")
eval_result_faithfulness_4 = faithfulness_evaluator.evaluate_response(response=response4)
eval_result_relevancy_4 = relevancy_evaluator.evaluate_response(query=query4, response=response4)
print(f"Faithfulness Score: {eval_result_faithfulness_4.score} (Reason: {eval_result_faithfulness_4.feedback})")
print(f"Relevancy Score: {eval_result_relevancy_4.score} (Reason: {eval_result_relevancy_4.feedback})")

print("\n--- Capstone Project Complete --- ")
print("This solution demonstrates multi-source indexing, intelligent routing, advanced retrieval (QueryFusion), agentic capabilities with custom tools, and basic evaluation.")
```
