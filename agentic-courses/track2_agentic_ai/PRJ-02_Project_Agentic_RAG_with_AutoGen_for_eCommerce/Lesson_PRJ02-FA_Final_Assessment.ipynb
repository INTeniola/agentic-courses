```python
import autogen
import json
import random
import re

# --- Simulated Data (for demonstration) ---
PRODUCT_CATALOG = {
    "Zenith 4K TV 55-inch": {
        "category": "TV", "screen_size": "55-inch", "resolution": "4K UHD",
        "smart_features": "Google TV, Voice Control", "price": 799.99,
        "warranty": "2 years", "colors": ["Black"], "brand": "Zenith"
    },
    "AuraLED Smart TV 50-inch": {
        "category": "TV", "screen_size": "50-inch", "resolution": "4K UHD",
        "smart_features": "WebOS, Smart Home Integration", "price": 649.99,
        "warranty": "1 year", "colors": ["Silver", "Black"], "brand": "AuraLED"
    },
    "EcoSmart Blender Pro": {
        "category": "Kitchen Appliance", "power": "1500W", "capacity": "2L",
        "features": "Variable Speed, Pulse Function", "price": 129.99,
        "warranty": "3 years", "colors": ["Green", "White"], "brand": "EcoSmart"
    },
    "PowerMix 5000": {
        "category": "Kitchen Appliance", "power": "1200W", "capacity": "1.8L",
        "features": "Pre-set Programs, Easy Clean", "price": 99.99,
        "warranty": "1 year", "colors": ["Red", "Black"], "brand": "PowerMix"
    },
    "SonicWave Soundbar": {
        "category": "Audio", "channels": "5.1", "features": "Dolby Atmos, Bluetooth",
        "price": 299.99, "warranty": "1 year", "colors": ["Black"], "brand": "SonicWave"
    },
    "AcousticFlow Soundbar": {
        "category": "Audio", "channels": "3.1", "features": "DTS:X, HDMI eARC",
        "price": 199.99, "warranty": "1 year", "colors": ["Grey"], "brand": "AcousticFlow"
    },
    "Zenith 4K TV 65-inch": {
        "category": "TV", "screen_size": "65-inch", "resolution": "4K UHD",
        "smart_features": "Google TV, Voice Control", "price": 1099.99,
        "warranty": "2 years", "colors": ["Black"], "brand": "Zenith"
    }
}

USER_PROFILES = {
    "user_123": {
        "past_purchases": [
            {"product_name": "Zenith 4K TV 65-inch", "date": "2024-01-15"},
            {"product_name": "EcoSmart Blender Pro", "date": "2023-11-01"}
        ],
        "preferences": {"brand_loyalty": {"Zenith": 0.8}, "category_interest": {"TV": 0.9, "Audio": 0.7}}
    },
    "user_456": {
        "past_purchases": [
            {"product_name": "PowerMix 5000", "date": "2024-03-20"}
        ],
        "preferences": {"brand_loyalty": {"PowerMix": 0.7}, "category_interest": {"Kitchen Appliance": 0.8}}
    }
}

# --- Configuration for AutoGen (adjust as needed) ---
# Ensure you have an OAI_CONFIG_LIST file or environment variables set up
# Example OAI_CONFIG_LIST:
# [
#     { "model": "gpt-4o", "api_key": "YOUR_OPENAI_API_KEY" },
#     { "model": "gpt-4-turbo", "api_key": "YOUR_OPENAI_API_KEY" }
# ]
config_list = autogen.config_list_from_json(
    "OAI_CONFIG_LIST",
    filter_dict={
        "model": ["gpt-4o", "gpt-4-turbo", "gpt-3.5-turbo"], # Use modern LLMs
    },
)

llm_config = {
    "config_list": config_list,
    "temperature": 0.1,
    "timeout": 120,
}

# --- Define Custom Tools (Implemented) ---

@autogen.tool
def get_product_details(product_name: str) -> dict:
    """Retrieves all details for a given product from the catalog.
    Returns a dictionary of product attributes or an empty dict if not found.
    """
    print(f"Tool Call: get_product_details for {product_name}")
    return PRODUCT_CATALOG.get(product_name, {})

@autogen.tool
def get_user_purchase_history(user_id: str) -> list:
    """Retrieves past purchases for a user.
    Returns a list of purchase dictionaries or an empty list if not found.
    """
    print(f"Tool Call: get_user_purchase_history for {user_id}")
    return USER_PROFILES.get(user_id, {}).get("past_purchases", [])

@autogen.tool
def find_related_products(category: str, exclude_products: list = None) -> list:
    """Finds products related to a category, optionally excluding certain items.
    Returns a list of product names.
    """
    print(f"Tool Call: find_related_products for category {category}, excluding {exclude_products}")
    related = []
    exclude_products_lower = [p.lower() for p in exclude_products] if exclude_products else []
    for name, details in PRODUCT_CATALOG.items():
        if details["category"].lower() == category.lower() and name.lower() not in exclude_products_lower:
            related.append(name)
    # Simple sorting for consistent output
    return sorted(related)

# --- Define Agents ---

# User Proxy Agent: Represents the user, can execute tools.
user_proxy = autogen.UserProxyAgent(
    name="User",
    human_input_mode="NEVER", # Set to 'ALWAYS' for interactive testing
    max_consecutive_auto_reply=10,
    is_termination_msg=lambda x: x.get("content", "").rstrip().endswith("TERMINATE"),
    code_execution_config={"work_dir": "coding", "use_docker": False}, # Use docker for security in production
    llm_config=llm_config,
    system_message="""
    You are a helpful user proxy. You will initiate the conversation and then let the agents handle the task.
    If the task is completed and a recommendation is provided, say 'TERMINATE'.
    """
)

# Product Comparison Agent: Specializes in fetching product details and comparing them.
product_comparison_agent = autogen.AssistantAgent(
    name="ProductComparisonAgent",
    llm_config=llm_config,
    system_message="""
    You are an expert in comparing e-commerce products. Your role is to:
    1. Identify products and attributes to compare from the user's query.
    2. Use the `get_product_details` tool to retrieve information for each product.
    3. Systematically compare the requested attributes and present the findings clearly in a markdown table.
    4. Once the comparison is complete, clearly state the comparison results and then explicitly ask the RecommendationAgent to provide a recommendation, passing along any relevant context like user_id or product categories.
    """
)

# Recommendation Agent: Specializes in personalized product recommendations.
recommendation_agent = autogen.AssistantAgent(
    name="RecommendationAgent",
    llm_config=llm_config,
    system_message="""
    You are an expert in personalized e-commerce recommendations. Your role is to:
    1. Receive the comparison results and the user's request for a recommendation from the ProductComparisonAgent.
    2. Extract the user_id from the conversation if available (e.g., "assume I am user_123").
    3. Use `get_user_purchase_history` to understand user preferences.
    4. Identify relevant product categories from the comparison or past purchases.
    5. Use `find_related_products` to suggest a relevant product based on past purchases or the context of the comparison, ensuring it's not one of the compared products.
    6. Provide a single, clear, and personalized recommendation in a friendly tone.
    7. Conclude the conversation with 'TERMINATE' after providing the recommendation.
    """
)

# --- Register Tools with Agents ---
# All agents that need to *call* a tool should have it registered for execution.
# All agents that need to *know about* a tool (for planning) should have it registered for LLM.

# User_proxy can execute tools on behalf of other agents
user_proxy.register_for_execution(get_product_details)
user_proxy.register_for_execution(get_user_purchase_history)
user_proxy.register_for_execution(find_related_products)

# ProductComparisonAgent needs to know about get_product_details to plan its actions
product_comparison_agent.register_for_llm(get_product_details)

# RecommendationAgent needs to know about get_user_purchase_history and find_related_products
recommendation_agent.register_for_llm(get_user_purchase_history)
recommendation_agent.register_for_llm(find_related_products)

# --- Create Group Chat ---
# Define the group chat and manager.
# Ensure the agents interact in a logical flow (e.g., comparison first, then recommendation).

groupchat = autogen.GroupChat(
    agents=[user_proxy, product_comparison_agent, recommendation_agent],
    messages=[],
    max_round=20, # Increased max_round for complex interactions
    speaker_selection_method="auto", # Let AutoGen decide who speaks next
    allow_repeat_speaker=False # Prevent agents from speaking consecutively without a response from another
)

manager = autogen.GroupChatManager(groupchat=groupchat, llm_config=llm_config)

# --- Initiate the Conversation ---
user_query = "Compare the 'Zenith 4K TV 55-inch' with the 'AuraLED Smart TV 50-inch' on screen size, resolution, and smart features. After that, recommend a soundbar based on my past purchases (assume I am user_123)."

print("\n--- Starting Conversation ---\n")
user_proxy.initiate_chat(manager, message=user_query)
print("\n--- Conversation Ended ---\n")
```
