## Tool Definition and JSON Schema for Function Calling

In the rapidly evolving landscape of AI agents, the ability to interact with the external world is paramount. While Large Language Models (LLMs) are incredibly powerful at understanding and generating human-like text, their knowledge is inherently limited to their training data. They cannot, for instance, fetch real-time weather, search the live internet, or send an email without external assistance.

This is where **tools** come into play. Think of an LLM as a brilliant, highly articulate brain, and tools as its specialized hands or gadgets. These tools allow the agent to perform actions, retrieve up-to-date information, or interact with APIs and services beyond its internal knowledge base. The process by which an LLM decides to use a tool and provides the necessary inputs is often called **function calling**.

### What is a Tool?

At its core, a tool is a function or an API endpoint that an AI agent can invoke. For an LLM to effectively use a tool, it needs a clear understanding of:

1.  **What the tool does:** A concise, human-readable description.
2.  **How to call it:** Its name.
3.  **What information it needs:** Its input parameters.

### The Role of JSON Schema

The most critical aspect of defining *what information a tool needs* is the use of **JSON Schema**. JSON Schema is a powerful, standardized way to describe the structure and constraints of JSON data. When defining tool parameters, we use JSON Schema to specify:

*   The expected data types (e.g., `string`, `number`, `boolean`, `object`, `array`).
*   Whether a parameter is `required` or optional.
*   Specific formats (e.g., `email`, `date-time`).
*   Value constraints (e.g., `minimum`, `maximum` for numbers, `minLength`, `maxLength` for strings).

**Why JSON Schema?**

Imagine you're giving instructions to a highly intelligent but literal assistant. If you just say, "Get me the weather for London," the assistant might ask, "Which London? What date? Do you want temperature in Celsius or Fahrenheit?" With JSON Schema, you provide a precise instruction manual:

```json
{
  "name": "get_current_weather",
  "description": "Get the current weather in a given location",
  "parameters": {
    "type": "object",
    "properties": {
      "location": {
        "type": "string",
        "description": "The city and state, e.g. San Francisco, CA"
      },
      "unit": {
        "type": "string",
        "enum": ["celsius", "fahrenheit"],
        "description": "The unit of temperature to use. Defaults to celsius."
      }
    },
    "required": ["location"]
  }
}
```

This schema tells the LLM:

*   The tool is `get_current_weather`.
*   It needs a `location` (which must be a string and is required).
*   It can optionally take a `unit` (which must be either `celsius` or `fahrenheit`).

When an LLM is presented with this definition and a user query like "What's the weather in Tokyo?", it can then generate a structured function call like `get_current_weather(location="Tokyo")`. The JSON Schema acts as a contract, ensuring the LLM provides valid and correctly typed arguments, which is crucial for robust agent execution. Modern LLM APIs (like OpenAI's Function Calling, Google's Gemini Function Calling, and Anthropic's Tools) natively understand and leverage these JSON Schema definitions to enable powerful agentic behaviors.


````python
import json

# --- 1. Define Tools using JSON Schema for parameters ---

# Tool 1: Get current weather
weather_tool_definition = {
    "name": "get_current_weather",
    "description": "Get the current weather in a given location",
    "parameters": {
        "type": "object",
        "properties": {
            "location": {
                "type": "string",
                "description": "The city and state, e.g. San Francisco, CA"
            },
            "unit": {
                "type": "string",
                "enum": ["celsius", "fahrenheit"],
                "description": "The unit of temperature to use. Defaults to celsius."
            }
        },
        "required": ["location"]
    }
}

# Tool 2: Search the web
web_search_tool_definition = {
    "name": "search_web",
    "description": "Searches the internet for a given query and returns relevant results.",
    "parameters": {
        "type": "object",
        "properties": {
            "query": {
                "type": "string",
                "description": "The search query to execute."
            },
            "num_results": {
                "type": "integer",
                "description": "The maximum number of search results to return. Defaults to 5.",
                "minimum": 1,
                "maximum": 10
            }
        },
        "required": ["query"]
    }
}

# Tool 3: Send an email
email_tool_definition = {
    "name": "send_email",
    "description": "Sends an email to one or more recipients.",
    "parameters": {
        "type": "object",
        "properties": {
            "to": {
                "type": "array",
                "items": {
                    "type": "string",
                    "format": "email"
                },
                "description": "A list of recipient email addresses."
            },
            "subject": {
                "type": "string",
                "description": "The subject line of the email."
            },
            "body": {
                "type": "string",
                "description": "The body content of the email."
            },
            "attachments": {
                "type": "array",
                "items": {
                    "type": "string"
                },
                "description": "Optional list of file paths to attach."
            }
        },
        "required": ["to", "subject", "body"]
    }
}

# Aggregate all tool definitions
all_tool_definitions = [
    weather_tool_definition,
    web_search_tool_definition,
    email_tool_definition
]

print("--- Defined Tools ---")
for tool_def in all_tool_definitions:
    print(f"Name: {tool_def['name']}")
    print(f"Description: {tool_def['description']}")
    print(f"Parameters (JSON Schema):\n{json.dumps(tool_def['parameters'], indent=2)}\n")

# --- 2. Simulate LLM Function Calling ---

# In a real scenario, an LLM would receive a user query and the tool definitions.
# Based on these, it would decide if a tool needs to be called and generate the call.
# We'll simulate this output for a few different user queries.

print("\n--- Simulated LLM Function Calls ---")

def simulate_llm_call(user_query, available_tools):
    print(f"User Query: '{user_query}'")
    # This is where a real LLM API call would happen.
    # For demonstration, we'll hardcode some expected outputs.

    if "weather in London" in user_query.lower():
        # LLM decides to call get_current_weather
        tool_call = {
            "tool_name": "get_current_weather",
            "arguments": {
                "location": "London, UK",
                "unit": "celsius"
            }
        }
    elif "search for AI agent frameworks" in user_query.lower():
        # LLM decides to call search_web
        tool_call = {
            "tool_name": "search_web",
            "arguments": {
                "query": "AI agent frameworks 2026",
                "num_results": 3
            }
        }
    elif "send an email to alice" in user_query.lower():
        # LLM decides to call send_email
        tool_call = {
            "tool_name": "send_email",
            "arguments": {
                "to": ["alice@example.com"],
                "subject": "Project Update",
                "body": "Hi Alice, the project is on track. More details soon."
            }
        }
    else:
        tool_call = None # LLM decides no tool is needed or generates a text response

    if tool_call:
        print(f"LLM generated tool call:\n{json.dumps(tool_call, indent=2)}\n")
        return tool_call
    else:
        print("LLM decided not to call a tool or generated a text response.\n")
        return None

# --- 3. Simulate Tool Execution ---

def execute_tool_call(tool_call):
    if not tool_call:
        return "No tool call to execute."

    tool_name = tool_call["tool_name"]
    arguments = tool_call["arguments"]

    print(f"Executing tool: {tool_name} with arguments: {arguments}")

    if tool_name == "get_current_weather":
        location = arguments.get("location")
        unit = arguments.get("unit", "celsius")
        # In a real scenario, this would call an external weather API
        return f"Simulating weather for {location}: 20 degrees {unit}, sunny."
    elif tool_name == "search_web":
        query = arguments.get("query")
        num_results = arguments.get("num_results", 5)
        # In a real scenario, this would call a search engine API
        return f"Simulating web search for '{query}' (top {num_results} results): [Result 1, Result 2, ...]"
    elif tool_name == "send_email":
        to = arguments.get("to")
        subject = arguments.get("subject")
        body = arguments.get("body")
        # In a real scenario, this would use an email sending library/API
        return f"Simulating email sent to {', '.join(to)} with subject '{subject}'."
    else:
        return f"Unknown tool: {tool_name}"

# Run simulations
queries = [
    "What's the weather like in London right now?",
    "Can you search for the latest AI agent frameworks in 2026, just 3 results?",
    "Send an email to alice@example.com with the subject 'Project Update' and body 'Hi Alice, the project is on track. More details soon.'",
    "Tell me a joke."
]

for query in queries:
    call = simulate_llm_call(query, all_tool_definitions)
    if call:
        result = execute_tool_call(call)
        print(f"Tool execution result: {result}\n")
````

### Interpreting the Code Output and Use Cases

The code demonstrates the full lifecycle of tool definition and simulated function calling:

1.  **Tool Definitions:** We first define three distinct tools (`get_current_weather`, `search_web`, `send_email`). Each definition includes a `name`, a `description` (crucial for the LLM to understand its purpose), and a `parameters` field. The `parameters` field is where **JSON Schema** shines. Notice how it specifies the `type` of the overall parameters object (`"type": "object"`), then lists individual `properties` (like `location`, `unit`, `query`, `to`, `subject`, `body`). For each property, we define its `type`, `description`, and sometimes `enum` (for allowed values) or `minimum`/`maximum` (for numerical constraints). The `required` array explicitly lists which parameters *must* be provided by the LLM.

2.  **Simulated LLM Function Calls:** The `simulate_llm_call` function mimics what a modern LLM API would do. Given a user query and the list of available tool definitions, the LLM's internal reasoning engine (which is trained on vast amounts of text and code) determines if any of the tools are relevant. If so, it generates a structured JSON object containing the `tool_name` and an `arguments` dictionary. Crucially, the LLM attempts to populate these `arguments` according to the JSON Schema provided in the tool definition. For example, when asked about "weather in London," the LLM correctly identifies `get_current_weather` and extracts "London, UK" for the `location` parameter.

3.  **Simulated Tool Execution:** The `execute_tool_call` function represents the agent's runtime. After the LLM generates a tool call, the agent takes this structured output and invokes the corresponding Python function or API. Our simulation simply returns a descriptive string, but in a real application, this would involve making an HTTP request to a weather API, interacting with a search engine, or sending an actual email via an SMTP client.

### Benefits and Trade-offs

**Benefits of JSON Schema for Tool Definition:**

*   **Robustness and Validation:** JSON Schema acts as a strict contract. It ensures that the arguments generated by the LLM conform to the expected types and constraints, preventing runtime errors in your tool functions. If the LLM tries to pass a string where a number is expected, the schema helps catch this. This is vital for building reliable agents.
*   **Clarity and Discoverability:** The `description` fields within the schema provide clear instructions to the LLM about what each parameter means and how it should be used. This improves the LLM's ability to correctly invoke tools.
*   **Interoperability:** JSON Schema is a widely adopted standard, making tool definitions portable across different LLM providers and agent frameworks.
*   **Reduced Hallucinations:** By providing a precise structure, the LLM is less likely to hallucinate incorrect parameter names or values.

**Performance Trade-offs:**

*   **Token Usage:** More complex tool definitions (with many parameters, nested objects, or extensive descriptions) consume more tokens when passed to the LLM. This can increase API costs and inference latency, especially for models with smaller context windows.
*   **LLM Reasoning Time:** While generally efficient, the LLM needs to parse the tool definitions and reason about which tool to use and how to populate its arguments. Extremely large numbers of tools or very intricate schemas can slightly increase the time it takes for the LLM to generate a tool call.

### Typical Use Cases for Function Calling:

*   **Real-time Information Retrieval:** Fetching current stock prices, news headlines, sports scores, or weather data.
*   **Data Manipulation and Calculation:** Performing complex calculations, querying databases, or transforming data using external libraries.
*   **External System Control:** Sending emails, scheduling calendar events, managing CRM records, controlling IoT devices, or interacting with enterprise software.
*   **Multi-step Workflows:** Orchestrating a sequence of tool calls to achieve a complex goal (e.g., "Find flights, book a hotel, and send me the itinerary").
*   **Personalized Experiences:** Accessing user-specific data from APIs to provide tailored responses or actions.
