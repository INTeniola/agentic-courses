## Prerequisites and Environment Setup

Before we embark on this journey, let's ensure you have the necessary foundation and a properly configured development environment.

### Prerequisites

*   **AG-02: Advanced Prompt Engineering & LLM Interaction:** A solid understanding of prompt engineering, LLM capabilities, and interaction patterns is crucial. We will build upon these concepts, assuming familiarity with effective LLM communication.
*   **Python Proficiency:** This course is heavily code-centric, utilizing Python 3.10+ for all implementations. You should be comfortable with Python syntax, object-oriented programming, and common data structures.
*   **Command Line Familiarity:** Basic command-line operations (e.g., `git clone`, `pip install`, `cd`) will be used frequently.

### Environment Setup

We recommend setting up a dedicated virtual environment for this course to manage dependencies cleanly. You can use `venv` (built-in Python module) or `conda`.

1.  **Clone the Course Repository:**
    All course materials, including notebooks, code examples, and `requirements.txt` files, are hosted in our dedicated GitHub repository. Open your terminal or command prompt and execute:
    ```bash
    git clone https://github.com/AgenticLabs/AG03-Building-AI-Agents-from-Scratch.git
    cd AG03-Building-AI-Agents-from-Scratch
    ```

2.  **Create and Activate a Virtual Environment:**
    *   **Using `venv` (recommended for most Python users):**
        ```bash
        python -m venv .venv
        # On macOS/Linux:
        source .venv/bin/activate
        # On Windows (Command Prompt):
        .venv\Scripts\activate.bat
        # On Windows (PowerShell):
        .venv\Scripts\Activate.ps1
        ```
    *   **Using `conda` (if you prefer Anaconda/Miniconda):**
        ```bash
        conda create -n ag03 python=3.11 -y
        conda activate ag03
        ```

3.  **Install Dependencies:**
    With your virtual environment activated, install all required Python packages:
    ```bash
    pip install -r requirements.txt
    ```

4.  **Jupyter Notebook/Lab:**
    Ensure you have Jupyter Notebook or JupyterLab installed within your environment. If not:
    ```bash
    pip install jupyterlab
    ```
    Then, launch Jupyter from the course repository root:
    ```bash
    jupyter lab
    ```
