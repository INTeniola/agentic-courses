4.  **Input Topic:** The crew should generate an article on the topic: "The Future of AI in Healthcare by 2030: Innovations and Ethical Challenges."

```python
# Ensure you have the necessary libraries installed:
# pip install crewai crewai-tools openai 'crewai[tools]'

import os
from crewai import Agent, Task, Crew, Process
from crewai_tools import SerperDevTool, BrowserbaseTool, FileTool
from langchain_openai import ChatOpenAI

# --- Set up Environment Variables ---
# Replace with your actual API keys or set them as environment variables
# For example, in your .env file or directly in your environment:
# OPENAI_API_KEY="your_openai_api_key"
# SERPER_API_KEY="your_serper_api_key"
# BROWSERBASE_API_KEY="your_browserbase_api_key"

# Load environment variables (if using python-dotenv)
# from dotenv import load_dotenv
# load_dotenv()

# Check if API keys are set
if not os.getenv("OPENAI_API_KEY"):
    raise ValueError("OPENAI_API_KEY not found. Please set it as an environment variable.")
if not os.getenv("SERPER_API_KEY"):
    raise ValueError("SERPER_API_KEY not found. Please set it as an environment variable.")
# Browserbase is optional for this exercise, but good to include for robust research
# if not os.getenv("BROWSERBASE_API_KEY"):
#     print("BROWSERBASE_API_KEY not found. BrowserbaseTool will not be available.")

# --- Initialize LLM ---
# Using a modern, cost-effective model for demonstration
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)

# --- Initialize Tools ---
# SerperDevTool for web search
serper_tool = SerperDevTool()

# BrowserbaseTool for advanced web browsing (optional, requires BROWSERBASE_API_KEY)
# browser_tool = BrowserbaseTool()

# FileTool for saving content (optional, but useful for persistent output)
file_tool = FileTool()

print("Setup complete. You can now proceed to define your agents, tasks, and crew.")

# --- Placeholder for student's implementation ---
# You will write your CrewAI agents, tasks, and crew definition here.
```

### Your Implementation

Now it's your turn! Based on the requirements outlined in the task description, implement the Researcher, Writer, and Editor agents, define their respective tasks, and assemble them into a CrewAI pipeline. Your goal is to generate a comprehensive and polished article on the specified topic.

Remember to:

*   Instantiate the `Crew` with `sequential` process and `verbose=True`.

```python
# --- Solution Code ---

# 1. Define Agents

# Researcher Agent
researcher = Agent(
    role='Senior Research Analyst',
    goal='Conduct comprehensive and accurate research on the given topic, identifying key facts, statistics, and trends.',
    backstory=(
        "A meticulous and detail-oriented researcher with a knack for uncovering hidden gems of information. "
        "Specializes in synthesizing complex data into digestible insights. "
        "Always strives for factual accuracy and breadth of coverage."
    ),
    verbose=True,
    allow_delegation=False, # Researcher focuses solely on research
    tools=[serper_tool], # Only web search is essential for research
    llm=llm
)

# Writer Agent
writer = Agent(
    role='Professional Content Strategist & Writer',
    goal='Draft an engaging, well-structured, and informative article based on the research findings, tailored for a general audience.',
    backstory=(
        "An experienced content creator known for transforming raw data into compelling narratives. "
        "Excels at clarity, conciseness, and audience engagement, ensuring the message resonates with readers. "
        "Always aims for a captivating and easy-to-understand style."
    ),
    verbose=True,
    allow_delegation=False, # Writer focuses solely on writing
    llm=llm
)

# Editor Agent
editor = Agent(
    role='Senior Editorial Reviewer',
    goal='Review, refine, and polish the drafted article for grammar, style, factual accuracy, coherence, and overall quality, ensuring it meets publication standards.',
    backstory=(
        "A seasoned editor with an eagle eye for detail and a deep understanding of journalistic integrity. "
        "Ensures every piece is flawless before publication, focusing on clarity, tone, and factual correctness. "
        "The final gatekeeper of quality."
    ),
    verbose=True,
    allow_delegation=False, # Editor focuses solely on editing
    llm=llm,
    tools=[file_tool] # Editor can save the final output
)

# 2. Define Tasks

# Research Task
research_task = Task(
    description=(
        "Conduct in-depth research on 'The Future of AI in Healthcare by 2030: Innovations and Ethical Challenges'. "
        "Identify key technological innovations (e.g., AI diagnostics, personalized medicine, drug discovery), "
        "potential benefits, and significant ethical considerations (e.g., data privacy, bias, accountability). "
        "Summarize findings into a comprehensive report, including statistics and expert opinions where available. "
        "Focus on providing actionable insights for a general audience."
    ),
    expected_output=(
        "A detailed research report, approximately 800-1000 words, covering key innovations, benefits, "
        "and ethical challenges of AI in healthcare by 2030. "
        "The report should be well-structured with clear headings and bullet points, ready for a writer to use."
    ),
    agent=researcher
)

# Writing Task
writing_task = Task(
    description=(
        "Using the research report provided, draft a compelling and informative article "
        "on 'The Future of AI in Healthcare by 2030: Innovations and Ethical Challenges'. "
        "The article should be engaging for a general audience, approximately 1200-1500 words. "
        "Structure it with an introduction, sections for innovations, benefits, ethical challenges, and a conclusion. "
        "Ensure a clear, concise, and persuasive writing style."
    ),
    expected_output=(
        "A well-structured, engaging, and informative article, approximately 1200-1500 words, "
        "ready for editorial review. It should flow logically and address all aspects of the topic."
    ),
    agent=writer,
    context=[research_task] # The output of the research task feeds into the writing task
)

# Editing Task
editing_task = Task(
    description=(
        "Review and meticulously edit the drafted article for grammar, spelling, punctuation, style, and factual accuracy. "
        "Ensure coherence, logical flow, and consistency in tone. "
        "Verify that all claims are supported by the research context and that the article meets high publication standards. "
        "Refine headings, transitions, and overall readability. "
        "Finally, save the polished article to a file named 'ai_healthcare_2030_article.md'."
    ),
    expected_output=(
        "A final, polished, and publication-ready article on 'The Future of AI in Healthcare by 2030: "
        "Innovations and Ethical Challenges', saved as 'ai_healthcare_2030_article.md'. "
        "The article should be flawless in grammar, style, and factual accuracy."
    ),
    agent=editor,
    context=[writing_task], # The output of the writing task feeds into the editing task
    output_file='ai_healthcare_2030_article.md' # Editor saves the final output
)

# 3. Configure and Run the Crew

# Instantiate your crew with a sequential process
content_crew = Crew(
    agents=[researcher, writer, editor],
    tasks=[research_task, writing_task, editing_task],
    process=Process.sequential, # Tasks run in a predefined order
    verbose=True # Enable verbose output to see agent thought processes
)

# Kick off the crew's work
print("\n### Starting the Content Generation Crew ###\n")
result = content_crew.kickoff()

print("\n### Content Generation Complete! ###\n")
print("Final Article Output:")
print(result)

# You can also verify the file was saved
if os.path.exists('ai_healthcare_2030_article.md'):
    print("\nArticle successfully saved to 'ai_healthcare_2030_article.md'")
    with open('ai_healthcare_2030_article.md', 'r') as f:
        print("\n--- Content of saved file ---")
        print(f.read()[:500]) # Print first 500 characters to confirm content
        print("\n--- End of file content preview ---")
else:
    print("\nError: Article file was not saved.")
```
