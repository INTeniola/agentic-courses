## Exercise: Build a Two-Agent Research Crew

In this exercise, you will apply the foundational concepts of CrewAI to build a simple, yet effective, two-agent research crew. Your goal is to create a system that can research a given topic and synthesize the findings into a concise report.

### Task Description

Your mission is to design and implement a CrewAI system consisting of two agents: a **Researcher** and a **Report Writer**. The crew should be able to take a research topic (e.g., "The impact of AI on the global job market by 2030") and produce a well-structured summary report.

### Requirements

1.  **Two Agents**: Define two distinct agents:
    *   **Researcher Agent**: Responsible for gathering information on the given topic.
    *   **Report Writer Agent**: Responsible for synthesizing the research findings and drafting a coherent report.
2.  **Distinct Roles, Goals, and Backstories**: Each agent must have a unique `role`, `goal`, and `backstory` that clearly defines its purpose and expertise.
3.  **At Least Two Tasks**: Define at least two tasks that leverage these agents:
    *   A research task assigned to the `Researcher` agent.
    *   A report writing/synthesis task assigned to the `Report Writer` agent.
4.  **Sequential Task Flow**: Ensure the tasks are executed in a logical sequence, where the output of the research task feeds into the report writing task.
5.  **Clear Output**: The final output of the crew should be a well-formatted string containing the research report.
6.  **LLM Integration**: Use a modern LLM (e.g., `OpenAIChat` with a recent model like `gpt-4o-2024-05-13`). Ensure your API key is loaded securely via environment variables.

### Evaluation Criteria

*   **Correct Agent Definition**: Agents are properly instantiated with `role`, `goal`, and `backstory`.
*   **Logical Task Assignment**: Tasks are assigned to the most appropriate agent based on their defined roles.
*   **Effective Task Chaining**: The output of the first task is successfully used as input for the subsequent task.
*   **Runnable Code**: The provided solution is complete and runs without errors.
*   **Clear and Concise Output**: The final report generated by the crew is relevant to the topic and well-structured.


```python
import os
from crewai import Agent, Task, Crew, Process
from langchain_openai import ChatOpenAI

# --- Configuration --- 
# Ensure your OpenAI API key is set as an environment variable.
# For example, in your shell: export OPENAI_API_KEY='your_api_key_here'
# Or, if running in a notebook, you can set it directly (for development only, not recommended for production):
# os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY"

if not os.getenv("OPENAI_API_KEY"):
    raise ValueError("OPENAI_API_KEY environment variable not set. Please set it before running the crew.")

# Initialize the LLM for all agents
# Using a modern, capable model for better performance in 2026.
llm = ChatOpenAI(model="gpt-4o-2024-05-13", temperature=0.7)

print("Setup complete. LLM initialized and API key check passed.")
```

### Your Implementation

Now it's your turn! In the code cell below, implement your two-agent research crew based on the requirements outlined above. Remember to define your agents, tasks, and then assemble and run the crew. Choose a research topic of your interest to make it more engaging!


```python
# --- Define Agents ---
# 1. Researcher Agent
researcher = Agent(
    role='Senior Research Analyst',
    goal='Uncover comprehensive and up-to-date information on the given topic.',
    backstory='An expert in data retrieval and analysis, skilled at finding relevant information from various sources and synthesizing key insights.',
    verbose=True,
    allow_delegation=False, # For a simple two-agent crew, delegation might not be necessary
    llm=llm
)

# 2. Report Writer Agent
report_writer = Agent(
    role='Professional Report Writer',
    goal='Draft a clear, concise, and well-structured report based on research findings.',
    backstory='A seasoned writer with a knack for transforming complex information into easily digestible and engaging reports, ensuring accuracy and clarity.',
    verbose=True,
    allow_delegation=False,
    llm=llm
)

# --- Define Tasks ---
research_topic = "The future of quantum computing in cybersecurity by 2030"

# Task 1: Research the topic
research_task = Task(
    description=f"Conduct a thorough investigation into '{research_topic}'. "
                "Identify key advancements, potential applications, challenges, and expert predictions. "
                "Focus on developments expected by 2030. "
                "Compile your findings into a detailed bullet-point summary, citing potential sources if applicable.",
    expected_output='A comprehensive bullet-point summary of key findings regarding the future of quantum computing in cybersecurity by 2030.',
    agent=researcher
)

# Task 2: Write the report based on research findings
# This task will receive the output of the research_task as its input implicitly
report_writing_task = Task(
    description=f"Based on the research findings provided, draft a professional, 500-word report on '{research_topic}'. "
                "The report should include an introduction, key findings, potential impacts/challenges, and a conclusion. "
                "Ensure the language is clear, concise, and suitable for a business audience.",
    expected_output='A well-structured, 500-word report summarizing the future of quantum computing in cybersecurity by 2030, based on the provided research.',
    agent=report_writer
)

# --- Assemble the Crew ---
# The tasks are ordered sequentially. The output of research_task will be passed to report_writing_task.
research_crew = Crew(
    agents=[researcher, report_writer],
    tasks=[research_task, report_writing_task],
    process=Process.sequential, # Tasks run in the order they are defined
    verbose=2 # Higher verbosity for detailed logs
)

# --- Run the Crew ---
print("\n### Initiating Research Crew...\n")
result = research_crew.kickoff()

print("\n### Research Crew Finished!\n")
print("\n### Final Report:\n")
print(result)
```
