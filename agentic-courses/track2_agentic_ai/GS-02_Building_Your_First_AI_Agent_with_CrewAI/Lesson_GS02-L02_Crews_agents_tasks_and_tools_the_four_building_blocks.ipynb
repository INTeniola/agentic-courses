## Exercise: Building Your First CrewAI Application

### Objective

In this exercise, you will build a simple CrewAI application that can generate a concise, engaging blog post outline on a given topic. This will solidify your understanding of Agents, Tasks, Tools, and Crews.

### Scenario

You are tasked with creating an automated workflow to assist a content marketing team. The workflow should take a blog post topic as input and produce a structured outline, ready for a human writer to expand upon.

### Requirements

1.  **Define at least two Agents**:
    *   A `Research Agent` responsible for gathering information on the given topic.
    *   An `Outline Architect Agent` responsible for structuring the gathered information into a blog post outline.
    *   Each agent must have a distinct `role`, `goal`, and `backstory`.
    *   Ensure agents are configured for `verbose` output to observe their thought process.

2.  **Define at least two Tasks**:
    *   A `Research Task` assigned to the `Research Agent` that focuses on finding key points and facts about the topic.
    *   An `Outline Task` assigned to the `Outline Architect Agent` that takes the research output and crafts a blog post outline.
    *   Each task must have a clear `description` and `expected_output`.

3.  **Define at least one Custom Tool**:
    *   Create a simple mock `SearchTool` that simulates searching for information. This tool will be used by the `Research Agent`.

4.  **Assemble and Execute the Crew**:
    *   Combine your defined Agents, Tasks, and Tool into a `Crew`.
    *   Configure the `Crew` to use a `sequential` process.
    *   Execute the `Crew` with a specific blog post topic (e.g., "The Future of AI in Healthcare by 2030").

### Evaluation Criteria

*   Correct instantiation and configuration of `Agent`, `Task`, `Tool`, and `Crew` objects.
*   Logical flow of tasks, with the `Outline Task` using the output of the `Research Task`.
*   Clear and descriptive `role`, `goal`, `backstory`, `description`, and `expected_output` for all components.
*   Successful execution of the crew, demonstrating the interaction between agents and the use of the tool.
*   The final output should be a well-structured blog post outline.


```python
import os
from crewai import Agent, Task, Crew, Process
from crewai_tools import tool # Using crewai_tools for modern tool definition

# --- Configuration and Setup ---

# Set your OpenAI API key as an environment variable.
# In a real-world scenario, you'd load this securely (e.g., from .env file).
# For this exercise, we'll mock it or assume it's set.
# os.environ["OPENAI_API_KEY"] = "YOUR_OPENAI_API_KEY"

# Mock API key for demonstration purposes if not set externally
if "OPENAI_API_KEY" not in os.environ:
    os.environ["OPENAI_API_KEY"] = "sk-mock-key-for-crewai-exercise"
    print("Warning: OPENAI_API_KEY not found. Using a mock key. Ensure you set a valid key for real use.")

# Define a simple mock tool
@tool("Search Tool")
def search_tool(query: str) -> str:
    """Simulates searching the web for a given query and returns relevant information."""
    print(f"\n--- Executing Search Tool for query: '{query}' ---\n")
    # In a real application, this would integrate with a search API (e.g., Google Search, Brave Search)
    # For this exercise, we return a canned response based on common topics.
    if "AI in Healthcare" in query or "future of AI" in query:
        return (
            "Key trends: AI-powered diagnostics (e.g., pathology, radiology), personalized medicine, drug discovery, "
            "robot-assisted surgery, predictive analytics for patient outcomes. "
            "Challenges: Data privacy, regulatory hurdles, ethical considerations, integration with legacy systems. "
            "Market growth: Expected to reach $100+ billion by 2030, driven by efficiency gains and improved patient care."
        )
    elif "quantum computing" in query:
        return (
            "Quantum computing uses quantum-mechanical phenomena like superposition and entanglement to perform computations. "
            "Potential applications: cryptography, drug discovery, materials science, financial modeling. "
            "Current state: Still in early stages, with significant research and development ongoing. "
            "Major players: IBM, Google, Microsoft, D-Wave."
        )
    else:
        return f"No specific information found for '{query}'. General web search results indicate broad applications and rapid development in AI."

print("Setup complete: Mock Search Tool defined and environment variable checked.")
```

1.  **Define the `Research Agent` and `Outline Architect Agent`**.
2.  **Define the `Research Task` and `Outline Task`**.
3.  **Create a `Crew` instance** using your agents and tasks.
4.  **Execute the `Crew`** with the topic "The Future of AI in Healthcare by 2030".

Feel free to add comments to your code to explain your design choices.


```python
# --- Reference Solution ---

# 1. Define the Agents

# Research Agent: Gathers information using the search tool
research_agent = Agent(
    role='Senior Research Analyst',
    goal='Gather comprehensive, up-to-date information on a given topic.',
    backstory=(
        "As a seasoned analyst, you excel at sifting through vast amounts of data "
        "to extract the most relevant and impactful insights. Your reports are known "
        "for their accuracy and depth, providing a solid foundation for any project."
    ),
    tools=[search_tool], # Integrate the mock search tool
    verbose=True,
    allow_delegation=False # This agent focuses solely on research
)

# Outline Architect Agent: Structures the research into a blog post outline
outline_architect_agent = Agent(
    role='Content Outline Architect',
    goal='Create a structured, engaging, and logical blog post outline from research findings.',
    backstory=(
        "You are a master of content structure, able to transform raw information "
        "into compelling narratives. Your outlines guide writers to produce high-quality, "
        "reader-friendly articles that capture attention and deliver value."
    ),
    verbose=True,
    allow_delegation=False # This agent focuses solely on outlining
)

# 2. Define the Tasks

# Research Task: Uses the Research Agent and the search tool
research_task = Task(
    description=(
        "Conduct a thorough investigation into '{topic}'. "
        "Identify key trends, challenges, market growth projections, and significant advancements. "
        "Focus on data and insights relevant to a blog post aimed at a general tech-savvy audience."
    ),
    expected_output=(
        "A detailed summary of research findings, including bullet points on key trends, "
        "challenges, market statistics, and future predictions related to the topic. "
        "The output should be comprehensive enough for an outline to be built upon."
    ),
    agent=research_agent,
    # No context needed initially, as it's the first task
)

# Outline Task: Uses the Outline Architect Agent and the output of the research task
outline_task = Task(
    description=(
        "Based on the research findings provided, create a compelling and logical blog post outline. "
        "The outline should include a catchy title, an introduction, 3-5 main sections with sub-points, "
        "and a conclusion. Ensure the structure flows well and addresses the key aspects of the topic."
    ),
    expected_output=(
        "A markdown-formatted blog post outline with a clear title, introduction, "
        "main headings, sub-headings, and a conclusion. "
        "Example format:\n\n# Blog Post Title\n\n## Introduction\n- Hook\n- Thesis Statement\n\n## Section 1: Main Point\n- Sub-point A\n- Sub-point B\n\n## Conclusion\n- Summary\n- Call to Action"
    ),
    agent=outline_architect_agent,
    context=[research_task] # This task depends on the output of the research_task
)

# 3. Assemble the Crew

# Define the topic for the blog post
blog_topic = "The Future of AI in Healthcare by 2030"

# Create the Crew instance
blog_outline_crew = Crew(
    agents=[research_agent, outline_architect_agent],
    tasks=[research_task, outline_task],
    process=Process.sequential, # Tasks will be executed one after another
    verbose=True, # Enable verbose output for the crew's execution
    full_output=True, # Get full output including intermediate steps
    manager_llm=None # For simple sequential process, manager_llm is not strictly needed, but can be set for more complex flows
)

# 4. Execute the Crew
print("\n--- Starting the Blog Post Outline Generation Crew ---\n")

# Kick off the crew with the topic as input to the first task
crew_result = blog_outline_crew.kickoff(inputs={'topic': blog_topic})

print("\n--- Crew Execution Finished ---\n")
print("\n### Generated Blog Post Outline ###\n")
print(crew_result['final_output'])

# You can also inspect the full result object for more details
# print("\n--- Full Crew Result Object ---\n")
# print(crew_result)
```
