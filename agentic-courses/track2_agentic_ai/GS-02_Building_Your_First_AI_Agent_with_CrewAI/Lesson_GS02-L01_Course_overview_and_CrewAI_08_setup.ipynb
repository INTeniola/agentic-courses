## Securely Managing API Keys

In 2026, secure handling of API keys is non-negotiable. We will use `python-dotenv` to load environment variables from a `.env` file, preventing sensitive information from being hardcoded or committed to version control.

2.  **Add your API Key**: Inside the `.env` file, add your OpenAI API key (or keys for other LLM providers) in the format:
    ```
    OPENAI_API_KEY='your_openai_api_key_here'

    ```

## Hello CrewAI 0.8! Your First Agentic System

Now that our environment is set up, let's run a minimal CrewAI example. This isn't a full-fledged agentic system yet, but it will introduce you to the core components:

*   **Agent**: An autonomous entity with a role, goal, and backstory.
*   **Task**: A specific piece of work assigned to an agent.
*   **Crew**: A collection of agents working together to accomplish a set of tasks.

Our first crew will be incredibly simple: a single agent with a single task to say hello. This confirms that CrewAI is correctly installed and can interact with your LLM.


````python
from crewai import Agent, Task, Crew, Process
from langchain_openai import ChatOpenAI

# Initialize the LLM for our agents
# We're using OpenAI's GPT-4o for its advanced capabilities as of 2026.
# Ensure OPENAI_API_KEY is set in your environment.
llm = ChatOpenAI(model="gpt-4o", temperature=0.7)

# 1. Define the Agent
# An agent needs a role, a goal, and a backstory to guide its behavior.
greeter_agent = Agent(
    role='Friendly Greeter',
    goal='Introduce the user to the CrewAI course and welcome them.',
    backstory='You are an AI assistant designed to make new users feel welcome and excited about learning CrewAI.',
    verbose=True, # Set to True to see the agent's thought process
    allow_delegation=False, # For this simple agent, delegation isn't needed
    llm=llm # Assign the LLM to the agent
)

# 2. Define the Task
# A task has a description, an expected output, and is assigned to an agent.
welcome_task = Task(
    description='Craft a warm and concise welcome message for a new student starting the CrewAI course.',
    expected_output='A friendly welcome message, no more than two sentences long.',
    agent=greeter_agent # Assign the task to our greeter agent
)

# 3. Form the Crew
# A crew consists of agents and tasks. We define the process flow.
# 'sequential' means tasks are executed one after another.
hello_crew = Crew(
    agents=[greeter_agent],
    tasks=[welcome_task],
    verbose=2, # Set to 2 for detailed logging of the crew's execution
    process=Process.sequential # Define how tasks are processed
)

# 4. Kickoff the Crew!
# This starts the execution of the agentic workflow.
print("\n--- Kicking off the Crew! ---\n")
result = hello_crew.kickoff()

print("\n--- Crew Execution Complete ---\n")
print("Final Output:")
print(result)
````
