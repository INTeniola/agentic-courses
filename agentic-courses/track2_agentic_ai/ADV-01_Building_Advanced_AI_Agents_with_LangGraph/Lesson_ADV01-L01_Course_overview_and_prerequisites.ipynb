## "Hello, LangGraph!" - Environment Check

Let's run a very simple LangGraph workflow to confirm that your environment is correctly set up and all necessary libraries are functioning as expected. This minimal example will define a state, a single node, and execute a basic graph.


```python
from langgraph.graph import StateGraph, END
from typing import TypedDict, Annotated
import operator

# 1. Define the Agent State
# This TypedDict defines the structure of the state that will be passed between nodes.
# 'messages' will accumulate outputs from nodes.
class AgentState(TypedDict):
    messages: Annotated[list, operator.add] # 'operator.add' means new messages are appended to the list

# 2. Define a Simple Node Function
# A node is a function that takes the current state and returns updates to the state.
def greet_node(state: AgentState) -> dict:
    print("Executing 'greet_node'...")
    # This node simply adds a greeting message to the state.
    return {"messages": ["Hello from LangGraph! Your environment is ready."]}

# 3. Build the LangGraph Workflow
# Initialize a StateGraph with our defined AgentState.
workflow = StateGraph(AgentState)

# Add our 'greet_node' to the graph, naming it 'greeter'.
workflow.add_node("greeter", greet_node)

# Set the entry point of the graph. Execution will start here.
workflow.set_entry_point("greeter")

# Add an edge from the 'greeter' node to the END of the graph.
# This means after 'greeter' executes, the graph terminates.
workflow.add_edge("greeter", END)

# 4. Compile and Run the Graph
# Compile the workflow into an executable LangGraph application.
app = workflow.compile()

print("--- Running 'Hello, LangGraph!' workflow ---")

# Invoke the graph with an initial empty state.
initial_state = {"messages": []}
final_state = app.invoke(initial_state)

print("\nFinal State:")
print(final_state)

print("\nCongratulations! Your LangGraph environment is set up and working. You're ready for advanced agent building!")
```
