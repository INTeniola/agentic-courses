## RAG03-L08: Exercise - Build a Full-Stack Document Assistant

### Task Description

Your mission is to develop a full-stack document assistant application. This application will allow users to upload PDF documents, process them to create a searchable knowledge base, and then ask questions based on the content of the uploaded documents. The entire application must be containerized using Docker and orchestrated with Docker Compose.

### Core Requirements

1.  **Backend (FastAPI):**
    *   **Document Upload:** An endpoint (`/upload_pdf`) to receive PDF files from the frontend.
    *   **Document Processing:** An endpoint (`/process_document`) that takes a document ID (or filename) and performs the following RAG steps:
        *   Loads the PDF content.
        *   Chunks the text into manageable segments.
        *   Generates embeddings for each chunk using a modern embedding model (e.g., `sentence-transformers/all-MiniLM-L6-v2`).
        *   Stores the chunks and their embeddings in a vector database (e.g., `ChromaDB`). Each document should have its own distinct collection or identifier within the vector store.
    *   **Chat Interface:** An endpoint (`/chat`) that accepts a document ID and a user question. It should retrieve relevant chunks from the vector database and use a Large Language Model (LLM) (e.g., `Ollama` with `llama3` or `mistral`) to generate an answer based *only* on the retrieved context.
    *   **Technology Stack:** FastAPI, LangChain (or LlamaIndex), PyPDF, Sentence Transformers, ChromaDB client, Ollama client.

2.  **Frontend (Streamlit):**
    *   **User Interface:** A clean and intuitive Streamlit application.
    *   **PDF Upload:** A file uploader component for users to select and upload PDF files.
    *   **Processing Trigger:** A button or mechanism to initiate the document processing on the backend after upload.
    *   **Chat Window:** A chat interface where users can type questions and view the assistant's responses.
    *   **Status Updates:** Display feedback to the user regarding document upload, processing status, and chat responses.
    *   **Technology Stack:** Streamlit, `requests` library for API calls.

3.  **Containerization (Docker & Docker Compose):**
    *   **Backend Dockerfile:** A `Dockerfile` for the FastAPI application.
    *   **Frontend Dockerfile:** A `Dockerfile` for the Streamlit application.
    *   **Docker Compose:** A `docker-compose.yml` file to orchestrate all services:
        *   FastAPI backend service.
        *   Streamlit frontend service.
        *   ChromaDB server service (for persistent vector storage).
        *   Ensure services can communicate with each other (e.g., frontend to backend, backend to ChromaDB).
        *   Use appropriate port mappings and environment variables.

````python
# --- Reference Solution --- 
# This section provides a complete, high-quality reference solution for the exercise.
# Review it after you've attempted your own implementation.

# To run this solution, you would typically create these files in their respective directories.
# For demonstration purposes in a Jupyter Notebook, we'll represent them as multi-line strings.

print("--- Reference Solution: File Contents ---\n")

print("### 1. `backend/requirements.txt`")
backend_requirements_txt = """
fastapi==0.111.0
uvicorn[standard]==0.30.1
python-multipart==0.0.9
langchain==0.2.5
langchain-community==0.2.5
langchain-core==0.2.10
pypdf==4.2.0
chromadb==0.5.3
sentence-transformers==2.7.0
ollama==0.2.1
"""
print(backend_requirements_txt)
print("\n" + "-"*80 + "\n")

print("### 2. `backend/Dockerfile`")
backend_dockerfile = """
# Use a slim Python image for smaller container size
FROM python:3.11-slim-bookworm

# Set the working directory inside the container
WORKDIR /app

# Copy the requirements file and install dependencies
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Copy the rest of the application code
COPY . .

# Expose the port FastAPI will run on
EXPOSE 8000

# Command to run the FastAPI application using Uvicorn
# --host 0.0.0.0 makes the server accessible from outside the container
CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]
"""
print(backend_dockerfile)
print("\n" + "-"*80 + "\n")

print("### 3. `backend/main.py`")
backend_main_py = """
import os
import uuid
import shutil
import tempfile
from typing import List, Dict

from fastapi import FastAPI, UploadFile, File, HTTPException, status
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import SentenceTransformerEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_community.chat_models import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# Initialize FastAPI app
app = FastAPI(
    title="Document Assistant Backend",
    description="FastAPI backend for RAG-based document assistant."
)

# Configure CORS to allow frontend (Streamlit) to communicate
app.add_middleware(
    CORSMiddleware,
    allow_origins=["http://localhost:8501", "http://frontend:8501"], # Allow Streamlit's default port
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"]
)

# --- Configuration --- 
# Environment variables for ChromaDB and Ollama host
CHROMA_HOST = os.getenv("CHROMA_HOST", "localhost")
CHROMA_PORT = os.getenv("CHROMA_PORT", "8000") # Default ChromaDB server port
OLLAMA_HOST = os.getenv("OLLAMA_HOST", "http://ollama:11434") # Ollama service name in docker-compose

# Initialize embedding model (using a smaller model for faster local processing)
embeddings = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")

# Initialize LLM (Ollama with llama3 model)
# Ensure 'llama3' model is pulled in your Ollama instance
llm = ChatOllama(model="llama3", base_url=OLLAMA_HOST)

# In-memory store for document metadata (for simplicity, in a real app use a database)
document_metadata = {}

# --- Pydantic Models for Request/Response --- 
class ProcessDocumentResponse(BaseModel):
    document_id: str
    message: str

class ChatRequest(BaseModel):
    document_id: str
    question: str

class ChatResponse(BaseModel):
    answer: str

# --- Helper Functions ---
def get_vectorstore_client(collection_name: str):
    """Returns a ChromaDB client connected to the specified collection."""
    try:
        # Connect to the ChromaDB server
        client = Chroma(client_url=f"http://{CHROMA_HOST}:{CHROMA_PORT}", collection_name=collection_name, embedding_function=embeddings)
        return client
    except Exception as e:
        raise HTTPException(status_code=status.HTTP_500_INTERNAL_SERVER_ERROR, detail=f"Failed to connect to ChromaDB: {e}")

# --- API Endpoints ---

@app.post("/upload_pdf", response_model=ProcessDocumentResponse)
async def upload_pdf(file: UploadFile = File(...)):
    """Uploads a PDF file and stores it temporarily."""
    if not file.filename.endswith('.pdf'):
        raise HTTPException(status_code=status.HTTP_400_BAD_REQUEST, detail="Only PDF files are allowed.")

    document_id = str(uuid.uuid4()) # Generate a unique ID for the document
    temp_dir = tempfile.mkdtemp() # Create a temporary directory
    file_path = os.path.join(temp_dir, file.filename) # Path to save the PDF

    try:
        with open(file_path, "wb") as buffer:
            shutil.copyfileobj(file.file, buffer)
        
        # Store the temporary file path and ID in metadata
        document_metadata[document_id] = {"filepath": file_path, "filename": file.filename, "temp_dir": temp_dir}
        return {"document_id": document_id, "message": f"PDF '{file.filename}' uploaded successfully. Ready for processing."
        }
    except Exception as e:
        # Clean up temp directory if something goes wrong
        if os.path.exists(temp_dir): shutil.rmtree(temp_dir)
        raise HTTPException(status_code=status.HTTP_500_INTERNAL_SERVER_ERROR, detail=f"Failed to upload PDF: {e}")

@app.post("/process_document", response_model=ProcessDocumentResponse)
async def process_document(document_id: str):
    """Processes an uploaded PDF, chunks it, embeds it, and stores it in ChromaDB."""
    doc_info = document_metadata.get(document_id)
    if not doc_info:
        raise HTTPException(status_code=status.HTTP_404_NOT_FOUND, detail="Document not found or expired.")

    file_path = doc_info["filepath"]
    filename = doc_info["filename"]
    temp_dir = doc_info["temp_dir"]

    try:
        # 1. Load PDF
        loader = PyPDFLoader(file_path)
        documents = loader.load()

        # 2. Split documents into chunks
        text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
        chunks = text_splitter.split_documents(documents)

        # 3. Create/Update ChromaDB collection for this document
        # Each document gets its own collection for isolation
        collection_name = f"doc-{document_id}"
        vectorstore = get_vectorstore_client(collection_name)
        
        # Add chunks to the vector store
        vectorstore.add_documents(chunks)
        
        # Clean up the temporary PDF file and directory after processing
        if os.path.exists(temp_dir): shutil.rmtree(temp_dir)
        document_metadata.pop(document_id, None) # Remove from in-memory store

        return {"document_id": document_id, "message": f"Document '{filename}' processed and indexed successfully."
        }
    except Exception as e:
        # Ensure cleanup even if processing fails
        if os.path.exists(temp_dir): shutil.rmtree(temp_dir)
        document_metadata.pop(document_id, None)
        raise HTTPException(status_code=status.HTTP_500_INTERNAL_SERVER_ERROR, detail=f"Failed to process document: {e}")

@app.post("/chat", response_model=ChatResponse)
async def chat_with_document(request: ChatRequest):
    """Answers a question based on the content of a specific document."""
    document_id = request.document_id
    question = request.question

    collection_name = f"doc-{document_id}"
    try:
        vectorstore = get_vectorstore_client(collection_name)
        retriever = vectorstore.as_retriever()
    except HTTPException as e:
        # Re-raise if ChromaDB connection failed (e.g., service not up)
        raise e
    except Exception as e:
        # This might happen if the collection doesn't exist yet, meaning document wasn't processed
        raise HTTPException(status_code=status.HTTP_404_NOT_FOUND, detail=f"Document collection for ID '{document_id}' not found. Please ensure the document was processed.")

    # Define the RAG prompt template
    template = """You are an AI assistant for answering questions about documents.
    Use the following context to answer the question. If you don't know the answer,
    just say that you don't know, don't try to make up an answer.
    Keep the answer concise and directly relevant to the question.

    Context: {context}
    Question: {question}
    """
    prompt = ChatPromptTemplate.from_template(template)

    # Construct the RAG chain
    rag_chain = (
        {"context": retriever, "question": RunnablePassthrough()} # Retrieve context based on question
        | prompt # Format prompt with context and question
        | llm # Pass to LLM
        | StrOutputParser() # Parse LLM output to string
    )

    try:
        answer = rag_chain.invoke(question)
        return {"answer": answer}
    except Exception as e:
        raise HTTPException(status_code=status.HTTP_500_INTERNAL_SERVER_ERROR, detail=f"Failed to generate answer: {e}")
"""
print(backend_main_py)
print("\n" + "-"*80 + "\n")

print("### 4. `frontend/requirements.txt`")
frontend_requirements_txt = """
streamlit==1.36.0
requests==2.32.3
"""
print(frontend_requirements_txt)
print("\n" + "-"*80 + "\n")

print("### 5. `frontend/Dockerfile`")
frontend_dockerfile = """
# Use a slim Python image
FROM python:3.11-slim-bookworm

# Set the working directory
WORKDIR /app

# Copy requirements and install dependencies
COPY requirements.txt .
RUN pip install --no-cache-dir -r requirements.txt

# Copy the Streamlit application
COPY . .

# Expose the port Streamlit will run on
EXPOSE 8501

# Command to run the Streamlit application
# --server.port and --server.address ensure it's accessible within Docker
ENTRYPOINT ["streamlit", "run", "app.py", "--server.port=8501", "--server.address=0.0.0.0"]
"""
print(frontend_dockerfile)
print("\n" + "-"*80 + "\n")

print("### 6. `frontend/app.py`")
frontend_app_py = """
import streamlit as st
import requests
import os
import time

# --- Configuration ---
# Get backend URL from environment variable (set in docker-compose.yml)
BACKEND_URL = os.getenv("BACKEND_URL", "http://localhost:8000")

# --- Streamlit UI --- 
st.set_page_config(page_title="Document Assistant", layout="centered")
st.title("📚 Document Assistant")
st.markdown("Upload a PDF, process it, and ask questions about its content!")

# Initialize chat history in session state
if "messages" not in st.session_state:
    st.session_state.messages = []
if "document_id" not in st.session_state:
    st.session_state.document_id = None
if "document_name" not in st.session_state:
    st.session_state.document_name = None

# Display current document status
if st.session_state.document_id:
    st.sidebar.success(f"Active Document: {st.session_state.document_name} (ID: {st.session_state.document_id[:8]}...)")
else:
    st.sidebar.info("No document loaded. Please upload a PDF.")

# --- PDF Upload Section ---
st.sidebar.header("Upload Document")
uploaded_file = st.sidebar.file_uploader("Choose a PDF file", type="pdf")

if uploaded_file is not None:
    if st.sidebar.button("Upload & Process PDF"):
        with st.spinner(f"Uploading '{uploaded_file.name}'..."):
            files = {"file": (uploaded_file.name, uploaded_file.getvalue(), "application/pdf")}
            try:
                upload_response = requests.post(f"{BACKEND_URL}/upload_pdf", files=files)
                upload_response.raise_for_status() # Raise an exception for HTTP errors
                upload_data = upload_response.json()
                st.session_state.document_id = upload_data["document_id"]
                st.session_state.document_name = uploaded_file.name
                st.sidebar.success(upload_data["message"])

                with st.spinner(f"Processing '{uploaded_file.name}'..."):
                    process_response = requests.post(f"{BACKEND_URL}/process_document", params={"document_id": st.session_state.document_id})
                    process_response.raise_for_status()
                    process_data = process_response.json()
                    st.sidebar.success(process_data["message"])
                    st.session_state.messages = [] # Clear chat history for new document
                    st.rerun()

            except requests.exceptions.RequestException as e:
                st.sidebar.error(f"Error communicating with backend: {e}")
            except Exception as e:
                st.sidebar.error(f"An unexpected error occurred: {e}")

# --- Chat Interface Section ---

# Display chat messages from history on app rerun
for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# React to user input
if prompt := st.chat_input("Ask a question about the document..."):
    if not st.session_state.document_id:
        st.error("Please upload and process a document first.")
    else:
        # Display user message in chat message container
        st.chat_message("user").markdown(prompt)
        # Add user message to chat history
        st.session_state.messages.append({"role": "user", "content": prompt})

        with st.spinner("Thinking..."):
            try:
                # Call backend /chat endpoint
                chat_payload = {"document_id": st.session_state.document_id, "question": prompt}
                chat_response = requests.post(f"{BACKEND_URL}/chat", json=chat_payload)
                chat_response.raise_for_status()
                chat_data = chat_response.json()
                assistant_response = chat_data["answer"]

                # Display assistant response in chat message container
                with st.chat_message("assistant"):
                    st.markdown(assistant_response)
                # Add assistant response to chat history
                st.session_state.messages.append({"role": "assistant", "content": assistant_response})

            except requests.exceptions.RequestException as e:
                st.error(f"Error communicating with backend: {e}")
                st.session_state.messages.append({"role": "assistant", "content": f"Error: Could not get a response from the backend. ({e})"})
            except Exception as e:
                st.error(f"An unexpected error occurred: {e}")
                st.session_state.messages.append({"role": "assistant", "content": f"Error: An unexpected error occurred. ({e})"})
"""
print(frontend_app_py)
print("\n" + "-"*80 + "\n")

print("### 7. `docker-compose.yml`")
docker_compose_yml = """
version: '3.8'

services:
  # Ollama service for local LLM inference
  ollama:
    image: ollama/ollama:latest
    ports:
      - "11434:11434"
    volumes:
      - ollama_data:/root/.ollama
    # Command to ensure llama3 model is available on startup
    # This will pull the model if it's not already present in the volume
    command: serve
    healthcheck:
      test: ["CMD", "curl", "-f", "http://localhost:11434/"]
      interval: 10s
      timeout: 5s
      retries: 5
    restart: always

  # ChromaDB vector store service
  chroma:
    image: chromadb/chroma:latest
    ports:
      - "8000:8000"
    volumes:
      - chroma_data:/chroma/data
    healthcheck:
      test: ["CMD", "curl", "-f", "http://localhost:8000/api/v1/heartbeat"]
      interval: 10s
      timeout: 5s
      retries: 5
    restart: always

  # FastAPI Backend service
  backend:
    build:
      context: ./backend
      dockerfile: Dockerfile
    ports:
      - "8000:8000"
    environment:
      # Connect to the ChromaDB service by its service name 'chroma'
      CHROMA_HOST: chroma
      CHROMA_PORT: 8000
      # Connect to the Ollama service by its service name 'ollama'
      OLLAMA_HOST: http://ollama:11434
    depends_on:
      ollama:
        condition: service_healthy
      chroma:
        condition: service_healthy
    restart: always

  # Streamlit Frontend service
  frontend:
    build:
      context: ./frontend
      dockerfile: Dockerfile
    ports:
      - "8501:8501"
    environment:
      # Connect to the FastAPI backend service by its service name 'backend'
      BACKEND_URL: http://backend:8000
    depends_on:
      backend:
        condition: service_started # Frontend only needs backend to be started, not necessarily healthy for initial load
    restart: always

# Define named volumes for persistent data
volumes:
  ollama_data:
  chroma_data:
"""
print(docker_compose_yml)
print("\n" + "-"*80 + "\n")

print("### Instructions to run the solution:")
print("1. Create the directory structure: `backend/`, `frontend/`.")
print("2. Inside `backend/`, create `requirements.txt`, `Dockerfile`, and `main.py` with the content above.")
print("3. Inside `frontend/`, create `requirements.txt`, `Dockerfile`, and `app.py` with the content above.")
print("4. In the root directory, create `docker-compose.yml` with the content above.")
print("5. Ensure you have a `sample.pdf` file in your project root for testing.")
print("6. Open your terminal in the project root and run: `docker compose up --build`")
print("7. Once all services are up, open your browser to `http://localhost:8501`.")
print("8. If you encounter issues with Ollama, ensure you've run `ollama pull llama3` on your host machine or adjust the `command` in `docker-compose.yml` to pull it within the container if needed (though the volume should persist it).")
````
