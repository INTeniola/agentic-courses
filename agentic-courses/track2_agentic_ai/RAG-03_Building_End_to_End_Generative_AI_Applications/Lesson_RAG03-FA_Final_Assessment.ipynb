## Capstone Project: Smart Document Assistant

**Objective**: Build a full-stack "Smart Document Assistant" that allows users to upload various document types (PDF, TXT, DOCX), process them into a RAG-ready format, and then ask questions to retrieve answers based *only* on the uploaded documents. The application should be containerized using Docker Compose.

**Requirements**:

1.  **FastAPI Backend (`backend` service)**:
    *   **Document Upload Endpoint**: An `async` endpoint (`/upload-document`) that accepts a file (PDF, TXT, DOCX). Upon upload, it should save the file temporarily.
    *   **Document Processing & Ingestion**: After upload, process the document:
        *   Load the document content using appropriate loaders (e.g., `PyPDFLoader`, `TextLoader`, `Docx2txtLoader`).
        *   Split the document into chunks using a recursive character text splitter.
        *   Generate embeddings for these chunks using a modern embedding model (e.g., `OpenAIEmbeddings`, `HuggingFaceEmbeddings`, or `OllamaEmbeddings`).
        *   Store the chunks and their embeddings in a persistent vector database (e.g., ChromaDB, LanceDB). Ensure the vector store is initialized to use a local directory for persistence.
    *   **RAG Query Endpoint**: An `async` endpoint (`/query`) that accepts a user question. It should:
        *   Retrieve relevant document chunks from the vector database based on the question.
        *   Construct a prompt for an LLM using the retrieved context and the user's question.
        *   Invoke an LLM (e.g., `ChatOpenAI`, `Ollama`, `HuggingFaceHub`) to generate an answer.
        *   Return the generated answer.
    *   **CORS**: Enable CORS for the frontend.

2.  **Streamlit Frontend (`frontend` service)**:
    *   **File Upload Interface**: A Streamlit file uploader widget to allow users to upload PDF, TXT, or DOCX files.
    *   **Processing Status**: Display a message indicating whether the document is being processed or has been successfully ingested.
    *   **Chat Interface**: A text input for users to ask questions and a display area to show the conversation history (questions and LLM answers).
    *   **Backend Communication**: Use `httpx` or `requests` to communicate with the FastAPI backend for document upload and RAG queries.
    *   **Session Management**: Use Streamlit's session state to maintain chat history and document processing status.

3.  **Containerization with Docker Compose**:
    *   Create a `docker-compose.yml` file that defines three services:
        *   `backend`: Runs the FastAPI application.
        *   `frontend`: Runs the Streamlit application.
        *   `vectordb`: A service for the persistent vector database (e.g., ChromaDB, LanceDB). This service should mount a volume to persist data.
    *   Each service should have its own `Dockerfile` and `requirements.txt`.
    *   Ensure services can communicate with each other using their service names (e.g., `http://backend:8000`).

```python
# Detailed Solution Code for Capstone Project

# --- Project Structure ---
# ./
# ├── docker-compose.yml
# ├── backend/
# │   ├── Dockerfile
# │   ├── requirements.txt
# │   └── main.py
# ├── frontend/
# │   ├── Dockerfile
# │   ├── requirements.txt
# │   └── app.py
# └── data/ (This directory will be created and used by Docker volumes)
#     ├── backend/ (for temporary files and ChromaDB persistence)
#     └── ollama/ (if using Ollama for model persistence)

# --- 1. docker-compose.yml ---
# Save this as `docker-compose.yml` in your project root
# Make sure to replace 'llama3' with your desired Ollama model if using Ollama.
# If using OpenAI, ensure OPENAI_API_KEY is set in your environment or .env file.

# version: '3.8'
# services:
#   backend:
#     build:
#       context: ./backend
#       dockerfile: Dockerfile
#     ports:
#       - "8000:8000"
#     environment:
#       OPENAI_API_KEY: ${OPENAI_API_KEY}
#       OLLAMA_HOST: http://ollama:11434 # Uncomment if using Ollama
#       # HUGGINGFACEHUB_API_TOKEN: ${HUGGINGFACEHUB_API_TOKEN} # Uncomment if using HuggingFace
#     volumes:
#       - ./data/backend:/app/data # Persistent storage for uploaded files and ChromaDB
#     depends_on:
#       # - ollama # Uncomment if using Ollama service
#       - frontend # Frontend depends on backend, so backend must be up first

#   frontend:
#     build:
#       context: ./frontend
#       dockerfile: Dockerfile
#     ports:
#       - "8501:8501"
#     environment:
#       FASTAPI_URL: http://backend:8000 # Internal Docker network communication
#     depends_on:
#       - backend

#   # Uncomment the 'ollama' service block below if you want to use a local Ollama LLM/Embeddings
#   # Make sure to pull the model (e.g., 'ollama pull llama3') before starting, or use the entrypoint.
#   # ollama:
#   #   image: ollama/ollama:latest
#   #   command: serve
#   #   ports:
#   #     - "11434:11434"
#   #   volumes:
#   #     - ./data/ollama:/root/.ollama # Persistent storage for Ollama models
#   #   environment:
#   #     OLLAMA_MODELS: llama3 # Specify the model to pull/use
#   #   # Optional: entrypoint to ensure model is pulled on first run
#   #   # entrypoint: ["/bin/bash", "-c", "ollama pull ${OLLAMA_MODELS} && ollama serve"]


# --- 2. backend/Dockerfile ---
# FROM python:3.11-slim-bookworm
# WORKDIR /app
# COPY requirements.txt .
# RUN pip install --no-cache-dir -r requirements.txt
# COPY . .
# CMD ["uvicorn", "main:app", "--host", "0.0.0.0", "--port", "8000"]


# --- 3. backend/requirements.txt ---
# fastapi
# uvicorn
# python-multipart
# langchain
# langchain-community
# langchain-openai # For OpenAI LLM and Embeddings
# # langchain-ollama # Uncomment if using Ollama
# # langchain-huggingface # Uncomment if using HuggingFace models
# pypdf
# python-docx
# chromadb # For vector store
# tiktoken # Required by langchain-openai for token counting


# --- 4. backend/main.py (FastAPI) ---
import os
import shutil
import asyncio
from typing import List, Dict, Any

from fastapi import FastAPI, UploadFile, File, HTTPException, BackgroundTasks
from fastapi.responses import JSONResponse
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel

# LangChain imports
from langchain_community.document_loaders import PyPDFLoader, TextLoader, Docx2txtLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OpenAIEmbeddings # Default to OpenAI
from langchain_community.vectorstores import Chroma
from langchain.chains import RetrievalQA
from langchain_openai import ChatOpenAI # Default to OpenAI LLM
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# Optional: For Ollama
# from langchain_community.embeddings import OllamaEmbeddings
# from langchain_community.llms import Ollama

# Optional: For HuggingFace
# from langchain_community.embeddings import HuggingFaceEmbeddings
# from langchain_huggingface import HuggingFaceEndpoint

app = FastAPI()

# Configure CORS for Streamlit frontend
origins = [
    "http://localhost:8501", # Streamlit default port for local development
    "http://frontend:8501"   # Docker internal frontend service name and port
]
app.add_middleware(
    CORSMiddleware,
    allow_origins=origins,
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"]
)

# Directory for storing uploaded files and ChromaDB persistence
DATA_DIR = "./data"
UPLOAD_DIR = os.path.join(DATA_DIR, "uploads")
VECTOR_DB_PATH = os.path.join(DATA_DIR, "chroma_db")

os.makedirs(UPLOAD_DIR, exist_ok=True)
os.makedirs(VECTOR_DB_PATH, exist_ok=True)

# Global variables for RAG components
# In a production system, these might be managed by a singleton pattern or dependency injection
vectorstore = None
llm = None
embeddings_model = None

# --- LLM and Embeddings Initialization ---
# Choose your LLM and Embeddings provider. Defaulting to OpenAI.
# Ensure you have the corresponding API key set as an environment variable.

def initialize_rag_components():
    global llm, embeddings_model, vectorstore

    # Initialize Embeddings Model
    # Option 1: OpenAI Embeddings
    if os.getenv("OPENAI_API_KEY"):
        embeddings_model = OpenAIEmbeddings(model="text-embedding-3-small")
        print("Initialized OpenAI Embeddings.")
    # Option 2: Ollama Embeddings (uncomment if using Ollama)
    # elif os.getenv("OLLAMA_HOST"):
    #     embeddings_model = OllamaEmbeddings(base_url=os.getenv("OLLAMA_HOST"), model="llama3")
    #     print("Initialized Ollama Embeddings.")
    # Option 3: HuggingFace Embeddings (requires token and model name)
    # elif os.getenv("HUGGINGFACEHUB_API_TOKEN"):
    #     embeddings_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
    #     print("Initialized HuggingFace Embeddings.")
    else:
        raise ValueError("No valid API key or Ollama host found for embeddings. Please set OPENAI_API_KEY or OLLAMA_HOST.")

    # Initialize LLM
    # Option 1: OpenAI Chat Model
    if os.getenv("OPENAI_API_KEY"):
        llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.1)
        print("Initialized OpenAI LLM.")
    # Option 2: Ollama LLM (uncomment if using Ollama)
    # elif os.getenv("OLLAMA_HOST"):
    #     llm = Ollama(base_url=os.getenv("OLLAMA_HOST"), model="llama3", temperature=0.1)
    #     print("Initialized Ollama LLM.")
    # Option 3: HuggingFace Endpoint LLM (requires token and repo_id)
    # elif os.getenv("HUGGINGFACEHUB_API_TOKEN"):
    #     llm = HuggingFaceEndpoint(repo_id="HuggingFaceH4/zephyr-7b-beta", temperature=0.1)
    #     print("Initialized HuggingFace LLM.")
    else:
        raise ValueError("No valid API key or Ollama host found for LLM. Please set OPENAI_API_KEY or OLLAMA_HOST.")

    # Initialize ChromaDB (or other vector store) with persistence
    # We'll re-initialize or load from disk when documents are processed
    # For now, ensure the directory exists.
    print(f"ChromaDB will persist to: {VECTOR_DB_PATH}")

# Initialize components on startup
@app.on_event("startup")
async def startup_event():
    try:
        initialize_rag_components()
    except ValueError as e:
        print(f"RAG component initialization failed: {e}")
        # In a real app, you might want to exit or disable RAG functionality


class QueryRequest(BaseModel):
    question: str

@app.post("/upload-document")
async def upload_document(background_tasks: BackgroundTasks, file: UploadFile = File(...)):
    if not file.filename:
        raise HTTPException(status_code=400, detail="No file uploaded.")

    file_extension = os.path.splitext(file.filename)[1].lower()
    if file_extension not in [".pdf", ".txt", ".docx"]:
        raise HTTPException(status_code=400, detail="Unsupported file type. Only PDF, TXT, and DOCX are allowed.")

    file_path = os.path.join(UPLOAD_DIR, file.filename)
    try:
        # Save the uploaded file temporarily
        with open(file_path, "wb") as buffer:
            shutil.copyfileobj(file.file, buffer)

        # Add document processing to background tasks
        background_tasks.add_task(process_document_and_ingest, file_path, file_extension)

        return JSONResponse(content={
            "message": f"Document '{file.filename}' uploaded successfully. Processing started in background."
        })
    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Failed to upload or start processing: {e}")

async def process_document_and_ingest(file_path: str, file_extension: str):
    global vectorstore

    if embeddings_model is None:
        print("Embeddings model not initialized. Cannot process document.")
        return

    try:
        # 1. Load document
        print(f"Loading document: {file_path}")
        if file_extension == ".pdf":
            loader = PyPDFLoader(file_path)
        elif file_extension == ".txt":
            loader = TextLoader(file_path)
        elif file_extension == ".docx":
            loader = Docx2txtLoader(file_path)
        else:
            print(f"Unsupported file type for processing: {file_extension}")
            return

        documents = loader.load()
        print(f"Loaded {len(documents)} pages/sections from {file_path}")

        # 2. Split documents into chunks
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=1000,
            chunk_overlap=200,
            length_function=len,
            is_separator_regex=False,
        )
        chunks = text_splitter.split_documents(documents)
        print(f"Split into {len(chunks)} chunks.")

        # 3. Generate embeddings and store in vector database
        print("Generating embeddings and ingesting into ChromaDB...")
        if vectorstore is None:
            # Initialize or load ChromaDB if it's the first document
            vectorstore = Chroma.from_documents(
                documents=chunks,
                embedding=embeddings_model,
                persist_directory=VECTOR_DB_PATH
            )
            print("ChromaDB initialized with first document.")
        else:
            # Add documents to existing ChromaDB
            vectorstore.add_documents(chunks)
            print("Documents added to existing ChromaDB.")

        vectorstore.persist() # Ensure changes are written to disk
        print(f"Finished ingesting {len(chunks)} chunks from {file_path} into ChromaDB.")

    except Exception as e:
        print(f"Error processing document {file_path}: {e}")
    finally:
        # Clean up the uploaded file after processing
        if os.path.exists(file_path):
            os.remove(file_path)
            print(f"Cleaned up temporary file: {file_path}")

@app.post("/query")
async def query_document(request: QueryRequest):
    global vectorstore, llm

    if vectorstore is None:
        # Attempt to load existing vectorstore if it exists on disk but not in memory
        if os.path.exists(VECTOR_DB_PATH) and len(os.listdir(VECTOR_DB_PATH)) > 0:
            try:
                if embeddings_model is None:
                    raise ValueError("Embeddings model not initialized. Cannot load vectorstore.")
                vectorstore = Chroma(persist_directory=VECTOR_DB_PATH, embedding_function=embeddings_model)
                print("Loaded existing ChromaDB from disk.")
            except Exception as e:
                print(f"Failed to load existing ChromaDB: {e}")
                raise HTTPException(status_code=400, detail="No documents have been ingested yet or failed to load existing data.")
        else:
            raise HTTPException(status_code=400, detail="No documents have been ingested yet. Please upload a document first.")

    if llm is None:
        raise HTTPException(status_code=500, detail="LLM not initialized. Server error.")

    try:
        # 1. Retrieve relevant documents
        retriever = vectorstore.as_retriever(search_kwargs={"k": 4}) # Retrieve top 4 relevant chunks

        # 2. Define prompt template
        template = """You are an AI assistant for answering questions based on the provided context. 
        Answer the question truthfully and concisely, using ONLY the following context. 
        If the answer is not in the context, state that you don't know. 

        Context: {context}

        Question: {question}
        Answer:"""
        prompt = ChatPromptTemplate.from_template(template)

        # 3. Create RAG chain
        rag_chain = (
            {"context": retriever, "question": RunnablePassthrough()} 
            | prompt 
            | llm 
            | StrOutputParser()
        )

        # 4. Invoke the chain asynchronously
        answer = await rag_chain.ainvoke(request.question)

        return {"answer": answer}
    except Exception as e:
        print(f"Error during RAG query: {e}")
        raise HTTPException(status_code=500, detail=f"Failed to retrieve answer: {e}")


# --- 5. frontend/Dockerfile ---
# FROM python:3.11-slim-bookworm
# WORKDIR /app
# COPY requirements.txt .
# RUN pip install --no-cache-dir -r requirements.txt
# COPY . .
# EXPOSE 8501
# ENTRYPOINT ["streamlit", "run", "app.py", "--server.port=8501", "--server.address=0.0.0.0"]


# --- 6. frontend/requirements.txt ---
# streamlit
# httpx


# --- 7. frontend/app.py (Streamlit) ---
import streamlit as st
import httpx
import os
import asyncio

# FastAPI backend URL (use service name for Docker internal communication)
# In local development, this might be http://localhost:8000
# In Docker Compose, it will be http://backend:8000
FASTAPI_URL = os.getenv("FASTAPI_URL", "http://localhost:8000")

st.set_page_config(page_title="Smart Document Assistant", layout="wide")
st.title("📚 Smart Document Assistant")

# Initialize session state
if "messages" not in st.session_state:
    st.session_state.messages = []
if "document_processed" not in st.session_state:
    st.session_state.document_processed = False
if "processing_status" not in st.session_state:
    st.session_state.processing_status = "No document uploaded."

# Document Upload Section
with st.sidebar:
    st.header("Upload Document")
    uploaded_file = st.file_uploader(
        "Choose a PDF, TXT, or DOCX file",
        type=["pdf", "txt", "docx"],
        accept_multiple_files=False
    )

    if uploaded_file is not None:
        if st.button("Process Document", key="process_doc_button"):
            st.session_state.processing_status = "Uploading and processing..."
            st.session_state.document_processed = False
            st.sidebar.info(st.session_state.processing_status)
            st.rerun() # Rerun to update status immediately

            try:
                # Use httpx for async requests
                files = {"file": (uploaded_file.name, uploaded_file.getvalue(), uploaded_file.type)}
                # Use a client with a timeout for potentially long uploads/processing starts
                with httpx.Client() as client:
                    response = client.post(f"{FASTAPI_URL}/upload-document", files=files, timeout=300)
                
                if response.status_code == 200:
                    st.session_state.processing_status = f"Successfully uploaded '{uploaded_file.name}'. Processing in background."
                    st.session_state.document_processed = True # Backend confirmed start of processing
                else:
                    st.session_state.processing_status = f"Error uploading document: {response.text}"
            except httpx.RequestError as e:
                st.session_state.processing_status = f"Network error: Could not connect to backend at {FASTAPI_URL}. Is it running? {e}"
            except Exception as e:
                st.session_state.processing_status = f"An unexpected error occurred: {e}"
            st.rerun() # Rerun to update status after processing attempt

    st.sidebar.info(st.session_state.processing_status)

# Chat Interface Section
st.header("Ask Questions")

# Display chat messages from history on app rerun
for message in st.session_state.messages:
    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# React to user input
if prompt := st.chat_input("Ask a question about the document..."):
    # Display user message in chat message container
    st.chat_message("user").markdown(prompt)
    # Add user message to chat history
    st.session_state.messages.append({"role": "user", "content": prompt})

    if not st.session_state.document_processed:
        response_content = "Please upload and process a document first before asking questions."
    else:
        with st.spinner("Thinking..."):
            try:
                # Use httpx for async requests
                with httpx.Client() as client:
                    response = client.post(f"{FASTAPI_URL}/query", json={"question": prompt}, timeout=300)
                
                if response.status_code == 200:
                    response_data = response.json()
                    response_content = response_data.get("answer", "No answer found.")
                else:
                    response_content = f"Error from backend: {response.text}"
            except httpx.RequestError as e:
                response_content = f"Network error: Could not connect to backend at {FASTAPI_URL}. Is it running? {e}"
            except Exception as e:
                response_content = f"An unexpected error occurred: {e}"

    # Display assistant response in chat message container
    with st.chat_message("assistant"):
        st.markdown(response_content)
    # Add assistant response to chat history
    st.session_state.messages.append({"role": "assistant", "content": response_content})

# --- How to Run This Solution ---
# 1. Create the directory structure:
#    my_rag_app/
#    ├── docker-compose.yml
#    ├── backend/
#    │   ├── Dockerfile
#    │   ├── requirements.txt
#    │   └── main.py
#    ├── frontend/
#    │   ├── Dockerfile
#    │   ├── requirements.txt
#    │   └── app.py
#    └── data/ (will be created by Docker)

# 2. Populate the files with the content above.
#    - For `docker-compose.yml`, uncomment the `ollama` service if you want to use it.
#    - For `backend/main.py`, ensure you have `OPENAI_API_KEY` set as an environment variable
#      or uncomment the Ollama/HuggingFace sections and set their respective environment variables.

# 3. Build and run the Docker containers:
#    Navigate to the `my_rag_app` directory in your terminal.
#    Run: `docker compose up --build`

# 4. Access the application:
#    Open your web browser and go to `http://localhost:8501`

# 5. Interact:
#    - Upload a PDF, TXT, or DOCX document using the sidebar uploader.
#    - Wait for the processing status to update.
#    - Ask questions related to the document content in the chat interface.
```
