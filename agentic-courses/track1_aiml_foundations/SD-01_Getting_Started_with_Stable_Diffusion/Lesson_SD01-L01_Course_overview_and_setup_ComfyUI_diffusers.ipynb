## Your First Image Generation with `diffusers`

Now for the exciting part! Let's generate our very first image using a pre-trained Stable Diffusion model. We'll use a popular and efficient model, `runwayml/stable-diffusion-v1-5`, which is a great starting point for understanding the basics.

This code block will:
1.  Load the `StableDiffusionPipeline` from the `diffusers` library.
2.  Download the model weights (this will happen only once).
3.  Move the model to the GPU for faster inference.
4.  Define a simple text prompt.
5.  Generate an image based on the prompt.
6.  Display the generated image.


```python
from diffusers import DiffusionPipeline
import torch
from IPython.display import display

# 1. Load the Stable Diffusion pipeline
# We specify torch_dtype=torch.float16 for memory efficiency and faster inference on modern GPUs.
# This model is hosted on Hugging Face Hub and will be downloaded automatically.
print("Loading Stable Diffusion v1.5 model...")
pipeline = DiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16)

# 2. Move the pipeline to the GPU (CUDA device)
# This is crucial for performance. If CUDA is not available, it will run on CPU (very slow).
if torch.cuda.is_available():
    pipeline.to("cuda")
    print("Model moved to CUDA (GPU).")
else:
    print("CUDA not available. Model running on CPU.")

# 3. Define your text prompt
prompt = "A futuristic city at sunset, highly detailed, digital painting, artstation, concept art, smooth, sharp focus, illustration, cinematic lighting"
print(f"\nGenerating image for prompt: '{prompt}'")

# 4. Generate the image
# The pipeline returns a list of images, we take the first one.
# Using num_inference_steps=25 is a good balance for quality and speed for a first try.
image = pipeline(prompt, num_inference_steps=25).images[0]

# 5. Display the generated image
print("\nImage generated successfully!")
display(image)

# Optionally, save the image to a file
image.save("my_first_stable_diffusion_image.png")
print("Image saved as 'my_first_stable_diffusion_image.png'")
```
