## Prompt Engineering for Images: Positive and Negative Prompts

Welcome to the exciting world of prompt engineering for generative AI! As developers and creators, understanding how to effectively communicate with AI models is paramount to achieving desired results. In text-to-image generation, our primary tools for communication are **prompts**.

Think of a prompt as a set of instructions you give to an artist. A traditional artist might interpret your words and draw based on their understanding and skill. An AI model, however, interprets your words based on its training data and internal representations. The more precise and clear your instructions, the better the outcome.

### The Power of Positive Prompts

A **positive prompt** is what you *want* to see in your generated image. It's your primary instruction set, describing the subject, style, setting, lighting, mood, and any other desired attributes. A well-crafted positive prompt is specific, rich in detail, and uses evocative keywords.

**Example:** Instead of just "a dog," a positive prompt might be: "A majestic golden retriever, running through a sunlit field of wildflowers, golden hour, volumetric lighting, hyperrealistic, cinematic, 8k, photorealistic."

### The Art of Negative Prompts

While positive prompts tell the AI what *to include*, **negative prompts** tell the AI what *to exclude*. This is a powerful technique to refine your images, remove common artifacts, correct stylistic inconsistencies, or simply prevent unwanted elements from appearing. It's like telling a chef, "Make me a delicious pasta dish, but *no olives* and *no mushrooms*."

Negative prompts are incredibly useful for:
*   **Removing undesirable elements:** e.g., "ugly, deformed, extra limbs, missing limbs."
*   **Correcting common AI generation flaws:** e.g., "bad anatomy, poorly drawn hands, disfigured, blurry, low resolution."
*   **Controlling style:** e.g., "cartoon, anime, 3D render, text, watermark."
*   **Enhancing quality:** By removing negative attributes, the model can focus its resources on generating higher-quality desired elements.

In essence, positive and negative prompts work in tandem, guiding the AI model through a process of inclusion and exclusion, allowing for a much finer degree of control over the final output. Modern diffusion models, especially those like Stable Diffusion XL, leverage these prompts to produce stunning and highly specific imagery. Let's dive into a practical example to see this in action.


```python
import torch
from diffusers import DiffusionPipeline
from PIL import Image

# --- Configuration --- 
# Set device to 'cuda' if a GPU is available, otherwise 'cpu'
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# --- Load the Stable Diffusion XL model --- 
# We'll use a modern, high-quality Stable Diffusion XL model.
# This might take a few minutes the first time as it downloads the model weights.
print("Loading Stable Diffusion XL model...")
pipeline = DiffusionPipeline.from_pretrained("stabilityai/stable-diffusion-xl-base-1.0", torch_dtype=torch.float16)
pipeline.to(device)
print("Model loaded successfully.")

# --- Define Prompts --- 
# A positive prompt describes what we want to see.
positive_prompt = "A futuristic city skyline at sunset, neon lights reflecting on wet streets, flying cars, cyberpunk aesthetic, highly detailed, cinematic, 8k, photorealistic."

# A negative prompt describes what we want to avoid.
# Common negative prompts include terms for quality issues, unwanted styles, or specific objects.
negative_prompt = "ugly, deformed, noisy, blurry, low resolution, bad anatomy, disfigured, poorly drawn face, poorly drawn hands, missing limbs, extra limbs, watermark, text, signature, cartoon, anime, 3D render, painting."

# --- Generate Images --- 
# For reproducibility, set a manual seed.
seed = 42
generator = torch.Generator(device=device).manual_seed(seed)

print("\nGenerating image with POSITIVE prompt only...")
# Generate an image using only the positive prompt
image_positive_only = pipeline(
    prompt=positive_prompt,
    generator=generator,
    num_inference_steps=30
).images[0]

print("Generating image with POSITIVE and NEGATIVE prompts...")
# Generate an image using both positive and negative prompts
image_with_negative = pipeline(
    prompt=positive_prompt,
    negative_prompt=negative_prompt,
    generator=generator,
    num_inference_steps=30
).images[0]

# --- Display Results --- 
print("\nDisplaying results...")

print("Image generated with POSITIVE prompt only:")
display(image_positive_only)

print("Image generated with POSITIVE and NEGATIVE prompts:")
display(image_with_negative)

# You can also save the images if you wish
# image_positive_only.save("city_positive_only.png")
# image_with_negative.save("city_with_negative.png")

print("\nComparison complete. Observe the differences in quality and unwanted elements.")
```

### Interpreting the Output and Practical Applications

After running the code, you should observe two images. The first image, generated with only the positive prompt, might already be quite good. However, the second image, generated with both positive and negative prompts, should exhibit noticeable improvements. You might see:

*   **Fewer artifacts:** The negative prompt helps reduce common issues like blurry areas, strange distortions, or pixelation.
*   **Improved clarity and detail:** By explicitly telling the model what *not* to do, it can better focus on rendering the desired elements with higher fidelity.
*   **Absence of unwanted stylistic elements:** If your negative prompt included terms like "cartoon" or "3D render," you should see a more photorealistic or specific style emerge.
*   **Better composition:** Sometimes, negative prompts can indirectly lead to better overall composition by removing distracting elements.

#### Performance Trade-offs

While adding negative prompts significantly enhances control and quality, there's a minor performance consideration. Longer and more complex prompts (both positive and negative) require more computational effort for the model to process. This might lead to slightly longer generation times. However, for most practical applications, the quality improvement far outweighs this minimal overhead, especially with modern GPUs and optimized libraries like Diffusers.

#### Typical Use Cases and Best Practices

1.  **Quality Enhancement:** Always start with a strong set of negative prompts to ensure baseline quality. Common terms include: `ugly, deformed, bad anatomy, disfigured, poorly drawn, extra limbs, missing limbs, blurry, low resolution, jpeg artifacts, watermark, text, signature, error, out of frame.`
2.  **Stylistic Control:** If you're aiming for a specific style (e.g., photorealistic), use negative prompts to exclude conflicting styles (e.g., `cartoon, anime, painting, sketch, 3D render`).
3.  **Removing Specific Objects:** If you want a landscape without people, add `people, crowd, human` to your negative prompt.
4.  **Iterative Refinement:** Prompt engineering is an iterative process. Generate an image, identify what you don't like, and add those elements to your negative prompt. Repeat until satisfied.
5.  **Prompt Weighting (Advanced):** Some models and interfaces allow you to assign weights to terms in your prompts (e.g., `(ugly:1.2)` to emphasize the negative impact of 'ugly'). While not explicitly covered in this basic example, it's a powerful technique for fine-tuning.

By mastering the interplay between positive and negative prompts, you gain unparalleled control over the creative output of text-to-image models, transforming them from unpredictable generators into powerful artistic collaborators.
