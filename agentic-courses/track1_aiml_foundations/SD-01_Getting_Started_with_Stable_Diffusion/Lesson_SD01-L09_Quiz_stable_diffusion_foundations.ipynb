# Quiz: Stable Diffusion Foundations

This quiz is designed to test your understanding of the fundamental concepts and components of Stable Diffusion, text-to-image generation, and basic prompting techniques.

### Questions

**Question 1: Diffusion Process**
Which of the following best describes the core process of a diffusion model during image generation (inference)?

A) It directly maps a text prompt to a pixel-perfect image.
B) It iteratively adds noise to an image until it becomes unrecognizable.
C) It iteratively removes noise from a random noise tensor, guided by a text prompt, to reveal an image.
D) It generates an image by combining pre-existing image segments based on a prompt.

**Question 2: Key Components**
In Stable Diffusion, what is the primary role of the U-Net architecture?

A) To encode the text prompt into a numerical representation.
B) To compress and decompress images between pixel and latent space.
C) To iteratively denoise the latent representation of an image.
D) To convert the final latent image into a viewable pixel image.

**Question 3: Latent Space**
Why do Stable Diffusion models primarily operate in latent space rather than directly in pixel space?

A) Latent space operations are computationally less expensive and more efficient.
B) Latent space allows for higher resolution image generation without memory constraints.
C) Latent space provides a more intuitive representation for human interaction.
D) Latent space is only used for training, not for inference.

**Question 4: Prompting**
Consider the prompt: "A futuristic city, neon lights, cyberpunk style, highly detailed, 8k, cinematic lighting. Negative prompt: blurry, low quality, deformed."
What is the primary purpose of the "Negative prompt" in this context?

A) To specify elements that *must* be included in the image.
B) To provide additional stylistic guidance for the positive prompt.
C) To instruct the model to *avoid* generating certain characteristics or elements.
D) To define the aspect ratio and resolution of the output image.

**Question 5: Modern Control Techniques**
Which of these modern techniques allows for fine-grained control over the composition and structure of images generated by Stable Diffusion, often by providing an input image (e.g., a sketch, depth map, or pose) alongside the text prompt?

A) LoRA (Low-Rank Adaptation)
B) Textual Inversion
C) ControlNet
D) Hypernetworks


### Answers and Explanations

**Question 1: Diffusion Process**
**Correct Answer: C) It iteratively removes noise from a random noise tensor, guided by a text prompt, to reveal an image.**
*   **Explanation:** Diffusion models start with pure noise and gradually denoise it over several steps. The text prompt provides the conditioning that guides this denoising process towards a desired image. Options A and D describe more direct generation or collage methods, while B describes the *forward* diffusion process (adding noise), not the *reverse* (generation) process.

**Question 2: Key Components**
**Correct Answer: C) To iteratively denoise the latent representation of an image.**
*   **Explanation:** The U-Net is the core component responsible for predicting the noise present in the latent representation at each step of the reverse diffusion process. It takes the noisy latent image and the text embeddings as input and outputs the predicted noise, which is then subtracted to refine the image. Option A is the Text Encoder, B is the VAE Encoder, and D is the VAE Decoder.

**Question 3: Latent Space**
**Correct Answer: A) Latent space operations are computationally less expensive and more efficient.**
*   **Explanation:** Stable Diffusion operates in a compressed latent space, which is a lower-dimensional representation of the image. This significantly reduces the computational resources (memory and processing power) required for the iterative denoising process compared to working directly with high-resolution pixel data. The VAE (Variational Autoencoder) handles the compression to and decompression from this latent space.

**Question 4: Prompting**
**Correct Answer: C) To instruct the model to *avoid* generating certain characteristics or elements.**
*   **Explanation:** A negative prompt is a crucial tool for steering the model away from undesirable features. By specifying terms like "blurry," "low quality," or "deformed," you tell the model what *not* to include, helping to improve the overall quality and adherence to the desired output. The positive prompt (the first part) specifies what *to* include.

**Question 5: Modern Control Techniques**
**Correct Answer: C) ControlNet**
*   **Explanation:** ControlNet is a revolutionary architecture that allows Stable Diffusion models to take an additional input image (e.g., a Canny edge map, depth map, human pose skeleton, or segmentation map) alongside the text prompt. This enables incredibly precise control over the spatial composition, structure, and pose of the generated image, making it a cornerstone for advanced generative workflows in 2026.
