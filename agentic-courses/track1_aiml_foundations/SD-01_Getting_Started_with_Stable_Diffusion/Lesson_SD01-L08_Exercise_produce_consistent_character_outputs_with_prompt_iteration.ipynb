## SD01-L08: Exercise - Produce Consistent Character Outputs with Prompt Iteration

### Task Description

One of the common challenges in text-to-image generation is maintaining the visual consistency of a specific character across multiple images, especially when that character is performing different actions or is placed in varying environments. This exercise challenges you to overcome this by leveraging advanced prompting techniques and iterative refinement.

Your goal is to generate a series of at least **three (3) distinct images** featuring the **same unique character**, performing **different actions or in different settings**, while ensuring the character's appearance remains as consistent as possible across all generations.

### Requirements

1.  **Define a Unique Character**: Create a detailed textual description of a fictional character. This description should be rich enough to allow the AI to consistently render them (e.g., specific hair color, clothing, accessories, facial features, body type).
2.  **Generate Multiple Scenarios**: For your defined character, create at least three distinct prompts that place the character in different situations or performing different actions.
3.  **Iterative Prompting**: You are encouraged to iterate on your prompts. Start with a base character description and then add scenario-specific details. You might need to adjust the base description or add negative prompts to improve consistency.
4.  **Simulated Generation**: Use the provided `MockStableDiffusionAPI` to simulate image generation. Focus on crafting effective prompts, as the mock API will simply return a string representing the generated image based on your prompt.
5.  **Consistency Focus**: The primary objective is visual consistency of the character. Pay attention to details like clothing, facial features, hair, and overall style.

### Evaluation Criteria

*   **Character Detail (30%)**: How well is the character described in the base prompt? Is it unique and specific?
*   **Prompt Iteration (30%)**: Are the scenario-specific prompts distinct and do they effectively integrate the character description?
*   **Simulated Consistency (40%)**: Based on the prompt structure, how likely is it that a real Stable Diffusion model would produce consistent outputs? (Consider how well the character's core attributes are maintained across different scenarios in your prompts).


```python
import random

# Mock Stable Diffusion API for demonstration purposes
# In a real scenario, this would interface with a cloud API or a local model.
class MockStableDiffusionAPI:
    def __init__(self, model_name="sdxl-turbo-v2026"): # Assuming a 2026 state-of-the-art model
        self.model_name = model_name
        print(f"Initialized MockStableDiffusionAPI with model: {self.model_name}")

    def generate_image(self, prompt: str, negative_prompt: str = "", seed: int = None, steps: int = 20, guidance_scale: float = 7.0) -> str:
        """
        Simulates image generation based on a prompt.
        Returns a string representing the 'generated' image description.
        """
        if seed is None:
            seed = random.randint(0, 2**32 - 1)

        print(f"\n--- Generating Image ---")
        print(f"Prompt: {prompt[:100]}...")
        if negative_prompt:
            print(f"Negative Prompt: {negative_prompt[:50]}...")
        print(f"Seed: {seed}")
        print(f"Model: {self.model_name}")
        
        # Simulate a complex generation process by returning a descriptive string
        generated_output = f"[Simulated Image Output - Seed: {seed}] A highly detailed image based on the prompt: '{prompt}'. This output represents a successful generation from {self.model_name}."
        
        return generated_output

# Initialize the mock API
sd_api = MockStableDiffusionAPI()

print("\nSetup complete. You can now use `sd_api.generate_image(prompt, ...)`")
```

### Your Implementation

Now it's your turn! 

1.  Define your unique character using a `base_character_prompt` string.
2.  Create a list of `scenarios` (strings) that your character will be in.
3.  Loop through your scenarios, combining them with your `base_character_prompt` to create full prompts.
4.  Use the `sd_api.generate_image()` function to simulate generating images for each prompt.
5.  Consider using a fixed `seed` for all generations to aid consistency, as this is a powerful technique in real Stable Diffusion workflows.
6.  Optionally, experiment with `negative_prompt` to further refine your character's appearance or remove undesirable elements.


```python
# --- Reference Solution --- 

# 1. Define your unique character with rich details
base_character_prompt = (
    "A young woman named Elara, 25 years old, with long, flowing auburn hair tied in a loose braid, "
    "striking emerald green eyes, a small silver crescent moon pendant around her neck, "
    "wearing a forest green tunic with leather bracers and dark brown practical trousers. "
    "She has a determined yet kind expression, athletic build, and a faint scar above her left eyebrow. "
    "Fantasy art style, highly detailed, cinematic lighting, volumetric fog, digital painting."
)

# 2. Create a list of scenarios for the character
scenarios = [
    "standing confidently on a rocky outcrop overlooking a vast, misty valley at dawn, holding a weathered map.",
    "sitting by a crackling campfire in a dense, ancient forest, sharpening a small dagger, a loyal wolfhound resting nearby.",
    "navigating a bustling, futuristic marketplace, examining exotic fruits, with neon signs reflecting in her eyes."
]

# 3. Define a common negative prompt to avoid common issues
negative_prompt = (
    "blurry, low quality, bad anatomy, deformed, extra limbs, ugly, disfigured, "
    "mutated, poorly drawn hands, missing fingers, duplicate, out of frame, watermark, text, signature"
)

# 4. Set a fixed seed for better consistency across generations
# In real Stable Diffusion, a fixed seed is crucial for consistency.
fixed_seed = 42 # A common choice, but any integer works

print(f"Attempting to generate consistent character images for Elara with fixed seed: {fixed_seed}\n")

generated_image_outputs = []

# 5. Loop through scenarios and generate images
for i, scenario in enumerate(scenarios):
    # Combine the base character prompt with the current scenario
    full_prompt = f"{base_character_prompt} {scenario}"
    
    print(f"--- Scenario {i+1}: {scenario[:50]}... ---")
    
    # Simulate image generation using the mock API
    # Pass the fixed_seed to encourage consistency
    output = sd_api.generate_image(
        prompt=full_prompt,
        negative_prompt=negative_prompt,
        seed=fixed_seed,
        steps=30, # Increased steps for potentially better detail in real models
        guidance_scale=8.0 # Slightly higher guidance for adherence to prompt
    )
    generated_image_outputs.append(output)
    print(f"Generated Output: {output}\n")

print("\n--- All Generations Complete ---")
print("Review the prompts and simulated outputs above. In a real environment, you would now visually inspect the generated images for character consistency.")
print("\nTips for improving consistency in a real Stable Diffusion model:")
print("- **More Detail**: Add even more specific details to your `base_character_prompt` (e.g., 'a specific type of leather on her bracers', 'the exact shade of green for her tunic').")
print("- **Reference Images (ControlNet)**: In advanced workflows, use ControlNet with a reference image of your character.")
print("- **Embeddings/LoRAs**: Train a custom embedding or LoRA on images of your character for superior consistency.")
print("- **Fixed Seed**: Always use a fixed seed for iterative generations of the same character.")
print("- **Negative Prompts**: Refine negative prompts to remove elements that might disrupt consistency (e.g., 'different clothing', 'different hair color').")
```
