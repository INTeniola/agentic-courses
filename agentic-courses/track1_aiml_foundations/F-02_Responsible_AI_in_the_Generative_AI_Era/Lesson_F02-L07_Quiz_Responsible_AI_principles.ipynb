B) To ensure the model's outputs align with ethical guidelines, prevent harmful content generation, and validate factual accuracy where critical.

### Question 5: Bias and Fairness

A) Privacy and Security
B) Robustness and Reliability
C) Fairness and Non-discrimination

### Answers and Explanations

**Question 1: Core Principles**
*   **Correct Answer: C) Profit Maximization**
*   **Explanation:** While businesses aim for profit, it is an organizational objective, not an ethical principle of Responsible AI itself. Core Responsible AI principles focus on societal impact, ethical considerations, and human well-being, such as fairness, transparency, accountability, safety, and privacy.

**Question 2: Generative AI Risks**
*   **Correct Answer: C) Hallucination**
*   **Explanation:** "Hallucination" refers to the phenomenon where generative AI models produce outputs that are plausible-sounding but factually incorrect, nonsensical, or fabricated. This is a significant challenge in ensuring the reliability and trustworthiness of generative AI.

**Question 3: AI Governance (Short Answer)**
*   **Correct Answer:** The primary purpose of an AI Governance Framework in an enterprise setting is to establish clear policies, processes, and responsibilities for the ethical, safe, compliant, and effective development, deployment, and use of AI systems. In the Generative AI era, this framework is crucial for managing novel risks like hallucination, deepfakes, intellectual property infringement, and ensuring alignment with organizational values and regulatory requirements (e.g., EU AI Act, NIST AI RMF).

**Question 4: Human Oversight**
*   **Correct Answer: B) To ensure the model's outputs align with ethical guidelines, prevent harmful content generation, and validate factual accuracy where critical.**
*   **Explanation:** Human-in-the-loop oversight is paramount for generative AI to catch and mitigate outputs that could be biased, harmful, factually incorrect (hallucinations), or otherwise inappropriate. While other options describe important aspects of AI deployment, the ethical and safety validation is the most critical role for human oversight in this context.

**Question 5: Bias and Fairness**
*   **Correct Answer: C) Fairness and Non-discrimination**
*   **Explanation:** When a generative AI model reinforces harmful stereotypes, it directly violates the principle of Fairness and Non-discrimination. This principle demands that AI systems treat all individuals and groups equitably, avoiding outcomes that disadvantage or discriminate against certain demographics based on protected characteristics.
