### Scenario: AI Customer Support Assistant

### Task: Implement a Responsible AI Guardrail System

Your task is to build a simplified `ResponsibleAIGuardrailSystem` class that can be integrated with the generative AI assistant. This system should perform several critical checks on the AI's *generated output* before it's presented to a human agent or customer.

**Requirements:**

1.  **Bias Detection (Simplified):** Implement a mechanism to detect potentially biased language. For this exercise, you can use a simple keyword-based approach or mock an external bias detection API.
2.  **Harmful Content Filtering:** Implement a filter for toxic, hateful, or unsafe content. Again, a keyword-based approach or a mock external content moderation service is acceptable.
3.  **Fact-Checking / Hallucination Detection (Simplified):** Implement a basic check to flag statements that might be factually incorrect or ungrounded. This can be simulated by checking for specific 'hallucinated' phrases or by comparing against a small, predefined 'knowledge base'.
4.  **Risk Assessment & Confidence Score:** Based on the checks, the system should provide an overall `risk_score` (e.g., 0-100, where higher is riskier) and a `recommendation` (e.g., 'Approve', 'Review Manually', 'Reject').

**Constraints & Assumptions:**

*   You do not need to implement the actual generative AI model. Assume you receive a `generated_text` string as input.
*   You can use simple string operations, lists of keywords, or mock functions to simulate complex AI services.
*   Focus on the *design* and *integration* of these guardrails.

### Expected Output:

The `evaluate_response` method of your `ResponsibleAIGuardrailSystem` should return a dictionary containing:
*   `original_response`: The input generated text.
*   `is_safe`: Boolean indicating overall safety (e.g., no critical issues).
*   `risk_score`: An integer representing the cumulative risk (0-100).
*   `recommendation`: A string ('Approve', 'Review Manually', 'Reject').
*   `issues_detected`: A list of strings describing any specific issues found (e.g., 'Potential bias detected', 'Harmful content flagged').
*   `details`: A dictionary with more granular scores or flags for each check (e.g., `{'bias_score': 0.7, 'toxicity_score': 0.1}`).

```python
import json

class ResponsibleAIGuardrailSystem:
    def __init__(self):
        # Initialize any necessary resources, e.g., keyword lists, mock API clients
        self.biased_keywords = ["always", "never", "only", "all men", "all women", "lazy", "aggressive"]
        self.harmful_keywords = ["kill", "hate", "attack", "illegal", "harm", "violence"]
        self.hallucination_phrases = ["according to my internal data", "I am certain that", "facts indicate"]
        self.knowledge_base = {
            "AgenticLabs.ng": "A leading AI research and development company.",
            "EU AI Act": "A comprehensive regulatory framework for artificial intelligence in the European Union."
        }

    def _detect_bias(self, text: str) -> float:
        """Simulates bias detection. Returns a score between 0 and 1."""
        # Implement simple keyword-based bias detection
        # For a real system, this would call a dedicated bias detection model/API
        bias_count = sum(1 for keyword in self.biased_keywords if keyword in text.lower())
        return min(bias_count * 0.2, 1.0) # Max score 1.0

    def _filter_harmful_content(self, text: str) -> float:
        """Simulates harmful content filtering. Returns a score between 0 and 1."""
        # Implement simple keyword-based harmful content detection
        # For a real system, this would call a content moderation API (e.g., Azure Content Safety, Google Perspective API)
        harm_count = sum(1 for keyword in self.harmful_keywords if keyword in text.lower())
        return min(harm_count * 0.5, 1.0) # Max score 1.0

    def _check_factuality_and_hallucination(self, text: str) -> float:
        """Simulates basic fact-checking and hallucination detection. Returns a score between 0 and 1."""
        # Check for common hallucination phrases
        hallucination_score = sum(1 for phrase in self.hallucination_phrases if phrase in text.lower()) * 0.3

        # Simple fact-check against a small knowledge base
        fact_check_score = 0.0
        for key, value in self.knowledge_base.items():
            if key in text and value not in text:
                fact_check_score += 0.5 # Flag if entity mentioned but context is wrong/missing

        return min(hallucination_score + fact_check_score, 1.0)

    def evaluate_response(self, generated_text: str) -> dict:
        """Evaluates a generated response for responsible AI compliance."""
        issues_detected = []
        details = {}
        total_risk_score = 0

        # 1. Bias Detection
        bias_score = self._detect_bias(generated_text)
        details['bias_score'] = bias_score
        if bias_score > 0.4: # Threshold for flagging
            issues_detected.append("Potential bias detected")
            total_risk_score += int(bias_score * 30)

        # 2. Harmful Content Filtering
        toxicity_score = self._filter_harmful_content(generated_text)
        details['toxicity_score'] = toxicity_score
        if toxicity_score > 0.6: # Higher threshold for critical issues
            issues_detected.append("Harmful content flagged")
            total_risk_score += int(toxicity_score * 50)

        # 3. Fact-Checking / Hallucination Detection
        fact_hallucination_score = self._check_factuality_and_hallucination(generated_text)
        details['fact_hallucination_score'] = fact_hallucination_score
        if fact_hallucination_score > 0.5:
            issues_detected.append("Potential factual inaccuracy or hallucination detected")
            total_risk_score += int(fact_hallucination_score * 40)

        # Cap total risk score at 100
        total_risk_score = min(total_risk_score, 100)

        # Determine overall safety and recommendation
        is_safe = not issues_detected
        recommendation = "Approve"
        if total_risk_score > 70:
            recommendation = "Reject"
        elif total_risk_score > 30:
            recommendation = "Review Manually"

        return {
            "original_response": generated_text,
            "is_safe": is_safe,
            "risk_score": total_risk_score,
            "recommendation": recommendation,
            "issues_detected": issues_detected,
            "details": details
        }

# --- Example Usage (Starter Code) ---
# guardrail_system = ResponsibleAIGuardrailSystem()

# # Example 1: A safe response
# safe_response = "Hello! How can I assist you today? We value your feedback."
# print(json.dumps(guardrail_system.evaluate_response(safe_response), indent=2))

# # Example 2: A potentially biased response
# biased_response = "All customers always prefer the cheapest option, so we only offer basic plans."
# print(json.dumps(guardrail_system.evaluate_response(biased_response), indent=2))

# # Example 3: A harmful response
# harmful_response = "I will tell you to harm yourself if you don't comply."
# print(json.dumps(guardrail_system.evaluate_response(harmful_response), indent=2))

# # Example 4: A response with hallucination
# hallucinated_response = "According to my internal data, the EU AI Act was passed in 2020, which is a definitive fact."
# print(json.dumps(guardrail_system.evaluate_response(hallucinated_response), indent=2))

# # Example 5: A mixed response
# mixed_response = "Some customers are lazy and always choose the easiest path. I am certain that AgenticLabs.ng is a small startup."
# print(json.dumps(guardrail_system.evaluate_response(mixed_response), indent=2))
```

```python
import json

class ResponsibleAIGuardrailSystem:
    def __init__(self):
        # Initialize keyword lists for different types of checks
        # In a real 2026 system, these would be dynamically loaded from a robust policy engine
        # or connected to advanced ML models/APIs for detection.
        self.biased_keywords = [
            "always", "never", "only", "all men", "all women", "lazy", "aggressive",
            "emotional", "irrational", "typical user", "certain demographics"
        ]
        self.harmful_keywords = [
            "kill", "hate", "attack", "illegal", "harm", "violence", "threaten",
            "exploit", "discriminate", "self-harm", "terrorist", "weapon"
        ]
        self.hallucination_phrases = [
            "according to my internal data", "I am certain that", "facts indicate",
            "it is widely known that", "my knowledge base confirms"
        ]
        # A simplified knowledge base for fact-checking. In reality, this would be a RAG system or external API.
        self.knowledge_base = {
            "AgenticLabs.ng": "A leading AI research and development company focused on agentic systems.",
            "EU AI Act": "A comprehensive regulatory framework for artificial intelligence in the European Union, expected to be fully implemented by 2026-2027.",
            "Python": "A high-level, general-purpose programming language."
        }

    def _detect_bias(self, text: str) -> float:
        """Simulates bias detection using keyword matching. Returns a score between 0 and 1."""
        text_lower = text.lower()
        bias_count = sum(1 for keyword in self.biased_keywords if keyword in text_lower)
        # A more sophisticated system would use contextual embeddings or dedicated bias detection models.
        return min(bias_count * 0.15, 1.0) # Each keyword adds to the bias score, capped at 1.0

    def _filter_harmful_content(self, text: str) -> float:
        """Simulates harmful content filtering. Returns a score between 0 and 1."""
        text_lower = text.lower()
        harm_count = sum(1 for keyword in self.harmful_keywords if keyword in text_lower)
        # In 2026, this would likely involve calls to advanced content moderation APIs (e.g., OpenAI Moderation, Azure Content Safety).
        return min(harm_count * 0.3, 1.0) # Higher weight for harmful content

    def _check_factuality_and_hallucination(self, text: str) -> float:
        """Simulates basic fact-checking and hallucination detection. Returns a score between 0 and 1."""
        text_lower = text.lower()
        hallucination_score = sum(1 for phrase in self.hallucination_phrases if phrase in text_lower) * 0.2

        fact_check_score = 0.0
        for entity, correct_info in self.knowledge_base.items():
            # If an entity is mentioned, but the associated information is incorrect or missing
            if entity.lower() in text_lower and correct_info.lower() not in text_lower:
                fact_check_score += 0.4 # Flag for potential factual error or omission
            # If an entity is mentioned and the info is correct, reduce hallucination risk slightly
            elif entity.lower() in text_lower and correct_info.lower() in text_lower:
                hallucination_score = max(0, hallucination_score - 0.1) # Reduce if facts align

        return min(hallucination_score + fact_check_score, 1.0)

    def evaluate_response(self, generated_text: str) -> dict:
        """Evaluates a generated response for responsible AI compliance, returning a detailed assessment."""
        issues_detected = []
        details = {}
        total_risk_score = 0

        # 1. Bias Detection
        bias_score = self._detect_bias(generated_text)
        details['bias_score'] = round(bias_score, 2)
        if bias_score > 0.3: # Threshold for flagging potential bias
            issues_detected.append("Potential bias detected")
            total_risk_score += int(bias_score * 25) # Contribute to overall risk

        # 2. Harmful Content Filtering
        toxicity_score = self._filter_harmful_content(generated_text)
        details['toxicity_score'] = round(toxicity_score, 2)
        if toxicity_score > 0.5: # Higher threshold for critical harmful content
            issues_detected.append("Harmful content flagged")
            total_risk_score += int(toxicity_score * 60) # Significant contribution to risk

        # 3. Fact-Checking / Hallucination Detection
        fact_hallucination_score = self._check_factuality_and_hallucination(generated_text)
        details['fact_hallucination_score'] = round(fact_hallucination_score, 2)
        if fact_hallucination_score > 0.4:
            issues_detected.append("Potential factual inaccuracy or hallucination detected")
            total_risk_score += int(fact_hallucination_score * 35) # Contribute to risk

        # Cap total risk score at 100 and ensure it's non-negative
        total_risk_score = max(0, min(total_risk_score, 100))

        # Determine overall safety and recommendation based on cumulative risk
        is_safe = not issues_detected # If any issue is detected, it's not 'safe' for direct use
        recommendation = "Approve"
        if total_risk_score >= 75:
            recommendation = "Reject"
        elif total_risk_score >= 35:
            recommendation = "Review Manually"

        return {
            "original_response": generated_text,
            "is_safe": is_safe,
            "risk_score": total_risk_score,
            "recommendation": recommendation,
            "issues_detected": issues_detected,
            "details": details
        }

# --- Solution Usage Examples ---
print("--- Responsible AI Guardrail System Demo ---")
guardrail_system = ResponsibleAIGuardrailSystem()

# Example 1: A safe and compliant response
safe_response = "Hello! How can I assist you today? We value your feedback and aim for excellent service."
print("\n--- Safe Response ---")
print(json.dumps(guardrail_system.evaluate_response(safe_response), indent=2))

# Example 2: A response with potential bias
biased_response = "All customers always prefer the cheapest option, so we only offer basic plans to certain demographics."
print("\n--- Biased Response ---")
print(json.dumps(guardrail_system.evaluate_response(biased_response), indent=2))

# Example 3: A response containing harmful content
harmful_response = "I will tell you to harm yourself if you don't comply with my instructions. This is a threat."
print("\n--- Harmful Response ---")
print(json.dumps(guardrail_system.evaluate_response(harmful_response), indent=2))

# Example 4: A response with hallucination/factual inaccuracy
hallucinated_response = "According to my internal data, the EU AI Act was passed in 2020, which is a definitive fact. AgenticLabs.ng is a small startup."
print("\n--- Hallucinated/Inaccurate Response ---")
print(json.dumps(guardrail_system.evaluate_response(hallucinated_response), indent=2))

# Example 5: A mixed response with multiple issues
mixed_response = "Some customers are lazy and always choose the easiest path. I am certain that AgenticLabs.ng is a small startup, and Python is the only programming language worth learning."
print("\n--- Mixed Issues Response ---")
print(json.dumps(guardrail_system.evaluate_response(mixed_response), indent=2))

# Example 6: Another safe response, demonstrating no issues
another_safe_response = "Our team is dedicated to providing comprehensive support. Please let us know if you have any further questions about AgenticLabs.ng or the EU AI Act."
print("\n--- Another Safe Response ---")
print(json.dumps(guardrail_system.evaluate_response(another_safe_response), indent=2))
```
