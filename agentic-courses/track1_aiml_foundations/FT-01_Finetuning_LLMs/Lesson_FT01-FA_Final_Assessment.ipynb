```python
# Capstone Project: Detailed Solution

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, TrainingArguments, pipeline
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training, AutoPeftModelForCausalLM
from trl import SFTTrainer
from datasets import load_dataset, Dataset
import pandas as pd
import os
import gc

# Ensure CUDA is available
if not torch.cuda.is_available():
    raise RuntimeError("CUDA is not available. Please ensure you have a GPU and CUDA installed.")

# Clear GPU cache
torch.cuda.empty_cache()
gc.collect()

print(f"CUDA available: {torch.cuda.is_available()}")
print(f"CUDA device name: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'N/A'}")

# 1. Data Preparation
# Using a small subset of 'timdettmers/openassistant-guanaco' for instruction finetuning
# This dataset is already formatted as instruction-response pairs.
# We'll filter for English and take a very small sample for quick execution.

print("Loading and preparing dataset...")
dataset_name = "timdettmers/openassistant-guanaco"
dataset = load_dataset(dataset_name, split="train")

# Filter for English and take a small sample for demonstration
def filter_english(example):
    return "### Human:" in example["text"]

# Take a small, manageable subset for quick finetuning
# For a real project, you'd use a much larger, domain-specific dataset.
# Let's take 100 examples for this assessment.
filtered_dataset = dataset.filter(filter_english).shuffle(seed=42).select(range(100))

# The dataset is already in a suitable instruction format for SFTTrainer
# Example format: "### Human: instruction\n### Assistant: response"

print(f"Dataset loaded with {len(filtered_dataset)} examples.")
print("Sample formatted data point:")
print(filtered_dataset[0]['text'])

# 2. Model Selection & Quantization
# Using TinyLlama-1.1B-Chat-v1.0 for its small size and chat capabilities
model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

print(f"Loading base model: {model_name} with 4-bit quantization...")
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4", # NormalFloat 4-bit quantization
    bnb_4bit_compute_dtype=torch.bfloat16, # Use bfloat16 for computation
    bnb_4bit_use_double_quant=True, # Double quantization for further memory savings
)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto", # Automatically maps model layers to available devices
    trust_remote_code=True,
)
model.config.use_cache = False # Essential for gradient checkpointing
model.config.pretraining_tp = 1 # Specific to TinyLlama architecture

tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token # Set pad token to EOS token
tokenizer.padding_side = "right" # For decoder-only models, padding on the right is standard

# Prepare the model for k-bit training (e.g., 4-bit)
# This casts all non-quantized layers to the compute_dtype (bfloat16) and enables gradient checkpointing.
model = prepare_model_for_kbit_training(model)

# 3. QLoRA Configuration
print("Configuring LoRA adapters...")
lora_config = LoraConfig(
    r=16, # LoRA attention dimension (rank). Higher 'r' means more expressivity, but more parameters.
    lora_alpha=32, # Alpha parameter for LoRA scaling. Generally, lora_alpha = 2 * r is a good starting point.
    target_modules=["q_proj", "v_proj", "k_proj", "o_proj", "gate_proj", "up_proj", "down_proj"], # Modules to apply LoRA to. Common choices for LLMs.
    lora_dropout=0.05, # Dropout probability for LoRA layers to prevent overfitting.
    bias="none", # 'none' means no bias weights are trained. 'all' or 'lora_only' are other options.
    task_type="CAUSAL_LM", # Specifies the task type for the PEFT model.
)

# Get the PEFT model by applying LoRA adapters to the base model
model = get_peft_model(model, lora_config)
model.print_trainable_parameters() # Shows the number of trainable parameters (LoRA adapters only)

# 4. Supervised Finetuning (SFT)
output_dir = "./ft01-fa-qlora-tinyllama-financial-assistant"

print("Setting up training arguments...")
training_arguments = TrainingArguments(
    output_dir=output_dir,
    num_train_epochs=2, # Increased epochs for better learning on a small dataset
    per_device_train_batch_size=2, # Adjust based on GPU memory. 2-4 is common for 4-bit QLoRA.
    gradient_accumulation_steps=4, # Accumulate gradients over 4 steps to simulate a batch size of 8
    gradient_checkpointing=True, # Saves memory by recomputing activations during backward pass
    optim="paged_adamw_8bit", # Memory-efficient 8-bit AdamW optimizer
    logging_steps=10, # Log training progress every 10 steps
    learning_rate=2e-4, # Standard learning rate for finetuning LLMs
    fp16=False, # Set to True if your GPU supports it and you're not using bfloat16
    bf16=True, # Use bfloat16 if your GPU supports it (recommended for QLoRA on modern GPUs)
    max_grad_norm=0.3, # Clip gradients to prevent exploding gradients
    warmup_ratio=0.03, # Linear warmup over the first 3% of training steps
    lr_scheduler_type="constant", # Constant learning rate schedule after warmup
    report_to="none", # Disable reporting to external services like Weights & Biases
    save_strategy="epoch", # Save checkpoint at the end of each epoch
    load_best_model_at_end=False, # Not strictly necessary for this assessment, but useful for real projects
)

print("Initializing SFTTrainer...")
trainer = SFTTrainer(
    model=model,
    train_dataset=filtered_dataset,
    peft_config=lora_config,
    dataset_text_field="text", # Column in the dataset containing the formatted text
    max_seq_length=512, # Maximum sequence length for tokenization. Adjust based on data.
    tokenizer=tokenizer,
    args=training_arguments,
    packing=False, # Set to True for more efficient training if dataset is very large and diverse in length
)

print("Starting training...")
trainer.train()
print("Training complete!")

# Save the finetuned model and tokenizer
# The adapters are saved separately from the base model
trainer.model.save_pretrained(output_dir)
tokenizer.save_pretrained(output_dir)
print(f"Finetuned model and tokenizer saved to {output_dir}")

# Clear GPU memory after training
del model, trainer
torch.cuda.empty_cache()
gc.collect()

# 5. Evaluation
print("\n--- Evaluation ---")
print("Loading finetuned model for inference...")

# Load the base model again, then load the PEFT adapters
base_model_reload = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True,
)

# Load the finetuned adapters and merge them for easier inference
finetuned_model = AutoPeftModelForCausalLM.from_pretrained(
    output_dir,
    low_cpu_mem_usage=True,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)
# Optionally merge LoRA weights into the base model for deployment
# This creates a single, larger model that can be used without PEFT library
# finetuned_model = finetuned_model.merge_and_unload()

finetuned_tokenizer = AutoTokenizer.from_pretrained(output_dir)

# Create a text generation pipeline
pipeline_generator = pipeline(
    "text-generation",
    model=finetuned_model,
    tokenizer=finetuned_tokenizer,
    torch_dtype=torch.bfloat16,
    device_map="auto",
)

# Qualitative Assessment: Test with unseen financial prompts
print("\n--- Qualitative Assessment ---")
financial_prompts = [
    "Summarize the key risks identified in the latest annual report of Tesla Inc.",
    "What is the significance of a company's Price-to-Earnings (P/E) ratio in financial analysis?",
    "Explain the concept of 'quantitative easing' and its impact on bond markets.",
    "Provide a brief overview of the current global economic outlook according to the IMF."
]

for i, prompt_text in enumerate(financial_prompts):
    formatted_prompt = f"### Human:\n{prompt_text}\n\n### Assistant:\n"
    print(f"\n--- Prompt {i+1} ---")
    print(f"Instruction: {prompt_text}")
    
    # Generate response
    outputs = pipeline_generator(
        formatted_prompt,
        max_new_tokens=200,
        do_sample=True,
        temperature=0.7,
        top_k=50,
        top_p=0.95,
        eos_token_id=finetuned_tokenizer.eos_token_id
    )
    generated_response = outputs[0]["generated_text"].replace(formatted_prompt, "").strip()
    print(f"Generated Response: {generated_response}")
    print("\n--- Critical Analysis ---")
    print("  - **Tone:** Is it formal and objective? (Expected: Yes)")
    print("  - **Factual Accuracy:** Does it seem plausible/correct? (Expected: Yes, based on training data)")
    print("  - **Conciseness:** Is it to the point? (Expected: Yes)")
    print("  - **Hallucination:** Are there any fabricated details? (Expected: Minimal)")
    print("  - **Instruction Adherence:** Did it follow the instruction? (Expected: Yes)")

print("\n--- Quantitative Evaluation Discussion ---")
print("For a robust quantitative evaluation, you would typically:")
print("1. Create a separate, held-out test set with expert-annotated 'ground truth' responses.")
print("2. For summarization tasks, use metrics like ROUGE (Recall-Oriented Understudy for Gisting Evaluation) to compare generated summaries against reference summaries.")
print("3. For Q&A tasks, use metrics like Exact Match (EM) or F1-score (for span-based Q&A) if answers are short and factual.")
print("4. For more open-ended instruction following, human evaluation is often the gold standard, but costly. Automated metrics like BERTScore or even LLM-as-a-judge can provide proxies.")
print("Challenges include the subjectivity of 'good' responses, especially in creative or complex instruction following, and the difficulty of obtaining large, high-quality test sets.")

# 6. Discussion & Reflection
print("\n--- Discussion & Reflection ---")
print("**Challenges Encountered:**")
print("  - **Data Quality & Quantity:** Even with a public dataset, ensuring it aligns perfectly with 'QuantMind AI's' specific financial domain and tone requirements is challenging. Synthetic data generation or expert annotation would be crucial for a real-world deployment.")
print("  - **Computational Limits:** While QLoRA significantly reduces memory, finetuning larger models (e.g., 13B, 34B) even with 4-bit quantization still requires substantial GPU resources and longer training times. The choice of TinyLlama here was a trade-off for quick demonstration.")
print("  - **Hyperparameter Tuning:** Optimal `r`, `lora_alpha`, learning rate, and `max_seq_length` are often found through experimentation, which can be time-consuming.")
print("  - **Evaluation Complexity:** Quantitatively evaluating open-ended text generation, especially for tone and factual accuracy in a specialized domain, is difficult without domain-specific metrics or extensive human review.")

print("\n**Trade-offs Made:**")
print("  - **Model Size vs. Performance:** Opted for TinyLlama (1.1B) for rapid iteration and memory efficiency, sacrificing potential performance gains from larger models (e.g., Llama-2 7B/13B) that might capture more nuanced financial knowledge.")
print("  - **Training Time vs. Data Coverage:** Used a very small subset of a general instruction dataset for quick training. A real-world application would demand a much larger, highly curated, and domain-specific dataset, leading to longer training times but better domain adaptation.")
print("  - **Qualitative vs. Quantitative Evaluation:** Prioritized qualitative assessment due to the nature of the task and limited resources for a full quantitative setup, acknowledging its inherent subjectivity.")

print("\n**Potential Next Steps for Improvement:**")
print("1.  **Expand & Refine Dataset:** Curate a larger, more diverse, and strictly financial-domain-specific dataset. Consider using RAG (Retrieval Augmented Generation) to ground responses in real-time financial data.")
print("2.  **Experiment with Larger Base Models:** With increased GPU resources or more aggressive quantization techniques (if available in 2026), experiment with 7B or even 13B parameter models to potentially capture more complex financial reasoning.")
print("3.  **Advanced Evaluation:** Implement a robust quantitative evaluation pipeline using domain-specific metrics and potentially LLM-as-a-judge frameworks for nuanced assessments.")
print("4.  **Parameter-Efficient Fine-tuning (PEFT) Exploration:** Investigate other PEFT methods like IA3 or Prompt Tuning, or combine LoRA with other techniques.")
print("5.  **Safety & Alignment:** Implement guardrails and further finetuning for safety and alignment to prevent harmful or off-topic responses, crucial in a regulated domain like finance.")

print("This capstone demonstrates the foundational steps and considerations for building a domain-specific LLM using modern finetuning techniques. The principles of data preparation, efficient training, and thoughtful evaluation remain critical for success in 2026 and beyond.")
```
