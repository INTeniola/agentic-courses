```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
from datasets import Dataset
import evaluate
import random
import uvicorn
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
import asyncio

# --- Part 1: Setup for Evaluation (Copied from above for self-contained solution) ---

print("Loading mock tokenizer...")
mock_tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")

def mock_model_generate(prompt: str, ground_truth: str, max_new_tokens: int = 50) -> str:
    """Simulates a finetuned LLM's generation by slightly altering the ground truth."""
    words = ground_truth.split()
    if len(words) > 5:
        idx = random.randint(0, len(words) - 1)
        words[idx] = words[idx] + " (modified)"
    
    generated_text = f"Response: {' '.join(words)}"
    
    if len(generated_text.split()) > max_new_tokens:
        generated_text = ' '.join(generated_text.split()[:max_new_tokens]) + "..."
        
    return generated_text

print("Mock model generation function defined.")

mock_test_data = [
    {
        "prompt": "What is the capital of France?",
        "ground_truth": "The capital of France is Paris."
    },
    {
        "prompt": "Explain the concept of quantum entanglement.",
        "ground_truth": "Quantum entanglement is a phenomenon where two or more particles become linked in such a way that they share the same fate, regardless of the distance separating them."
    },
    {
        "prompt": "Summarize the benefits of a Mediterranean diet.",
        "ground_truth": "The Mediterranean diet emphasizes plant-based foods, healthy fats like olive oil, and lean proteins, leading to improved heart health, weight management, and reduced risk of chronic diseases."
    },
    {
        "prompt": "Describe the process of photosynthesis.",
        "ground_truth": "Photosynthesis is the process used by plants, algae, and certain bacteria to convert light energy into chemical energy, producing glucose and oxygen from carbon dioxide and water."
    },
    {
        "prompt": "What are the main challenges in deploying LLMs?",
        "ground_truth": "Deploying LLMs involves challenges such as high computational resource requirements, latency, cost, data privacy, model drift, and ensuring responsible AI practices."
    }
]

print("Mock test dataset created.")

rouge = evaluate.load("rouge")
bleu = evaluate.load("bleu")

print("Evaluation metrics (ROUGE and BLEU) loaded.")

# --- Part 1: Reference Solution - Model Evaluation ---

print("\n--- Starting Model Evaluation ---")

predictions = []
references = []

for item in mock_test_data:
    prompt = item["prompt"]
    ground_truth = item["ground_truth"]
    
    # Simulate model inference
    # In a real scenario, you'd pass the prompt to your actual model.generate()
    generated_text = mock_model_generate(prompt, ground_truth, max_new_tokens=100)
    
    predictions.append(generated_text)
    references.append(ground_truth)
    
    print(f"Prompt: {prompt[:50]}...")
    print(f"Generated: {generated_text[:70]}...")
    print(f"Ground Truth: {ground_truth[:70]}...\n")

# Calculate ROUGE scores
# ROUGE (Recall-Oriented Understudy for Gisting Evaluation) is commonly used for summarization and text generation.
# It compares an automatically produced summary or translation against a reference (human-produced) summary or translation.
rouge_results = rouge.compute(predictions=predictions, references=references)
print("ROUGE Scores:")
for key, value in rouge_results.items():
    print(f"  {key}: {value:.4f}")

# Calculate BLEU score
# BLEU (Bilingual Evaluation Understudy) is a metric for evaluating the quality of text which has been machine-translated
# from one natural language to another. It's also applicable for other text generation tasks.
# Note: BLEU expects references as a list of lists of tokens, or a list of strings.
# The 'evaluate' library handles this conversion if strings are provided.
bleu_results = bleu.compute(predictions=predictions, references=references)
print("\nBLEU Score:")
print(f"  BLEU: {bleu_results['bleu']:.4f}")

print("--- Model Evaluation Complete ---\n")

# --- Part 2: Reference Solution - Model Serving ---

# FastAPI application instance (copied for self-contained solution)
app = FastAPI(title="Finetuned LLM Serving API", version="1.0.0")

# Pydantic model for request body validation (copied for self-contained solution)
class GenerationRequest(BaseModel):
    prompt: str
    max_new_tokens: int = 100 # Default value for generation length

# In a real-world scenario, the model and tokenizer would be loaded globally
# to avoid reloading on each request.
# For this exercise, we'll use the mock_model_generate function directly.

@app.post("/generate")
async def generate_text(request: GenerationRequest):
    """Endpoint to generate text from the finetuned LLM."""
    if not request.prompt or len(request.prompt.strip()) == 0:
        raise HTTPException(status_code=400, detail="Prompt cannot be empty.")
    
    # Simulate model inference with the mock function.
    # In a production setup, you'd call your actual model here:
    # inputs = mock_tokenizer(request.prompt, return_tensors="pt").to("cuda")
    # outputs = mock_model.generate(**inputs, max_new_tokens=request.max_new_tokens)
    # generated_text = mock_tokenizer.decode(outputs[0], skip_special_tokens=True)
    
    # For this exercise, we don't have a direct 'ground_truth' for arbitrary prompts
    # in the serving context. We'll make the mock model generate a generic response.
    # A more sophisticated mock could use a simple rule-based system or a small pre-defined set of responses.
    
    # Let's make the mock model generate a response that acknowledges the prompt.
    # This simulates a model trying to answer a new, unseen prompt.
    mock_response_template = "Thank you for your question. Based on the information provided, here is a generated response: "
    
    # Simple mock logic: append a generic answer or rephrase the prompt
    if "capital of" in request.prompt.lower():
        mock_generated_text = mock_response_template + "The capital you are asking about is a major city in its country."
    elif "explain" in request.prompt.lower() or "describe" in request.prompt.lower():
        mock_generated_text = mock_response_template + "This is a complex topic that involves several key concepts."
    else:
        mock_generated_text = mock_response_template + f"I've processed your request: '{request.prompt[:50]}...'."
        
    # Truncate to max_new_tokens for consistency, even for mock
    if len(mock_generated_text.split()) > request.max_new_tokens:
        mock_generated_text = ' '.join(mock_generated_text.split()[:request.max_new_tokens]) + "..."

    return {"generated_text": mock_generated_text}

print("FastAPI /generate endpoint defined.")

# --- Instructions for running the API ---
print("\n--- API Serving Instructions ---")
print("To run the FastAPI application, save this code as a Python file (e.g., `main.py`).")
print("Then, open your terminal in the same directory and run the command:")
print("  `uvicorn main:app --host 0.0.0.0 --port 8000 --reload`")
print("\nOnce running, you can test the endpoint using `curl` or a tool like Postman/Insomnia:")
print("  `curl -X POST http://localhost:8000/generate -H 'Content-Type: application/json' -d '{"prompt": "What is the weather like today?"}'`")
print("  `curl -X POST http://localhost:8000/generate -H 'Content-Type: application/json' -d '{"prompt": "Explain machine learning."}'`")
print("\nFor production deployments, consider using high-performance inference servers like vLLM or NVIDIA Triton Inference Server for better throughput and lower latency.")

# Optional: Uncomment the following block to run the FastAPI app directly from the notebook
# Note: Running uvicorn directly in a Jupyter notebook can sometimes block the kernel.
# It's generally better to run it from a separate terminal.
# if __name__ == "__main__":
#     print("\nAttempting to run FastAPI app directly from notebook (may block kernel)...")
#     uvicorn.run(app, host="0.0.0.0", port=8000)
```
