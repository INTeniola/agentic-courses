### Answer Key and Explanations

### Question 1: LoRA/QLoRA Advantages

**Correct Answer: B)** QLoRA enables finetuning of models with billions of parameters on consumer-grade GPUs by quantizing the base model and finetuning low-rank adapters.

**Explanation:**
*   **A) Incorrect.** While finetuning can improve accuracy, QLoRA's primary innovation is memory efficiency, not novel attention mechanisms. The accuracy gains come from effective finetuning, not the QLoRA method itself. 
*   **B) Correct.** QLoRA (Quantized LoRA) is a breakthrough technique that quantizes the base LLM to 4-bit precision and then finetunes only a small set of low-rank adapter matrices. This drastically reduces memory footprint and computational requirements, making it feasible to finetune very large models (e.g., 70B parameters) on GPUs with limited VRAM (e.g., 24GB consumer cards). This is a cornerstone for democratizing LLM finetuning in 2026.
*   **C) Incorrect.** QLoRA is a finetuning technique, not a pre-training method. Pre-training involves training a model from scratch on a massive corpus, which is far more resource-intensive than finetuning.
*   **D) Incorrect.** QLoRA, like other finetuning methods, still requires a dataset for supervised finetuning. It's a parameter-efficient finetuning approach, not a data-efficient one in terms of needing no data.

---

### Question 2: Supervised Finetuning (SFT) Data Quality

**Correct Answer: B)** The data should be diverse, high-quality, and closely aligned with the target task's input/output format and style.

**Explanation:**
*   **A) Incorrect.** While more data is generally better, the *quality* and *relevance* of the data are far more important than a raw quantity threshold. A smaller, high-quality dataset can outperform a much larger, noisy, or irrelevant one. The specific number 10 million is arbitrary and often unnecessary for effective SFT.
*   **B) Correct.** High-quality, diverse, and task-aligned data is paramount for successful SFT. The model learns from the patterns, style, and content of the finetuning data. If the data is noisy, inconsistent, or doesn't match the desired output format (e.g., conversational vs. factual summaries), the finetuned model will reflect those deficiencies. Data curation and cleaning are often the most time-consuming but impactful parts of an SFT project.
*   **C) Incorrect.** While synthetic data can augment real datasets, relying *exclusively* on it can lead to models that hallucinate or lack grounding in reality, especially if the synthetic data generation process itself is flawed. A balanced approach often involves a mix of real and carefully curated synthetic data.
*   **D) Incorrect.** While tokenization is crucial, using a 


universal tokenizer


regardless of the base model is incorrect. Each base LLM comes with its own specific tokenizer (e.g., Llama's tokenizer, GPT's tokenizer). Using the *correct* tokenizer for the *specific base model* is essential for proper tokenization and model performance. Mismatched tokenizers will lead to gibberish outputs.

### Question 3: LLM Evaluation in Production

**Correct Answer: B)** Human evaluation for relevance, coherence, and helpfulness, complemented by automated metrics like ROUGE or BERTScore for specific tasks, and latency/throughput benchmarks.

**Explanation:**
*   **A) Incorrect.** Perplexity is a measure of how well a probability model predicts a sample, often used during pre-training or for general language modeling, but it doesn't directly correlate with the quality of generated text for a specific task like customer support. BLEU (Bilingual Evaluation Understudy) is primarily for machine translation and struggles with the open-ended, diverse nature of chatbot responses, often penalizing semantically correct but lexically different answers.
*   **B) Correct.** For a customer support chatbot, human evaluation is indispensable. Humans can assess subjective qualities like relevance, coherence, helpfulness, tone, and safety, which automated metrics often miss. Automated metrics like ROUGE (Recall-Oriented Understudy for Gisting Evaluation) or BERTScore can provide quantitative signals for specific aspects (e.g., factual recall, semantic similarity to a reference) and are useful for tracking progress, but they are not sufficient on their own. Furthermore, in a production setting, operational metrics like latency (response time) and throughput (requests per second) are critical for user experience and system scalability.
*   **C) Incorrect.** Accuracy on a multiple-choice dataset is too narrow for evaluating a generative chatbot. Chatbots need to generate free-form text, not just select options. This metric would only be suitable for specific classification tasks.
*   **D) Incorrect.** F1-score is for classification tasks (like sentiment analysis), not generative text. Cosine similarity can measure semantic similarity but doesn't capture the full spectrum of quality for a generated response (e.g., grammar, fluency, helpfulness).

---

### Question 4: Deployment of QLoRA Finetuned Models

**Correct Answer: B)** Ensuring the serving infrastructure can dynamically merge the LoRA adapters with the quantized base model and efficiently handle inference requests.

**Explanation:**
*   **A) Incorrect.** Reverting QLoRA adapters to full precision would defeat the purpose of QLoRA's memory efficiency. The base model remains quantized (e.g., 4-bit), and the LoRA adapters are typically kept at a higher precision (e.g., 16-bit) but are much smaller. The benefit of QLoRA is maintaining the quantized base model for inference.
*   **B) Correct.** Modern LLM serving frameworks (e.g., vLLM, TGI, TensorRT-LLM) are designed to efficiently load a quantized base model and dynamically merge the small LoRA adapter weights during inference. This allows for rapid switching between different finetuned versions of the same base model (adapter swapping) and minimizes VRAM usage, which is crucial for serving multiple models or handling high throughput on limited hardware. This is a key capability for efficient LLM deployment in 2026.
*   **C) Incorrect.** While possible for very small models or low-traffic scenarios, deploying large LLMs (even QLoRA-finetuned ones) on a single CPU instance is generally not feasible for production due to extremely high latency and low throughput. GPUs are almost always required for practical LLM inference.
*   **D) Incorrect.** The LoRA adapters are specifically trained to modify the weights of a *particular* base model. Using a different base model for inference would result in completely nonsensical outputs, as the adapter weights would not align with the new model's architecture or learned representations.
