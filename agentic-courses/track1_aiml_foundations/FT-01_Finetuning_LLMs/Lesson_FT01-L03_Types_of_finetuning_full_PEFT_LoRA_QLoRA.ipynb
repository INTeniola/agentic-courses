## Types of Finetuning: Full, PEFT, LoRA, QLoRA

Finetuning is the process of taking a pre-trained Large Language Model (LLM) and further training it on a smaller, task-specific dataset to adapt its capabilities. This lesson explores the spectrum of finetuning approaches, from computationally intensive full finetuning to highly efficient parameter-efficient methods like LoRA and QLoRA.

### 1. Full Finetuning: The Traditional Approach

Imagine you have a magnificent, pre-painted masterpiece (your pre-trained LLM). Full finetuning is like taking that entire masterpiece and repainting every single brushstroke to adapt it to a new style or subject. In the context of LLMs, this means updating *all* the parameters (weights and biases) of the pre-trained model on your new dataset.

**How it works:**
*   The entire model architecture is loaded.
*   All layers are unfrozen and made trainable.
*   The model is trained end-to-end on the new dataset.

**Pros:**
*   **Maximum Adaptability:** Potentially achieves the highest performance on the target task as the entire model can learn new representations.
*   **No Compromises:** The model has full capacity to adjust to the new data.

**Cons:**
*   **High Computational Cost:** Requires significant GPU memory and computational power, often demanding multiple high-end GPUs for large LLMs (e.g., 70B+ parameters).
*   **Slow Training:** Training can take days or weeks.
*   **Catastrophic Forgetting:** The model might forget some of its general knowledge acquired during pre-training, especially if the finetuning dataset is small or very different from the pre-training data.
*   **Storage:** Requires storing a full copy of the finetuned model, which can be hundreds of gigabytes.

### 2. Parameter-Efficient Finetuning (PEFT): A Smarter Approach

Given the prohibitive costs of full finetuning, especially for models with hundreds of billions of parameters, researchers developed **Parameter-Efficient Finetuning (PEFT)** methods. Instead of repainting the entire masterpiece, PEFT is like adding a few strategic, new brushstrokes or a small, specialized element to subtly alter its meaning or style without touching the original work extensively.

**Core Idea:** Only finetune a small subset of the model's parameters, or introduce a small number of new, trainable parameters, while keeping the vast majority of the pre-trained weights frozen.

**Benefits of PEFT (in general):**
*   **Reduced Computational Cost:** Significantly less GPU memory and compute required.
*   **Faster Training:** Training completes much quicker.
*   **Mitigates Catastrophic Forgetting:** By keeping most pre-trained weights frozen, the model retains its general knowledge.
*   **Smaller Storage:** Only the small set of finetuned parameters (or adapters) needs to be stored, not the entire model.
