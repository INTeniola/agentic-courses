## Hello World: First Stable Diffusion Inference

To confirm our environment is fully operational, let's run a simple text-to-image generation using a lightweight Stable Diffusion model. This will serve as our "Hello World" for SD-02, ensuring all components are working as expected.

We'll use a distilled version of Stable Diffusion for speed, but the principles apply to larger models like SDXL.


```python
from diffusers import DiffusionPipeline
import torch
from PIL import Image

# Ensure device is set from previous cell
if 'device' not in locals():
    device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Loading model on {device}...")

# Load a lightweight Stable Diffusion model for quick testing
# Using a distilled SDXL model for 2026 readiness and efficiency
model_id = "stabilityai/sdxl-turbo"

# Initialize the pipeline
# For production, consider using `torch.compile` for further optimization (PyTorch 2.0+)
pipe = DiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16 if device == "cuda" else torch.float32)
pipe = pipe.to(device)

# Optional: Enable memory efficient attention if xformers is installed and on CUDA
if device == "cuda" and 'xformers' in globals() and xformers.__version__:
    try:
        pipe.enable_xformers_memory_efficient_attention()
        print("XFormers memory efficient attention enabled.")
    except Exception as e:
        print(f"Could not enable xformers memory efficient attention: {e}")

# Define a simple prompt
prompt = "A futuristic city skyline at sunset, highly detailed, cinematic lighting, 8k"

print(f"Generating image for prompt: '{prompt}'")

# Generate the image
# For SDXL-Turbo, fewer steps are needed for good results
with torch.no_grad(): # Disable gradient calculation for inference
    image = pipe(prompt, num_inference_steps=2, guidance_scale=0.0).images[0]

# Display the image
print("Image generated successfully!")
image.save("sd_hello_world.png")
display(image)

print("Environment check complete. You are ready for advanced Stable Diffusion!")
```
