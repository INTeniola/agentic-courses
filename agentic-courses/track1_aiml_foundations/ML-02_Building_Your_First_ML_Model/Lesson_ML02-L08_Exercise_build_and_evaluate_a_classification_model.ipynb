```python
# Import necessary libraries
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier # Alternative model
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns

# Set a random seed for reproducibility across the notebook
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

print("Libraries imported successfully.")

# --- Setup: Generate a synthetic classification dataset ---
# We'll create a dataset with 1000 samples, 20 features (10 informative, 5 redundant), and 2 classes.
X, y = make_classification(
    n_samples=1000,
    n_features=20,
    n_informative=10,
    n_redundant=5,
    n_classes=2,
    random_state=RANDOM_SEED
)

# Convert to DataFrame for easier inspection (optional but good practice)
df = pd.DataFrame(X, columns=[f'feature_{i}' for i in range(X.shape[1])])
df['target'] = y

print(f"\nDataset generated: {X.shape[0]} samples, {X.shape[1]} features.")
print("First 5 rows of the dataset (features and target):")
print(df.head())
print("\nTarget distribution:")
print(df['target'].value_counts())

# You can add more setup code here if needed, e.g., helper functions
# For this exercise, the dataset generation is the primary setup.
```

## Your Turn! Implement the Classification Workflow

Now it's your turn to complete the exercise. Below, you'll find a code cell where you should implement the remaining steps of the machine learning workflow:

1.  **Split the data** into training and testing sets (`X_train`, `X_test`, `y_train`, `y_test`). Use a test size of `0.2` (20%) and `random_state=RANDOM_SEED`.
2.  **Scale the features** using `StandardScaler`. Remember to fit the scaler only on the training data and then transform both training and testing data.
3.  **Initialize and train a classification model**. You can start with `LogisticRegression` or try `RandomForestClassifier`. Use `random_state=RANDOM_SEED` for the model.
4.  **Make predictions** on the scaled test set.
5.  **Evaluate the model** by calculating and printing:
    *   Accuracy Score
    *   Precision Score
    *   Recall Score
    *   F1 Score
    *   A full `classification_report`
    *   Visualize the `confusion_matrix` using `seaborn.heatmap`.

Good luck!


```python
# --- Reference Solution --- 
# This cell contains the complete, high-quality solution for the exercise.

# 1. Split the data into training and testing sets
# We use a test size of 20% and the predefined RANDOM_SEED for reproducibility.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_SEED)

print(f"\nData split complete:")
print(f"  Training features shape: {X_train.shape}")
print(f"  Testing features shape: {X_test.shape}")
print(f"  Training target shape: {y_train.shape}")
print(f"  Testing target shape: {y_test.shape}")

# 2. Feature Preprocessing: Scale the features using StandardScaler
# StandardScaler transforms features to have a mean of 0 and a standard deviation of 1.
# It's crucial to fit the scaler ONLY on the training data to prevent data leakage.
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) # Fit on training data and transform
X_test_scaled = scaler.transform(X_test)     # Transform test data using the fitted scaler

print("\nFeatures scaled using StandardScaler.")

# 3. Model Training: Initialize and train a classification model
# We'll use Logistic Regression, a robust baseline for binary classification.
# The 'solver' and 'max_iter' parameters are often set for convergence and modern scikit-learn versions.
model = LogisticRegression(random_state=RANDOM_SEED, solver='liblinear', max_iter=1000)

# Train the model using the scaled training data
model.fit(X_train_scaled, y_train)

print("\nLogistic Regression model trained successfully.")

# 4. Prediction: Make predictions on the scaled test set
y_pred = model.predict(X_test_scaled)

print("Predictions made on the test set.")

# 5. Model Evaluation: Calculate and display key classification metrics
print("\n--- Model Evaluation ---")

# Accuracy: Proportion of correctly classified instances
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")

# Precision: Of all positive predictions, how many were actually positive?
# 'weighted' average is used for multi-class, but works for binary too, considering class imbalance.
precision = precision_score(y_test, y_pred, average='weighted')
print(f"Precision: {precision:.4f}")

# Recall: Of all actual positive instances, how many were correctly predicted?
recall = recall_score(y_test, y_pred, average='weighted')
print(f"Recall: {recall:.4f}")

# F1-Score: Harmonic mean of precision and recall, balancing both.
f1 = f1_score(y_test, y_pred, average='weighted')
print(f"F1-Score: {f1:.4f}")

# Classification Report: Provides a text summary of the precision, recall, F1-score for each class.
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Confusion Matrix: A table showing the number of correct and incorrect predictions for each class.
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix (raw):")
print(cm)

# Visualize the Confusion Matrix using Seaborn for better readability
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Predicted 0', 'Predicted 1'],
            yticklabels=['Actual 0', 'Actual 1'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()

print("\n--- Evaluation Complete ---")
print("The model's performance metrics and confusion matrix provide a comprehensive view of its effectiveness.")
print("For this synthetic dataset, the Logistic Regression model performs quite well, indicating good separation between classes.")
```
