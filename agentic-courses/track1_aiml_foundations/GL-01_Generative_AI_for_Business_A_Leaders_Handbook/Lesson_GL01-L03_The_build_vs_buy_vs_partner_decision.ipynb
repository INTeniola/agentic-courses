### Scenario: GlobalGadgets Inc.'s GenAI Ambition

Their current customer support relies on a traditional FAQ chatbot and a human support team, leading to slow resolution times and inconsistent customer experiences. GlobalGadgets' leadership has identified two key areas where Generative AI could provide a significant competitive edge:

````python
import pandas as pd

# --- Setup Code: Decision Framework Template ---

# This dictionary defines the decision criteria and their initial weights.
# Weights should reflect the importance of each factor for GlobalGadgets Inc.
# (e.g., 1 = low importance, 5 = high importance)

decision_criteria = {
    "Cost (Initial & Ongoing)": {"weight": 4, "description": "Financial investment and operational expenses."},
    "Time-to-Market": {"weight": 5, "description": "Speed of deployment and achieving business impact."},
    "Control & Customization": {"weight": 3, "description": "Ability to tailor the solution precisely to brand and needs."},
    "Internal Expertise Required": {"weight": 4, "description": "Level of in-house GenAI skills and resources needed."},
    "Data Sensitivity & Security": {"weight": 5, "description": "Handling proprietary customer data and compliance."},
    "Strategic Importance": {"weight": 4, "description": "How core is this GenAI capability to long-term competitive advantage?"},
    "Scalability & Future-Proofing": {"weight": 3, "description": "Ability to grow with business needs and adapt to new tech."}
}

# This dictionary will store the scores for each option against each criterion.
# Scores should reflect how well an option performs against a criterion (e.g., 1 = poor, 5 = excellent).

decision_options = {
    "Build (In-house)": {},
    "Buy (Off-the-shelf/SaaS)": {},
    "Partner (Vendor/Consultancy)": {}
}

# Helper function to calculate weighted scores
def calculate_weighted_scores(criteria, options):
    results = {}
    for option_name, scores in options.items():
        total_weighted_score = 0
        for criterion_name, criterion_data in criteria.items():
            weight = criterion_data["weight"]
            score = scores.get(criterion_name, 0) # Get score for this option/criterion pair
            total_weighted_score += weight * score
        results[option_name] = total_weighted_score
    return results

print("Setup complete. You can now use 'decision_criteria' and 'decision_options' to structure your analysis.")
print("The 'calculate_weighted_scores' function will help you sum up your findings.")
````

````python
# --- Student Implementation (Reference Solution) ---

# 1. Adjust Weights (if necessary) - Based on GlobalGadgets' scenario, these weights seem appropriate.
#    - Time-to-Market and Data Sensitivity are critical (5).
#    - Cost, Internal Expertise, Strategic Importance are very important (4).
#    - Control and Scalability are important but perhaps slightly less urgent (3).

# No changes to decision_criteria weights for this solution, as the initial setup reflects the scenario well.
# For example, Time-to-Market is crucial for GlobalGadgets to catch up with competitors.
# Data Sensitivity is high due to customer purchase history.
# Internal Expertise is a challenge, making it a high-weight factor.

# 2. Populate Scores for each option against each criterion (1=poor, 5=excellent)

# --- Build (In-house) --- 
# GlobalGadgets has a data science team but lacks GenAI expertise, making this challenging.
decision_options["Build (In-house)"] = {
    "Cost (Initial & Ongoing)": 2, # High initial investment in talent, infrastructure, R&D. Ongoing maintenance is also high.
    "Time-to-Market": 1,         # Very slow. Requires hiring, training, development from scratch.
    "Control & Customization": 5,  # Maximum control and ability to tailor precisely to brand voice and specific needs.
    "Internal Expertise Required": 1, # Requires significant upskilling or hiring of specialized GenAI engineers.
    "Data Sensitivity & Security": 4, # High control over internal data security, but internal team must implement best practices.
    "Strategic Importance": 5,     # Builds core GenAI capability, offering long-term strategic advantage.
    "Scalability & Future-Proofing": 3 # Scalability depends on internal infrastructure. Future-proofing requires continuous R&D.
}

# --- Buy (Off-the-shelf/SaaS) --- 
# Many GenAI platforms exist, offering quick deployment but less customization.
decision_options["Buy (Off-the-shelf/SaaS)"] = {
    "Cost (Initial & Ongoing)": 3, # Moderate subscription costs, potentially lower initial investment than build.
    "Time-to-Market": 5,         # Fastest deployment. Plug-and-play solutions are readily available.
    "Control & Customization": 2,  # Limited customization. Brand voice might be hard to fully integrate. Vendor lock-in.
    "Internal Expertise Required": 4, # Low to moderate. Focus on integration and prompt engineering, not core model development.
    "Data Sensitivity & Security": 3, # Depends heavily on vendor's security posture and data handling policies. Less direct control.
    "Strategic Importance": 2,     # Provides immediate utility but doesn't build unique internal capability. Generic solutions.
    "Scalability & Future-Proofing": 4 # Generally good scalability provided by vendor. Future-proofing depends on vendor's roadmap.
}

# --- Partner (Vendor/Consultancy) --- 
# Collaborating with a specialized GenAI firm to co-develop or integrate.
decision_options["Partner (Vendor/Consultancy)"] = {
    "Cost (Initial & Ongoing)": 3, # Potentially higher than 'buy' but lower than 'build'. Project-based fees.
    "Time-to-Market": 4,         # Faster than build, slower than buy. Involves collaboration and integration.
    "Control & Customization": 4,  # Good balance. Customization possible through collaboration, shared ownership.
    "Internal Expertise Required": 3, # Moderate. Requires internal team to manage the partnership and integrate solutions.
    "Data Sensitivity & Security": 4, # Shared responsibility. Requires strong contracts and trust with partner, but can be tailored.
    "Strategic Importance": 4,     # Builds some internal knowledge through collaboration, potentially leading to future in-house capabilities.
    "Scalability & Future-Proofing": 4 # Scalability can be planned with the partner. Access to partner's expertise for future trends.
}

# 3. Calculate Weighted Scores
final_scores = calculate_weighted_scores(decision_criteria, decision_options)

# Convert to DataFrame for better readability
scores_df = pd.DataFrame(index=decision_criteria.keys())
for option, scores in decision_options.items():
    scores_df[option] = [scores.get(criterion, 0) for criterion in decision_criteria.keys()]

scores_df['Weight'] = [data['weight'] for data in decision_criteria.values()]

# Calculate weighted scores for each criterion and option
for option in decision_options.keys():
    scores_df[f'{option} (Weighted)'] = scores_df[option] * scores_df['Weight']

print("\n--- Detailed Scorecard ---")
print(scores_df)

print("\n--- Final Weighted Scores ---")
for option, score in final_scores.items():
    print(f"{option}: {score}")

# Determine the recommended strategy
recommended_strategy = max(final_scores, key=final_scores.get)
print(f"\nRecommended Strategy: {recommended_strategy}")
````
