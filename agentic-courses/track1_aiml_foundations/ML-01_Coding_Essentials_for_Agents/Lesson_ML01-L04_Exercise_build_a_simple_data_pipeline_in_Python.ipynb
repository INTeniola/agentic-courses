## Exercise: Build a Simple Data Pipeline in Python

### Task Description

In the world of AI and ML, data is the fuel. Before any model can be trained, data often needs to be collected, cleaned, transformed, and prepared. This process is known as a data pipeline. Your task in this exercise is to build a simple, end-to-end data pipeline using Python and the `pandas` library.

You will be provided with a mock dataset representing sensor readings from various devices. Your pipeline should perform the following operations:

1.  **Load Data**: Read the raw data from a CSV file.
2.  **Clean Data**: Handle missing values in a specific column.
3.  **Transform Data**: Create a new feature by combining existing ones.
4.  **Filter Data**: Select only the data that meets certain criteria.
5.  **Save Data**: Store the processed data into a new CSV file.

### Requirements

*   **Cleaning**: Fill any missing values (`NaN`) in the `temperature_celsius` column with the mean of that column.
*   **Transformation**: Create a new column called `energy_index` which is calculated as `(temperature_celsius * humidity_percent) / 100`.
*   **Filtering**: Keep only the rows where `device_status` is 'operational' AND `energy_index` is greater than 50.
*   **Output Data**: Save the final processed DataFrame to a new CSV file named `processed_sensor_data.csv` without the DataFrame index.
*   Your code should be well-commented and easy to understand.
*   Print the head of the DataFrame after each major step (load, clean, transform, filter) to demonstrate the changes.

```python
import pandas as pd
import numpy as np
import os

# --- Setup Code: Generate Mock Dataset ---
# This code block prepares the input data for your exercise.
# You do not need to modify this block.

# Define the path for the mock input CSV file
input_csv_path = 'mock_sensor_data.csv'
output_csv_path = 'processed_sensor_data.csv'

# Create a mock DataFrame with some missing values and varied data
data = {
    'device_id': [f'DEV-{i:03d}' for i in range(1, 11)],
    'timestamp': pd.to_datetime(['2026-01-01 10:00:00', '2026-01-01 10:05:00', '2026-01-01 10:10:00', 
                                 '2026-01-01 10:15:00', '2026-01-01 10:20:00', '2026-01-01 10:25:00',
                                 '2026-01-01 10:30:00', '2026-01-01 10:35:00', '2026-01-01 10:40:00',
                                 '2026-01-01 10:45:00']),
    'temperature_celsius': [22.5, 23.1, np.nan, 21.9, 24.0, 22.8, np.nan, 23.5, 21.0, 24.2],
    'humidity_percent': [60, 62, 58, 65, 70, 61, 63, 68, 59, 72],
    'pressure_hpa': [1012, 1011, 1013, 1010, 1015, 1012, 1014, 1011, 1013, 1016],
    'device_status': ['operational', 'operational', 'maintenance', 'operational', 'operational',
                      'operational', 'maintenance', 'operational', 'operational', 'operational']
}
mock_df = pd.DataFrame(data)

# Introduce an additional NaN for demonstration in a non-target column
mock_df.loc[3, 'pressure_hpa'] = np.nan

# Save the mock DataFrame to a CSV file
mock_df.to_csv(input_csv_path, index=False)

print(f"Mock dataset '{input_csv_path}' created successfully.")
print("Initial mock data head:")
print(mock_df.head())
print("\nInitial mock data info (note NaNs):")
mock_df.info()

# Clean up previous output file if it exists
if os.path.exists(output_csv_path):
    os.remove(output_csv_path)
    print(f"Cleaned up previous output file: {output_csv_path}")
```

### Your Implementation

Now it's your turn! Implement the data pipeline steps as described in the requirements. Use the `input_csv_path` variable to load your data and save your final processed data to `output_csv_path`.

Remember to print the head of your DataFrame after each major step to observe the changes.


```python
import pandas as pd
import numpy as np
import os

# Define file paths (these are already set up in the previous cell, but good practice to re-declare or ensure access)
input_csv_path = 'mock_sensor_data.csv'
output_csv_path = 'processed_sensor_data.csv'

# --- Step 1: Load Data ---
# Load the dataset from the specified CSV file into a pandas DataFrame.
print("\n--- Step 1: Loading Data ---")
raw_df = pd.read_csv(input_csv_path)
print("DataFrame loaded. Head of raw data:")
print(raw_df.head())
print("\nInfo after loading (check for NaNs):")
raw_df.info()

# Create a working copy to perform operations on
df = raw_df.copy()

# --- Step 2: Clean Data ---
# Fill missing values (NaN) in the 'temperature_celsius' column with its mean.
print("\n--- Step 2: Cleaning Data ---")

# Calculate the mean of 'temperature_celsius' before filling NaNs
mean_temperature = df['temperature_celsius'].mean()
print(f"Mean temperature for filling NaNs: {mean_temperature:.2f} C")

# Fill NaNs in 'temperature_celsius' column
df['temperature_celsius'] = df['temperature_celsius'].fillna(mean_temperature)

print("DataFrame after filling missing 'temperature_celsius' values. Head:")
print(df.head())
print("\nInfo after cleaning (check 'temperature_celsius' for NaNs):")
df.info()

# --- Step 3: Transform Data ---
# Create a new column 'energy_index' based on 'temperature_celsius' and 'humidity_percent'.
print("\n--- Step 3: Transforming Data ---")

df['energy_index'] = (df['temperature_celsius'] * df['humidity_percent']) / 100

print("DataFrame after creating 'energy_index'. Head:")
print(df.head())

# --- Step 4: Filter Data ---
# Filter the DataFrame to keep only 'operational' devices with 'energy_index' > 50.
print("\n--- Step 4: Filtering Data ---")

# Define the filtering conditions
condition_status = df['device_status'] == 'operational'
condition_energy = df['energy_index'] > 50

# Apply both conditions to filter the DataFrame
filtered_df = df[condition_status & condition_energy]

print("DataFrame after filtering. Head:")
print(filtered_df.head())
print(f"Original rows: {len(df)}, Filtered rows: {len(filtered_df)}")

# --- Step 5: Save Data ---
# Save the final processed DataFrame to a new CSV file without the index.
print("\n--- Step 5: Saving Data ---")

filtered_df.to_csv(output_csv_path, index=False)

print(f"Processed data saved successfully to '{output_csv_path}'.")

# --- Verification (Optional) ---
# You can optionally load the saved file to verify its content.
print("\n--- Verification: Loading saved file ---")
verified_df = pd.read_csv(output_csv_path)
print("Head of the verified (saved) DataFrame:")
print(verified_df.head())
print("Info of the verified (saved) DataFrame:")
verified_df.info()

# Clean up the generated files after the exercise is complete
# Uncomment the lines below if you want to automatically remove the files
# if os.path.exists(input_csv_path):
#     os.remove(input_csv_path)
#     print(f"Cleaned up input file: {input_csv_path}")
# if os.path.exists(output_csv_path):
#     os.remove(output_csv_path)
#     print(f"Cleaned up output file: {output_csv_path}")
```
