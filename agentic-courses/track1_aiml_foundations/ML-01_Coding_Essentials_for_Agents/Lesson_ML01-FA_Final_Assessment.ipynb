## Part 3: Capstone Project - Detailed Solution


````python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import io
import numpy as np

class DataProcessor:
    def __init__(self, 
                 imputation_strategy: str = 'mean', 
                 numerical_cols: list = None, 
                 categorical_cols: list = None):
        
        self.imputation_strategy = imputation_strategy
        self.numerical_cols = numerical_cols if numerical_cols is not None else []
        self.categorical_cols = categorical_cols if categorical_cols is not None else []
        self.preprocessor = None # Will store the fitted ColumnTransformer
        self.data = None # Stores the original loaded DataFrame
        self.processed_data = None # Stores the processed DataFrame

    def load_data(self, data_source: str = None, df: pd.DataFrame = None):
        """
        Loads data from a CSV file path or accepts a pandas DataFrame directly.
        """
        if data_source:
            try:
                self.data = pd.read_csv(data_source)
                print(f"Data loaded successfully from {data_source}")
            except FileNotFoundError:
                print(f"Error: File not found at {data_source}")
                self.data = None
            except pd.errors.EmptyDataError:
                print(f"Error: CSV file at {data_source} is empty.")
                self.data = None
            except Exception as e:
                print(f"An unexpected error occurred while loading data: {e}")
                self.data = None
        elif df is not None:
            self.data = df.copy() # Use .copy() to avoid modifying the original DataFrame outside the class
            print("Data loaded successfully from DataFrame.")
        else:
            raise ValueError("Either 'data_source' (CSV path) or 'df' (DataFrame) must be provided.")
        
        return self.data

    def preprocess(self) -> pd.DataFrame:
        """
        Applies imputation, scaling, and one-hot encoding to the loaded data.
        Returns the transformed pandas DataFrame.
        """
        if self.data is None:
            raise ValueError("No data loaded. Please call load_data() first.")

        # Define preprocessing steps for numerical features
        numerical_transformer = Pipeline(steps=[
            ('imputer', SimpleImputer(strategy=self.imputation_strategy)),
            ('scaler', StandardScaler())
        ])

        # Define preprocessing steps for categorical features
        # handle_unknown='ignore' is crucial for unseen categories during inference
        categorical_transformer = OneHotEncoder(handle_unknown='ignore')

        # Create a ColumnTransformer to apply different transformers to different columns
        # 'remainder="passthrough"' keeps columns not specified in numerical_cols or categorical_cols
        self.preprocessor = ColumnTransformer(
            transformers=[
                ('num', numerical_transformer, self.numerical_cols),
                ('cat', categorical_transformer, self.categorical_cols)
            ], 
            remainder='passthrough' # Keep other columns (like 'sensor_id') as they are
        )

        # Fit and transform the data
        # The output of ColumnTransformer is a NumPy array
        processed_data_array = self.preprocessor.fit_transform(self.data)
        
        # Get feature names after transformation for the new DataFrame
        # This is a bit tricky with ColumnTransformer and OneHotEncoder
        new_column_names = []
        
        # Add numerical column names (they remain the same after scaling/imputation)
        new_column_names.extend(self.numerical_cols)
        
        # Add one-hot encoded categorical column names
        # get_feature_names_out() is the modern way to get column names from transformers
        if 'cat' in self.preprocessor.named_transformers:
            ohe_feature_names = self.preprocessor.named_transformers_['cat'].get_feature_names_out(self.categorical_cols)
            new_column_names.extend(ohe_feature_names)
            
        # Add names for 'passthrough' columns (columns not transformed)
        # Identify columns that were not part of numerical_cols or categorical_cols
        all_input_cols = set(self.data.columns)
        transformed_cols = set(self.numerical_cols + self.categorical_cols)
        passthrough_cols = list(all_input_cols - transformed_cols)
        new_column_names.extend(passthrough_cols)

        # Convert the processed array back to a DataFrame with correct column names
        # Ensure the order of columns matches the order in new_column_names
        # The ColumnTransformer output order is: numerical, categorical, remainder
        self.processed_data = pd.DataFrame(processed_data_array, columns=new_column_names)
        
        print("Data preprocessing complete.")
        return self.processed_data

    def save_data(self, output_path: str):
        """
        Saves the processed data to a CSV file.
        """
        if self.processed_data is None:
            raise ValueError("No processed data to save. Please load and preprocess data first.")
        
        try:
            self.processed_data.to_csv(output_path, index=False)
            print(f"Processed data saved successfully to {output_path}")
        except Exception as e:
            print(f"An error occurred while saving data to {output_path}: {e}")

# --- Example Usage --- 

# Simulate raw CSV data using io.StringIO
raw_csv_data = """
sensor_id,temperature,humidity,pressure,device_type,status
sensor_001,25.3,60.1,1012.5,TypeA,operational
sensor_002,NaN,62.5,1011.8,TypeB,warning
sensor_003,24.8,59.9,NaN,TypeA,operational
sensor_004,26.1,NaN,1013.2,TypeC,operational
sensor_005,25.5,61.0,1012.9,TypeB,error
"""

# Create a DataFrame from the raw data string
raw_df = pd.read_csv(io.StringIO(raw_csv_data))

print("Original Data:\n", raw_df)
print("\nOriginal Data Info:")
raw_df.info()

# Define columns for preprocessing
numerical_features = ['temperature', 'humidity', 'pressure']
categorical_features = ['device_type', 'status']

# Initialize the DataProcessor
processor = DataProcessor(
    imputation_strategy='mean', # Can also be 'median' or 'constant'
    numerical_cols=numerical_features,
    categorical_cols=categorical_features
)

# Load data into the processor
processor.load_data(df=raw_df)

# Preprocess data
processed_df = processor.preprocess()

# Display processed data
print("\nProcessed Data (first 5 rows):\n", processed_df.head())
print("\nProcessed Data Info:")
processed_df.info()

# Verify imputation and scaling (e.g., mean of temperature should be close to 0, std dev close to 1)
print("\nDescriptive statistics for processed numerical features:")
print(processed_df[numerical_features].describe())

# Save processed data to a dummy CSV file
output_file = 'processed_sensor_data.csv'
processor.save_data(output_file)

# Optional: Load and display the saved data to confirm
# loaded_processed_df = pd.read_csv(output_file)
# print(f"\nData loaded from {output_file}:\n", loaded_processed_df.head())
````
