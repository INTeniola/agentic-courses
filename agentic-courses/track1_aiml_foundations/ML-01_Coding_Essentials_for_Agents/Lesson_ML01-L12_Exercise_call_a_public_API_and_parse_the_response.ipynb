```python
# Solution: Call a Public API and Parse the Response

import requests

# Define the API endpoint and parameters (already defined in setup, but good practice to include)
API_URL = "https://jsonplaceholder.typicode.com/posts"
NUM_POSTS_TO_FETCH = 5

def fetch_and_parse_posts(url: str, limit: int) -> None:
    """
    Fetches posts from a given URL, parses the JSON response,
    and prints details for a specified number of posts.
    """
    print(f"Attempting to fetch {limit} posts from {url}...")
    try:
        # 1. Make a GET Request
        # Using a timeout is crucial for robust network interactions in 2026 and beyond.
        # It prevents your agent from hanging indefinitely on unresponsive servers.
        response = requests.get(url, timeout=10) 

        # 2. Error Handling: Check for HTTP status codes
        # raise_for_status() will raise an HTTPError for bad responses (4xx or 5xx)
        response.raise_for_status()

        # 3. Parse JSON Response
        posts = response.json()

        print("Successfully fetched data. Parsing...")
        print("\n--- Extracted Posts ---")

        # 4. Extract Specific Data & 5. Formatted Output
        # Ensure we don't try to access more posts than available
        posts_to_display = posts[:limit]

        if not posts_to_display:
            print("No posts found or available to display.")
            return

        for i, post in enumerate(posts_to_display):
            post_id = post.get('id', 'N/A')
            title = post.get('title', 'N/A')
            body = post.get('body', 'N/A')

            print(f"\nPost #{i + 1} (ID: {post_id})")
            print(f"  Title: {title}")
            # Truncate body for cleaner output if it's too long
            print(f"  Body: {body[:100]}..." if len(body) > 100 else f"  Body: {body}")
            print("----------------------")

    except requests.exceptions.Timeout:
        print(f"Error: The request timed out after 10 seconds while connecting to {url}.")
        print("This might indicate a slow server or network issues.")
    except requests.exceptions.ConnectionError as e:
        print(f"Error: Could not connect to the API at {url}.")
        print(f"Please check your internet connection or the URL. Details: {e}")
    except requests.exceptions.HTTPError as e:
        print(f"Error: HTTP request failed with status code {e.response.status_code} for {url}.")
        print(f"Details: {e}")
    except requests.exceptions.RequestException as e:
        print(f"An unexpected error occurred during the API request: {e}")
    except ValueError as e:
        print(f"Error: Could not decode JSON response from {url}. Details: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Execute the function
fetch_and_parse_posts(API_URL, NUM_POSTS_TO_FETCH)
```
