### Environment Choices: Colab vs. VS Code

We recommend two primary environments, each with its own advantages:

1.  **Google Colaboratory (Colab)**: A free, cloud-based Jupyter notebook environment that requires zero setup. It's perfect for quick experiments, sharing code, and leveraging free access to GPUs/TPUs, which are essential for many AI tasks. Colab comes pre-installed with most popular ML libraries, making it incredibly convenient.

2.  **Visual Studio Code (VS Code)**: A powerful, open-source code editor that runs locally on your machine. It's highly customizable, supports a vast ecosystem of extensions (especially for Python development), and is an industry standard for professional software development. For AI work, VS Code with the Python extension and proper virtual environment management (like `venv` or `conda`, or even modern `dev containers` for consistent team environments) offers unparalleled control and flexibility.

**For this course, you can choose either, but we will primarily demonstrate code in a way that is compatible with both.** If you're unsure, **start with Google Colab** for its ease of use.

### Colab Specific Setup (Minimal)

If you're using Google Colab, most of your environment is already set up! Python and essential libraries are pre-installed. We'll just run a quick check to confirm your Python version.

### VS Code Specific Setup (Guidance)

If you prefer VS Code:
1.  **Install VS Code**: Download from [code.visualstudio.com](https://code.visualstudio.com/).
2.  **Install Python**: Ensure you have a recent version of Python (3.9+) installed on your system. We recommend using a tool like `pyenv` or `conda` for managing multiple Python versions.
3.  **Install VS Code Python Extension**: Search for "Python" in the VS Code Extensions view and install the official Microsoft extension.
4.  **Create a Virtual Environment**: It's best practice to create a virtual environment for each project to manage dependencies. Open your project folder in VS Code, open the integrated terminal (Ctrl+` or Cmd+`), and run:
    ```bash
    python -m venv .venv
    # Activate it (Linux/macOS)
    source .venv/bin/activate
    # Activate it (Windows PowerShell)
    .venv\Scripts\Activate.ps1
    ```
    VS Code will usually detect and prompt you to select this environment.
5.  **Install Jupyter Extension**: If you plan to run `.ipynb` notebooks directly in VS Code, install the Jupyter extension.
