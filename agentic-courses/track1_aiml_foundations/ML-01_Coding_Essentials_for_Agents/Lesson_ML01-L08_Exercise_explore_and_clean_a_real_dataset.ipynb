```python
import pandas as pd
import numpy as np

# --- Setup: Generate a Synthetic Dataset ---
# This dataset simulates agent performance logs with common data quality issues.

np.random.seed(42) # for reproducibility

num_records = 1000

data = {
    'agent_id': ['AGENT_' + str(i % 100).zfill(3) for i in range(num_records)],
    'timestamp': pd.to_datetime('2025-01-01') + pd.to_timedelta(np.random.randint(0, 365, num_records), unit='D') + pd.to_timedelta(np.random.randint(0, 24*60*60, num_records), unit='s'),
    'task_completed': np.random.randint(10, 100, num_records),
    'response_time_ms': np.random.normal(loc=200, scale=50, size=num_records),
    'task_type': np.random.choice(['Chat', 'Email', 'Voice', 'Bot', 'Other'], num_records, p=[0.3, 0.25, 0.2, 0.15, 0.1]),
    'success_rate': np.random.uniform(0.7, 0.99, num_records),
    'region': np.random.choice(['North', 'South', 'East', 'West'], num_records, p=[0.3, 0.25, 0.25, 0.2]),
    'feedback_score': np.random.randint(1, 6, num_records)
}

df_raw = pd.DataFrame(data)

# Introduce data quality issues

# 1. Missing Values
missing_indices = np.random.choice(df_raw.index, size=int(num_records * 0.05), replace=False)
df_raw.loc[missing_indices[:20], 'task_completed'] = np.nan
df_raw.loc[missing_indices[20:40], 'response_time_ms'] = np.nan
df_raw.loc[missing_indices[40:60], 'task_type'] = np.nan
df_raw.loc[missing_indices[60:80], 'success_rate'] = np.nan
df_raw.loc[missing_indices[80:100], 'region'] = np.nan
df_raw.loc[missing_indices[100:120], 'timestamp'] = np.nan

# 2. Duplicates
duplicate_rows = df_raw.sample(n=10, random_state=1).copy()
df_raw = pd.concat([df_raw, duplicate_rows], ignore_index=True)

# 3. Outliers
df_raw.loc[np.random.choice(df_raw.index, 5), 'response_time_ms'] = np.random.uniform(1000, 5000, 5) # Very high response times
df_raw.loc[np.random.choice(df_raw.index, 3), 'task_completed'] = np.random.randint(200, 500, 3) # Very high tasks completed
df_raw.loc[np.random.choice(df_raw.index, 2), 'success_rate'] = 1.5 # Invalid success rate
df_raw.loc[np.random.choice(df_raw.index, 2), 'success_rate'] = -0.2 # Invalid success rate

# 4. Incorrect Data Types / Inconsistent Formats
df_raw['timestamp'] = df_raw['timestamp'].astype(str) # Convert to string to simulate incorrect type
df_raw.loc[np.random.choice(df_raw.index, 5), 'timestamp'] = 'invalid_date'
df_raw.loc[np.random.choice(df_raw.index, 5), 'task_type'] = 'chat ' # Trailing space
df_raw.loc[np.random.choice(df_raw.index, 5), 'task_type'] = 'EMAIL' # Different casing
df_raw.loc[np.random.choice(df_raw.index, 5), 'region'] = 'east '

# 5. Negative values where not expected
df_raw.loc[np.random.choice(df_raw.index, 5), 'response_time_ms'] = np.random.uniform(-100, -1, 5)

print(f"Raw DataFrame shape: {df_raw.shape}")
print("Raw DataFrame head:")
print(df_raw.head())
print("\nRaw DataFrame info:")
df_raw.info()
print("\nRaw DataFrame missing values:")
print(df_raw.isnull().sum())

# Make a copy for students to work on
df_student = df_raw.copy()
```

```python
import pandas as pd
import numpy as np

# Assuming df_student is already loaded from the setup cell
# df_student = df_raw.copy() # Uncomment if running this cell independently

print("--- Starting Data Cleaning Process ---")
print(f"Initial DataFrame shape: {df_student.shape}")

# 1. Initial Exploration (Re-confirming before cleaning)
print("\n--- Initial Data Exploration ---")
print("First 5 rows of the raw data:")
print(df_student.head())
print("\nDataFrame Info (before cleaning):")
df_student.info()
print("\nDescriptive Statistics (before cleaning):")
print(df_student.describe())
print("\nMissing values count (before cleaning):")
print(df_student.isnull().sum())

# Make a copy to work on, preserving the original student copy for comparison if needed
df_cleaned = df_student.copy()

# 2. Handle Duplicates
# Identify and remove duplicate rows. This is usually one of the first steps
# to ensure that subsequent analyses are not skewed by redundant entries.
initial_rows = df_cleaned.shape[0]
duplicate_rows_count = df_cleaned.duplicated().sum()
df_cleaned.drop_duplicates(inplace=True)
print(f"\n--- Handling Duplicates ---")
print(f"Found {duplicate_rows_count} duplicate rows. Removed them.")
print(f"DataFrame shape after removing duplicates: {df_cleaned.shape}")

# 3. Correct Data Types and Handle Invalid Entries
# Convert 'timestamp' to datetime objects. Invalid dates will be coerced to NaT (Not a Time).
# 'errors='coerce'' is crucial here to handle the 'invalid_date' strings gracefully.
print("\n--- Correcting Data Types ---")
df_cleaned['timestamp'] = pd.to_datetime(df_cleaned['timestamp'], errors='coerce')
print("Converted 'timestamp' to datetime, coercing invalid entries to NaT.")

# Convert numerical columns to appropriate types if they were objects due to NaNs or errors
# This is often handled implicitly by fillna or other operations, but good to be explicit.
# For 'task_completed' and 'feedback_score', ensure they are integer types after handling NaNs.
# For 'response_time_ms' and 'success_rate', ensure they are float types.
# We'll handle NaNs for these in the next step, then convert to int where appropriate.

# 4. Handle Missing Values
print("\n--- Handling Missing Values ---")
print("Missing values before handling:\n", df_cleaned.isnull().sum())

# For 'timestamp': Drop rows where timestamp is NaT, as a valid timestamp is critical for logs.
# Alternatively, one could impute with a common time or the mean, but for logs, invalid times are often unrecoverable.
initial_timestamp_nan = df_cleaned['timestamp'].isnull().sum()
df_cleaned.dropna(subset=['timestamp'], inplace=True)
print(f"Dropped {initial_timestamp_nan} rows with missing or invalid 'timestamp'.")

# For 'task_completed' (numerical, count): Impute with the median.
# Median is preferred over mean for skewed distributions or presence of outliers, which is common for counts.
median_task_completed = df_cleaned['task_completed'].median()
df_cleaned['task_completed'].fillna(median_task_completed, inplace=True)
print(f"Imputed 'task_completed' missing values with median: {median_task_completed}.")

# For 'response_time_ms' (numerical, continuous): Impute with the mean.
# Mean is suitable if the distribution is relatively symmetric and outliers are handled separately.
mean_response_time = df_cleaned['response_time_ms'].mean()
df_cleaned['response_time_ms'].fillna(mean_response_time, inplace=True)
print(f"Imputed 'response_time_ms' missing values with mean: {mean_response_time:.2f}.")

# For 'task_type' (categorical): Impute with the mode (most frequent category).
# This is a common strategy for categorical data to preserve distribution.
mode_task_type = df_cleaned['task_type'].mode()[0]
df_cleaned['task_type'].fillna(mode_task_type, inplace=True)
print(f"Imputed 'task_type' missing values with mode: '{mode_task_type}'.")

# For 'success_rate' (numerical, ratio): Impute with the mean.
# Success rates are often well-behaved, so mean is a reasonable choice.
mean_success_rate = df_cleaned['success_rate'].mean()
df_cleaned['success_rate'].fillna(mean_success_rate, inplace=True)
print(f"Imputed 'success_rate' missing values with mean: {mean_success_rate:.2f}.")

# For 'region' (categorical): Impute with the mode.
mode_region = df_cleaned['region'].mode()[0]
df_cleaned['region'].fillna(mode_region, inplace=True)
print(f"Imputed 'region' missing values with mode: '{mode_region}'.")

print("Missing values after handling:\n", df_cleaned.isnull().sum())

# 5. Standardize Categorical Data
print("\n--- Standardizing Categorical Data ---")
# Convert 'task_type' and 'region' to lowercase and strip whitespace to ensure consistency.
# This helps in accurate grouping and analysis of categorical features.
df_cleaned['task_type'] = df_cleaned['task_type'].str.lower().str.strip()
df_cleaned['region'] = df_cleaned['region'].str.lower().str.strip()
print("Standardized 'task_type' and 'region' to lowercase and stripped whitespace.")

# 6. Handle Outliers and Invalid Values
print("\n--- Handling Outliers and Invalid Values ---")

# For 'response_time_ms': Cap extreme values.
# Negative response times are physically impossible. High response times might be outliers.
# We'll cap negative values at 0 and cap upper outliers using IQR method.
Q1_rt = df_cleaned['response_time_ms'].quantile(0.25)
Q3_rt = df_cleaned['response_time_ms'].quantile(0.75)
IQR_rt = Q3_rt - Q1_rt
upper_bound_rt = Q3_rt + 1.5 * IQR_rt

# Cap negative values at 0
initial_negative_rt = (df_cleaned['response_time_ms'] < 0).sum()
df_cleaned['response_time_ms'] = np.where(df_cleaned['response_time_ms'] < 0, 0, df_cleaned['response_time_ms'])
print(f"Capped {initial_negative_rt} negative 'response_time_ms' values to 0.")

# Cap upper outliers
initial_outlier_rt = (df_cleaned['response_time_ms'] > upper_bound_rt).sum()
df_cleaned['response_time_ms'] = np.where(df_cleaned['response_time_ms'] > upper_bound_rt, upper_bound_rt, df_cleaned['response_time_ms'])
print(f"Capped {initial_outlier_rt} upper outliers in 'response_time_ms' using IQR method (upper bound: {upper_bound_rt:.2f}).")

# For 'success_rate': Ensure values are within [0, 1].
# Values outside this range are invalid and should be capped.
initial_invalid_sr = ((df_cleaned['success_rate'] < 0) | (df_cleaned['success_rate'] > 1)).sum()
df_cleaned['success_rate'] = np.clip(df_cleaned['success_rate'], 0, 1)
print(f"Capped {initial_invalid_sr} 'success_rate' values to be within [0, 1].")

# 7. Final Data Type Conversions (after handling NaNs and outliers)
# Convert 'task_completed' and 'feedback_score' to integer type.
# This is safe now that NaNs have been handled.
df_cleaned['task_completed'] = df_cleaned['task_completed'].astype(int)
df_cleaned['feedback_score'] = df_cleaned['feedback_score'].astype(int)
print("Converted 'task_completed' and 'feedback_score' to integer types.")

# Convert categorical columns to 'category' dtype for memory efficiency and better performance with some ML models.
df_cleaned['task_type'] = df_cleaned['task_type'].astype('category')
df_cleaned['region'] = df_cleaned['region'].astype('category')
print("Converted 'task_type' and 'region' to 'category' dtype.")

# 8. Feature Engineering
print("\n--- Feature Engineering ---")
# Create a 'performance_score' based on tasks completed and inverse response time.
# A higher task_completed and lower response_time_ms should result in a higher score.
# Add a small constant to response_time_ms to avoid division by zero.
# Normalize response_time_ms to prevent it from dominating the score.

# Simple approach: (tasks_completed / log(response_time_ms + 1)) * success_rate
# Or a more direct weighted sum/product

# Let's create a simple 'efficiency_score': tasks_completed / (response_time_ms / 1000) * success_rate
# The division by 1000 for response_time_ms converts it to seconds, making the ratio more interpretable.
# Add a small epsilon to avoid division by zero if response_time_ms could be 0.
epsilon = 1e-6
df_cleaned['efficiency_score'] = (df_cleaned['task_completed'] / (df_cleaned['response_time_ms'] / 1000 + epsilon)) * df_cleaned['success_rate']
print("Created new feature 'efficiency_score'.")

# Another feature: 'hour_of_day' from timestamp
df_cleaned['hour_of_day'] = df_cleaned['timestamp'].dt.hour
print("Created new feature 'hour_of_day' from timestamp.")

# 9. Final Review
print("\n--- Final Review of Cleaned Data ---")
print(f"Final DataFrame shape: {df_cleaned.shape}")
print("\nFirst 5 rows of the cleaned data:")
print(df_cleaned.head())
print("\nDataFrame Info (after cleaning):")
df_cleaned.info()
print("\nDescriptive Statistics (after cleaning):")
print(df_cleaned.describe())
print("\nMissing values count (after cleaning):")
print(df_cleaned.isnull().sum())

print("\n--- Data Cleaning Process Completed ---")
```
