````python
# Part 2: Capstone Coding Project - Detailed Solution

import torch
import torchvision
from torchvision.models.detection import FasterRCNN
from torchvision.models.detection.rpn import AnchorGenerator
from torchvision.transforms import functional as F
from torch.utils.data import DataLoader, Dataset
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
import os
from PIL import Image
import random
import json
from tqdm.notebook import tqdm # For progress bars

# Ensure torchmetrics is installed: pip install torchmetrics
from torchmetrics.detection import MeanAveragePrecision

# --- Configuration --- #
# Number of classes: 4 product classes + 1 background class
NUM_CLASSES = 5  
BATCH_SIZE = 4
LEARNING_RATE = 0.005
NUM_EPOCHS = 10 # Increased epochs for better training simulation
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# --- 1. Custom Dataset and DataLoader --- #

# Helper function to generate dummy COCO-like annotations
def generate_dummy_coco_annotations(num_images, img_width=800, img_height=600, num_classes=NUM_CLASSES-1):
    annotations = {
        "images": [],
        "annotations": [],
        "categories": []
    }
    
    for i in range(1, num_classes + 1):
        annotations["categories"].append({"id": i, "name": f"Product_{i}"})

    ann_id_counter = 0
    for img_id in range(num_images):
        img_filename = f'image_{img_id:05d}.jpg'
        annotations["images"].append({"id": img_id, "file_name": img_filename, "width": img_width, "height": img_height})
        
        num_objects = random.randint(1, 8) # Simulate 1 to 8 objects per image
        for _ in range(num_objects):
            x_min = random.randint(0, img_width - 100)
            y_min = random.randint(0, img_height - 100)
            width = random.randint(50, min(200, img_width - x_min))
            height = random.randint(50, min(200, img_height - y_min))
            
            bbox = [x_min, y_min, width, height] # COCO format: [x, y, width, height]
            category_id = random.randint(1, num_classes)
            
            annotations["annotations"].append({
                "id": ann_id_counter,
                "image_id": img_id,
                "bbox": bbox,
                "area": width * height,
                "category_id": category_id,
                "iscrowd": 0
            })
            ann_id_counter += 1
            
    return annotations

class RetailShelfDataset(Dataset):
    def __init__(self, root_dir, coco_annotations, transforms=None):
        self.root_dir = root_dir
        self.transforms = transforms
        self.coco_annotations = coco_annotations
        
        self.img_ids = [img['id'] for img in coco_annotations['images']]
        self.img_id_to_info = {img['id']: img for img in coco_annotations['images']}
        
        # Map image ID to its annotations
        self.img_id_to_anns = {img_id: [] for img_id in self.img_ids}
        for ann in coco_annotations['annotations']:
            self.img_id_to_anns[ann['image_id']].append(ann)

    def __len__(self):
        return len(self.img_ids)

    def __getitem__(self, idx):
        img_id = self.img_ids[idx]
        img_info = self.img_id_to_info[img_id]
        img_filename = img_info['file_name']
        
        # In a real scenario, load image from self.root_dir / img_filename
        # For this solution, we'll generate a dummy image for demonstration.
        # In a real setup, you'd have actual image files.
        # Example: img = Image.open(os.path.join(self.root_dir, img_filename)).convert("RGB")
        img = Image.fromarray(np.random.randint(0, 255, (img_info['height'], img_info['width'], 3), dtype=np.uint8))

        # Parse annotations for the current image
        annotations = self.img_id_to_anns[img_id]
        
        boxes = []
        labels = []
        for ann in annotations:
            x, y, w, h = ann['bbox']
            boxes.append([x, y, x + w, y + h]) # Convert COCO [x,y,w,h] to [x_min, y_min, x_max, y_max]
            labels.append(ann['category_id'])
        
        boxes = torch.as_tensor(boxes, dtype=torch.float32)
        labels = torch.as_tensor(labels, dtype=torch.int64)
        
        # Create target dictionary for PyTorch Faster R-CNN
        target = {}
        target["boxes"] = boxes
        target["labels"] = labels
        target["image_id"] = torch.tensor([img_id])
        
        # Apply transformations
        if self.transforms:
            img, target = self.transforms(img, target)

        return img, target

# Define transformations
class Compose(object):
    def __init__(self, transforms):
        self.transforms = transforms

    def __call__(self, image, target):
        for t in self.transforms:
            image, target = t(image, target)
        return image, target

class ToTensor(object):
    def __call__(self, image, target):
        image = F.to_tensor(image)
        return image, target

class RandomHorizontalFlip(object):
    def __init__(self, prob=0.5):
        self.prob = prob

    def __call__(self, image, target):
        if random.random() < self.prob:
            width, height = image.size # PIL Image size (width, height)
            image = F.hflip(image)
            
            bbox = target["boxes"]
            # Flip x coordinates: x_min becomes width - x_max, x_max becomes width - x_min
            bbox_flipped = bbox.clone()
            bbox_flipped[:, 0] = width - bbox[:, 2] # new x_min
            bbox_flipped[:, 2] = width - bbox[:, 0] # new x_max
            target["boxes"] = bbox_flipped
        return image, target

def get_transform(train):
    transforms = []
    transforms.append(ToTensor())
    if train:
        transforms.append(RandomHorizontalFlip(0.5))
    return Compose(transforms)

# Collate function for DataLoader
def collate_fn(batch):
    return tuple(zip(*batch))

# --- 2. Model Definition --- #

def get_object_detection_model(num_classes):
    # Load a pre-trained Faster R-CNN model with a ResNet-50 FPN backbone
    # This model is robust and a good starting point for transfer learning.
    model = torchvision.models.detection.fasterrcnn_resnet50_fpn(pretrained=True)

    # Get the number of input features for the classifier head
    in_features = model.roi_heads.box_predictor.cls_score.in_features
    # Replace the pre-trained head with a new one for our specific number of classes.
    # The +1 is for the background class, which Faster R-CNN expects.
    model.roi_heads.box_predictor = torchvision.models.detection.faster_rcnn.FastRCNNPredictor(in_features, num_classes)

    return model

# --- 3. Training Function --- #

def train_one_epoch(model, optimizer, data_loader, device, epoch):
    model.train()
    total_loss = 0
    # Use tqdm for a nice progress bar
    for i, (images, targets) in enumerate(tqdm(data_loader, desc=f"Epoch {epoch+1} Training")):
        images = list(image.to(device) for image in images)
        targets = [{k: v.to(device) for k, v in t.items()} for t in targets]

        # Forward pass: model returns a dictionary of losses
        loss_dict = model(images, targets)
        losses = sum(loss for loss in loss_dict.values())

        # Backward pass and optimization
        optimizer.zero_grad()
        losses.backward()
        optimizer.step()

        total_loss += losses.item()

    avg_loss = total_loss / len(data_loader)
    return avg_loss

# --- 4. Evaluation Function --- #

def evaluate_model(model, data_loader, device):
    model.eval()
    # Initialize MeanAveragePrecision metric from torchmetrics
    metric = MeanAveragePrecision(box_format='xyxy', iou_type='bbox', class_metrics=True)
    
    with torch.no_grad():
        for images, targets in tqdm(data_loader, desc="Evaluating"): 
            images = list(image.to(device) for image in images)
            
            # Model returns predictions in a list of dicts
            predictions = model(images)
            
            # Move targets to CPU for metric calculation if needed, or ensure they are tensors
            targets_cpu = [{k: v.cpu() for k, v in t.items()} for t in targets]
            predictions_cpu = [{k: v.cpu() for k, v in p.items()} for p in predictions]
            
            # Update the metric with current batch's predictions and targets
            metric.update(predictions_cpu, targets_cpu)
            
    # Compute the final mAP results
    map_results = metric.compute()
    print("\n--- Evaluation Results ---")
    print(f"mAP: {map_results['map'].item():.4f}")
    print(f"mAP_50: {map_results['map_50'].item():.4f}")
    print(f"mAP_75: {map_results['map_75'].item():.4f}")
    # You can also inspect class-specific mAP if class_metrics=True
    # print(f"Class-specific mAP: {map_results['map_per_class']}")
    print("--------------------------\n")
    return map_results

# --- 5. Inference and Visualization Function --- #

def visualize_predictions(model, image_data, class_names, device, threshold=0.7):
    model.eval()
    
    # image_data can be a PIL Image or a path to a dummy image
    if isinstance(image_data, str):
        if not os.path.exists(image_data):
            print(f"Warning: Image '{image_data}' not found. Generating a dummy image for visualization.")
            image = Image.fromarray(np.random.randint(0, 255, (600, 800, 3), dtype=np.uint8))
        else:
            image = Image.open(image_data).convert("RGB")
    else: # Assume it's already a PIL Image
        image = image_data

    img_tensor = F.to_tensor(image).to(device)

    with torch.no_grad():
        # Model expects a list of tensors
        prediction = model([img_tensor])

    # Process predictions from the first (and only) image in the batch
    boxes = prediction[0]['boxes'].cpu().numpy()
    labels = prediction[0]['labels'].cpu().numpy()
    scores = prediction[0]['scores'].cpu().numpy()

    fig, ax = plt.subplots(1, figsize=(12, 12))
    ax.imshow(image)

    for box, label, score in zip(boxes, labels, scores):
        if score > threshold:
            xmin, ymin, xmax, ymax = box
            # Create a Rectangle patch
            rect = patches.Rectangle((xmin, ymin), xmax - xmin, ymax - ymin, 
                                     linewidth=2, edgecolor='r', facecolor='none')
            ax.add_patch(rect)
            # Add text label with score
            ax.text(xmin, ymin - 10, f'{class_names[label]}: {score:.2f}', 
                    color='red', fontsize=12, bbox=dict(facecolor='white', alpha=0.7))
    
    ax.axis('off')
    plt.title(f"Detected Objects (Threshold: {threshold})")
    plt.show()

# --- Main Execution Block --- #
if __name__ == '__main__':
    # Create dummy directories for simulated dataset
    os.makedirs('data/images', exist_ok=True)
    # In a real scenario, populate 'data/images' with actual images.

    # Define class names (index 0 is background, 1-4 are products)
    CLASS_NAMES = {
        0: 'background',
        1: 'Cereal Box',
        2: 'Milk Carton',
        3: 'Soda Can',
        4: 'Snack Bag'
    }

    # Generate simulated COCO-like annotations
    print("Generating simulated dataset...")
    num_train_images = 100 # More images for better simulation
    num_val_images = 20
    train_coco_annotations = generate_dummy_coco_annotations(num_train_images)
    val_coco_annotations = generate_dummy_coco_annotations(num_val_images)

    # Initialize datasets and dataloaders
    train_dataset = RetailShelfDataset(root_dir='data/images', coco_annotations=train_coco_annotations, transforms=get_transform(train=True))
    val_dataset = RetailShelfDataset(root_dir='data/images', coco_annotations=val_coco_annotations, transforms=get_transform(train=False))

    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate_fn, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_fn, num_workers=2)

    # Initialize model
    # Model Choice Justification: Faster R-CNN with ResNet-50 FPN is a robust and widely used two-stage detector.
    # It offers a good balance between accuracy and speed, making it suitable for retail monitoring where
    # precise localization and classification are important. Its pre-trained weights on COCO provide an excellent
    # starting point for transfer learning on our custom product dataset.
    model = get_object_detection_model(NUM_CLASSES)
    model.to(DEVICE)

    # Initialize optimizer and learning rate scheduler
    # We optimize all parameters since we're fine-tuning the entire network.
    params = [p for p in model.parameters() if p.requires_grad]
    optimizer = torch.optim.SGD(params, lr=LEARNING_RATE, momentum=0.9, weight_decay=0.0005)
    # StepLR reduces the learning rate by a factor of gamma every step_size epochs.
    # This helps in fine-tuning the model more carefully in later stages.
    lr_scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=3, gamma=0.1)

    print(f"Starting training on {DEVICE} for {NUM_EPOCHS} epochs...")
    for epoch in range(NUM_EPOCHS):
        avg_loss = train_one_epoch(model, optimizer, train_loader, DEVICE, epoch)
        lr_scheduler.step()
        print(f"Epoch {epoch+1} Average Loss: {avg_loss:.4f}")

        # Evaluate on validation set after each epoch
        evaluate_model(model, val_loader, DEVICE)

    print("Training complete!")

    # --- Inference Example --- #
    print("\n--- Running Inference Example ---")
    # Create a dummy image for inference visualization
    dummy_inference_image = Image.fromarray(np.random.randint(0, 255, (600, 800, 3), dtype=np.uint8))
    visualize_predictions(model, dummy_inference_image, CLASS_NAMES, DEVICE, threshold=0.7)

    # Save the trained model state dictionary
    model_save_path = 'retail_shelf_detector.pth'
    torch.save(model.state_dict(), model_save_path)
    print(f"Model saved to {model_save_path}")

    # --- Conceptual Discussion for Deployment --- #
    print("\n--- Conceptual Discussion: Deployment and Next Steps ---")
    print("Observed Performance: (Based on simulated data, actual performance would vary)")
    print("During training, we expect the average loss to decrease over epochs, indicating the model is learning.")
    print("Evaluation metrics like mAP_50 and mAP_75 on the validation set would show the model's detection and localization accuracy.")
    print("A good mAP_50 (e.g., >0.7) would suggest reasonable performance for many retail applications.")
    print("\nNext Steps for Deployment:")
    print("1.  **Real Data Collection & Annotation:** The most crucial step. Collect a large, diverse dataset of actual retail shelf images with high-quality annotations.")
    print("2.  **Hyperparameter Tuning:** Systematically tune learning rate, batch size, optimizer, and scheduler parameters using techniques like grid search or Bayesian optimization.")
    print("3.  **Advanced Augmentation:** Implement more sophisticated augmentation (e.g., CutMix, Mosaic, or generative augmentations) to improve robustness.")
    print("4.  **Model Optimization:** For edge deployment (e.g., on a shelf camera), consider model quantization (e.g., INT8), pruning, or using smaller, faster models (e.g., MobileNet-SSD, YOLO-Nano). Tools like ONNX Runtime or TensorRT would be used for deployment optimization.")
    print("5.  **Robustness Testing:** Test the model under various lighting conditions, occlusions, product variations, and camera angles.")
    print("6.  **Integration with Inventory Systems:** Develop APIs to integrate the model's output (detected products, empty slots) with existing inventory management and replenishment systems.")
    print("7.  **Monitoring & Retraining:** Implement continuous monitoring of model performance in production and set up a pipeline for periodic retraining with new data to adapt to changing product layouts or new items.")
    print("8.  **Ethical Review:** Conduct a thorough ethical review to ensure fairness, privacy (e.g., no facial recognition of shoppers), and transparency in the system's operation.")
````
