### Answer Key and Explanations

### Question 1: Vision Transformer Image Processing
**Correct Answer: B)** Divides the image into fixed-size patches, linearly projects each patch into an embedding, and adds positional embeddings.

**Explanation:** Unlike CNNs that use convolutions, ViTs treat images as sequences of patches. The image is first split into non-overlapping patches. Each patch is then flattened and linearly projected into a higher-dimensional embedding space. Crucially, positional embeddings are added to these patch embeddings to retain spatial information, as the self-attention mechanism itself is permutation-invariant.

---

### Question 2: ViT vs. CNN - Global Context
**Correct Answer: B)** ViTs use self-attention mechanisms that allow each patch embedding to interact directly with every other patch embedding, regardless of spatial distance.

**Explanation:** CNNs build up global context hierarchically through successive layers with increasing receptive fields. In contrast, the self-attention mechanism in ViTs allows any patch to directly attend to any other patch in the input sequence, enabling the model to capture long-range dependencies and global contextual information in a single layer, which is a significant architectural difference and advantage for understanding relationships across an entire image.

---

### Question 3: Multimodal Learning - CLIP's Mechanism
**Correct Answer: C)** Contrastive learning, where the model learns to maximize the similarity of correct image-text pairs and minimize the similarity of incorrect pairs in a joint embedding space.

**Explanation:** CLIP employs a contrastive learning objective. It trains separate image and text encoders to project their respective inputs into a shared, high-dimensional embedding space. During training, it aims to pull embeddings of matching image-text pairs closer together and push embeddings of non-matching pairs further apart. This allows for zero-shot generalization and robust cross-modal understanding.

---

### Question 4: Multimodal CV Application
**Correct Answer: C)** Employing a model like CLIP to embed both images and text into a common latent space, allowing for direct similarity comparison between query embeddings and image embeddings.

**Explanation:** For natural language image retrieval, models like CLIP are ideal. They create a shared embedding space where the semantic meaning of images and text are aligned. A natural language query can be encoded into this space, and then a simple cosine similarity (or similar metric) can be used to find the most semantically relevant images whose embeddings are closest to the query embedding. This is highly efficient for large databases as image embeddings can be pre-computed.
