```python
# Part 2: Capstone Project - Starter Template

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer, BitsAndBytesConfig
from datasets import load_dataset
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
import accelerate
import os

# 1. Configuration
MODEL_NAME = "mistralai/Mistral-7B-Instruct-v0.3" # Or "meta-llama/Llama-3-8B-Instruct"
DATASET_NAME = "tatsu-lab/alpaca" # A small instruction dataset
OUTPUT_DIR = "./ft02-fa-finetuned-model"

# LoRA configuration
LORA_R = 16
LORA_ALPHA = 32
LORA_DROPOUT = 0.05

# Training arguments
BATCH_SIZE = 4
GRADIENT_ACCUMULATION_STEPS = 4
LEARNING_RATE = 2e-4
NUM_TRAIN_EPOCHS = 1
MAX_SEQ_LENGTH = 512

# 2. Load Model and Tokenizer (Quantized)
# Define BitsAndBytesConfig for 4-bit quantization
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
)

# Load base model
print(f"Loading model: {MODEL_NAME} with 4-bit quantization...")
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=quantization_config,
    device_map="auto", # Automatically maps model to available devices
    torch_dtype=torch.bfloat16 # Use bfloat16 for computation
)
model.config.use_cache = False # Disable cache for training
model.config.pretraining_tp = 1 # Recommended for Mistral/Llama

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token # Set pad token
tokenizer.padding_side = "right" # Pad to the right for causal models

# 3. Prepare Model for QLoRA
# Prepare model for k-bit training (e.g., 4-bit)
model = prepare_model_for_kbit_training(model)

# Configure LoRA
peft_config = LoraConfig(
    lora_alpha=LORA_ALPHA,
    lora_dropout=LORA_DROPOUT,
    r=LORA_R,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"], # Common for Llama/Mistral
)

# Get PEFT model
model = get_peft_model(model, peft_config)
model.print_trainable_parameters() # Display trainable parameters

# 4. Load and Preprocess Dataset
print(f"Loading dataset: {DATASET_NAME}...")
dataset = load_dataset(DATASET_NAME, split="train")

# Select a small subset for quick testing (e.g., 1000 samples)
# dataset = dataset.shuffle(seed=42).select(range(1000))

# Define a formatting function for instruction tuning
def format_instruction_dataset(sample):
    # This function should convert your dataset sample into a single string
    # suitable for instruction tuning, e.g., "### Instruction:\n{instruction}\n### Response:\n{response}"
    # Example for Alpaca dataset:
    instruction = sample["instruction"]
    input_text = sample["input"]
    output_text = sample["output"]

    if input_text:
        full_text = f"### Instruction:\n{instruction}\n### Input:\n{input_text}\n### Response:\n{output_text}{tokenizer.eos_token}"
    else:
        full_text = f"### Instruction:\n{instruction}\n### Response:\n{output_text}{tokenizer.eos_token}"
    return {"text": full_text}

# Apply formatting
formatted_dataset = dataset.map(format_instruction_dataset, remove_columns=list(dataset.features))

# Tokenize the dataset
def tokenize_function(examples):
    return tokenizer(examples["text"], truncation=True, max_length=MAX_SEQ_LENGTH, padding="max_length")

tokenized_dataset = formatted_dataset.map(tokenize_function, batched=True, remove_columns=["text"])

# 5. Configure Training Arguments
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=BATCH_SIZE,
    gradient_accumulation_steps=GRADIENT_ACCUMULATION_STEPS,
    learning_rate=LEARNING_RATE,
    num_train_epochs=NUM_TRAIN_EPOCHS,
    logging_steps=10,
    save_steps=500, # Save checkpoint every 500 steps
    fp16=False, # Use bfloat16 instead of fp16 for better stability with 4-bit
    bf16=True, # Enable bfloat16
    optim="paged_adamw_8bit", # Use 8-bit AdamW optimizer for memory efficiency
    lr_scheduler_type="cosine",
    warmup_ratio=0.03,
    report_to="none", # Disable reporting to W&B or other services for simplicity
    push_to_hub=False,
    gradient_checkpointing=True, # Enable gradient checkpointing for memory saving
    gradient_checkpointing_kwargs={'use_reentrant': False}, # Recommended for some models
)

# 6. Initialize Trainer
trainer = Trainer(
    model=model,
    train_dataset=tokenized_dataset,
    args=training_args,
    data_collator=lambda data: {'input_ids': torch.stack([f['input_ids'] for f in data]),
                                'attention_mask': torch.stack([f['attention_mask'] for f in data]),
                                'labels': torch.stack([f['input_ids'] for f in data])} # Labels are input_ids for causal LM
)

# 7. Train the model
print("Starting training...")
# trainer.train()
print("Training complete. Uncomment trainer.train() to run.")

# 8. Save the fine-tuned model and tokenizer
# trainer.save_model(OUTPUT_DIR)
# tokenizer.save_pretrained(OUTPUT_DIR)
print(f"Model and tokenizer would be saved to {OUTPUT_DIR}. Uncomment save_model/save_pretrained to run.")

# 9. Inference (after training and saving)
# To perform inference, you would typically reload the base model and then load the PEFT adapters.
# For demonstration, let's show how to load and merge adapters.

# from peft import PeftModel
# from transformers import pipeline

# # Load the base model again (or use the one from memory if not restarted)
# base_model = AutoModelForCausalLM.from_pretrained(
#     MODEL_NAME,
#     torch_dtype=torch.bfloat16,
#     device_map="auto",
# )
# base_model.config.use_cache = True # Enable cache for inference

# # Load the PEFT adapters
# peft_model = PeftModel.from_pretrained(base_model, OUTPUT_DIR)

# # Merge LoRA layers into the base model for easier deployment (optional)
# merged_model = peft_model.merge_and_unload()

# # Create a pipeline for inference
# pipe = pipeline(
#     "text-generation",
#     model=merged_model,
#     tokenizer=tokenizer,
#     torch_dtype=torch.bfloat16,
#     device_map="auto",
# )

# # Example inference
# instruction = "Explain the concept of quantum entanglement in simple terms."
# prompt = f"### Instruction:\n{instruction}\n### Response:\n"

# outputs = pipe(prompt, max_new_tokens=200, do_sample=True, temperature=0.7, top_k=50, top_p=0.95)
# print("\n--- Inference Result ---")
# print(outputs[0]["generated_text"])

# instruction_2 = "Write a short, creative story about a robot who discovers emotions."
# prompt_2 = f"### Instruction:\n{instruction_2}\n### Response:\n"
# outputs_2 = pipe(prompt_2, max_new_tokens=200, do_sample=True, temperature=0.7, top_k=50, top_p=0.95)
# print("\n--- Inference Result 2 ---")
# print(outputs_2[0]["generated_text"])
```

```python
# Part 2: Capstone Project - Detailed Solution

# This solution provides a complete, runnable example for fine-tuning a Mistral-7B-Instruct-v0.3
# model using QLoRA on a subset of the Alpaca dataset. It demonstrates the full pipeline
# from data preparation to efficient training and inference.

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer, BitsAndBytesConfig
from datasets import load_dataset
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training, PeftModel
import accelerate
import os

# Ensure reproducibility
torch.manual_seed(42)

# --- 1. Configuration --- 
# Define model, dataset, and output paths. Using Mistral-7B-Instruct-v0.3 as a good balance
# of performance and resource requirements for a capstone project.
MODEL_NAME = "mistralai/Mistral-7B-Instruct-v0.3"
DATASET_NAME = "tatsu-lab/alpaca" # A classic instruction dataset
OUTPUT_DIR = "./ft02-fa-mistral-qlora-finetuned"

# LoRA (Low-Rank Adaptation) configuration parameters.
# These values are chosen for efficient fine-tuning while maintaining good performance.
# r: The rank of the update matrices. A higher rank means more expressivity but more parameters.
# lora_alpha: Scaling factor for the LoRA weights. Typically twice 'r'.
# lora_dropout: Dropout probability for LoRA layers.
LORA_R = 16
LORA_ALPHA = 32
LORA_DROPOUT = 0.05

# Training arguments for the Hugging Face Trainer.
# batch_size: Number of samples per batch. Small for QLoRA to fit in memory.
# gradient_accumulation_steps: Accumulate gradients over multiple steps to simulate a larger batch size.
# learning_rate: Standard learning rate for QLoRA.
# num_train_epochs: Keep it low for a quick assessment run.
# max_seq_length: Maximum sequence length for tokenization.
BATCH_SIZE = 4
GRADIENT_ACCUMULATION_STEPS = 4 # Effective batch size = BATCH_SIZE * GRADIENT_ACCUMULATION_STEPS = 16
LEARNING_RATE = 2e-4
NUM_TRAIN_EPOCHS = 1
MAX_SEQ_LENGTH = 512

# Create output directory if it doesn't exist
os.makedirs(OUTPUT_DIR, exist_ok=True)

print(f"Configuration loaded. Model: {MODEL_NAME}, Dataset: {DATASET_NAME}")

# --- 2. Load Model and Tokenizer with 4-bit Quantization (QLoRA Setup) ---
# BitsAndBytesConfig is crucial for QLoRA, enabling 4-bit quantization.
# load_in_4bit=True: Activates 4-bit quantization.
# bnb_4bit_quant_type="nf4": Uses NormalFloat4 quantization, which is optimal for transformers.
# bnb_4bit_compute_dtype=torch.bfloat16: Computations are performed in bfloat16 for stability and speed.
# bnb_4bit_use_double_quant=True: Applies a second quantization to the first quantization constants, saving more memory.
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
)

# Load the base model from Hugging Face Hub.
# device_map="auto": Automatically distributes the model across available GPUs.
# torch_dtype=torch.bfloat16: Ensures the model's weights are loaded in bfloat16 where possible.
print(f"Loading base model '{MODEL_NAME}' with 4-bit quantization...")
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=quantization_config,
    device_map="auto",
    torch_dtype=torch.bfloat16,
    trust_remote_code=True # Required for some models like Falcon
)

# Disable cache during training for memory efficiency and to avoid issues with gradient checkpointing.
model.config.use_cache = False
# pretraining_tp (tensor parallelism) is often set to 1 for fine-tuning to avoid issues.
model.config.pretraining_tp = 1

# Load the tokenizer. Set pad_token to eos_token and padding_side to 'right' for causal models.
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

print("Model and Tokenizer loaded successfully.")

# --- 3. Prepare Model for QLoRA Fine-tuning ---
# prepare_model_for_kbit_training: This function wraps the model to enable gradient computation
# for the 4-bit quantized layers, which is essential for QLoRA.
model = prepare_model_for_kbit_training(model)

# LoraConfig: Defines how LoRA adapters are attached to the model.
# target_modules: Specifies which layers (e.g., query, key, value projections) will have LoRA adapters.
#                 These are typically the attention projection layers.
peft_config = LoraConfig(
    lora_alpha=LORA_ALPHA,
    lora_dropout=LORA_DROPOUT,
    r=LORA_R,
    bias="none", # Generally 'none' for LLMs
    task_type="CAUSAL_LM",
    target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"], # Common for Llama/Mistral
)

# get_peft_model: Applies the LoRA configuration to the base model, making it a PEFT model.
model = get_peft_model(model, peft_config)

# Print the number of trainable parameters. This should be a very small fraction of the total parameters,
# demonstrating the efficiency of LoRA/QLoRA.
model.print_trainable_parameters()
print("Model prepared for QLoRA fine-tuning.")

# --- 4. Load and Preprocess Dataset ---
print(f"Loading dataset '{DATASET_NAME}'...")
dataset = load_dataset(DATASET_NAME, split="train")

# For a quick assessment, we'll use a small subset of the dataset.
# In a real scenario, you'd use the full dataset or a larger, curated one.
print(f"Original dataset size: {len(dataset)} samples. Using a subset of 1000 samples for demonstration.")
dataset = dataset.shuffle(seed=42).select(range(1000))

# Define a formatting function to convert raw dataset samples into instruction-following format.
# This format is crucial for instruction tuning, teaching the model to follow specific prompts.
# The `tokenizer.eos_token` is added to explicitly mark the end of a response.
def format_instruction_dataset(sample):
    instruction = sample["instruction"]
    input_text = sample["input"]
    output_text = sample["output"]

    if input_text:
        # Format for samples with both instruction and input
        full_text = f"### Instruction:\n{instruction}\n### Input:\n{input_text}\n### Response:\n{output_text}{tokenizer.eos_token}"
    else:
        # Format for samples with only instruction
        full_text = f"### Instruction:\n{instruction}\n### Response:\n{output_text}{tokenizer.eos_token}"
    return {"text": full_text}

# Apply the formatting function to the dataset.
formatted_dataset = dataset.map(format_instruction_dataset, remove_columns=list(dataset.features))

# Tokenize the formatted dataset.
# truncation=True: Truncates sequences longer than MAX_SEQ_LENGTH.
# padding="max_length": Pads shorter sequences to MAX_SEQ_LENGTH.
# remove_columns=["text"]: Removes the raw text column after tokenization.
def tokenize_function(examples):
    return tokenizer(examples["text"], truncation=True, max_length=MAX_SEQ_LENGTH, padding="max_length")

tokenized_dataset = formatted_dataset.map(tokenize_function, batched=True, remove_columns=["text"])

print(f"Dataset preprocessed. Tokenized dataset size: {len(tokenized_dataset)} samples.")

# --- 5. Configure Training Arguments ---
# TrainingArguments defines all the hyperparameters for the training process.
# optim="paged_adamw_8bit": Uses an 8-bit AdamW optimizer with memory paging, further reducing memory footprint.
# bf16=True: Enables bfloat16 training, which is stable and faster on modern GPUs.
# gradient_checkpointing=True: Saves memory by recomputing activations during the backward pass.
# gradient_checkpointing_kwargs={'use_reentrant': False}: Recommended for compatibility with some models/setups.
training_args = TrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=BATCH_SIZE,
    gradient_accumulation_steps=GRADIENT_ACCUMULATION_STEPS,
    learning_rate=LEARNING_RATE,
    num_train_epochs=NUM_TRAIN_EPOCHS,
    logging_steps=10,
    save_steps=500,
    fp16=False, # Explicitly disable fp16 if bf16 is used
    bf16=True,
    optim="paged_adamw_8bit",
    lr_scheduler_type="cosine",
    warmup_ratio=0.03,
    report_to="none",
    push_to_hub=False,
    gradient_checkpointing=True,
    gradient_checkpointing_kwargs={'use_reentrant': False},
    # Optional: Add evaluation strategy if you have a validation set
    # evaluation_strategy="steps",
    # eval_steps=500,
    # load_best_model_at_end=True,
)

print("Training arguments configured.")

# --- 6. Initialize Trainer ---
# The Hugging Face Trainer simplifies the training loop.
# data_collator: A function that takes a list of samples and forms a batch.
#                For causal language modeling, labels are typically the input_ids themselves.
trainer = Trainer(
    model=model,
    train_dataset=tokenized_dataset,
    args=training_args,
    data_collator=lambda data: {
        'input_ids': torch.stack([f['input_ids'] for f in data]),
        'attention_mask': torch.stack([f['attention_mask'] for f in data]),
        'labels': torch.stack([f['input_ids'] for f in data])
    }
)

# Disable cache for the model within the Trainer as well, just to be safe.
model.config.use_cache = False

print("Trainer initialized.")

# --- 7. Train the model ---
print("Starting training... (This might take a while depending on your hardware and dataset size)")
# To run the training, uncomment the line below.
# trainer.train()
print("Training complete. (trainer.train() was commented out for assessment submission. Uncomment to run.)")

# --- 8. Save the fine-tuned model and tokenizer ---
# After training, save the PEFT adapters and the tokenizer.
# The base model is not saved as it remains unchanged (only adapters are trained).
# To save, uncomment the lines below.
# trainer.save_model(OUTPUT_DIR)
# tokenizer.save_pretrained(OUTPUT_DIR)
print(f"Model adapters and tokenizer would be saved to {OUTPUT_DIR}. (Uncomment save_model/save_pretrained to run.)")

# --- 9. Inference with the Fine-tuned Model ---
# To perform inference, we load the base model and then load the PEFT adapters on top of it.
# Optionally, the adapters can be merged into the base model for easier deployment.

print("\n--- Demonstrating Inference Setup (requires trained model) ---")

# Load the base model again (or use the one from memory if not restarted)
# Ensure use_cache is True for efficient inference.
base_model_for_inference = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.bfloat16,
    device_map="auto",
    trust_remote_code=True
)
base_model_for_inference.config.use_cache = True

# Load the PEFT adapters from the saved directory
# This step assumes the model adapters were saved to OUTPUT_DIR
# peft_model_path = OUTPUT_DIR # Assuming adapters are saved directly in OUTPUT_DIR
# if os.path.exists(peft_model_path):
#     print(f"Loading PEFT adapters from {peft_model_path}...")
#     peft_model = PeftModel.from_pretrained(base_model_for_inference, peft_model_path)

#     # Merge LoRA layers into the base model for easier deployment (optional but recommended for inference)
#     print("Merging LoRA adapters into the base model...")
#     merged_model = peft_model.merge_and_unload()
#     print("Model merged successfully.")

#     # Create a pipeline for inference
#     from transformers import pipeline
#     pipe = pipeline(
#         "text-generation",
#         model=merged_model,
#         tokenizer=tokenizer,
#         torch_dtype=torch.bfloat16,
#         device_map="auto",
#     )

#     # Example inference prompts
#     instruction_1 = "Explain the concept of quantum entanglement in simple terms, suitable for a high school student."
#     prompt_1 = f"### Instruction:\n{instruction_1}\n### Response:\n"

#     instruction_2 = "Write a short, creative story about a robot who discovers emotions for the first time."
#     prompt_2 = f"### Instruction:\n{instruction_2}\n### Response:\n"

#     print("\n--- Generating Response 1 ---")
#     outputs_1 = pipe(prompt_1, max_new_tokens=200, do_sample=True, temperature=0.7, top_k=50, top_p=0.95)
#     print(outputs_1[0]["generated_text"])

#     print("\n--- Generating Response 2 ---")
#     outputs_2 = pipe(prompt_2, max_new_tokens=200, do_sample=True, temperature=0.7, top_k=50, top_p=0.95)
#     print(outputs_2[0]["generated_text"])

# else:
#     print(f"Warning: PEFT adapters not found at {peft_model_path}. Please run trainer.train() and save_model() first.")
#     print("Skipping inference demonstration as no fine-tuned model is available.")

print("Inference setup complete. Uncomment the inference block to run after training and saving.")

# --- Discussion of Expected Performance and Challenges ---
print("\n--- Discussion ---")
print("**Expected Performance Improvements:**")
print("1.  **Domain Adaptation:** The fine-tuned model is expected to generate responses that are more aligned with the instruction-following format and style present in the Alpaca dataset. This means better adherence to prompt structures (e.g., '### Instruction:', '### Response:').")
print("2.  **Task-Specific Quality:** For tasks similar to those in the fine-tuning dataset, the model should exhibit improved coherence, relevance, and accuracy compared to the base model, which is more general-purpose.")
print("3.  **Efficiency:** QLoRA significantly reduces memory footprint and training time compared to full fine-tuning, making it feasible to fine-tune large models on consumer-grade GPUs or smaller clusters. The use of `paged_adamw_8bit` and `gradient_checkpointing` further enhances this efficiency.")

print("\n**Challenges Encountered/Considerations:**")
print("1.  **Hyperparameter Tuning:** Optimal LoRA parameters (r, alpha, dropout) and training arguments (learning rate, batch size) are crucial and often require experimentation. Suboptimal choices can lead to underfitting or catastrophic forgetting.")
print("2.  **Data Quality and Quantity:** While QLoRA is data-efficient, the quality and diversity of the instruction-tuning dataset are paramount. A small, biased, or low-quality dataset can lead to a model that performs poorly or inherits undesirable traits.")
print("3.  **Evaluation:** Qualitative evaluation (manual inspection of generated text) is often necessary but subjective. Quantitative evaluation for instruction following can be challenging without specific metrics or human preference scores.")
print("4.  **Catastrophic Forgetting:** Fine-tuning, especially with small datasets or aggressive learning rates, can sometimes lead to the model 'forgetting' some of its general knowledge acquired during pretraining. Careful monitoring and validation are needed.")
print("5.  **Hardware Limitations:** Even with QLoRA, fine-tuning larger models (e.g., 70B+) still requires significant GPU memory, potentially necessitating multi-GPU setups with `accelerate` and FSDP/ZeRO for practical training times.")
```
