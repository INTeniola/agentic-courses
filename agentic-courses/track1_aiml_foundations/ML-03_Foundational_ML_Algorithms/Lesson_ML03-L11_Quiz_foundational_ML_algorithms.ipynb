## Quiz: Foundational ML Algorithms

This quiz is designed to test your understanding of the foundational machine learning algorithms covered in this course. Read each question carefully and select the best answer.

---

### Question 1

A tech company wants to predict whether a user will click on an ad based on their browsing history and demographics. Which type of machine learning problem is this, and which algorithm is most suitable?

A) Regression; Linear Regression
B) Classification; Logistic Regression
C) Clustering; K-Means
D) Dimensionality Reduction; PCA

---

### Question 2

In the context of K-Means clustering, what does the 'K' represent?

A) The number of features in the dataset.
B) The number of clusters to be formed.
C) The maximum number of iterations for the algorithm.
D) The distance metric used (e.g., Euclidean, Manhattan).

---

### Question 3

Principal Component Analysis (PCA) is primarily used for which of the following tasks in machine learning?

A) Predicting a continuous target variable.
B) Classifying data points into predefined categories.
C) Reducing the dimensionality of a dataset while retaining most of its variance.
D) Grouping similar data points into clusters without prior labels.

---

### Question 4

Which of the following statements accurately distinguishes supervised learning from unsupervised learning?

A) Supervised learning requires labeled data, while unsupervised learning works with unlabeled data.
B) Unsupervised learning uses a target variable for training, while supervised learning does not.
C) Supervised learning is primarily used for clustering, whereas unsupervised learning is for regression.
D) Both supervised and unsupervised learning always require feature engineering before model training.

---

### Question 5

A financial institution wants to identify distinct groups of customers based on their transaction history and spending patterns to tailor marketing strategies. Which foundational ML algorithm would be most appropriate for this task?

A) Linear Regression
B) Support Vector Machine (SVM)
C) K-Means Clustering
D) Decision Tree Classifier

---

### Question 1
**Correct Answer: B) Classification; Logistic Regression**

*   **Explanation:** Predicting whether a user *will click* (a binary outcome: yes/no) is a classification problem. Logistic Regression is a foundational algorithm specifically designed for binary classification tasks, predicting the probability of an event occurring.

---

### Question 2
**Correct Answer: B) The number of clusters to be formed.**

*   **Explanation:** In K-Means clustering, 'K' is a hyperparameter that must be specified by the user, indicating the desired number of distinct clusters the algorithm should identify within the data.

---

### Question 3
**Correct Answer: C) Reducing the dimensionality of a dataset while retaining most of its variance.**

*   **Explanation:** PCA is a powerful unsupervised dimensionality reduction technique. It transforms the data into a new set of orthogonal variables (principal components) that capture the maximum variance, effectively reducing the number of features while preserving essential information.

---

### Question 4
**Correct Answer: A) Supervised learning requires labeled data, while unsupervised learning works with unlabeled data.**

*   **Explanation:** This is the fundamental distinction. Supervised learning models learn from input-output pairs (labeled data) to make predictions, while unsupervised learning models find patterns or structures in input data without explicit output labels.

---

### Question 5
**Correct Answer: C) K-Means Clustering**

*   **Explanation:** The task of identifying *distinct groups* or segments of customers based on their behavior without prior labels is a classic clustering problem. K-Means Clustering is a widely used unsupervised algorithm for this purpose, grouping similar data points together.
