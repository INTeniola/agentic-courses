```python
# Import necessary libraries
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report

# --- Setup Code: Generate Synthetic Dataset ---

# Generate a synthetic binary classification dataset
# n_samples: total number of samples
# n_features: total number of features
# n_informative: number of informative features (contribute to the target)
# n_redundant: number of redundant features (linear combinations of informative features)
# n_classes: number of target classes
# random_state: for reproducibility
X, y = make_classification(
    n_samples=1000,        # 1000 data points
    n_features=20,         # 20 features in total
    n_informative=10,      # 10 features are informative
    n_redundant=5,         # 5 features are redundant
    n_classes=2,           # Binary classification
    random_state=42,       # Seed for reproducibility
    class_sep=0.8          # How separated the classes are
)

print(f"Dataset generated: X shape {X.shape}, y shape {y.shape}")
print(f"Number of positive class samples: {np.sum(y == 1)}")
print(f"Number of negative class samples: {np.sum(y == 0)}")

# --- Setup Code: Split Data ---

# Split the dataset into training and testing sets
# test_size=0.3: 30% of data for testing, 70% for training
# stratify=y: Ensures that the proportion of target classes is the same in both train and test sets
# random_state=42: Seed for reproducibility
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)

print(f"\nTraining data shape: X_train {X_train.shape}, y_train {y_train.shape}")
print(f"Testing data shape: X_test {X_test.shape}, y_test {y_test.shape}")

print("\nSetup complete. You are ready to implement the models!")
```

## Your Turn! Implement and Compare Tree Models

Now it's your turn to apply what you've learned. In the code cell below, implement the three tree-based classification models as described in the requirements:

1.  **Decision Tree Classifier**
2.  **Random Forest Classifier**
3.  **Gradient Boosting Classifier**

For each model:
*   Instantiate the model, setting a `random_state` for reproducibility and a basic hyperparameter like `max_depth` or `n_estimators`.
*   Train the model using `X_train` and `y_train`.
*   Make predictions on `X_test`.
*   Calculate and print the `accuracy_score` and `classification_report`.

After evaluating all three, add a markdown cell or print statements to discuss your findings.


```python
# --- Reference Solution: Implement and Compare Tree Models ---

print("--- Starting Model Training and Evaluation ---\n")

# 1. Decision Tree Classifier
print("### Decision Tree Classifier ###")
# Initialize Decision Tree Classifier with a max_depth to prevent overfitting
# random_state ensures reproducibility of the tree's splits
dt_classifier = DecisionTreeClassifier(max_depth=7, random_state=42)

# Train the model on the training data
dt_classifier.fit(X_train, y_train)

# Make predictions on the test data
y_pred_dt = dt_classifier.predict(X_test)

# Evaluate the Decision Tree model
print(f"Accuracy Score (Decision Tree): {accuracy_score(y_test, y_pred_dt):.4f}")
print("Classification Report (Decision Tree):\n", classification_report(y_test, y_pred_dt))
print("\n" + "-"*50 + "\n")

# 2. Random Forest Classifier
print("### Random Forest Classifier ###")
# Initialize Random Forest Classifier
# n_estimators: number of trees in the forest (more trees generally improve performance but increase computation)
# max_depth: maximum depth of individual trees (can be left None, but setting it can prevent overfitting)
# random_state ensures reproducibility
rf_classifier = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)

# Train the model on the training data
rf_classifier.fit(X_train, y_train)

# Make predictions on the test data
y_pred_rf = rf_classifier.predict(X_test)

# Evaluate the Random Forest model
print(f"Accuracy Score (Random Forest): {accuracy_score(y_test, y_pred_rf):.4f}")
print("Classification Report (Random Forest):\n", classification_report(y_test, y_pred_rf))
print("\n" + "-"*50 + "\n")

# 3. Gradient Boosting Classifier
print("### Gradient Boosting Classifier ###")
# Initialize Gradient Boosting Classifier
# n_estimators: number of boosting stages (weak learners)
# learning_rate: shrinks the contribution of each tree (prevents overfitting)
# max_depth: maximum depth of the individual regression estimators
# random_state ensures reproducibility
gb_classifier = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3, random_state=42)

# Train the model on the training data
gb_classifier.fit(X_train, y_train)

# Make predictions on the test data
y_pred_gb = gb_classifier.predict(X_test)

# Evaluate the Gradient Boosting model
print(f"Accuracy Score (Gradient Boosting): {accuracy_score(y_test, y_pred_gb):.4f}")
print("Classification Report (Gradient Boosting):\n", classification_report(y_test, y_pred_gb))
print("\n" + "-"*50 + "\n")

print("--- Model Comparison and Discussion ---\n")

# Discussion of results
print("**Summary of Model Performance:**")
print(f"- Decision Tree Accuracy: {accuracy_score(y_test, y_pred_dt):.4f}")
print(f"- Random Forest Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"- Gradient Boosting Accuracy: {accuracy_score(y_test, y_pred_gb):.4f}")

print("\n**Observations and Insights:**")
print("1.  **Decision Tree:** Typically serves as a baseline. Its performance is often lower than ensemble methods because it's prone to overfitting (high variance) if `max_depth` is not carefully controlled, or underfitting (high bias) if it's too shallow. Here, with `max_depth=7`, it performs reasonably well but is generally outperformed.")
print("2.  **Random Forest:** As an ensemble of decision trees, Random Forest significantly improves performance by reducing variance. It builds multiple trees independently and averages their predictions. This usually leads to higher accuracy and better generalization than a single decision tree, as seen in the results.")
print("3.  **Gradient Boosting:** Another powerful ensemble method, Gradient Boosting builds trees sequentially, where each new tree corrects the errors of the previous ones. This iterative error correction often leads to very high accuracy, frequently outperforming Random Forests, especially on complex datasets. The `learning_rate` and `n_estimators` are crucial hyperparameters for balancing bias and variance.")
print("\nIn this specific synthetic dataset, both Random Forest and Gradient Boosting demonstrate superior performance compared to the single Decision Tree, which is a common outcome in real-world scenarios. Gradient Boosting often achieves the highest performance due to its error-correcting nature, but it can be more sensitive to hyperparameter tuning and more prone to overfitting if not regularized properly.")
```
