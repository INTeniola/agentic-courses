## Exercise: Strategic AI Team Structure Decision Framework

    *   The calculated suitability score for the 'AI-Native Team' strategy.

````python
from typing import Dict, Any, Tuple

# Define mappings for categorical inputs to numerical values for easier calculation.
# These mappings reflect a scale where higher numbers generally mean 'more' of that attribute.
# For 'budget_constraint', 'Low' constraint means more budget flexibility, so it's mapped to a higher number.
INPUT_MAPPINGS: Dict[str, Dict[str, int]] = {
    "existing_team_ai_maturity": {"Low": 1, "Medium": 2, "High": 3},
    "project_complexity": {"Simple": 1, "Moderate": 2, "Complex": 3},
    "budget_constraint": {"Low": 3, "Medium": 2, "High": 1}, # Low constraint = more budget flexibility
    "time_to_market_pressure": {"Low": 1, "Medium": 2, "High": 3},
    "organizational_culture_adaptability": {"Low": 1, "Medium": 2, "High": 3},
}

# Define weights and inverse flags for how each factor influences the suitability score for each strategy.
# 'weight': The importance of the factor.
# 'inverse': If True, (4 - numerical_value) is used, meaning lower input value leads to higher score contribution.
#            If False, numerical_value is used directly, meaning higher input value leads to higher score contribution.
STRATEGY_FACTORS: Dict[str, Dict[str, Dict[str, Any]]] = {
    "ai_native": {
        "existing_team_ai_maturity": {"weight": 1.0, "inverse": True},  # Lower existing maturity favors AI-native (fills gap)
        "project_complexity": {"weight": 2.5, "inverse": False}, # Higher complexity strongly favors AI-native
        "budget_constraint": {"weight": 2.0, "inverse": False},   # Lower budget constraint (more budget) favors AI-native
        "time_to_market_pressure": {"weight": 1.5, "inverse": False}, # Higher pressure can favor AI-native (if quick talent acquisition)
        "organizational_culture_adaptability": {"weight": 1.0, "inverse": False} # Higher adaptability helps integrate new team
    },
    "retrofit_existing": {
        "existing_team_ai_maturity": {"weight": 3.0, "inverse": False}, # Higher existing maturity is a huge advantage for upskilling
        "project_complexity": {"weight": 2.0, "inverse": True}, # Lower complexity is better for retrofitting
        "budget_constraint": {"weight": 1.5, "inverse": True},  # Higher budget constraint (less budget) favors retrofitting
        "time_to_market_pressure": {"weight": 1.0, "inverse": True}, # Lower pressure (more time) favors retrofitting
        "organizational_culture_adaptability": {"weight": 1.0, "inverse": False} # Higher adaptability is crucial for upskilling and change
    }
}

# Base scores to ensure scores are not negative and provide a baseline for comparison.
BASE_SCORE = 10.0
# Threshold for considering scores 'balanced' (i.e., neither strategy is significantly better).
BALANCE_THRESHOLD = 2.0

def evaluate_ai_team_strategy(
    existing_team_ai_maturity: str,
    project_complexity: str,
    budget_constraint: str,
    time_to_market_pressure: str,
    organizational_culture_adaptability: str
) -> Tuple[float, float, str]:
    """
    Evaluates the suitability of building an AI-native team versus retrofitting an existing team
    for a given AI initiative based on strategic organizational parameters.

    Args:
        existing_team_ai_maturity (str): Current AI proficiency of existing workforce ('Low', 'Medium', 'High').
        project_complexity (str): Technical difficulty and scope of the AI initiative ('Simple', 'Moderate', 'Complex').
        budget_constraint (str): Financial limitations ('Low', 'Medium', 'High').
        time_to_market_pressure (str): Urgency to deliver the AI solution ('Low', 'Medium', 'High').
        organizational_culture_adaptability (str): Organization's openness to change ('Low', 'Medium', 'High').

    Returns:
        Tuple[float, float, str]: A tuple containing:
            - The calculated suitability score for the AI-Native Team strategy.
            - The calculated suitability score for the Retrofit Existing Team strategy.
            - A string recommendation ('Favor AI-Native Team', 'Favor Retrofit Existing Team', 'Balanced Approach').
    """
    # Initialize scores with a base value to ensure they are not negative and provide a starting point.
    ai_native_score = BASE_SCORE
    retrofit_score = BASE_SCORE

    # 1. Convert categorical inputs to numerical values using the predefined mappings.
    params_numerical: Dict[str, int] = {
        "existing_team_ai_maturity": INPUT_MAPPINGS["existing_team_ai_maturity"][existing_team_ai_maturity],
        "project_complexity": INPUT_MAPPINGS["project_complexity"][project_complexity],
        "budget_constraint": INPUT_MAPPINGS["budget_constraint"][budget_constraint],
        "time_to_market_pressure": INPUT_MAPPINGS["time_to_market_pressure"][time_to_market_pressure],
        "organizational_culture_adaptability": INPUT_MAPPINGS["organizational_culture_adaptability"][organizational_culture_adaptability],
    }

    # 2. Calculate ai_native_score based on the numerical inputs and defined factors.
    for param, factor_props in STRATEGY_FACTORS["ai_native"].items():
        value = params_numerical[param]
        weight = factor_props["weight"]
        is_inverse = factor_props["inverse"]

        # Apply inverse logic if specified: (4 - value) effectively inverts the scale (1->3, 2->2, 3->1).
        score_contribution = (4 - value) * weight if is_inverse else value * weight
        ai_native_score += score_contribution

    # 3. Calculate retrofit_score based on the numerical inputs and defined factors.
    for param, factor_props in STRATEGY_FACTORS["retrofit_existing"].items():
        value = params_numerical[param]
        weight = factor_props["weight"]
        is_inverse = factor_props["inverse"]

        # Apply inverse logic if specified.
        score_contribution = (4 - value) * weight if is_inverse else value * weight
        retrofit_score += score_contribution

    # 4. Determine recommendation based on the calculated scores.
    recommendation: str
    score_difference = ai_native_score - retrofit_score

    if score_difference > BALANCE_THRESHOLD:
        recommendation = "Favor AI-Native Team"
    elif score_difference < -BALANCE_THRESHOLD:
        recommendation = "Favor Retrofit Existing Team"
    else:
        recommendation = "Balanced Approach (Consider Hybrid or Further Analysis)"

    return ai_native_score, retrofit_score, recommendation

# --- Example Usage Scenarios ---
print("--- Scenario 1: High-stakes, complex AI project with good budget and adaptable culture ---")
# This scenario strongly favors an AI-native team due to complexity, budget, and pressure.
native_score_1, retrofit_score_1, recommendation_1 = evaluate_ai_team_strategy(
    existing_team_ai_maturity="Low",
    project_complexity="Complex",
    budget_constraint="Low", # Low constraint means high budget flexibility
    time_to_market_pressure="High",
    organizational_culture_adaptability="High"
)
print(f"AI-Native Team Score: {native_score_1:.2f}")
print(f"Retrofit Existing Team Score: {retrofit_score_1:.2f}")
print(f"Recommendation: {recommendation_1}\n")

print("--- Scenario 2: Simpler AI project, tight budget, existing team has some AI maturity ---")
# This scenario leans towards retrofitting due to budget constraints and simpler project.
native_score_2, retrofit_score_2, recommendation_2 = evaluate_ai_team_strategy(
    existing_team_ai_maturity="Medium",
    project_complexity="Simple",
    budget_constraint="High", # High constraint means low budget flexibility
    time_to_market_pressure="Medium",
    organizational_culture_adaptability="Medium"
)
print(f"AI-Native Team Score: {native_score_2:.2f}")
print(f"Retrofit Existing Team Score: {retrofit_score_2:.2f}")
print(f"Recommendation: {recommendation_2}\n")

print("--- Scenario 3: Moderate project, balanced conditions ---")
# This scenario might result in a balanced approach.
native_score_3, retrofit_score_3, recommendation_3 = evaluate_ai_team_strategy(
    existing_team_ai_maturity="Medium",
    project_complexity="Moderate",
    budget_constraint="Medium",
    time_to_market_pressure="Medium",
    organizational_culture_adaptability="Medium"
)
print(f"AI-Native Team Score: {native_score_3:.2f}")
print(f"Retrofit Existing Team Score: {retrofit_score_3:.2f}")
print(f"Recommendation: {recommendation_3}\n")

print("--- Scenario 4: High existing AI maturity, low pressure, complex project ---")
# This scenario tests if high existing maturity can make retrofit viable even for complex projects.
native_score_4, retrofit_score_4, recommendation_4 = evaluate_ai_team_strategy(
    existing_team_ai_maturity="High",
    project_complexity="Complex",
    budget_constraint="Medium",
    time_to_market_pressure="Low",
    organizational_culture_adaptability="High"
)
print(f"AI-Native Team Score: {native_score_4:.2f}")
print(f"Retrofit Existing Team Score: {retrofit_score_4:.2f}")
print(f"Recommendation: {recommendation_4}\n")

print("--- Scenario 5: Low existing AI maturity, high pressure, simple project ---")
# This scenario might still favor retrofit if the project is simple and native hiring is slow.
native_score_5, retrofit_score_5, recommendation_5 = evaluate_ai_team_strategy(
    existing_team_ai_maturity="Low",
    project_complexity="Simple",
    budget_constraint="Medium",
    time_to_market_pressure="High",
    organizational_culture_adaptability="Medium"
)
print(f"AI-Native Team Score: {native_score_5:.2f}")
print(f"Retrofit Existing Team Score: {retrofit_score_5:.2f}")
print(f"Recommendation: {recommendation_5}\n")
````
