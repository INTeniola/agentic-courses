```python
# Capstone Project: Smart Farm Crop Health Classifier - Starter Template

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, Subset
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np
import matplotlib.pyplot as plt

# 1. Device Configuration
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# 2. Data Loading and Preprocessing
# Define transformations for the dataset
# TODO: Add data augmentation transforms for training
transform_train = transforms.Compose([
    transforms.Resize((64, 64)), # Resize images to a common size
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)) # Normalize pixel values
])

transform_test = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

# Load CIFAR-10 dataset as a stand-in for 'Crop Health'
# We'll filter it to create a binary classification problem
full_trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform_train)
full_testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform_test)

# Map CIFAR-10 classes to our binary problem:
# Class 0 (airplane) -> Healthy (new label 0)
# Class 1 (automobile) -> Diseased (new label 1)
# We will filter out other classes.

def filter_cifar_classes(dataset, class_map):
    indices = []
    new_targets = []
    for i, (_, label) in enumerate(dataset):
        if label in class_map:
            indices.append(i)
            new_targets.append(class_map[label])
    
    # Create a subset with filtered indices and update targets
    filtered_dataset = Subset(dataset, indices)
    # This is a bit hacky for Subset, usually you'd modify the original dataset's targets
    # For this exercise, we'll assume the model learns the new mapping implicitly
    # or we can create a custom dataset class if needed for strict target remapping.
    # For now, we'll just ensure the DataLoader gets the right data.
    
    # A more robust way for Subset to have new targets:
    class CustomFilteredDataset(torch.utils.data.Dataset):
        def __init__(self, original_dataset, indices, new_targets):
            self.original_dataset = original_dataset
            self.indices = indices
            self.new_targets = new_targets
        
        def __len__(self):
            return len(self.indices)
        
        def __getitem__(self, idx):
            img, _ = self.original_dataset[self.indices[idx]]
            return img, self.new_targets[idx]
            
    return CustomFilteredDataset(dataset, indices, new_targets)

# Define the mapping for our binary classification
cifar_to_crop_health_map = {
    0: 0, # airplane -> Healthy
    1: 1  # automobile -> Diseased
}

train_dataset = filter_cifar_classes(full_trainset, cifar_to_crop_health_map)
test_dataset = filter_cifar_classes(full_testset, cifar_to_crop_health_map)

# Create DataLoaders
# TODO: Set appropriate batch_size and num_workers
batch_size = 64
num_workers = 4 # Adjust based on your system's CPU cores

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=num_workers, pin_memory=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, num_workers=num_workers, pin_memory=True)

print(f"Number of training samples: {len(train_dataset)}")
print(f"Number of test samples: {len(test_dataset)}")

# 3. CNN Architecture Design
class CropHealthCNN(nn.Module):
    def __init__(self, num_classes=2):
        super(CropHealthCNN, self).__init__()
        # TODO: Define your CNN layers here
        # Example structure: Conv -> ReLU -> MaxPool -> Conv -> ReLU -> MaxPool -> Flatten -> FC -> FC
        self.features = nn.Sequential(
            # First convolutional block
            nn.Conv2d(3, 32, kernel_size=3, padding=1), # Input: 3 channels, Output: 32 channels, 64x64 -> 64x64
            nn.ReLU(),
            nn.BatchNorm2d(32),
            nn.MaxPool2d(kernel_size=2, stride=2), # 64x64 -> 32x32
            
            # Second convolutional block
            nn.Conv2d(32, 64, kernel_size=3, padding=1), # Input: 32 channels, Output: 64 channels, 32x32 -> 32x32
            nn.ReLU(),
            nn.BatchNorm2d(64),
            nn.MaxPool2d(kernel_size=2, stride=2), # 32x32 -> 16x16
            
            # Third convolutional block
            nn.Conv2d(64, 128, kernel_size=3, padding=1), # Input: 64 channels, Output: 128 channels, 16x16 -> 16x16
            nn.ReLU(),
            nn.BatchNorm2d(128),
            nn.MaxPool2d(kernel_size=2, stride=2) # 16x16 -> 8x8
        )
        
        # Calculate the size of the flattened features for the first fully connected layer
        # For 64x64 input, after 3 MaxPool2d(kernel_size=2, stride=2) layers, the spatial dimension becomes 64 / (2*2*2) = 8
        # So, the output of features is 128 channels * 8 * 8
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 8 * 8, 512), # First fully connected layer
            nn.ReLU(),
            nn.Dropout(0.5), # Dropout for regularization
            nn.Linear(512, num_classes) # Output layer with 2 classes
        )

    def forward(self, x):
        # TODO: Implement the forward pass
        x = self.features(x)
        x = self.classifier(x)
        return x

# Instantiate the model and move it to the device
model = CropHealthCNN(num_classes=2).to(device)

# Optional: Compile the model for performance (PyTorch 2.0+)
# model = torch.compile(model)

# 4. Training Loop Implementation
# Define hyperparameters
num_epochs = 10
learning_rate = 0.001

# Define Loss function and Optimizer
criterion = nn.CrossEntropyLoss() # Appropriate for multi-class classification
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# Training loop
print("\nStarting training...")
for epoch in range(num_epochs):
    model.train() # Set model to training mode
    running_loss = 0.0
    for i, (images, labels) in enumerate(train_loader):
        images = images.to(device)
        labels = labels.to(device)

        # Forward pass
        outputs = model(images)
        loss = criterion(outputs, labels)

        # Backward and optimize
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    # Print training loss for the epoch
    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {running_loss/len(train_loader):.4f}")

    # 5. Validation (within training loop)
    model.eval() # Set model to evaluation mode
    val_loss = 0.0
    correct = 0
    total = 0
    with torch.no_grad(): # Disable gradient calculation for validation
        for images, labels in test_loader:
            images = images.to(device)
            labels = labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            val_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    
    val_accuracy = 100 * correct / total
    print(f"Validation Loss: {val_loss/len(test_loader):.4f}, Validation Accuracy: {val_accuracy:.2f}%")

print("Training complete!")

# 6. Final Evaluation on Test Set
model.eval() # Set model to evaluation mode
all_labels = []
all_predictions = []

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        
        all_labels.extend(labels.cpu().numpy())
        all_predictions.extend(predicted.cpu().numpy())

# Calculate metrics
accuracy = accuracy_score(all_labels, all_predictions)
precision = precision_score(all_labels, all_predictions, average='binary') # 'binary' for 2 classes
recall = recall_score(all_labels, all_predictions, average='binary')
f1 = f1_score(all_labels, all_predictions, average='binary')

print("\n--- Final Test Set Evaluation ---")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")

# Optional: Visualize some predictions
def imshow(img):
    img = img / 2 + 0.5  # unnormalize
    npimg = img.numpy()
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.show()

# Get a batch of test images
dataiter = iter(test_loader)
images, labels = next(dataiter)

# Make predictions
outputs = model(images.to(device))
_, predicted = torch.max(outputs, 1)

print('GroundTruth: ', ' '.join(f'{label.item()}' for label in labels))
print('Predicted:   ', ' '.join(f'{p.item()}' for p in predicted))

# Show images
# imshow(torchvision.utils.make_grid(images))
```

```python
# Capstone Project: Smart Farm Crop Health Classifier - Detailed Solution

import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
from torch.utils.data import DataLoader, Subset
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Device Configuration ---
# Automatically detect and use GPU if available, otherwise fall back to CPU.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# --- 2. Data Loading and Preprocessing ---
# Define transformations for the dataset.
# For training, we include data augmentation techniques to improve model generalization.
transform_train = transforms.Compose([
    transforms.Resize((64, 64)),          # Resize all images to 64x64 pixels
    transforms.RandomHorizontalFlip(),    # Randomly flip images horizontally
    transforms.RandomRotation(10),        # Randomly rotate images by up to 10 degrees
    transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2), # Randomly change brightness, contrast, saturation
    transforms.ToTensor(),                # Convert images to PyTorch tensors
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)) # Normalize pixel values to [-1, 1]
])

# For testing/validation, we only resize and normalize, no augmentation.
transform_test = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])

# Load CIFAR-10 dataset as a stand-in for 'Crop Health' dataset.
# We download it if not already present.
full_trainset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform_train)
full_testset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform_test)

# Define a mapping from CIFAR-10 classes to our binary 'Crop Health' problem.
# We'll use 'airplane' (label 0) as 'Healthy' (new label 0)
# and 'automobile' (label 1) as 'Diseased' (new label 1).
# All other CIFAR-10 classes will be filtered out.
cifar_to_crop_health_map = {
    0: 0, # CIFAR-10 'airplane' -> 'Healthy'
    1: 1  # CIFAR-10 'automobile' -> 'Diseased'
}

# Helper function to filter the dataset and remap labels.
class CustomFilteredDataset(torch.utils.data.Dataset):
    def __init__(self, original_dataset, class_map):
        self.original_dataset = original_dataset
        self.indices = []
        self.new_targets = []
        
        for i, (_, label) in enumerate(original_dataset):
            if label in class_map:
                self.indices.append(i)
                self.new_targets.append(class_map[label])
        
    def __len__(self):
        return len(self.indices)
    
    def __getitem__(self, idx):
        # Retrieve original image and apply its transform
        img, _ = self.original_dataset[self.indices[idx]] 
        # Return the transformed image and the new remapped target
        return img, self.new_targets[idx]

# Create the filtered training and testing datasets.
train_dataset = CustomFilteredDataset(full_trainset, cifar_to_crop_health_map)
test_dataset = CustomFilteredDataset(full_testset, cifar_to_crop_health_map)

# Define DataLoader parameters.
batch_size = 128 # Larger batch size for faster training on GPU
num_workers = 4  # Number of subprocesses to use for data loading. Adjust based on CPU cores.

# Create DataLoaders for efficient batching and shuffling.
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=num_workers, pin_memory=True)
test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, num_workers=num_workers, pin_memory=True)

print(f"Number of training samples: {len(train_dataset)}")
print(f"Number of test samples: {len(test_dataset)}")

# --- 3. CNN Architecture Design ---
# Define the Convolutional Neural Network (CNN) model.
class CropHealthCNN(nn.Module):
    def __init__(self, num_classes=2):
        super(CropHealthCNN, self).__init__()
        # Feature extraction layers (Convolutional blocks)
        self.features = nn.Sequential(
            # Block 1: Input 3x64x64
            nn.Conv2d(3, 32, kernel_size=3, padding=1), # Output: 32x64x64
            nn.ReLU(),
            nn.BatchNorm2d(32), # Batch Normalization for stable training
            nn.MaxPool2d(kernel_size=2, stride=2), # Output: 32x32x32
            
            # Block 2: Input 32x32x32
            nn.Conv2d(32, 64, kernel_size=3, padding=1), # Output: 64x32x32
            nn.ReLU(),
            nn.BatchNorm2d(64),
            nn.MaxPool2d(kernel_size=2, stride=2), # Output: 64x16x16
            
            # Block 3: Input 64x16x16
            nn.Conv2d(64, 128, kernel_size=3, padding=1), # Output: 128x16x16
            nn.ReLU(),
            nn.BatchNorm2d(128),
            nn.MaxPool2d(kernel_size=2, stride=2) # Output: 128x8x8
        )
        
        # Classifier layers (Fully Connected blocks)
        # The output of the last MaxPool2d is 128 channels with 8x8 spatial dimensions.
        self.classifier = nn.Sequential(
            nn.Flatten(), # Flatten the 128x8x8 tensor into a 1D vector (128 * 8 * 8 = 8192 features)
            nn.Linear(128 * 8 * 8, 512), # First fully connected layer with 512 output features
            nn.ReLU(),
            nn.Dropout(0.5), # Dropout for regularization to prevent overfitting
            nn.Linear(512, num_classes) # Output layer with `num_classes` (2 for Healthy/Diseased)
        )

    def forward(self, x):
        x = self.features(x)   # Pass input through convolutional feature extractor
        x = self.classifier(x) # Pass features through the classifier
        return x

# Instantiate the model and move it to the configured device (GPU/CPU).
model = CropHealthCNN(num_classes=2).to(device)

# --- 5. Performance Optimization (Bonus): Integrate torch.compile ---
# This feature (PyTorch 2.0+) compiles the model into optimized kernels,
# potentially offering significant speedups for both training and inference.
# It's a 'just-in-time' compiler that optimizes PyTorch code graphs.
# Benefits: Faster execution, reduced memory usage, especially on GPUs.
# Note: The first pass through the compiled model might be slower due to compilation overhead.
# Uncomment the line below to enable torch.compile.
# model = torch.compile(model)
print(f"Model compiled with torch.compile: {'Yes' if isinstance(model, torch._dynamo.eval_frame.OptimizedModule) else 'No'}")

# --- 4. Training Loop Implementation ---
# Define hyperparameters for training.
num_epochs = 20         # Number of full passes over the training dataset
learning_rate = 0.001   # Step size for the optimizer

# Define Loss function and Optimizer.
# CrossEntropyLoss is suitable for multi-class classification problems.
# Adam optimizer is a popular choice for its adaptive learning rates.
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# Lists to store training and validation metrics for plotting later.
train_losses = []
val_losses = []
val_accuracies = []

print("\nStarting training...")
for epoch in range(num_epochs):
    model.train() # Set the model to training mode (enables dropout, batchnorm updates)
    running_loss = 0.0
    for i, (images, labels) in enumerate(train_loader):
        images = images.to(device) # Move images to the specified device
        labels = labels.to(device) # Move labels to the specified device

        # Forward pass: Compute model output for the current batch.
        outputs = model(images)
        # Calculate the loss between model outputs and true labels.
        loss = criterion(outputs, labels)

        # Backward and optimize:
        optimizer.zero_grad() # Clear previous gradients
        loss.backward()       # Compute gradients of the loss with respect to model parameters
        optimizer.step()      # Update model parameters using the optimizer

        running_loss += loss.item() # Accumulate batch loss

    # Calculate average training loss for the epoch.
    avg_train_loss = running_loss / len(train_loader)
    train_losses.append(avg_train_loss)
    
    # --- 5. Validation (within training loop) ---
    model.eval() # Set the model to evaluation mode (disables dropout, uses running stats for batchnorm)
    val_loss = 0.0
    correct = 0
    total = 0
    all_val_labels = []
    all_val_predictions = []
    
    with torch.no_grad(): # Disable gradient calculation during validation to save memory and speed up computation
        for images, labels in test_loader:
            images = images.to(device)
            labels = labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            val_loss += loss.item()
            
            _, predicted = torch.max(outputs.data, 1) # Get the class with the highest probability
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            
            all_val_labels.extend(labels.cpu().numpy())
            all_val_predictions.extend(predicted.cpu().numpy())
    
    # Calculate average validation loss and accuracy for the epoch.
    avg_val_loss = val_loss / len(test_loader)
    val_accuracy = 100 * correct / total
    val_losses.append(avg_val_loss)
    val_accuracies.append(val_accuracy)
    
    print(f"Epoch [{epoch+1}/{num_epochs}], Train Loss: {avg_train_loss:.4f}, Val Loss: {avg_val_loss:.4f}, Val Accuracy: {val_accuracy:.2f}%")

print("Training complete!")

# Plotting training and validation loss/accuracy
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(train_losses, label='Training Loss')
plt.plot(val_losses, label='Validation Loss')
plt.title('Loss over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(val_accuracies, label='Validation Accuracy', color='orange')
plt.title('Validation Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.legend()
plt.show()

# --- 6. Final Evaluation on Test Set ---
# Evaluate the model's performance on the unseen test dataset using various metrics.
model.eval() # Ensure model is in evaluation mode
all_labels = []
all_predictions = []

with torch.no_grad(): # No need to compute gradients for inference
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1) # Get the predicted class
        
        all_labels.extend(labels.cpu().numpy())         # Store true labels
        all_predictions.extend(predicted.cpu().numpy()) # Store predicted labels

# Calculate and print evaluation metrics.
accuracy = accuracy_score(all_labels, all_predictions)
precision = precision_score(all_labels, all_predictions, average='binary') # 'binary' for 2 classes
recall = recall_score(all_labels, all_predictions, average='binary')
f1 = f1_score(all_labels, all_predictions, average='binary')

print("\n--- Final Test Set Evaluation ---")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")

# Optional: Visualize some predictions from the test set
def imshow(img):
    # Unnormalize image for display (reverse the normalization transform)
    img = img / 2 + 0.5  
    npimg = img.numpy()
    # Transpose from (C, H, W) to (H, W, C) for matplotlib
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.axis('off') # Hide axes
    plt.show()

# Get a batch of test images and their labels
dataiter = iter(test_loader)
images, labels = next(dataiter)

# Make predictions on this batch
outputs = model(images.to(device))
_, predicted = torch.max(outputs, 1)

# Define class names for better readability
class_names = {0: 'Healthy', 1: 'Diseased'}

print('\nGround Truth: ', ' '.join(f'{class_names[label.item()]}' for label in labels[:8])) # Show first 8
print('Predicted:    ', ' '.join(f'{class_names[p.item()]}' for p in predicted[:8])) # Show first 8

# Display the first few images with their true and predicted labels
fig, axes = plt.subplots(1, 8, figsize=(16, 2))
for i in range(8):
    ax = axes[i]
    img = images[i]
    # Unnormalize and display
    img = img / 2 + 0.5
    npimg = img.numpy()
    ax.imshow(np.transpose(npimg, (1, 2, 0)))
    ax.set_title(f'True: {class_names[labels[i].item()]}\nPred: {class_names[predicted[i].item()]}', fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
