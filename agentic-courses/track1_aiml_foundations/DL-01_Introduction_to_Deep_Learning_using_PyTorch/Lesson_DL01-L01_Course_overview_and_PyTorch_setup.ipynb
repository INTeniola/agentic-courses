```python
# --- Hello, PyTorch! Basic Tensor Operations ---

# This simple code block will verify your PyTorch installation and introduce you to tensors,
# the fundamental data structure in PyTorch.

import torch

print("\n--- Running 'Hello, PyTorch!' demo ---")

# 1. Create a PyTorch tensor
# Tensors are multi-dimensional arrays, similar to NumPy arrays, but with GPU acceleration capabilities.
scalar_tensor = torch.tensor(7)
print(f"Scalar Tensor: {scalar_tensor}, Type: {scalar_tensor.dtype}, Shape: {scalar_tensor.shape}")

vector_tensor = torch.tensor([1, 2, 3])
print(f"Vector Tensor: {vector_tensor}, Type: {vector_tensor.dtype}, Shape: {vector_tensor.shape}")

matrix_tensor = torch.tensor([[1, 2], [3, 4]])
print(f"Matrix Tensor:\n{matrix_tensor}, Type: {matrix_tensor.dtype}, Shape: {matrix_tensor.shape}")

# 2. Perform a simple operation
# Tensors support various mathematical operations.
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])
z = x + y
print(f"\nTensor Addition (x + y): {z}")

# 3. Move a tensor to GPU (if available)
if torch.cuda.is_available():
    device = torch.device("cuda")
    x_gpu = x.to(device)
    y_gpu = y.to(device)
    z_gpu = x_gpu + y_gpu
    print(f"\nTensor Addition on GPU (x_gpu + y_gpu): {z_gpu}, Device: {z_gpu.device}")
else:
    print("\nGPU not available, skipping GPU tensor operations.")

print("\n'Hello, PyTorch!' demo complete. Your environment is ready!")
```
