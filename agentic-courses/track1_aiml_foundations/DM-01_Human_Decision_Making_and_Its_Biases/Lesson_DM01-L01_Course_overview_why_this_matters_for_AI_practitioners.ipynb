## Setting Up Your Environment

While this introductory lesson focuses on conceptual understanding, subsequent lessons will involve practical Python exercises. To ensure a smooth learning experience, let's set up our development environment. No prior knowledge of cognitive science is required, but basic familiarity with Python is beneficial.

### Recommended Tools (as of 2026):

*   **Python 3.10+**: The standard for modern AI/ML development.
*   **`pip` (or `conda`)**: For package management.
*   **Jupyter Environment**: We recommend Google Colab for a zero-setup cloud experience, or a local setup using JupyterLab, VS Code with Jupyter extensions, or similar.
*   **Virtual Environments**: Best practice dictates using `venv` (Python's built-in virtual environment tool) or `conda env` to manage project dependencies and avoid conflicts.

If you are using Google Colab, most of the core libraries are already pre-installed. For local setups, follow the instructions below to create a virtual environment and install the necessary packages.


```python
# Check Python version
import sys
print(f"Python Version: {sys.version.split(' ')[0]}")

# --- For Local Environments (Skip if using Google Colab) ---
# It's highly recommended to use a virtual environment.
# If you haven't already, create and activate one:
# 1. Create: python -m venv dm01_env
# 2. Activate (Linux/macOS): source dm01_env/bin/activate
# 3. Activate (Windows): .\dm01_env\Scripts\activate

# Install core libraries. We'll use these throughout the course.
# As of 2026, these are standard for data manipulation and scientific computing.
# We'll install them here to ensure your environment is ready.

# Define a list of required packages
required_packages = [
    "numpy>=1.26",      # Numerical computing
    "pandas>=2.2",      # Data manipulation and analysis
    "scipy>=1.12",      # Scientific computing
    "matplotlib>=3.8",  # Plotting and visualization
    "seaborn>=0.13"     # Statistical data visualization
]

# Install packages using pip
# The try-except block handles cases where packages might already be installed
# or if there are minor version conflicts, providing a robust installation.
print("\nInstalling/Upgrading required packages...")
import subprocess
import importlib.util

for package in required_packages:
    package_name = package.split('>')[0].split('=')[0] # Extract base name for check
    try:
        # Check if package is already installed and meets version requirements
        spec = importlib.util.find_spec(package_name)
        if spec is None:
            # Package not found, install it
            subprocess.check_call([sys.executable, "-m", "pip", "install", package])
            print(f"Successfully installed {package}")
        else:
            # Package found, check version or upgrade if needed
            # For simplicity, we'll just try to upgrade to ensure latest compatible version
            subprocess.check_call([sys.executable, "-m", "pip", "install", "--upgrade", package])
            print(f"Successfully upgraded/ensured {package}")
    except subprocess.CalledProcessError as e:
        print(f"Error installing {package}: {e}")
    except Exception as e:
        print(f"An unexpected error occurred for {package}: {e}")

print("\nEnvironment setup complete (or verified) for core libraries.")

# For best practice, you would typically save these to a requirements.txt file:
# pip freeze > requirements.txt
```

## Verifying Your Setup

To confirm that our environment is correctly configured and the necessary libraries are accessible, let's run a simple "hello world" style check. This will import a couple of the core libraries and perform a basic operation, ensuring everything is ready for our deeper dives into human decision-making and AI biases.


```python
# Import core libraries to verify installation
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Print versions to confirm successful import and installation
print(f"NumPy Version: {np.__version__}")
print(f"Pandas Version: {pd.__version__}")
print(f"Matplotlib Version: {plt.__version__}")
print(f"Seaborn Version: {sns.__version__}")

# Perform a simple operation to demonstrate functionality
# Create a NumPy array
data_array = np.array([10, 20, 30, 40, 50])
print(f"\nSample NumPy array: {data_array}")
print(f"Mean of array: {np.mean(data_array)}")

# Create a Pandas DataFrame
data_df = pd.DataFrame({
    'Category': ['A', 'B', 'C', 'D', 'E'],
    'Value': data_array
})
print(f"\nSample Pandas DataFrame:\n{data_df}")

# Basic plot to ensure plotting libraries are working
plt.figure(figsize=(6, 4))
sns.barplot(x='Category', y='Value', data=data_df)
plt.title('Environment Check: Basic Bar Plot')
plt.ylabel('Value')
plt.xlabel('Category')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

print("\nCongratulations! Your environment is successfully set up and ready for DM-01. Let's dive into the fascinating world of human decision-making and its implications for AI!")
```
