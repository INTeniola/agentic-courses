## Quiz: Recognizing and Countering Biases in AI

### Question 1: Identifying Bias Type

A new AI-powered medical diagnostic tool, trained predominantly on health data from individuals in high-income countries, consistently misdiagnoses rare diseases that are more prevalent in low-income regions. What type of bias is most evident in this scenario?

A) Confirmation Bias
B) Selection Bias
C) Automation Bias
D) Historical Bias

### Question 2: Proactive Mitigation Strategy

Which of the following is the most effective *proactive* strategy to reduce selection bias in a new AI model's training data for a critical application like loan approvals?

A) Post-hoc re-weighting of underrepresented groups in the dataset after initial model training.
B) Implementing a robust data collection plan that ensures diverse and representative sampling across all relevant demographic groups.
C) Relying solely on transfer learning from a pre-trained model on a large, publicly available dataset without further validation.
D) Conducting A/B testing on the deployed model to identify performance disparities among different user segments.

### Question 3: Ethical Implications of Bias

An AI-powered hiring tool consistently ranks male candidates higher than equally qualified female candidates, despite gender-neutral resumes. What is the most significant ethical implication of this bias?

A) Reduced computational efficiency of the hiring process.
B) Perpetuation of systemic gender inequality in the workforce.
C) Increased need for human oversight in the final hiring decision.
D) Potential for legal challenges due to data privacy violations.

### Question 4: Human vs. Algorithmic Bias

Which statement best describes the relationship between human cognitive biases and algorithmic biases in AI systems?

A) Algorithmic biases are entirely distinct from human cognitive biases and arise solely from technical errors.
B) Algorithmic biases are often a reflection and amplification of human cognitive biases present in the data or design choices.
C) Human cognitive biases are irrelevant to AI systems, as algorithms operate purely on logic and mathematical principles.
D) Algorithmic biases always correct for human cognitive biases, leading to inherently fairer and more objective outcomes.

### Answers and Explanations

**Question 1: Identifying Bias Type**

**Correct Answer: B) Selection Bias**

*   **Explanation:** Selection bias occurs when the data used to train an AI model does not accurately represent the population or conditions the model will encounter in the real world. In this case, training data from high-income countries (a specific subset) leads to poor performance on populations from low-income regions (another subset), indicating that the training data was not representative of the full operational domain. While historical societal inequalities might contribute to data availability (leading to D), the direct issue described is the *selection* of unrepresentative data.

**Question 2: Proactive Mitigation Strategy**

**Correct Answer: B) Implementing a robust data collection plan that ensures diverse and representative sampling across all relevant demographic groups.**

*   **Explanation:** This is the most proactive and fundamental approach. Addressing bias at the data collection stage prevents it from being embedded into the model from the outset. Options A and D are reactive (post-training or post-deployment). Option C is risky as pre-trained models can carry their own biases, and relying solely on them without validation is not a proactive mitigation strategy for *your specific* application's bias.

**Question 3: Ethical Implications of Bias**

**Correct Answer: B) Perpetuation of systemic gender inequality in the workforce.**

*   **Explanation:** The most profound ethical implication of biased AI in hiring is its potential to reinforce and exacerbate existing societal inequalities. By systematically disadvantaging certain groups, the AI contributes to unfair outcomes and hinders progress towards a more equitable society. While C (increased human oversight) might be a necessary *response* to the bias, and D (legal challenges) a *consequence*, B describes the core ethical harm.

**Question 4: Human vs. Algorithmic Bias**

**Correct Answer: B) Algorithmic biases are often a reflection and amplification of human cognitive biases present in the data or design choices.**

*   **Explanation:** This statement accurately captures the intricate relationship. AI systems learn from data created by humans or reflecting human decisions, which inherently contain human cognitive biases (e.g., implicit bias, confirmation bias). Algorithms can then amplify these biases due to their pattern-matching nature, leading to biased outcomes at scale. They are not distinct from human biases (A), nor do they always correct them (D), and human biases are highly relevant (C).
