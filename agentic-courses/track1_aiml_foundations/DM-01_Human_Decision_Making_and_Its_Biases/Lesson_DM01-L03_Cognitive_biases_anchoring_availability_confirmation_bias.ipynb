## Cognitive Biases: Anchoring, Availability, and Confirmation Bias

Welcome to Lesson DM01-L03, where we delve into three fundamental cognitive biases that profoundly influence human decision-making: **Anchoring Bias**, the **Availability Heuristic**, and **Confirmation Bias**. Understanding these biases is crucial not only for comprehending human behavior but also for recognizing how they can inadvertently infect and degrade the performance and fairness of AI systems.

Cognitive biases are systematic errors in thinking that occur when people process and interpret information in the world around them. They are mental shortcuts (heuristics) that our brains use to make decisions quickly, often leading to deviations from rational judgment. While sometimes efficient, these shortcuts can lead to predictable and significant errors.

---

### 1. Anchoring Bias

**Definition:** Anchoring bias describes our tendency to rely too heavily on the first piece of information offered (the "anchor") when making decisions. Once an anchor is set, subsequent judgments are made by adjusting away from that anchor, but typically, the adjustment is insufficient.

**Intuitive Analogy:** Imagine you're trying to guess the number of jelly beans in a large jar. If someone first suggests "10,000" (a high anchor), your guess will likely be higher than if they first suggested "1,000" (a low anchor), even if you try to be objective.

**Real-World Examples:**
*   **Negotiations:** In a salary negotiation, the first offer made (the anchor) significantly influences the final agreed-upon salary, even if it's an extreme value.
*   **Retail Pricing:** A product initially priced very high, then put on "sale" at a lower but still expensive price, appears more attractive due to the high initial anchor.
*   **Legal Judgments:** Initial damage requests in lawsuits can anchor jury decisions, even if they are arbitrary.

**How it Infects AI:**
*   **Initial Parameters/Defaults:** Default settings or initial weights in a machine learning model can act as anchors, influencing the optimization process and final model performance, especially in complex, non-convex optimization landscapes.
*   **First Impressions in Data:** If an AI system is exposed to a particular type of data or a specific outcome early in its training, it might be unduly influenced by this initial exposure, making it harder to adjust to new, contradictory information later.
*   **Human-in-the-Loop Labeling:** Human annotators, when labeling data, might be anchored by previous labels or by instructions that implicitly suggest a certain distribution of labels.

---

### 2. Availability Heuristic

**Definition:** The availability heuristic is a mental shortcut that relies on immediate examples that come to a given person's mind when evaluating a specific topic, concept, method, or decision. We tend to overestimate the likelihood or frequency of events that are more easily recalled from memory.

**Intuitive Analogy:** Think about news coverage. Dramatic, rare events like plane crashes or shark attacks receive extensive media attention. Because these events are so vivid and easily recalled, people often overestimate their actual probability compared to more common but less dramatic events like car accidents or heart disease.

**Real-World Examples:**
*   **Risk Assessment:** People often fear flying more than driving, despite statistics showing driving is far more dangerous. Plane crashes are highly publicized and easily recalled, making them seem more probable.
*   **Product Reviews:** A few extremely positive or negative reviews, if easily recalled, can disproportionately influence a purchasing decision, even if the overall average rating is moderate.
*   **Medical Diagnoses:** Doctors might be more likely to diagnose a rare disease if they've recently encountered a similar case, even if common ailments are statistically more probable.

**How it Infects AI:**
*   **Skewed Datasets:** If a training dataset disproportionately represents certain events or outcomes (even if they are rare in reality), an AI model will learn to predict those events as more probable.
*   **Recent Data Bias:** Models that are continuously updated might over-index on recent, easily available data, potentially forgetting or downplaying older, but still relevant, patterns.
*   **Search & Recommendation Systems:** If certain content is more easily indexed or frequently interacted with, recommendation systems might over-promote it, creating a feedback loop that reinforces its availability.

---

### 3. Confirmation Bias

**Definition:** Confirmation bias is the tendency to search for, interpret, favor, and recall information in a way that confirms one's pre-existing beliefs or hypotheses. It's a form of selective thinking where we look for evidence that supports what we already believe and ignore evidence that contradicts it.

**Intuitive Analogy:** Imagine you believe that a certain type of weather always brings bad luck. You'll likely remember all the times something bad happened on that weather day, while conveniently forgetting all the times nothing happened, or when bad things happened on other weather days.

**Real-World Examples:**
*   **Political Views:** People tend to consume news and social media that aligns with their political ideology, reinforcing their existing beliefs and dismissing opposing viewpoints.
*   **Scientific Research (unintentional):** Researchers might unconsciously design experiments or interpret results in a way that confirms their initial hypothesis, even with the best intentions.
*   **Hiring Decisions:** An interviewer who believes a candidate from a certain university is superior might unconsciously ask leading questions or interpret ambiguous answers in a way that confirms their initial positive impression.

**How it Infects AI:**
*   **Data Labeling:** Human annotators, influenced by their own biases, might label data in a way that confirms their existing beliefs, leading to biased training sets.
*   **Feature Selection:** AI developers might unconsciously select features that align with their preconceived notions about what's important, ignoring other potentially relevant features.
*   **Model Evaluation:** When evaluating models, developers might focus on metrics or scenarios that make their model look good, while downplaying or ignoring failures that contradict their belief in the model's efficacy.
*   **Recommendation Systems & Filter Bubbles:** These systems can create "echo chambers" by continuously recommending content similar to what a user has already consumed, reinforcing existing beliefs and limiting exposure to diverse perspectives.


```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# --- 1. Simulating Anchoring Bias ---
print("--- Anchoring Bias Simulation ---")
# Scenario: Estimating the number of countries in Africa (actual ~54)
actual_value = 54

# Simulate two groups of people asked to estimate
np.random.seed(2026) # For reproducibility

# Group A: Given a low anchor (e.g., "Is it more or less than 10?")
low_anchor = 10
# People tend to adjust upwards from a low anchor, but not enough
estimates_low_anchor = np.random.normal(loc=actual_value * 0.7, scale=15, size=100) # Mean pulled towards anchor
estimates_low_anchor = np.clip(estimates_low_anchor, 15, 70) # Realistic range

# Group B: Given a high anchor (e.g., "Is it more or less than 100?")
high_anchor = 100
# People tend to adjust downwards from a high anchor, but not enough
estimates_high_anchor = np.random.normal(loc=actual_value * 1.3, scale=15, size=100) # Mean pulled towards anchor
estimates_high_anchor = np.clip(estimates_high_anchor, 30, 90) # Realistic range

print(f"Actual value: {actual_value}")
print(f"Mean estimate (Low Anchor Group): {np.mean(estimates_low_anchor):.2f}")
print(f"Mean estimate (High Anchor Group): {np.mean(estimates_high_anchor):.2f}")

plt.figure(figsize=(10, 6))
plt.hist(estimates_low_anchor, bins=15, alpha=0.6, label=f'Low Anchor ({low_anchor})', color='skyblue')
plt.hist(estimates_high_anchor, bins=15, alpha=0.6, label=f'High Anchor ({high_anchor})', color='salmon')
plt.axvline(actual_value, color='green', linestyle='dashed', linewidth=2, label=f'Actual Value ({actual_value})')
plt.title('Simulated Estimates Influenced by Anchoring Bias')
plt.xlabel('Estimated Number of Countries')
plt.ylabel('Frequency')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()

print("\n" + "-"*50 + "\n")

# --- 2. Simulating Availability Heuristic ---
print("--- Availability Heuristic Simulation ---")
# Scenario: Perceived risk of rare vs. common events based on 'news coverage'

# True underlying probabilities of events
true_events = {
    'Car Accident': 0.60,
    'Heart Attack': 0.30,
    'Shark Attack': 0.001, # Very rare
    'Lightning Strike': 0.009,
    'Flu Complications': 0.09
}

# Simulate a large number of actual events over time
num_actual_events = 10000
actual_event_stream = np.random.choice(
    list(true_events.keys()),
    size=num_actual_events,
    p=list(true_events.values())
)

# Calculate true frequencies
true_frequencies = pd.Series(actual_event_stream).value_counts(normalize=True)
print("True Frequencies of Events (Actual Probabilities):\n", true_frequencies.sort_values(ascending=False))

# Simulate 'news coverage' where rare, dramatic events are over-reported
# We'll artificially boost 'Shark Attack' and 'Lightning Strike' in the reported stream
reported_events_list = list(actual_event_stream)

# Add extra 'Shark Attack' and 'Lightning Strike' mentions to simulate media focus
for _ in range(500): # Add 500 extra mentions of shark attacks
    reported_events_list.append('Shark Attack')
for _ in range(200): # Add 200 extra mentions of lightning strikes
    reported_events_list.append('Lightning Strike')

# Calculate perceived frequencies based on 'reported' events
perceived_frequencies = pd.Series(reported_events_list).value_counts(normalize=True)
print("\nPerceived Frequencies of Events (Based on 'News Coverage'):\n", perceived_frequencies.sort_values(ascending=False))

plt.figure(figsize=(12, 6))
bar_width = 0.35
index = np.arange(len(true_frequencies))

plt.bar(index, true_frequencies.values, bar_width, label='True Probability', color='lightgray')
plt.bar(index + bar_width, perceived_frequencies.reindex(true_frequencies.index).values, bar_width, label='Perceived Probability (Availability)', color='darkblue')

plt.xlabel('Event Type')
plt.ylabel('Probability / Frequency')
plt.title('Availability Heuristic: True vs. Perceived Event Probabilities')
plt.xticks(index + bar_width / 2, true_frequencies.index, rotation=45, ha='right')
plt.legend()
plt.tight_layout()
plt.show()

print("\n" + "-"*50 + "\n")

# --- 3. Simulating Confirmation Bias ---
print("--- Confirmation Bias Simulation ---")
# Scenario: A hiring model biased towards candidates from 'Elite University'

# Generate synthetic data for hiring decisions
# Features: 'Years_Experience', 'Interview_Score', 'Elite_University' (0 or 1)
# Target: 'Hired' (0 or 1)

num_samples = 1000
np.random.seed(2026)

data = {
    'Years_Experience': np.random.randint(1, 15, num_samples),
    'Interview_Score': np.random.randint(50, 100, num_samples),
    'Elite_University': np.random.choice([0, 1], size=num_samples, p=[0.7, 0.3]) # 30% from elite uni
}
df = pd.DataFrame(data)

# True underlying hiring logic (unbiased, for comparison):
# High experience AND high interview score are good indicators
# Elite university might give a slight edge, but not dominant
df['True_Hired'] = ((df['Years_Experience'] >= 5) & (df['Interview_Score'] >= 75)).astype(int)
# Add a slight boost for elite university, but not enough to dominate
df.loc[df['Elite_University'] == 1, 'True_Hired'] = df.loc[df['Elite_University'] == 1, 'True_Hired'] | (np.random.rand(len(df[df['Elite_University'] == 1])) < 0.2).astype(int)

# Simulate a 'biased' data labeling process (confirmation bias):
# The human labeler *believes* Elite University candidates are always better.
# So, they tend to label Elite University candidates as 'Hired' more often,
# even if their other qualifications aren't outstanding.

df['Biased_Hired'] = df['True_Hired'].copy() # Start with true labels

# Introduce bias: If Elite_University == 1, increase chance of being labeled 'Hired'
# Even if True_Hired was 0, there's a chance it's flipped to 1 due to bias
biased_elite_indices = df[(df['Elite_University'] == 1) & (df['True_Hired'] == 0)].index
flip_to_hired_count = int(len(biased_elite_indices) * 0.6) # 60% of non-hired elite candidates get flipped
flip_indices = np.random.choice(biased_elite_indices, size=flip_to_hired_count, replace=False)
df.loc[flip_indices, 'Biased_Hired'] = 1

# Also, if Elite_University == 0, and True_Hired was 1, there's a chance it's flipped to 0 (downplaying non-elite)
biased_non_elite_indices = df[(df['Elite_University'] == 0) & (df['True_Hired'] == 1)].index
flip_to_not_hired_count = int(len(biased_non_elite_indices) * 0.2) # 20% of hired non-elite candidates get flipped
flip_indices_non_elite = np.random.choice(biased_non_elite_indices, size=flip_to_not_hired_count, replace=False)
df.loc[flip_indices_non_elite, 'Biased_Hired'] = 0

print("Original 'True_Hired' distribution:\n", df['True_Hired'].value_counts(normalize=True))
print("\nBiased 'Biased_Hired' distribution (due to human confirmation bias):\n", df['Biased_Hired'].value_counts(normalize=True))

# Train a model on the *biased* labels
X = df[['Years_Experience', 'Interview_Score', 'Elite_University']]
y_biased = df['Biased_Hired']

X_train, X_test, y_train_biased, y_test_biased = train_test_split(X, y_biased, test_size=0.3, random_state=2026)

model_biased = LogisticRegression(solver='liblinear', random_state=2026)
model_biased.fit(X_train, y_train_biased)

y_pred_biased = model_biased.predict(X_test)

print("\nModel trained on Biased Labels Performance:")
print(f"Accuracy: {accuracy_score(y_test_biased, y_pred_biased):.2f}")
print("Classification Report:\n", classification_report(y_test_biased, y_pred_biased))

# Let's see how this biased model predicts for new, unseen candidates
# Candidate 1: High experience, high score, NOT elite uni
candidate1 = pd.DataFrame([{'Years_Experience': 10, 'Interview_Score': 90, 'Elite_University': 0}])
# Candidate 2: Low experience, low score, IS elite uni
candidate2 = pd.DataFrame([{'Years_Experience': 3, 'Interview_Score': 60, 'Elite_University': 1}])
# Candidate 3: Moderate experience, moderate score, NOT elite uni
candidate3 = pd.DataFrame([{'Years_Experience': 6, 'Interview_Score': 70, 'Elite_University': 0}])
# Candidate 4: Moderate experience, moderate score, IS elite uni
candidate4 = pd.DataFrame([{'Years_Experience': 6, 'Interview_Score': 70, 'Elite_University': 1}])

print("\nPredictions from Biased Model:")
print(f"Candidate 1 (Exp:10, Score:90, Elite:No) -> Hired: {model_biased.predict(candidate1)[0]}")
print(f"Candidate 2 (Exp:3, Score:60, Elite:Yes) -> Hired: {model_biased.predict(candidate2)[0]}")
print(f"Candidate 3 (Exp:6, Score:70, Elite:No) -> Hired: {model_biased.predict(candidate3)[0]}")
print(f"Candidate 4 (Exp:6, Score:70, Elite:Yes) -> Hired: {model_biased.predict(candidate4)[0]}")

# For comparison, train a model on the *true* labels (if we had them)
X_train, X_test, y_train_true, y_test_true = train_test_split(X, df['True_Hired'], test_size=0.3, random_state=2026)
model_true = LogisticRegression(solver='liblinear', random_state=2026)
model_true.fit(X_train, y_train_true)

print("\nPredictions from Unbiased (True Labels) Model (for comparison):")
print(f"Candidate 1 (Exp:10, Score:90, Elite:No) -> Hired: {model_true.predict(candidate1)[0]}")
print(f"Candidate 2 (Exp:3, Score:60, Elite:Yes) -> Hired: {model_true.predict(candidate2)[0]}")
print(f"Candidate 3 (Exp:6, Score:70, Elite:No) -> Hired: {model_true.predict(candidate3)[0]}")
print(f"Candidate 4 (Exp:6, Score:70, Elite:Yes) -> Hired: {model_true.predict(candidate4)[0]}")
```

### Interpreting the Simulations and AI Implications

Our simulations vividly demonstrate how cognitive biases, even when simplified, can lead to skewed perceptions and decisions, mirroring their impact on both human and AI systems.

#### 1. Anchoring Bias Interpretation

*   **Code Output:** You observed two distinct distributions of estimates, one pulled towards a low anchor and another towards a high anchor, both deviating from the actual value. The mean estimates for each group were significantly different, despite the underlying true value being the same.
*   **AI Implications:** This illustrates how initial conditions or default values in AI can act as anchors. For instance, if a model's initial weights are set in a certain range, or if an optimization algorithm starts from a particular point, it might converge to a suboptimal solution that is "anchored" to that starting point, even if better solutions exist elsewhere in the parameter space. In recommendation systems, an initial set of highly-rated items (an anchor) might disproportionately influence subsequent recommendations, making it hard for new, diverse items to gain traction.
*   **Mitigation:** Techniques like diverse initialization strategies, ensemble methods (averaging multiple models started from different anchors), and robust hyperparameter tuning can help reduce anchoring effects in AI.

#### 2. Availability Heuristic Interpretation

*   **Code Output:** The "True Frequencies" showed the actual probabilities of events, with common events like 'Car Accident' and 'Heart Attack' being most frequent. However, the "Perceived Frequencies" (simulating news coverage) dramatically inflated the perceived likelihood of rare, dramatic events like 'Shark Attack' and 'Lightning Strike'. This discrepancy highlights how easily recalled information disturbs our perception of reality.
*   **AI Implications:** This bias directly translates to issues with data representation. If an AI model is trained on a dataset where certain events are over-represented (e.g., due to selective data collection, media bias in text corpora, or specific user interaction patterns), the model will learn to assign higher probabilities or importance to those events, regardless of their true prevalence. This can lead to: 
    *   **Misleading Predictions:** A medical diagnostic AI might over-diagnose a rare condition if its training data contained an unusually high number of such cases.
    *   **Unfair Resource Allocation:** An AI system allocating emergency services might prioritize responses to highly publicized (but statistically rare) incidents over more common, yet less reported, emergencies.
    *   **Filter Bubbles:** Recommendation systems can fall prey to this by promoting content that is already popular or easily accessible, reinforcing its availability and making it seem more important or relevant than it truly is.
*   **Mitigation:** Careful data collection strategies, re-sampling techniques (oversampling rare classes, undersampling common ones), synthetic data generation, and robust evaluation metrics that account for class imbalance are crucial.

#### 3. Confirmation Bias Interpretation

*   **Code Output:** You saw how a human's pre-existing belief ("Elite University candidates are better") led to a *biased* labeling of the training data. The model trained on this biased data then reflected this bias in its predictions. Specifically, the biased model was more likely to hire candidates from 'Elite University' even with moderate qualifications, and less likely to hire highly qualified candidates *not* from an elite university, compared to the 'True Labels' model.
*   **AI Implications:** This is perhaps one of the most insidious biases in AI, as it often originates from human biases embedded in the data or the development process:
    *   **Biased Training Data:** If human annotators or data collectors have confirmation bias, they will create datasets that reinforce existing stereotypes or beliefs. This is common in areas like hiring, loan applications, and criminal justice, where historical human decisions (often biased) are used to train AI.
    *   **Feature Selection & Model Design:** Developers might unconsciously select features or design models that confirm their initial hypotheses, ignoring contradictory evidence or alternative approaches.
    *   **Evaluation Metrics:** Choosing evaluation metrics that favor a certain outcome or focusing only on performance aspects that confirm the model's perceived success can hide underlying biases.
    *   **Algorithmic Feedback Loops:** In systems like social media feeds, if a user interacts with content that confirms their beliefs, the algorithm will show them more of that content, creating an echo chamber that reinforces the bias.
*   **Mitigation:** Diverse and representative data collection, blind labeling processes, fairness-aware machine learning algorithms, explainable AI (XAI) to understand model decisions, and rigorous, multi-faceted evaluation by diverse teams are essential to combat confirmation bias.

### Performance Trade-offs and Use Cases

While these simulations don't directly measure computational performance, the presence of these biases in AI systems has significant performance trade-offs:

*   **Reduced Generalization:** Biased models often perform poorly on data outside their biased training distribution, leading to brittle and unreliable systems.
*   **Unfair Outcomes:** This is the most critical trade-off. Biased AI can perpetuate and amplify societal inequalities, leading to discriminatory decisions in critical domains like healthcare, finance, and employment.
*   **Lack of Trust:** Systems perceived as biased lose user trust, leading to low adoption and negative societal impact.
*   **Suboptimal Solutions:** Anchoring can prevent models from finding the truly optimal solution, leading to less efficient or effective systems.

**Typical Use Cases (where these biases are critical to address):**
*   **Hiring and Recruitment:** AI used to screen resumes or conduct interviews.
*   **Loan and Credit Scoring:** AI determining creditworthiness.
*   **Criminal Justice:** Predictive policing, recidivism risk assessment.
*   **Healthcare:** Diagnostic tools, treatment recommendations.
*   **Recommendation Systems:** Content suggestions, product recommendations.
*   **Autonomous Systems:** Decision-making in self-driving cars or drones.

Understanding these biases is the first step towards building more robust, fair, and ethical AI systems. It requires a conscious effort to identify potential sources of bias in data, models, and human processes, and to implement proactive mitigation strategies.
