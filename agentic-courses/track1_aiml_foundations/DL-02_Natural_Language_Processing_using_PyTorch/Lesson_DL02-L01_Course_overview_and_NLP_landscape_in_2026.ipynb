## Hello World: Your First NLP Task with Hugging Face Transformers

To confirm our environment is fully operational and to get a taste of modern NLP, let's perform a simple sentiment analysis task using a pre-trained model from the Hugging Face `transformers` library. This will demonstrate how easily you can leverage powerful models with just a few lines of PyTorch-backed code.


```python
import torch
from transformers import pipeline

# Define the device to use (GPU if available, otherwise CPU)
device = 0 if torch.cuda.is_available() else -1
print(f"\nUsing device: {'GPU' if device == 0 else 'CPU'}")

# Load a pre-trained sentiment analysis pipeline
# We'll use a small, efficient model for this 'hello world' example.
# 'distilbert-base-uncased-finetuned-sst-2-english' is a good choice for quick sentiment analysis.
print("Loading sentiment analysis pipeline...")
sentiment_pipeline = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english", device=device)
print("Pipeline loaded successfully!")

# Define some text inputs
texts = [
    "This course is absolutely fantastic and incredibly insightful!",
    "I am not sure about the content, it seems a bit generic.",
    "The future of AI is bright, and NLP is leading the charge."
]

print("\nPerforming sentiment analysis:")
for text in texts:
    result = sentiment_pipeline(text)
    label = result[0]['label']
    score = result[0]['score']
    print(f"- Text: '{text}'")
    print(f"  Sentiment: {label} (Score: {score:.4f})")

print("\nEnvironment check complete! You're ready to dive into NLP with PyTorch.")
```
