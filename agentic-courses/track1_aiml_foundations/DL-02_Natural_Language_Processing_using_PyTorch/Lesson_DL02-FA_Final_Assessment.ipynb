```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, random_split
from torchtext.data.utils import get_tokenizer
from torchtext.vocab import build_vocab_from_iterator
import random
from collections import Counter
import math
from sklearn.metrics import f1_score, accuracy_score
import numpy as np

# --- Configuration --- #
NUM_CLASSES = 5 # e.g., Politics, Sports, Technology, Finance, Health
MAX_SEQ_LEN = 32
EMBED_DIM = 128
NUM_HEADS = 4
NUM_ENCODER_LAYERS = 2
BATCH_SIZE = 64
NUM_EPOCHS = 10
LEARNING_RATE = 1e-4
DROPOUT = 0.1

# --- 1. Data Generation (Synthetic) --- #
def generate_synthetic_data(num_samples=2000):
    topics = ['politics', 'sports', 'technology', 'finance', 'health']
    common_words = {
        'politics': ['government', 'election', 'president', 'congress', 'policy', 'vote', 'party'],
        'sports': ['team', 'game', 'win', 'lose', 'championship', 'athlete', 'score'],
        'technology': ['AI', 'software', 'startup', 'innovation', 'data', 'cyber', 'robot'],
        'finance': ['market', 'stock', 'economy', 'invest', 'bank', 'profit', 'trade'],
        'health': ['medical', 'disease', 'hospital', 'research', 'vaccine', 'wellness', 'patient']
    }
    headlines = []
    labels = []

    for _ in range(num_samples):
        num_labels = random.randint(0, NUM_CLASSES) # 0 to NUM_CLASSES labels
        current_labels_idx = random.sample(range(NUM_CLASSES), num_labels)
        current_labels_one_hot = [0] * NUM_CLASSES
        text_parts = []

        if num_labels == 0:
            # If no specific labels, pick some general words
            text_parts.extend(random.sample(['news', 'report', 'update', 'story', 'breaking'], random.randint(1, 3)))
        else:
            for idx in current_labels_idx:
                current_labels_one_hot[idx] = 1
                topic = topics[idx]
                text_parts.extend(random.sample(common_words[topic], random.randint(1, 3)))

        # Add some random filler words to make it more natural
        filler_words = ['the', 'a', 'and', 'in', 'for', 'with', 'new', 'big', 'small', 'global']
        text_parts.extend(random.sample(filler_words, random.randint(0, 2)))
        random.shuffle(text_parts)
        headline = ' '.join(text_parts).strip()

        headlines.append(headline)
        labels.append(current_labels_one_hot)

    return list(zip(headlines, labels))

# --- 2. Tokenization & Vocabulary --- #
tokenizer = get_tokenizer('basic_english')

def yield_tokens(data_iter):
    for text, _ in data_iter:
        yield tokenizer(text)

def build_vocabulary(data_iter):
    vocab = build_vocab_from_iterator(yield_tokens(data_iter), specials=['<unk>', '<pad>'], min_freq=2)
    vocab.set_default_index(vocab['<unk>'])
    return vocab

# --- 3. Dataset & DataLoader --- #
class TextClassificationDataset(Dataset):
    def __init__(self, data, vocab, tokenizer, max_seq_len):
        self.texts = [item[0] for item in data]
        self.labels = [item[1] for item in data]
        self.vocab = vocab
        self.tokenizer = tokenizer
        self.max_seq_len = max_seq_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        text = self.texts[idx]
        label = torch.tensor(self.labels[idx], dtype=torch.float32)

        token_ids = self.vocab(self.tokenizer(text))

        # Pad or truncate to max_seq_len
        if len(token_ids) < self.max_seq_len:
            padding = [self.vocab['<pad>']] * (self.max_seq_len - len(token_ids))
            token_ids.extend(padding)
        else:
            token_ids = token_ids[:self.max_seq_len]

        return torch.tensor(token_ids, dtype=torch.long), label

# --- 4. Transformer Model --- #
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=5000):
        super().__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0) # Add batch dimension
        self.register_buffer('pe', pe)

    def forward(self, x):
        # x is (batch_size, seq_len, embed_dim)
        # self.pe is (1, max_len, embed_dim)
        # We need to slice pe to match the sequence length of x
        return x + self.pe[:, :x.size(1)]

class TransformerClassifier(nn.Module):
    def __init__(self, vocab_size, embed_dim, num_heads, num_encoder_layers, num_classes, max_seq_len, dropout=0.1):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim)
        self.pos_encoder = PositionalEncoding(embed_dim, max_len)

        encoder_layer = nn.TransformerEncoderLayer(d_model=embed_dim, nhead=num_heads, dropout=dropout, batch_first=True)
        self.transformer_encoder = nn.TransformerEncoder(encoder_layer, num_layers=num_encoder_layers)

        # Classification head: takes the output of the TransformerEncoder
        # We'll use the mean of the sequence output for classification
        self.classifier = nn.Sequential(
            nn.Linear(embed_dim, embed_dim // 2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(embed_dim // 2, num_classes)
        )

    def forward(self, text):
        # text: (batch_size, seq_len)

        # 1. Embed text: (batch_size, seq_len, embed_dim)
        embedded = self.embedding(text) * math.sqrt(self.embedding.embedding_dim) # Scale embeddings

        # 2. Add positional encoding: (batch_size, seq_len, embed_dim)
        embedded = self.pos_encoder(embedded)

        # 3. Pass through TransformerEncoder: (batch_size, seq_len, embed_dim)
        # No attention mask needed for simple classification where all tokens can attend to each other
        transformer_output = self.transformer_encoder(embedded)

        # 4. Apply classification head
        # For classification, we can average the sequence output or use a special token (like CLS)
        # Here, we'll average the token embeddings across the sequence dimension
        pooled_output = transformer_output.mean(dim=1) # (batch_size, embed_dim)

        logits = self.classifier(pooled_output) # (batch_size, num_classes)
        return logits

# --- 5. Training Loop --- #
def train_model(model, train_dataloader, val_dataloader, criterion, optimizer, num_epochs, device):
    model.train()
    for epoch in range(num_epochs):
        total_loss = 0
        for batch_idx, (text_batch, label_batch) in enumerate(train_dataloader):
            text_batch, label_batch = text_batch.to(device), label_batch.to(device)

            optimizer.zero_grad()
            logits = model(text_batch)
            loss = criterion(logits, label_batch)
            loss.backward()
            optimizer.step()

            total_loss += loss.item()

        avg_train_loss = total_loss / len(train_dataloader)

        # Validation step
        val_loss, val_accuracy, val_f1_macro = evaluate_model(model, val_dataloader, criterion, device)
        print(f"Epoch {epoch+1}/{num_epochs}, Train Loss: {avg_train_loss:.4f}, Val Loss: {val_loss:.4f}, Val Acc: {val_accuracy:.4f}, Val F1 (macro): {val_f1_macro:.4f}")

# --- 6. Evaluation --- #
def evaluate_model(model, dataloader, criterion, device):
    model.eval()
    total_loss = 0
    all_preds = []
    all_labels = []
    with torch.no_grad():
        for text_batch, label_batch in dataloader:
            text_batch, label_batch = text_batch.to(device), label_batch.to(device)
            logits = model(text_batch)
            loss = criterion(logits, label_batch)
            total_loss += loss.item()

            # Convert logits to predictions for multi-label classification
            # Apply sigmoid and then a threshold (e.g., 0.5)
            preds = (torch.sigmoid(logits) > 0.5).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(label_batch.cpu().numpy())

    avg_loss = total_loss / len(dataloader)
    all_preds = np.array(all_preds)
    all_labels = np.array(all_labels)

    # Calculate accuracy (exact match) and F1-score (macro average)
    accuracy = accuracy_score(all_labels, all_preds)
    f1_macro = f1_score(all_labels, all_preds, average='macro', zero_division=0)

    return avg_loss, accuracy, f1_macro

# --- Main Execution Block --- #
if __name__ == '__main__':
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Using device: {device}")

    # 1. Generate data
    full_data = generate_synthetic_data(num_samples=2000)
    print(f"Generated {len(full_data)} synthetic samples.")

    # Split data into train, validation, test
    train_size = int(0.7 * len(full_data))
    val_size = int(0.15 * len(full_data))
    test_size = len(full_data) - train_size - val_size
    train_data, val_data, test_data = random_split(full_data, [train_size, val_size, test_size])

    # 2. Build vocabulary
    # Need to convert Subset objects to iterables of (text, label) tuples
    train_data_list = [full_data[i] for i in train_data.indices]
    vocab = build_vocabulary(train_data_list)
    vocab_size = len(vocab)
    print(f"Vocabulary size: {vocab_size}")

    # 3. Create datasets and dataloaders
    train_dataset = TextClassificationDataset(train_data_list, vocab, tokenizer, MAX_SEQ_LEN)
    val_dataset = TextClassificationDataset([full_data[i] for i in val_data.indices], vocab, tokenizer, MAX_SEQ_LEN)
    test_dataset = TextClassificationDataset([full_data[i] for i in test_data.indices], vocab, tokenizer, MAX_SEQ_LEN)

    train_dataloader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
    val_dataloader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False)
    test_dataloader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)

    # 4. Initialize model, loss, and optimizer
    model = TransformerClassifier(vocab_size, EMBED_DIM, NUM_HEADS, NUM_ENCODER_LAYERS, NUM_CLASSES, MAX_SEQ_LEN, DROPOUT).to(device)
    criterion = nn.BCEWithLogitsLoss() # Suitable for multi-label classification
    optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE)

    print(f"Model parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad)}")

    # 5. Train the model
    print("\nStarting training...")
    train_model(model, train_dataloader, val_dataloader, criterion, optimizer, NUM_EPOCHS, device)

    # 6. Final evaluation on test set
    print("\nEvaluating on test set...")
    test_loss, test_accuracy, test_f1_macro = evaluate_model(model, test_dataloader, criterion, device)
    print(f"\nTest Loss: {test_loss:.4f}, Test Accuracy: {test_accuracy:.4f}, Test F1 (macro): {test_f1_macro:.4f}")

    print("\nCapstone project complete!")
```
