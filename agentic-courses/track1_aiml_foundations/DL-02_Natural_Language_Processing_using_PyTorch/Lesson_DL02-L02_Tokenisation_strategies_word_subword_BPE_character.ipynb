# Tokenisation Strategies: Word, Subword (BPE), Character

Welcome to DL02-L02, where we delve into the crucial first step of almost any Natural Language Processing (NLP) pipeline: **tokenisation**. Just as a chef needs to chop ingredients before cooking, an AI model needs to break down raw text into manageable, meaningful units before it can 'understand' or process it. These units are called **tokens**.

Imagine you're trying to teach a child to read. You wouldn't just show them a whole book and expect them to grasp it. Instead, you'd start by teaching them letters, then words, and then how words form sentences. Tokenisation is a similar process for computers.

Why is tokenisation so important?

1.  **Computers don't understand raw text**: They operate on numbers. Tokens are the bridge, converting text into numerical representations (embeddings) that models can process.
2.  **Vocabulary Management**: Models have a finite vocabulary. Tokenisation helps manage this vocabulary, especially for handling new or rare words.
3.  **Sequence Length**: Breaking text into smaller units helps control the input sequence length, which is critical for models with fixed input sizes (like many transformer architectures).

Let's explore the three primary tokenisation strategies:

### 1. Word Tokenisation

This is the most intuitive approach: splitting text into individual words. It's like separating a sentence into its constituent words based on spaces and punctuation. For example, the sentence "Hello, world!" would become `['Hello', ',', 'world', '!']`.

**Analogy**: Think of a child learning to read by identifying whole words. They see "cat" and understand it as a single unit.

**Pros**: Simple, human-readable tokens, often aligns with semantic units.
**Cons**: 
*   **Out-Of-Vocabulary (OOV) words**: What if the model encounters a word it has never seen during training (e.g., a new slang term or a proper noun)? It won't have an embedding for it, leading to issues.
*   **Large Vocabulary**: Languages have vast vocabularies, including morphological variations (e.g., "run", "running", "ran"). This leads to very large embedding tables, consuming significant memory and computation.
*   **Language-specific rules**: Punctuation, contractions (e.g., "don't"), and hyphenated words require complex, language-specific rules.

### 2. Character Tokenisation

At the other extreme, character tokenisation breaks text down into individual characters. The sentence "Hello, world!" would become `['H', 'e', 'l', 'l', 'o', ',', ' ', 'w', 'o', 'r', 'l', 'd', '!']`.

**Analogy**: This is like teaching a child to spell out every word letter by letter, even if they know the word as a whole.

**Pros**: 
*   **Handles OOV words perfectly**: Any word can be represented as a sequence of characters, so there are no OOV issues at the character level.
*   **Smallest vocabulary**: Typically just the alphabet, numbers, and common symbols.
*   **Robust to typos**: Minor spelling errors might still allow the model to infer meaning.

**Cons**: 
*   **Long sequences**: A single word becomes multiple tokens, leading to much longer input sequences for the model. This increases computational cost and can hit sequence length limits faster.
*   **Less semantic meaning per token**: Individual characters carry very little semantic information compared to words.

### 3. Subword Tokenisation (e.g., Byte Pair Encoding - BPE)

Subword tokenisation strikes a balance between word and character tokenisation. It breaks words into smaller, frequently occurring units (subwords) rather than whole words or single characters. The most popular algorithm for this is **Byte Pair Encoding (BPE)**, which is widely used in modern transformer models like GPT, BERT, and RoBERTa.

**How BPE works (intuitively)**:
1.  Start with a vocabulary of all individual characters in your text corpus.
2.  Repeatedly merge the most frequent adjacent pair of characters (or character sequences) into a new, single subword token.
3.  Continue this process for a predefined number of merges or until a desired vocabulary size is reached.

For example, if "_ing" and "_ed" are very common suffixes, BPE will learn to treat them as single tokens. A word like "tokenisation" might be broken into `['token', 'is', 'ation']` or `['token', 'isation']` depending on the learned subword units.

**Analogy**: Think of learning common syllables, prefixes, and suffixes. Instead of just 'cat' or 'c-a-t', you learn 'un-', '-ing', 're-', etc. This allows you to construct many words from a smaller set of building blocks.

**Pros**: 
*   **Excellent OOV handling**: Rare words can be broken down into known subwords and characters (e.g., "unfamiliarword" -> `['un', 'familiar', 'word']` or `['un', 'fam', 'iliar', 'word']`).
*   **Manages vocabulary size**: The vocabulary is much smaller than word-level but larger than character-level, offering a good balance.
*   **Reduced sequence length**: Shorter sequences than character tokenisation, improving computational efficiency.
*   **Captures morphological information**: Common prefixes/suffixes often become subword tokens.

**Cons**: 
*   **More complex to implement**: Requires training a tokenizer on a corpus.
*   **Less intuitive tokens**: Subword units might not always align with linguistic units.

In the era of large language models (LLMs) and transformers, subword tokenisation, particularly BPE and its variants (WordPiece, SentencePiece), has become the de-facto standard due to its efficiency and effectiveness in handling diverse linguistic phenomena. Let's see these in action!

## Modern Tools for Tokenisation

In 2026, the `Hugging Face tokenizers` library is the gold standard for efficient, production-ready tokenisation, supporting all major subword algorithms and offering blazing fast performance. For simpler tasks or legacy systems, `NLTK` still provides robust word and sentence tokenizers.


```python
# Ensure you have the necessary libraries installed:
# pip install nltk tokenizers

import nltk
from nltk.tokenize import word_tokenize
from tokenizers import BytePairEncoding, Tokenizer
from tokenizers.models import BPE
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.trainers import BpeTrainer

# Download NLTK data if not already present
try:
    nltk.data.find('tokenizers/punkt')
except nltk.downloader.DownloadError:
    nltk.download('punkt')

# --- Sample Text --- #
text = "AgenticLabs.ng is pioneering AI research and development in 2026, pushing the boundaries of what's possible with advanced models. It's truly revolutionary!"

print(f"Original Text:\n{text}\n")

# --- 1. Word Tokenisation (using NLTK) ---
print("--- Word Tokenisation (NLTK) ---")
word_tokens = word_tokenize(text)
print(f"Tokens: {word_tokens}")
print(f"Number of tokens: {len(word_tokens)}\n")

# --- 2. Character Tokenisation ---
print("--- Character Tokenisation ---")
# Simple character split, including spaces and punctuation
char_tokens = list(text)
print(f"Tokens: {char_tokens[:30]}... (showing first 30)") # Print only a portion for readability
print(f"Number of tokens: {len(char_tokens)}\n")

# --- 3. Subword Tokenisation (BPE using Hugging Face tokenizers) ---
print("--- Subword Tokenisation (Hugging Face BPE) ---")

# 3.1. Train a simple BPE tokenizer
# For real-world applications, you'd train on a much larger corpus.
# Here, we'll use a small, illustrative corpus.
corpus = [
    "AgenticLabs.ng is pioneering AI research and development in 2026, pushing the boundaries of what's possible with advanced models.",
    "It's truly revolutionary! AgenticLabs is leading the charge.",
    "Deep learning engineers are excited about new NLP advancements.",
    "Tokenisation is a fundamental step in NLP pipelines."
]

# Initialize a BPE tokenizer model
# The `unk_token` is for unknown tokens, `suffix` helps identify subwords
bpe_model = BPE(unk_token="[UNK]")
tokenizer = Tokenizer(bpe_model)

# Set a pre-tokenizer (how to split initial words before BPE merges)
# Whitespace pre-tokenizer is common
tokenizer.pre_tokenizer = Whitespace()

# Define the trainer for BPE
trainer = BpeTrainer(
    vocab_size=100, # Small vocab size for demonstration; typically much larger (e.g., 30k-50k)
    min_frequency=2, # Minimum frequency for a token to be added to the vocabulary
    special_tokens=["[UNK]", "[CLS]", "[SEP]", "[PAD]", "[MASK]"] # Common special tokens for transformer models
)

# Train the tokenizer
print("Training BPE tokenizer...")
tokenizer.train_from_iterator(corpus, trainer=trainer)
print("BPE tokenizer trained.")

# Get the vocabulary to see learned subwords
print(f"BPE Vocabulary size: {tokenizer.get_vocab_size()}")
# print(f"BPE Vocabulary: {tokenizer.get_vocab()}\n") # Uncomment to see the full vocab

# 3.2. Use the trained BPE tokenizer on our sample text
encoded_text = tokenizer.encode(text)

print(f"Tokens: {encoded_text.tokens}")
print(f"Number of tokens: {len(encoded_text.tokens)}\n")

# Example of how a new, unseen word might be handled by BPE
new_word_text = "AgenticLabs.ng also explores QuantumAI."
encoded_new_word = tokenizer.encode(new_word_text)
print(f"New word 'QuantumAI' tokenised by BPE: {encoded_new_word.tokens}")
print(f"Number of tokens for new word: {len(encoded_new_word.tokens)}")
```

### Interpreting the Output and Trade-offs

Let's break down what we observed from the code execution:

1.  **Word Tokenisation (NLTK)**:
    *   **Output**: `['AgenticLabs.ng', 'is', 'pioneering', 'AI', 'research', 'and', 'development', 'in', '2026', ',', 'pushing', 'the', 'boundaries', 'of', 'what', '\'s', 'possible', 'with', 'advanced', 'models', '.', 'It', '\'s', 'truly', 'revolutionary', '!']`
    *   **Number of tokens**: 26
    *   **Interpretation**: NLTK's `word_tokenize` is quite sophisticated, handling punctuation and contractions (`what's`, `It's`) intelligently. Each word and significant punctuation mark becomes a distinct token. This is generally easy for humans to read and understand.
    *   **Trade-offs**: If `AgenticLabs.ng` or `pioneering` were not in the training data, they would be OOV. The vocabulary size for a large corpus would be very high.

2.  **Character Tokenisation**:
    *   **Output**: `['A', 'g', 'e', 'n', 't', 'i', 'c', 'L', 'a', 'b', 's', '.', 'n', 'g', ' ', 'i', 's', ' ', 'p', 'i', 'o', 'n', 'e', 'e', 'r', 'i', 'n', 'g', ' ', 'A', ...] `
    *   **Number of tokens**: 144
    *   **Interpretation**: Every single character, including spaces and punctuation, is a token. This results in a significantly larger number of tokens for the same input text compared to word tokenisation.
    *   **Trade-offs**: While it perfectly handles OOV words (as any word is just a sequence of characters), the resulting sequences are very long. This increases computational load for models and can quickly exceed typical maximum sequence lengths (e.g., 512 or 1024 tokens for transformers). Each token carries minimal semantic information.

3.  **Subword Tokenisation (Hugging Face BPE)**:
    *   **Output**: `['Agentic', 'Labs', '.', 'ng', 'Ġis', 'Ġpioneering', 'ĠAI', 'Ġresearch', 'Ġand', 'Ġdevelopment', 'Ġin', 'Ġ2026', ',', 'Ġpushing', 'Ġthe', 'Ġboundaries', 'Ġof', 'Ġwhat', '\'s', 'Ġpossible', 'Ġwith', 'Ġadvanced', 'Ġmodels', '.', 'ĠIt', '\'s', 'Ġtruly', 'Ġrevolutionary', '!']`
    *   **Number of tokens**: 29
    *   **Interpretation**: Notice how BPE breaks words into meaningful subword units. `AgenticLabs.ng` is split into `['Agentic', 'Labs', '.', 'ng']`. Common words like `is`, `pioneering`, `research` are kept as single tokens (often prefixed with `Ġ` to indicate a word boundary, a common BPE convention). The number of tokens is closer to word tokenisation but slightly higher in this specific example due to the splitting of `AgenticLabs.ng` and `what's`/`It's` into multiple subwords/characters. The `Ġ` prefix is a common way for BPE to denote that the token starts a new word, which helps reconstruct the original text without ambiguity.
    *   **Handling OOV**: When we tokenized `QuantumAI`, it was broken down into `['Qu', 'ant', 'um', 'AI']`. This demonstrates BPE's ability to represent unseen words using known subword units, effectively mitigating the OOV problem without resorting to character-level granularity.
    *   **Trade-offs**: BPE offers a fantastic balance. It keeps sequence lengths manageable (much shorter than character-level), handles OOV words gracefully by breaking them into known subwords, and maintains a reasonable vocabulary size. This is why it's the dominant strategy for modern LLMs.
