### Answers and Explanations

**Question 1 (Multiple Choice): Self-Attention Mechanism**

**Correct Answer: B)** It allows the model to weigh the importance of different words in the input sequence when encoding a specific word, regardless of their distance.

**Explanation:**
Self-attention is the core innovation of the Transformer. Unlike RNNs, which process tokens sequentially and struggle with long-range dependencies due to vanishing/exploding gradients, self-attention computes a weighted sum of all input tokens for each output token. This allows the model to directly capture relationships between any two words in the sequence, no matter how far apart they are, making it highly effective for understanding context across long texts. Options A and C describe characteristics of RNNs and CNNs, respectively, while D is incorrect as Transformers can be computationally intensive and parameter-heavy.

---

**Explanation:**
Positional Encoding is crucial because the Transformer architecture, particularly the self-attention mechanism, processes all tokens in a sequence simultaneously without any inherent notion of order or position. If positional encoding were absent, the model would treat a sequence like a bag of words, losing all information about word order, which is vital for understanding grammar, syntax, and meaning in natural language.
