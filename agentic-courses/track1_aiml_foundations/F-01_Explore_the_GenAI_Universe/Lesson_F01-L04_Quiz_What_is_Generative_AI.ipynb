### Answer Key & Explanations

**Question 1: Primary Goal**
*   **Correct Answer: C) To create novel, original content that didn't exist before.**
*   **Explanation:** The defining characteristic of Generative AI is its ability to *generate* new, unique content. Options A, B, and D describe tasks typically performed by discriminative or predictive AI models, which analyze existing data rather than creating new data.

**Question 2: Not an Application**
*   **Correct Answer: C) Diagnosing a medical condition from patient symptoms.**
*   **Explanation:** While AI is heavily used in healthcare, diagnosing medical conditions is primarily a task for *discriminative* AI (e.g., classification models) that analyze symptoms, test results, and patient history to predict a diagnosis. Generative AI excels at creating new content like music, images, or stories. While Generative AI might assist in generating synthetic patient data for training or creating personalized treatment plans, direct diagnosis from symptoms is not its primary generative function.
