## 🤔 Section 2: What is a Large Language Model (LLM)?

Think of an LLM like **autocomplete on steroids**. It's a giant brain trained on a massive amount of text and code from the internet. Its main job is to predict the next most logical word in a sequence.

When you give it a prompt (a question or instruction), it starts generating a response, one word at a time, based on the patterns it learned during its training. It's not "thinking" or "understanding" like a human, but it's incredibly good at mimicking human language and reasoning.

Let's see it in action!


```python
# 1. Select the model we want to use
# 'gemini-1.5-flash' is a fast and versatile model
model = genai.GenerativeModel('gemini-1.5-flash')

# 2. Create our first prompt
prompt = "Tell me a fun fact about Nigeria."

# 3. Send the prompt to the model and get a response
response = model.generate_content(prompt)

# 4. Print the response!
print(response.text)
```

## 📝 Section 3: The Prompting Blueprint

Getting the AI to do what you want is a skill called **prompt engineering**. A vague prompt leads to a vague answer. A specific, well-structured prompt leads to a specific, high-quality answer.

To get reliable results, use the **4-Step Prompting Blueprint**:

-   **ROLE:** Who should the AI act as? (e.g., "You are a helpful assistant," "You are a professional copywriter.")
-   **TASK:** What is the specific action you want the AI to perform? (e.g., "Summarize this text," "Write an email.")
-   **CONTEXT:** What background information, constraints, or details does the AI need to know? (e.g., "The audience is beginners," "The tone should be friendly.")
-   **FORMAT:** How should the output be structured? (e.g., "Use bullet points," "Provide the answer in a JSON format," "Write a single paragraph.")

Let's compare a weak prompt to a strong one.


### Weak Prompt (Vague Instructions)


```python
# This prompt is too simple and lacks detail.
weak_prompt = "Explain AI."

print(f"--- WEAK PROMPT ---\n{weak_prompt}\n")

response = model.generate_content(weak_prompt)
print("--- AI RESPONSE ---")
print(response.text)
```

### Strong Prompt (Using the Blueprint)


```python
# This prompt uses our 4-step blueprint for a much better result.

# 1. Define the components of our prompt
role = "You are a science communicator who is an expert at making complex topics simple."
task = "Explain the concept of Artificial Intelligence (AI)."
context = "Your explanation is for a 10-year-old child who is curious about technology. Avoid jargon and use a simple analogy."
format = "Structure your answer in two short paragraphs."

# 2. Combine them into a single, strong prompt
strong_prompt = f"ROLE: {role}\nTASK: {task}\nCONTEXT: {context}\nFORMAT: {format}"

print(f"--- STRONG PROMPT ---\n{strong_prompt}\n")

response = model.generate_content(strong_prompt)
print("--- AI RESPONSE ---")
print(response.text)
```

## 🌡️ Section 4: Controlling Creativity with Temperature

The `temperature` is a setting that controls how "creative" or "random" the AI's responses are. It's a number, usually between 0.0 and 2.0.

-   **Low Temperature (e.g., 0.1):** The model becomes more focused and deterministic. It will almost always choose the most likely next word. This is great for factual, logical tasks like writing code, summarizing facts, or performing math.

-   **High Temperature (e.g., 1.5):** The model becomes more creative and surprising. It might choose less common words, leading to more diverse and imaginative outputs. This is perfect for brainstorming, writing stories, or creating marketing copy.


### Temperature Demo: Low (Factual & Predictable)


```python
# We configure the model with a low temperature
low_temp_model = genai.GenerativeModel(
    'gemini-1.5-flash',
    generation_config={'temperature': 0.1}
)

prompt = "What is the capital of France?"

print(f"--- PROMPT ---\n{prompt}\n")

response = low_temp_model.generate_content(prompt)

print("--- AI RESPONSE (Temp: 0.1) ---")
print(response.text)
print("\n(Run this cell again - the answer will likely be identical.)")
```

### Temperature Demo: High (Creative & Unpredictable)


```python
# Now we configure the model with a high temperature
high_temp_model = genai.GenerativeModel(
    'gemini-1.5-flash',
    generation_config={'temperature': 1.5}
)

prompt = "Write a one-sentence slogan for a new brand of coffee."

print(f"--- PROMPT ---\n{prompt}\n")

response = high_temp_model.generate_content(prompt)

print("--- AI RESPONSE (Temp: 1.5) ---")
print(response.text)
print("\n(Run this cell again - you'll get a different creative slogan each time!)")
```

## 🚀 Section 5: Your Mini-Capstone Project

It's time to put everything together! Your task is to create a specialized AI assistant.

Instead of putting the **ROLE** in every prompt, we can give the model a `system_instruction`. This sets its core personality and behavior for the entire conversation.

**Your Goal:** Configure a model to act as a **Research Assistant** that always follows a specific format.


```python
# 1. Define the system instruction for our Research Assistant
system_instruction = """
You are a world-class research assistant. Your purpose is to find and summarize information on any topic given to you. 

When a user asks a question, you MUST follow this format exactly:
1.  **Topic:** Restate the user's topic.
2.  **Summary:** Provide a concise, three-bullet-point summary.
3.  **Keywords:** List 5 relevant keywords for further research.

You must not deviate from this format. Do not add any conversational text before or after your response.
"""

# 2. Create the model with the system instruction
research_assistant_model = genai.GenerativeModel(
    'gemini-1.5-flash',
    system_instruction=system_instruction
)

# 3. Ask your new assistant a question!
user_question = "What are the benefits of learning a new language?"

print(f"--- USER QUESTION ---\n{user_question}\n")

response = research_assistant_model.generate_content(user_question)

print("--- RESEARCH ASSISTANT RESPONSE ---")
print(response.text)

print("\n\n✅ Congratulations! You've configured your first specialized AI assistant.")
```
